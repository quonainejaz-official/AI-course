---\n## 4. Concurrency vs Parallelism\n\n**Concurrency:** Ek time but multiple tasks but progress (can work on single core via task switching).\n**Parallelism:** Ek time but multiple tasks literally ek with chal rahe are (multiple cores required).\n\n**Python Options:**\n- Threading: Concurrency for I/O-bound (GIL limits CPU-bound)\n- Multiprocessing: True parallelism for CPU-bound (separate processes)\n- Asyncio: Cooperative concurrency for I/O-bound (single thread, less overhead)\n

```python
# --- Concurrency vs Parallelism ---\nimport time\nimport threading\n\ndef cpu_task(n):\n    return sum(i*i for i in range(n))\n\ndef io_task(delay):\n    time.sleep(delay)\n    return delay\n\nprint("--- CPU-bound: Threading is NOT faster (GIL) ---")\nn = 5000000\nstart = time.time()\ncpu_task(n)\ncpu_task(n)\nprint(f"Sequential: {time.time()-start:.2f}s")\n\nprint("\n--- I/O-bound: Threading IS faster ---")\ndelays = [0.5, 1, 1.5, 2]\nstart = time.time()\nthreads = [threading.Thread(target=io_task, args=(d,)) for d in delays]\nfor t in threads: t.start()\nfor t in threads: t.join()\nprint(f"Threading: {time.time()-start:.1f}s (vs 5s sequential)")\n\nprint("\n--- Best Uses ---")\nprint("CPU-bound  -> Multiprocessing (Pool.map)")\nprint("I/O-bound  -> Asyncio or Threading")\nprint("Mixed      -> Asyncio + run_in_executor")\n
```

### problem 1: Median of Two Sorted Arrays (LeetCode 4)\n\n**problem:** Do sorted arrays of median find do in O(log(min(m,n))) time.\n\n**Approach:** withoutry search on smaller array. Partition dono arrays in do taake left part in saare elements right part from chhote hon.\n

```python
# Median of Two Sorted Arrays\ndef find_median_sorted_arrays(nums1, nums2):\n    A, B = nums1, nums2\n    if len(A) > len(B): A, B = B, A\n    m, n = len(A), len(B)\n    total = m + n\n    half = (total + 1) // 2\n    l, r = 0, m\n    \n    while l <= r:\n        i = (l + r) // 2\n        j = half - i\n        \n        A_left = A[i-1] if i > 0 else float("-inf")\n        A_right = A[i] if i < m else float("inf")\n        B_left = B[j-1] if j > 0 else float("-inf")\n        B_right = B[j] if j < n else float("inf")\n        \n        if A_left <= B_right and B_left <= A_right:\n            if total % 2 == 0:\n                return (max(A_left, B_left) + min(A_right, B_right)) / 2\n            return max(A_left, B_left)\n        elif A_left > B_right:\n            r = i - 1\n        else:\n            l = i + 1\n    return 0.0\n\nprint("--- Median of Two Sorted Arrays ---")\nprint(f"[1,3], [2] -> {find_median_sorted_arrays([1,3], [2])}")\nprint(f"[1,2], [3,4] -> {find_median_sorted_arrays([1,2], [3,4])}")\nprint(f"[0,0], [0,0] -> {find_median_sorted_arrays([0,0], [0,0])}")\n
```

### problem 2: Regular Expression Matching (LeetCode 10)\n\n**problem:** String s and pattern p. '.' matches any single char, '*' matches 0+ occurrences of preceding char.\n\n**Approach:** Dynamic Programming. dp[i][j] = s[i:] matches p[j:]\n

```python
# Regular Expression Matching\ndef is_match(s, p):\n    m, n = len(s), len(p)\n    dp = [[False] * (n + 1) for _ in range(m + 1)]\n    dp[m][n] = True  # Empty matches empty\n    \n    for j in range(n - 1, -1, -1):\n        if j + 1 < n and p[j + 1] == '*':\n            dp[m][j] = dp[m][j + 2]\n    \n    for i in range(m - 1, -1, -1):\n        for j in range(n - 1, -1, -1):\n            match = s[i] == p[j] or p[j] == '.'\n            \n            if j + 1 < n and p[j + 1] == '*':\n                dp[i][j] = dp[i][j + 2] or (match and dp[i + 1][j])\n            else:\n                dp[i][j] = match and dp[i + 1][j + 1]\n    \n    return dp[0][0]\n\nprint("--- Regular Expression Matching ---")\nprint(f"is_match('aa', 'a'): {is_match('aa', 'a')}")\nprint(f"is_match('aa', 'a*'): {is_match('aa', 'a*')}")\nprint(f"is_match('ab', '.*'): {is_match('ab', '.*')}")\nprint(f"is_match('aab', 'c*a*b'): {is_match('aab', 'c*a*b')}")\n
```

### problem 3: Merge k Sorted Lists (LeetCode 23)\n\n**problem:** k sorted linked lists then merge to do is ek sorted list in.\n\n**Approach:** Min-heap (priority queue). Har list of first node then heap in daalo. Smallest nikaalo, uske next then heap in daalo. O(n log k) time, O(k) space.\n

```python
# Merge k Sorted Lists\nimport heapq\n\nclass ListNode:\n    def __init__(self, val=0, next=None):\n        self.val = val\n        self.next = next\n\ndef merge_k_lists(lists):\n    dummy = ListNode()\n    curr = dummy\n    heap = []\n    \n    for i, lst in enumerate(lists):\n        if lst:\n            heapq.heappush(heap, (lst.val, i, lst))\n    \n    while heap:\n        val, i, node = heapq.heappop(heap)\n        curr.next = node\n        curr = curr.next\n        if node.next:\n            heapq.heappush(heap, (node.next.val, i, node.next))\n    \n    return dummy.next\n\ndef to_list(head):\n    res = []\n    while head: res.append(head.val); head = head.next\n    return res\n\ndef from_list(arr):\n    if not arr: return None\n    h = ListNode(arr[0]); c = h\n    for v in arr[1:]: c.next = ListNode(v); c = c.next\n    return h\n\nprint("--- Merge k Sorted Lists ---")\nl1 = from_list([1, 4, 5])\nl2 = from_list([1, 3, 4])\nl3 = from_list([2, 6])\nmerged = merge_k_lists([l1, l2, l3])\nprint(f"Merged: {to_list(merged)}")\n
```

### problem 4: First Missing Positive (LeetCode 41)\n\n**problem:** Unsorted array in sabse chhota missing positive integer find do. O(n) time, O(1) space in.\n\n**Approach:** Cyclic sort. Har positive number then uski sahi position (index = value-1) but rakhdo. Phir check do konse index but sahi number not is.\n

```python
# First Missing Positive\ndef first_missing_positive(nums):\n    n = len(nums)\n    \n    # Cyclic sort: place each positive number at correct index\n    for i in range(n):\n        while 1 <= nums[i] <= n and nums[nums[i] - 1] != nums[i]:\n            correct_pos = nums[i] - 1\n            nums[i], nums[correct_pos] = nums[correct_pos], nums[i]\n    \n    # Find first index where number is wrong\n    for i in range(n):\n        if nums[i] != i + 1:\n            return i + 1\n    \n    return n + 1\n\nprint("--- First Missing Positive ---")\nprint(f"[1,2,0] -> {first_missing_positive([1,2,0])}")\nprint(f"[3,4,-1,1] -> {first_missing_positive([3,4,-1,1])}")\nprint(f"[7,8,9,11,12] -> {first_missing_positive([7,8,9,11,12])}")\n
```

### problem 5: Trapping Rain Water (LeetCode 42)\n\n**problem:** Height array diya gaya is jo elevation map represent does is. rainy season of paani kitna trap is can is?\n\n**Approach:** Two pointers. Left max and right max track do. Chhoti side from paani count do. O(n) time, O(1) space.\n

```python
# Trapping Rain Water\ndef trap(height):\n    if not height: return 0\n    \n    l, r = 0, len(height) - 1\n    left_max, right_max = height[l], height[r]\n    water = 0\n    \n    while l < r:\n        if left_max < right_max:\n            l += 1\n            left_max = max(left_max, height[l])\n            water += left_max - height[l]\n        else:\n            r -= 1\n            right_max = max(right_max, height[r])\n            water += right_max - height[r]\n    \n    return water\n\nprint("--- Trapping Rain Water ---")\nh1 = [0,1,0,2,1,0,1,3,2,1,2,1]\nprint(f"Example 1: {trap(h1)}")\nh2 = [4,2,0,3,2,5]\nprint(f"Example 2: {trap(h2)}")\n
```

### problem 6: N-Queens (LeetCode 51)\n\n**problem:** NxN chessboard but N queens rakho taake koi do queens ek doosre then attack no do. Saare unique solutions return do.\n\n**Approach:** Backtracking. Har row in ek queen rakho. Columns, diagonals track do. if safe position is then place do and next row but jao.\n

```python
# N-Queens\ndef solve_n_queens(n):\n    cols, diag1, diag2 = set(), set(), set()\n    result = []\n    board = ["." * n for _ in range(n)]\n    \n    def backtrack(row):\n        if row == n:\n            result.append(board[:])\n            return\n        \n        for col in range(n):\n            d1, d2 = row - col, row + col\n            if col in cols or d1 in diag1 or d2 in diag2:\n                continue\n            \n            cols.add(col); diag1.add(d1); diag2.add(d2)\n            board[row] = "." * col + "Q" + "." * (n - col - 1)\n            backtrack(row + 1)\n            cols.remove(col); diag1.remove(d1); diag2.remove(d2)\n            board[row] = "." * n\n    \n    backtrack(0)\n    return result\n\nprint("--- N-Queens (n=4) ---")\nsolutions = solve_n_queens(4)\nprint(f"Total solutions for n=4: {len(solutions)}")\nfor sol in solutions:\n    for row in sol:\n        print(row)\n    print()\n
```

### problem 7: Word Laddelay (LeetCode 127)\n\n**problem:** beginWord from endWord tak shortest transformation sequence find do. Har step in ek letter change kar can are and wordList in hona should.\n\n**Approach:** BFS (shortest path in unweighted graph). Har word from connected words find do (1 letter change) and BFS do.\n

```python
# Word Laddelay\nfrom collections import deque\n\ndef laddelay_length(begin_word, end_word, word_list):\n    word_set = set(word_list)\n    if end_word not in word_set:\n        return 0\n    \n    q = deque([(begin_word, 1)])\n    visited = {begin_word}\n    \n    while q:\n        word, length = q.popleft()\n        if word == end_word:\n            return length\n        \n        # Try changing each character\n        for i in range(len(word)):\n            for c in "abcdefghijklmnopqrstuvwxyz":\n                if c == word[i]:\n                    continue\n                new_word = word[:i] + c + word[i+1:]\n                if new_word in word_set and new_word not in visited:\n                    visited.add(new_word)\n                    q.append((new_word, length + 1))\n    \n    return 0\n\nprint("--- Word Laddelay ---")\nwordList = ["hot", "dot", "dog", "lot", "log", "cog"]\nprint(f"hit -> cog: {laddelay_length('hit', 'cog', wordList)}")\nprint(f"hit -> cog (no cog): {laddelay_length('hit', 'cog', ['hot','dot','dog'])}")\n
```
