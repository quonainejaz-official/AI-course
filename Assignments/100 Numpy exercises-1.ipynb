````python
# Question 47: Given two arrays X and Y, construct the Cauchy matrix C (Cij = 1/(xi - yj))
# The Cauchy matrix is a famous matrix in linear algebra where each element is
# the reciprocal of the difference between x_i and y_j.
X = np.arange(5)
Y = X + 0.5
# Using broadcasting: 1.0 / (X[:, np.newaxis] - Y[np.newaxis, :])
# X[:, np.newaxis] makes X a column vector, Y[np.newaxis, :] makes Y a row vector.
# Their subtraction produces a 5x5 matrix via broadcasting.
C = 1.0 / (X[:, np.newaxis] - Y)
print(C)
````

#### 52. Consider a random vector with shape (100,2) representing coordinates, find point by point distances (★★☆)

````python
# Question 52: Consider a random vector with shape (100,2) representing coordinates,
# find point by point distances
# We compute pairwise Euclidean distances between all 100 points.
# Using broadcasting: differences[None,:,:] - differences[:,None,:] gives all pairwise diffs.
coordinates = np.random.random((100, 2))
# Compute differences between all pairs of points
diff = coordinates[:, np.newaxis, :] - coordinates[np.newaxis, :, :]
# Euclidean distance is sqrt of sum of squared differences
distances = np.sqrt(np.sum(diff**2, axis=2))
print("Shape of distance matrix:", distances.shape)  # (100, 100)
````

#### 78. Consider 2 sets of points P0,P1 describing lines (2d) and a point p, how to compute distance from p to each line i  (P0\[i\],P1\[i\])? (★★★)

````python
# Question 78: Consider 2 sets of points P0,P1 describing lines (2d) and a point p,
# how to compute distance from p to each line i (P0[i],P1[i])?
# The distance from a point to a line defined by two points uses the formula:
# d = |cross(P1-P0, P0-p)| / |P1-P0|
P0 = np.array([[0, 0], [1, 1], [2, 2]])
P1 = np.array([[3, 0], [3, 1], [0, 3]])
p = np.array([1, 1])

# Vector from P0 to P1
d_line = P1 - P0
# Vector from P0 to p
d_point = P0 - p
# Cross product (2D cross product gives scalar)
cross = d_line[:, 0] * d_point[:, 1] - d_line[:, 1] * d_point[:, 0]
# Distance = |cross| / |d_line|
distances = np.abs(cross) / np.sqrt(d_line[:, 0]**2 + d_line[:, 1]**2)
print("Distances from point to each line:", distances)
````

#### 80. Consider an arbitrary array, write a function that extract a subpart with a fixed shape and centered on a given element (pad with a `fill` value when necessary) (★★★)

````python
# Question 80: Consider an arbitrary array, write a function that extracts a subpart with a
# fixed shape and centered on a given element (pad with a `fill` value when necessary)
# We calculate the slicing bounds and use np.pad for edges that go out of bounds.

def extract_subpart(array, center, shape, fill=0):
    """Extract a sub-array of given shape centered on 'center' index, pad with fill if needed."""
    half = np.array(shape) // 2
    starts = np.array(center) - half
    ends = starts + np.array(shape)

    # Calculate how much padding is needed on each side
    pad_before = np.maximum(-starts, 0)
    pad_after = np.maximum(ends - np.array(array.shape), 0)

    # Calculate valid slicing ranges
    src_starts = np.maximum(starts, 0)
    src_ends = np.minimum(ends, np.array(array.shape))

    # Extract the valid portion
    sliced = array[tuple(slice(s, e) for s, e in zip(src_starts, src_ends))]

    # Pad if needed
    if pad_before.any() or pad_after.any():
        sliced = np.pad(sliced, tuple(zip(pad_before, pad_after)),
                        mode='constant', constant_values=fill)
    return sliced

z = np.arange(25).reshape(5, 5)
print("Full array:")
print(z)
print("Subpart centered at (2,2) with shape (3,3):")
print(extract_subpart(z, (2, 2), (3, 3)))
print("Subpart centered at (0,0) with shape (3,3) (needs padding):")
print(extract_subpart(z, (0, 0), (3, 3), fill=-1))
````

#### 88. How to implement the Game of Life using numpy arrays? (★★★)

````python
# Question 88: How to implement the Game of Life using numpy arrays?
# Conway's Game of Life rules:
# 1. A live cell with <2 or >3 live neighbors dies (underpopulation/overpopulation)
# 2. A live cell with 2-3 live neighbors survives
# 3. A dead cell with exactly 3 live neighbors becomes alive (reproduction)
# We count neighbors using convolution with a 3x3 kernel (excluding center).

def step(Z):
    # Count number of neighbors using slicing (equivalent to convolution)
    N = np.zeros(Z.shape, dtype=int)
    N += Z[np.ix_([1, -1], [1, -1])]  # diagonals
    N += Z[1:, 1:-1] + Z[:-1, 1:-1]  # vertical neighbors
    N += Z[1:-1, 1:] + Z[1:-1, :-1]  # horizontal neighbors
    # Apply rules: alive cell stays alive with 2 or 3 neighbors,
    # dead cell comes alive with exactly 3 neighbors
    birth = (Z == 0) & (N == 3)
    survive = (Z == 1) & ((N == 2) | (N == 3))
    Z[:] = 0
    Z[birth | survive] = 1
    return Z

# Initialize a random board
Z = np.random.randint(0, 2, (10, 10))
print("Initial state:")
print(Z)
Z = step(Z)
print("After one step:")
print(Z)
````

#### 90. Given an arbitrary number of vectors, build the cartesian product (every combinations of every item) (★★★)

````python
# Question 90: Given an arbitrary number of vectors, build the cartesian product
# (every combination of every item)
# np.meshgrid creates all combinations, then reshape them into pairs.
def cartesian_product(*arrays):
    """Compute the cartesian product of multiple arrays."""
    grids = np.meshgrid(*arrays, indexing='ij')
    return np.stack(grids, axis=-1).reshape(-1, len(arrays))

A = np.array([1, 2, 3])
B = np.array([4, 5])
C = np.array([6, 7, 8])
result = cartesian_product(A, B, C)
print("Cartesian product of A, B, C:")
print(result)
````

#### 98. Considering a path described by two vectors (X,Y), how to sample it using equidistant samples (★★★)?

````python
# Question 98: Considering a path described by two vectors (X,Y), how to sample it using equidistant samples?
# We compute the cumulative arc length along the path, then interpolate at
# evenly spaced intervals using np.interp.
X = np.array([0, 1, 2, 3, 5, 7])   # x-coordinates
Y = np.array([0, 3, 1, 2, 4, 1])   # y-coordinates

# Compute cumulative distance along the path
dX = np.diff(X)
dY = np.diff(Y)
distances = np.sqrt(dX**2 + dY**2)
cumulative_dist = np.concatenate(([0], np.cumsum(distances)))

# Create equidistant sample points
n_samples = 10
equidistant_dist = np.linspace(0, cumulative_dist[-1], n_samples)

# Interpolate X and Y at equidistant distances
X_interp = np.interp(equidistant_dist, cumulative_dist, X)
Y_interp = np.interp(equidistant_dist, cumulative_dist, Y)

print("Equidistant X:", np.round(X_interp, 3))
print("Equidistant Y:", np.round(Y_interp, 3))
````

#### 100. Compute bootstrapped 95% confidence intervals for the mean of a 1D array X (i.e., resample the elements of an array with replacement N times, compute the mean of each sample, and then compute percentiles over the means). (★★★)

````python
# Question 100: Compute bootstrapped 95% confidence intervals for the mean of a 1D array X
# Bootstrap: resample with replacement N times, compute mean of each sample,
# then find the 2.5th and 97.5th percentiles for the 95% confidence interval.
X = np.random.normal(5, 2, 100)  # 100 samples from normal distribution
N = 10000  # Number of bootstrap resamples

# Generate N bootstrap samples, each of the same size as X
# np.random.choice with replacement=True creates bootstrap samples
bootstrap_means = np.array([np.mean(np.random.choice(X, size=len(X), replace=True))
                            for _ in range(N)])

# Compute 95% confidence interval using percentiles
lower = np.percentile(bootstrap_means, 2.5)
upper = np.percentile(bootstrap_means, 97.5)

print(f"Sample mean: {np.mean(X):.4f}")
print(f"95% Confidence Interval: [{lower:.4f}, {upper:.4f}]")
````
