## 1. What is OOP? <a id="what-is-oop"></a>

Object-Oriented Programming (OOP) is a programming paradigm based on the concept of **objects** that contain data and code.

### Key Concepts:
- **Object**: An instance of a class that bundles data (attributes) and behavior (methods)
- **Class**: A blueprint or template for creating objects

### Why OOP?
- **Modularity**: Code is organized into reusable, independent objects
- **Reusability**: Classes can be reused to create multiple objects
- **Maintainability**: Changes to one object don't affect others
- **Scalability**: Easier to manage complex applications
- **Abstraction**: Hide complex implementation details
- **Encapsulation**: Bundle data and methods together

### Real-world Analogy:
Think of a **car** blueprint (class) vs an actual **car** (object). The blueprint defines the structure, but each car built from it is a separate instance with its own properties.

## 2. Procedural vs OOP <a id="procedural-vs-oop"></a>

| Feature | Procedural Programming | Object-Oriented Programming |
|---------|----------------------|----------------------------|
| Structure | Functions and procedures | Classes and objects |
| Data Access | Global data accessible everywhere | Data encapsulated within objects |
| Data Security | Less secure, data is exposed | More secure, data is protected |
| Code Reusability | Limited | High (inheritance, polymorphism) |
| Complexity | Harder to manage as code grows | Easier to manage and scale |
| Focus | On functions | On data and its behavior |

### Example Comparison:

````python
# Procedural approach
def create_student(name, age, grade):
    return {'name': name, 'age': age, 'grade': grade}

def display_student(student):
    print(f"Name: {student['name']}, Age: {student['age']}, Grade: {student['grade']}")

student1 = create_student('Alice', 20, 'A')
display_student(student1)
````

````python
# Object-Oriented approach
class Student:
    def __init__(self, name, age, grade):
        self.name = name
        self.age = age
        self.grade = grade

    def display(self):
        print(f"Name: {self.name}, Age: {self.age}, Grade: {self.grade}")

student1 = Student('Alice', 20, 'A')
student1.display()
````

## 3. Class and Object <a id="class-and-object"></a>

A **class** is a blueprint for creating objects. An **object** is an instance of a class.

```python
class ClassName:
    # class body
    pass
```

### Key Points:
- Classes are defined using the `class` keyword
- Class names should use **PascalCase** convention
- The class body contains attributes (data) and methods (functions)
- Objects are created by calling the class name as a function

````python
class Dog:
    """A simple Dog class."""
    
    # Class variable (shared by all instances)
    species = "Canis lupus"
    
    def __init__(self, name, age):
        self.name = name
        self.age = age

# Creating objects (instances)
dog1 = Dog('Buddy', 3)
dog2 = Dog('Max', 5)

print(f"Dog 1: {dog1.name}, {dog1.age} years old")
print(f"Dog 2: {dog2.name}, {dog2.age} years old")
print(f"Species: {Dog.species}")
````

## 4. The `__init__` Method (Constructor) <a id="init-method"></a>

The `__init__` method is a **special method** (also called a **dunder method**) that Python calls automatically when a new object is created.

### Key Points:
- It's also known as the **constructor**
- It initializes the object's attributes
- It runs automatically when you create a new instance
- It must have at least `self` as its first parameter
- You can pass additional parameters to customize object creation

### Syntax:
```python
def __init__(self, param1, param2, ...):
    self.attribute1 = param1
    self.attribute2 = param2
```

````python
class Car:
    def __init__(self, make, model, year=2024):
        """Initialize a new Car instance."""
        self.make = make
        self.model = model
        self.year = year
        self.mileage = 0  # Default value

# Create instances with different parameters
car1 = Car('Tesla', 'Model S', 2023)
car2 = Car('Toyota', 'Camry')  # Uses default year

print(f"Car 1: {car1.year} {car1.make} {car1.model}")
print(f"Car 2: {car2.year} {car2.make} {car2.model}")
````

## 5. Instance vs Class Variables <a id="instance-vs-class-variables"></a>

### Instance Variables
- Defined inside `__init__` using `self`
- Each object has its **own copy**
- Changes to one object don't affect others

### Class Variables
- Defined directly inside the class (outside methods)
- Shared by **all instances** of the class
- Changes affect all objects

### Example:

````python
class BankAccount:
    # Class variable - shared by all instances
    interest_rate = 0.02
    bank_name = "Python Bank"
    
    def __init__(self, owner, balance=0):
        # Instance variables - unique to each instance
        self.owner = owner
        self.balance = balance

# Create two accounts
acc1 = BankAccount('Alice', 1000)
acc2 = BankAccount('Bob', 500)

# Instance variables are different
print(f"Account 1 owner: {acc1.owner}, balance: {acc1.balance}")
print(f"Account 2 owner: {acc2.owner}, balance: {acc2.balance}")

# Class variables are shared
print(f"Bank name (acc1): {acc1.bank_name}")
print(f"Bank name (acc2): {acc2.bank_name}")
print(f"Interest rate: {BankAccount.interest_rate}")

# Modifying class variable affects all instances
BankAccount.interest_rate = 0.03
print(f"New interest rate (acc1): {acc1.interest_rate}")
print(f"New interest rate (acc2): {acc2.interest_rate}")
````

## 6. Instance Methods <a id="instance-methods"></a>

Instance methods are functions defined inside a class that operate on instances.

### Key Points:
- The first parameter is always `self` (refers to the instance)
- Can access and modify instance variables and class variables
- Can call other instance methods

### Example:

````python
class Calculator:
    def __init__(self, value=0):
        self.value = value
    
    def add(self, x):
        self.value += x
        return self
    
    def subtract(self, x):
        self.value -= x
        return self
    
    def multiply(self, x):
        self.value *= x
        return self
    
    def get_value(self):
        return self.value

# Method chaining
calc = Calculator(10)
result = calc.add(5).subtract(3).multiply(2).get_value()
print(f"Result: {result}")  # (10 + 5 - 3) * 2 = 24
````

## 7. The `self` Parameter <a id="self-parameter"></a>

The `self` parameter is a reference to the **current instance** of the class.

### Key Points:
- It's the **first parameter** in every instance method
- It's **not a keyword** in Python - you can name it anything (but `self` is the convention)
- It allows you to access the instance's attributes and methods
- When you call `obj.method()`, Python automatically passes `obj` as `self`

### How it works internally:
```python
obj.method(arg1, arg2)
# Python internally translates to:
Class.method(obj, arg1, arg2)
```

````python
class Person:
    def __init__(self, name):
        self.name = name
    
    def greet(self):
        return f"Hello, my name is {self.name}"

person = Person('Alice')
print(person.greet())

# You can also call it explicitly:
print(Person.greet(person))

# Demonstrating that self is just a reference
print(f"Person object id: {id(person)}")
print(f"Self in greet refers to the same object: {id(person)}")
````

## 8. Encapsulation <a id="encapsulation"></a>

Encapsulation is the bundling of data and methods within a single unit (class), and controlling access to them.

### Access Modifiers in Python:

| Modifier | Syntax | Access Level |
|----------|--------|-------------|
| Public | `name` | Accessible from anywhere |
| Protected | `_name` | Accessible within class and subclasses |
| Private | `__name` | Accessible only within the class (name mangling) |

### Important Notes:
- Python doesn't have true private members like Java/C++
- The underscore convention is a **signal** to other developers
- Private attributes use **name mangling**: `__name` becomes `_ClassName__name`

````python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner          # Public
        self._account_type = 'savings'  # Protected
        self.__balance = balance   # Private (name mangled)

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            return f"Deposited ${amount}"
        return "Invalid amount"

    def get_balance(self):
        return self.__balance

account = BankAccount('Alice', 1000)

# Public access
print(f"Owner: {account.owner}")

# Protected access (should be accessed by subclasses, but works)
print(f"Account type: {account._account_type}")

# Private access - can't access directly
# print(account.__balance)  # AttributeError

# But can access via name mangling
print(f"Balance (via name mangling): {account._BankAccount__balance}")

# Best practice: use getter/setter methods
print(f"Balance: {account.get_balance()}")
````

## 9. Getters and Setters <a id="getters-and-setters"></a>

Getters and setters are methods used to **read** and **modify** private attributes.

### Why use them?
- **Validation**: Check data before setting
- **Logging**: Track changes to attributes
- **Computed properties**: Calculate values on the fly
- **Backward compatibility**: Change internal implementation without breaking code

### Example:

````python
class Temperature:
    def __init__(self, celsius=0):
        self._celsius = celsius
    
    # Getter
    def get_celsius(self):
        return self._celsius
    
    # Setter with validation
    def set_celsius(self, value):
        if value < -273.15:
            raise ValueError("Temperature cannot be below absolute zero!")
        self._celsius = value
    
    # Computed property
    def get_fahrenheit(self):
        return self._celsius * 9/5 + 32

temp = Temperature(25)
print(f"Celsius: {temp.get_celsius()}")
print(f"Fahrenheit: {temp.get_fahrenheit()}")

temp.set_celsius(100)
print(f"Boiling point - Celsius: {temp.get_celsius()}, Fahrenheit: {temp.get_fahrenheit()}")

# This will raise an error
try:
    temp.set_celsius(-300)
except ValueError as e:
    print(f"Error: {e}")
````

## 10. Inheritance <a id="inheritance"></a>

Inheritance allows a class (child/derived class) to inherit attributes and methods from another class (parent/base class).

### Key Points:
- The child class **inherits** all attributes and methods from the parent
- The child class can **override** parent methods
- The child class can **extend** the parent by adding new attributes/methods
- Use `super()` to call parent methods

### Syntax:
```python
class ParentClass:
    pass

class ChildClass(ParentClass):
    pass
```

````python
class Animal:
    def __init__(self, name):
        self.name = name
    
    def speak(self):
        pass
    
    def info(self):
        return f"I am {self.name}"

class Dog(Animal):
    def speak(self):
        return "Woof!"
    
    def fetch(self):
        return f"{self.name} is fetching the ball!"

class Cat(Animal):
    def speak(self):
        return "Meow!"
    
    def climb(self):
        return f"{self.name} is climbing a tree!"

dog = Dog('Buddy')
cat = Cat('Whiskers')

print(dog.info())
print(dog.speak())
print(dog.fetch())
print()
print(cat.info())
print(cat.speak())
print(cat.climb())
````

## 11. The `super()` Function <a id="super-function"></a>

The `super()` function allows you to call methods from the parent class.

### Key Points:
- Used to extend (not replace) parent class methods
- Avoids hardcoding the parent class name
- Essential for **cooperative inheritance** (especially with multiple inheritance)
- In Python 3, `super()` doesn't need arguments

### Example:

````python
class Vehicle:
    def __init__(self, make, model):
        self.make = make
        self.model = model
        self.started = False
    
    def start(self):
        self.started = True
        return f"{self.make} {self.model} is starting..."

class Car(Vehicle):
    def __init__(self, make, model, doors):
        # Call parent's __init__ using super()
        super().__init__(make, model)
        self.doors = doors
        self.fuel = 100
    
    def start(self):
        # Extend parent's start method
        parent_msg = super().start()
        return f"{parent_msg} Engine running. Fuel: {self.fuel}%"

car = Car('Tesla', 'Model S', 4)
print(car.start())
print(f"Doors: {car.doors}")
````

## 12. Method Overriding <a id="method-overriding"></a>

Method overriding is when a child class provides a **specific implementation** of a method that is already defined in its parent class.

### Rules:
- The method name must be the same
- The signature should be compatible
- Use `super()` to call the parent's version if needed
- If you don't call `super()`, the parent's method is completely replaced

### Example:

````python
class Shape:
    def __init__(self, color):
        self.color = color
    
    def area(self):
        raise NotImplementedError("Subclass must implement this method")
    
    def describe(self):
        return f"A {self.color} shape"

class Rectangle(Shape):
    def __init__(self, color, width, height):
        super().__init__(color)
        self.width = width
        self.height = height
    
    def area(self):
        return self.width * self.height
    
    def describe(self):
        parent_desc = super().describe()
        return f"{parent_desc} with area {self.area()}"

class Circle(Shape):
    def __init__(self, color, radius):
        super().__init__(color)
        self.radius = radius
    
    def area(self):
        import math
        return math.pi * self.radius ** 2

rect = Rectangle('red', 5, 3)
circle = Circle('blue', 4)

print(rect.describe())
print(f"Rectangle area: {rect.area()}")
print()
print(circle.describe())
print(f"Circle area: {circle.area():.2f}")
````

## 13. Multiple Inheritance <a id="multiple-inheritance"></a>

Multiple inheritance allows a class to inherit from **more than one** parent class.

### Key Points:
- Python supports multiple inheritance
- The **MRO (Method Resolution Order)** determines which parent's method is called
- Use `super()` for cooperative inheritance
- Can lead to complexity - use with caution

### Diamond Problem:
When a class inherits from two classes that both inherit from a common parent.

````python
class Flyable:
    def move(self):
        return "Flying through the air"

class Swimmable:
    def move(self):
        return "Swimming in water"

class Duck(Flyable, Swimmable):
    def move(self):
        fly_msg = Flyable.move(self)
        swim_msg = Swimmable.move(self)
        return f"{fly_msg} and {swim_msg}"

duck = Duck()
print(duck.move())

# Check MRO
print(f"MRO: {Duck.__mro__}")
````

## 14. Multilevel Inheritance <a id="multilevel-inheritance"></a>

Multilevel inheritance is when a class inherits from a class that itself inherits from another class.

### Structure:
```
Grandparent
    ↓
Parent
    ↓
Child
```

### Example:

````python
class Organism:
    def __init__(self, name):
        self.name = name
        self.alive = True
    
    def breathe(self):
        return f"{self.name} is breathing"

class Animal(Organism):
    def __init__(self, name, species):
        super().__init__(name)
        self.species = species
    
    def eat(self):
        return f"{self.name} is eating"

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name, species='Canis lupus')
        self.breed = breed
    
    def bark(self):
        return f"{self.name} says Woof!"

dog = Dog('Buddy', 'Golden Retriever')
print(dog.breathe())
print(dog.eat())
print(dog.bark())
print(f"Species: {dog.species}")
print(f"Breed: {dog.breed}")
print(f"Alive: {dog.alive}")
print(f"MRO: {[cls.__name__ for cls in Dog.__mro__]}")
````

## 15. Polymorphism <a id="polymorphism"></a>

Polymorphism means "many forms" - the ability to use a common interface for different data types.

### Types of Polymorphism:
1. **Method Overriding**: Child class provides its own implementation
2. **Duck Typing**: If it walks like a duck and quacks like a duck, it's a duck
3. **Operator Overloading**: Using operators with custom objects

### Example:

````python
class Bird:
    def __init__(self, name):
        self.name = name
    
    def fly(self):
        return f"{self.name} is flying"

class Penguin(Bird):
    def fly(self):
        return f"{self.name} can't fly, but swims!"

class Eagle(Bird):
    def fly(self):
        return f"{self.name} is soaring high in the sky!"

# Polymorphism: same interface, different implementations
birds = [Bird('Sparrow'), Penguin('Pingu'), Eagle('Talon')]

for bird in birds:
    print(bird.fly())

# Duck typing example
class Duck:
    def quack(self):
        return "Quack!"

class Person:
    def quack(self):
        return "I'm pretending to be a duck! Quack!"

def make_it_quack(thing):
    print(thing.quack())

make_it_quack(Duck())
make_it_quack(Person())
````

## 16. `__str__` and `__repr__` <a id="str-and-repr"></a>

These are **dunder methods** used for string representation of objects.

### `__str__`:
- Called by `str()` and `print()`
- Returns a **human-readable** string
- Intended for end users

### `__repr__`:
- Called by `repr()` and in interactive interpreter
- Returns an **unambiguous** string representation
- Intended for developers
- Should ideally be able to recreate the object

### Example:

````python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages
    
    def __str__(self):
        return f'"{self.title}" by {self.author}'
    
    def __repr__(self):
        return f'Book(title="{self.title}", author="{self.author}", pages={self.pages})'

book = Book('1984', 'George Orwell', 328)

print(str(book))   # Human-readable
print(repr(book))  # Developer-friendly
print(book)        # Uses __str__

# In interactive mode, repr is used
print(f"In f-string: {book}")  # Uses __str__
print(f"In repr: {repr(book)}")  # Uses __repr__"
````

## 17. `@classmethod` <a id="classmethod"></a>

A class method is a method that is bound to the **class** rather than the instance.

### Key Points:
- Decorated with `@classmethod`
- First parameter is `cls` (refers to the class, not instance)
- Can be called on both class and instances
- Can modify class state (class variables)
- Often used as **alternative constructors**

### Example:

````python
class Person:
    population = 0  # Class variable
    
    def __init__(self, name, age):
        self.name = name
        self.age = age
        Person.population += 1
    
    @classmethod
    def get_population(cls):
        return f"Total population: {cls.population}"
    
    @classmethod
    def from_birth_year(cls, name, birth_year):
        """Alternative constructor"""
        current_year = 2024
        age = current_year - birth_year
        return cls(name, age)
    
    def __str__(self):
        return f"{self.name}, {self.age} years old"

# Using classmethod
print(Person.get_population())

# Using alternative constructor
person1 = Person('Alice', 30)
person2 = Person.from_birth_year('Bob', 1995)

print(person1)
print(person2)
print(Person.get_population())
````

## 18. `@staticmethod` <a id="staticmethod"></a>

A static method is a method that doesn't access the class or instance.

### Key Points:
- Decorated with `@staticmethod`
- No `self` or `cls` parameter
- Behaves like a regular function defined inside a class
- Can be called on both class and instances
- Used for utility functions related to the class

### Example:

````python
class MathUtils:
    @staticmethod
    def is_prime(n):
        if n < 2:
            return False
        for i in range(2, int(n**0.5) + 1):
            if n % i == 0:
                return False
        return True
    
    @staticmethod
    def factorial(n):
        if n <= 1:
            return 1
        return n * MathUtils.factorial(n - 1)
    
    @staticmethod
    def gcd(a, b):
        while b:
            a, b = b, a % b
        return a

# Can call without creating an instance
print(f"Is 17 prime? {MathUtils.is_prime(17)}")
print(f"Factorial of 5: {MathUtils.factorial(5)}")
print(f"GCD of 48 and 18: {MathUtils.gcd(48, 18)}")

# Can also call through an instance
utils = MathUtils()
print(f"Is 10 prime? {utils.is_prime(10)}")
````

## 19. Comparing Objects <a id="comparing-objects"></a>

Python allows you to define how objects of a class should be compared using **dunder methods**.

### Comparison Methods:

| Method | Operator | Description |
|--------|----------|-------------|
| `__eq__` | `==` | Equality |
| `__ne__` | `!=` | Inequality |
| `__lt__` | `<` | Less than |
| `__le__` | `<=` | Less than or equal |
| `__gt__` | `>` | Greater than |
| `__ge__` | `>=` | Greater than or equal |

### Example:

````python
class Student:
    def __init__(self, name, grade):
        self.name = name
        self.grade = grade
    
    def __eq__(self, other):
        if not isinstance(other, Student):
            return NotImplemented
        return self.grade == other.grade
    
    def __lt__(self, other):
        if not isinstance(other, Student):
            return NotImplemented
        return self.grade < other.grade
    
    def __le__(self, other):
        if not isinstance(other, Student):
            return NotImplemented
        return self.grade <= other.grade
    
    def __gt__(self, other):
        if not isinstance(other, Student):
            return NotImplemented
        return self.grade > other.grade
    
    def __ge__(self, other):
        if not isinstance(other, Student):
            return NotImplemented
        return self.grade >= other.grade
    
    def __str__(self):
        return f"{self.name} (Grade: {self.grade})"

s1 = Student('Alice', 95)
s2 = Student('Bob', 87)
s3 = Student('Charlie', 95)

print(f"{s1} == {s3}: {s1 == s3}")
print(f"{s1} > {s2}: {s1 > s2}")
print(f"{s1} < {s2}: {s1 < s2}")
print(f"{s1} >= {s3}: {s1 >= s3}")
print(f"{s2} <= {s1}: {s2 <= s1}")

# Sorting students
students = [s1, s2, s3]
sorted_students = sorted(students)
print(f"\nSorted by grade: {[str(s) for s in sorted_students]}")
````

### Problem 1: Design HashSet (LeetCode #705)

Design a HashSet without using any built-in hash table library.

Implement `MyHashSet` class:
- `void add(key)` Inserts the value `key` into the HashSet.
- `bool contains(key)` Returns whether the value `key` exists in the HashSet or not.
- `void remove(key)` Removes the value `key` in the HashSet. If `key` does not exist, do nothing.

````python
class MyHashSet:
    def __init__(self):
        """Initialize the HashSet with a fixed-size bucket array."""
        self.bucket_count = 769  # A prime number to reduce collisions
        self.buckets = [[] for _ in range(self.bucket_count)]
    
    def _get_hash(self, key):
        """Hash function to determine the bucket index."""
        return key % self.bucket_count
    
    def add(self, key: int) -> None:
        """Add a key to the HashSet."""
        hash_key = self._get_hash(key)
        if key not in self.buckets[hash_key]:
            self.buckets[hash_key].append(key)
    
    def remove(self, key: int) -> None:
        """Remove a key from the HashSet."""
        hash_key = self._get_hash(key)
        if key in self.buckets[hash_key]:
            self.buckets[hash_key].remove(key)
    
    def contains(self, key: int) -> bool:
        """Check if the HashSet contains the key."""
        hash_key = self._get_hash(key)
        return key in self.buckets[hash_key]

# Test the HashSet
hash_set = MyHashSet()
hash_set.add(1)
hash_set.add(2)
print(f"Contains 1: {hash_set.contains(1)}")  # True
print(f"Contains 3: {hash_set.contains(3)}")  # False
hash_set.add(2)
print(f"Contains 2: {hash_set.contains(2)}")  # True
hash_set.remove(2)
print(f"Contains 2 after remove: {hash_set.contains(2)}")  # False
````

### Problem 2: Design HashMap (LeetCode #706)

Design a HashMap without using any built-in hash table library.

Implement `MyHashMap` class:
- `void put(int key, int value)` Inserts a (key, value) pair into the HashMap.
- `int get(int key)` Returns the value to which the key is mapped, or -1 if not found.
- `void remove(int key)` Removes the key and its value from the HashMap.

````python
class MyHashMap:
    def __init__(self):
        """Initialize the HashMap with a fixed-size bucket array."""
        self.bucket_count = 769
        self.buckets = [[] for _ in range(self.bucket_count)]
    
    def _get_hash(self, key):
        """Hash function to determine the bucket index."""
        return key % self.bucket_count
    
    def put(self, key: int, value: int) -> None:
        """Insert or update a key-value pair."""
        hash_key = self._get_hash(key)
        for i, (k, v) in enumerate(self.buckets[hash_key]):
            if k == key:
                self.buckets[hash_key][i] = (key, value)
                return
        self.buckets[hash_key].append((key, value))
    
    def get(self, key: int) -> int:
        """Get the value for a key, or -1 if not found."""
        hash_key = self._get_hash(key)
        for k, v in self.buckets[hash_key]:
            if k == key:
                return v
        return -1
    
    def remove(self, key: int) -> None:
        """Remove a key-value pair from the HashMap."""
        hash_key = self._get_hash(key)
        for i, (k, v) in enumerate(self.buckets[hash_key]):
            if k == key:
                self.buckets[hash_key].pop(i)
                return

# Test the HashMap
hash_map = MyHashMap()
hash_map.put(1, 10)
hash_map.put(2, 20)
print(f"Get 1: {hash_map.get(1)}")  # 10
print(f"Get 3: {hash_map.get(3)}")  # -1
hash_map.put(2, 30)
print(f"Get 2 after update: {hash_map.get(2)}")  # 30
hash_map.remove(2)
print(f"Get 2 after remove: {hash_map.get(2)}")  # -1
````

### Problem 3: Min Stack (LeetCode #155)

Design a stack that supports push, pop, top, and retrieving the minimum element in constant time.

Implement `MinStack` class:
- `void push(int val)` Pushes the element onto the stack.
- `void pop()` Removes the element from the top of the stack.
- `int top()` Returns the top element.
- `int getMin()` Retrieves the minimum element in the stack.

````python
class MinStack:
    def __init__(self):
        """Initialize the stack with a main stack and a min stack."""
        self.stack = []
        self.min_stack = []
    
    def push(self, val: int) -> None:
        """Push a value onto the stack."""
        self.stack.append(val)
        # Track the minimum value
        if not self.min_stack or val <= self.min_stack[-1]:
            self.min_stack.append(val)
    
    def pop(self) -> None:
        """Remove the top element from the stack."""
        if self.stack:
            val = self.stack.pop()
            if val == self.min_stack[-1]:
                self.min_stack.pop()
    
    def top(self) -> int:
        """Return the top element without removing it."""
        if self.stack:
            return self.stack[-1]
        return None
    
    def getMin(self) -> int:
        """Return the minimum element in the stack."""
        if self.min_stack:
            return self.min_stack[-1]
        return None

# Test the MinStack
min_stack = MinStack()
min_stack.push(-2)
min_stack.push(0)
min_stack.push(-3)
print(f"getMin(): {min_stack.getMin()}")  # -3
min_stack.pop()
print(f"top(): {min_stack.top()}")  # 0
print(f"getMin(): {min_stack.getMin()}")  # -2
````

### Problem 4: Implement Queue using Stacks (LeetCode #232)

Implement a first-in-first-out (FIFO) queue using only two stacks.

Implement `MyQueue` class:
- `void push(int x)` Pushes element to the back of the queue.
- `int pop()` Removes the element from the front of the queue and returns it.
- `int peek()` Returns the element at the front of the queue.
- `boolean empty()` Returns true if the queue is empty, false otherwise.

````python
class MyQueue:
    def __init__(self):
        """Initialize the queue with two stacks."""
        self.in_stack = []    # For push operations
        self.out_stack = []   # For pop/peek operations
    
    def push(self, x: int) -> None:
        """Push element to the back of the queue."""
        self.in_stack.append(x)
    
    def _transfer(self):
        """Transfer elements from in_stack to out_stack."""
        while self.in_stack:
            self.out_stack.append(self.in_stack.pop())
    
    def pop(self) -> int:
        """Remove and return the front element."""
        if not self.out_stack:
            self._transfer()
        return self.out_stack.pop() if self.out_stack else None
    
    def peek(self) -> int:
        """Return the front element without removing it."""
        if not self.out_stack:
            self._transfer()
        return self.out_stack[-1] if self.out_stack else None
    
    def empty(self) -> bool:
        """Check if the queue is empty."""
        return not self.in_stack and not self.out_stack

# Test the MyQueue
my_queue = MyQueue()
my_queue.push(1)
my_queue.push(2)
my_queue.push(3)
print(f"Peek: {my_queue.peek()}")  # 1
print(f"Pop: {my_queue.pop()}")    # 1
print(f"Empty: {my_queue.empty()}")  # False
print(f"Pop: {my_queue.pop()}")    # 2
````

### Problem 5: Implement Stack using Queues (LeetCode #225)

Implement a last-in-first-out (LIFO) stack using only two queues.

Implement `MyStack` class:
- `void push(int x)` Pushes element onto the stack.
- `int pop()` Removes and returns the top element.
- `int top()` Returns the top element without removing it.
- `boolean empty()` Returns true if the stack is empty, false otherwise.

````python
from collections import deque

class MyStack:
    def __init__(self):
        """Initialize the stack with two queues."""
        self.queue1 = deque()  # Main queue
        self.queue2 = deque()  # Helper queue
    
    def push(self, x: int) -> None:
        """Push element onto the stack."""
        # Add to the non-empty queue, or queue1 if both empty
        if self.queue1:
            self.queue1.append(x)
        else:
            self.queue2.append(x)
    
    def pop(self) -> int:
        """Remove and return the top element."""
        if not self.queue1 and not self.queue2:
            return None
        
        # Determine which queue has elements
        if self.queue1:
            source, dest = self.queue1, self.queue2
        else:
            source, dest = self.queue2, self.queue1
        
        # Move all elements except the last one
        while len(source) > 1:
            dest.append(source.popleft())
        
        return source.popleft()
    
    def top(self) -> int:
        """Return the top element without removing it."""
        if not self.queue1 and not self.queue2:
            return None
        
        # Determine which queue has elements
        if self.queue1:
            source, dest = self.queue1, self.queue2
        else:
            source, dest = self.queue2, self.queue1
        
        # Move all elements except the last one, then move them back
        while len(source) > 1:
            dest.append(source.popleft())
        
        top_val = source.popleft()
        dest.append(top_val)
        
        return top_val
    
    def empty(self) -> bool:
        """Check if the stack is empty."""
        return not self.queue1 and not self.queue2

# Test the MyStack
my_stack = MyStack()
my_stack.push(1)
my_stack.push(2)
my_stack.push(3)
print(f"Top: {my_stack.top()}")  # 3
print(f"Pop: {my_stack.pop()}")  # 3
print(f"Empty: {my_stack.empty()}")  # False
print(f"Pop: {my_stack.pop()}")  # 2
````
