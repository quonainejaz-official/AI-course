## 1. What is Pandas?

Pandas is a powerful Python data analysis and manipulation library. It provides:
- **Data structures**: Series (1D) and DataFrame (2D)
- **Data analysis tools**: Filtering, grouping, aggregation
- **I/O capabilities**: Read/write CSV, Excel, SQL, JSON, and more
- **Time series analysis**: Date range generation, frequency conversion

### Why Pandas for Data Analysis?
1. **Easy handling of missing data**
2. **Flexible reshaping and pivoting** of data
3. **Powerful group-by functionality**
4. **Excellent integration** with NumPy, Matplotlib, Scikit-learn
5. **Industry standard** in data science and analytics

## 3. Series

A Series is a one-dimensional labeled array capable of holding any data type.

```python
# Creating a Series from a list
s1 = pd.Series([10, 20, 30, 40, 50])
print(s1)
print(f"\nDtype: {s1.dtype}")
print(f"Shape: {s1.shape}")
```

```python
# Creating a Series with custom index
s2 = pd.Series([100, 200, 300, 400], index=['a', 'b', 'c', 'd'])
print(s2)
print(f"\nIndex: {s2.index.tolist()}")
```

```python
# Creating a Series from a dictionary
s3 = pd.Series({'apple': 5, 'banana': 10, 'cherry': 7})
print(s3)
print(f"\nValues: {s3.values}")
print(f"Index: {s3.index.tolist()}")
```

```python
# Series indexing
print(f"s2['b'] = {s2['b']}")
print(f"s2[1] = {s2[1]}")
print(f"s2[1:3] = {s2[1:3].tolist()}")
print(f"s2[['a', 'c']] = {s2[['a', 'c']].tolist()}")
```

```python
# Series operations
s4 = pd.Series([1, 2, 3, 4, 5])
s5 = pd.Series([10, 20, 30, 40, 50])

print(f"Addition: {(s4 + s5).tolist()}")
print(f"Multiplication: {(s4 * 2).tolist()}")
print(f"Square root: {np.sqrt(s4).tolist()}")

# Boolean indexing
print(f"\nValues > 3: {s4[s4 > 3].tolist()}")
```

## 4. DataFrame

A DataFrame is a two-dimensional, size-mutable, potentially heterogeneous tabular data structure with labeled axes (rows and columns).

```python
# Creating a DataFrame from a dictionary
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Age': [25, 30, 35, 28, 32],
    'City': ['New York', 'London', 'Tokyo', 'Paris', 'Sydney'],
    'Salary': [75000, 80000, 90000, 65000, 85000]
}
df = pd.DataFrame(data)
print(df)
print(f"\nShape: {df.shape}")
print(f"Columns: {df.columns.tolist()}")
print(f"Index: {df.index.tolist()}")
print(f"Dtypes:\n{df.dtypes}")
```

```python
# DataFrame from a list of dictionaries
data2 = [
    {'Product': 'Laptop', 'Price': 999, 'Stock': 50},
    {'Product': 'Mouse', 'Price': 25, 'Stock': 200},
    {'Product': 'Keyboard', 'Price': 75, 'Stock': 100}
]
df2 = pd.DataFrame(data2)
print(df2)
```

```python
# DataFrame from a NumPy array
arr = np.random.randint(1, 100, size=(5, 3))
df3 = pd.DataFrame(arr, columns=['A', 'B', 'C'])
print(df3)
print(f"\nDescribe:\n{df3.describe()}")
```

```python
# head() and tail() methods
print("First 3 rows:")
print(df.head(3))
print("\nLast 2 rows:")
print(df.tail(2))
```

```python
# DataFrame attributes
print(f"Shape: {df.shape}")
print(f"Size: {df.size}")
print(f"Number of dimensions: {df.ndim}")
print(f"Column count: {len(df.columns)}")
print(f"Row count: {len(df)}")
```

## 5. Column and Row Selection

```python
# Column selection
print("Select 'Name' column:")
print(df['Name'])
print(f"\nType: {type(df['Name'])}")

print("\nSelect multiple columns:")
print(df[['Name', 'Age']])
```

```python
# Row selection with loc (label-based)
print("Row at index 0:")
print(df.loc[0])

print("\nRows 1 to 3:")
print(df.loc[1:3])

print("\nSpecific rows and columns:")
print(df.loc[[0, 2], ['Name', 'City']])
```

```python
# Row selection with iloc (position-based)
print("Row at position 0:")
print(df.iloc[0])

print("\nRows at positions 1 to 3:")
print(df.iloc[1:4])

print("\nSpecific rows and columns by position:")
print(df.iloc[[0, 2], [0, 2]])
```

```python
# Boolean indexing
print("Rows where Age > 30:")
print(df[df['Age'] > 30])

print("\nRows where City is 'Tokyo' or 'Paris':")
print(df[df['City'].isin(['Tokyo', 'Paris'])])
```

## 6. Adding and Removing Columns

```python
# Adding a new column
df['Department'] = ['Engineering', 'Marketing', 'Sales', 'HR', 'Finance']
print(df)

# Adding a computed column
df['Salary_Bonus'] = df['Salary'] * 1.10
print("\nWith bonus column:")
print(df[['Name', 'Salary', 'Salary_Bonus']])
```

```python
# Inserting a column at a specific position
df.insert(2, 'Experience', [3, 5, 8, 2, 6])
print(df)

# Removing a column
df.drop('Salary_Bonus', axis=1, inplace=True)
print("\nAfter dropping Salary_Bonus:")
print(df)

# Alternative: pop() method
dept = df.pop('Department')
print("\nPopped Department:")
print(dept)
print(f"\nRemaining columns: {df.columns.tolist()}")
```

```python
# Removing rows
print("Original DataFrame:")
print(df)

df_dropped = df.drop(2, axis=0)  # Drop row at index 2
print("\nAfter dropping row at index 2:")
print(df_dropped)

# Drop multiple rows
df_multi = df.drop([0, 1], axis=0)
print("\nAfter dropping rows at index 0 and 1:")
print(df_multi)
```

## 7. Renaming Columns

```python
# Rename a single column
df.rename(columns={'Name': 'Full_Name'}, inplace=True)
print(df)

# Rename multiple columns
df.rename(columns={'Age': 'Years', 'City': 'Location'}, inplace=True)
print("\nAfter renaming multiple columns:")
print(df)
```

```python
# Rename all columns at once
df.columns = ['Employee_Name', 'Age_Years', 'Location', 'Annual_Salary', 'Years_Exp']
print(df)

# Rename using a function
df.columns = df.columns.str.lower()
print("\nLowercase columns:")
print(df)

# Rename index
df.rename(index={0: 'first', 1: 'second'}, inplace=True)
print("\nWith renamed index:")
print(df)
```

## 8. Sorting

```python
# Create a fresh DataFrame for sorting examples
df_sort = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Age': [25, 30, 35, 28, 32],
    'Score': [88, 92, 76, 95, 87]
})

# Sort by index
print("Sort by index (ascending):")
print(df_sort.sort_index())

print("\nSort by index (descending):")
print(df_sort.sort_index(ascending=False))
```

```python
# Sort by values
print("Sort by Age:")
print(df_sort.sort_values('Age'))

print("\nSort by Score (descending):")
print(df_sort.sort_values('Score', ascending=False))

print("\nSort by multiple columns:")
print(df_sort.sort_values(['Age', 'Score'], ascending=[True, False]))
```

```python
# Sort with NaN handling
df_nan = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'Age': [25, np.nan, 35, 28]
})
print(df_nan)
print("\nSort with NaN at end:")
print(df_nan.sort_values('Age', na_position='last'))
print("\nSort with NaN at beginning:")
print(df_nan.sort_values('Age', na_position='first'))
```

## 10. Handling Missing Values

```python
# Create a DataFrame with missing values
df_missing = pd.DataFrame({
    'A': [1, 2, np.nan, 4, 5],
    'B': ['x', np.nan, 'z', 'w', 'v'],
    'C': [10.5, 20.3, 30.1, np.nan, 50.0],
    'D': [True, False, np.nan, True, False]
})
print(df_missing)
print(f"\nMissing values per column:")
print(df_missing.isnull().sum())
```

```python
# Check for missing values
print("isnull():")
print(df_missing.isnull())

print("\nnotnull():")
print(df_missing.notnull())

print(f"\nTotal missing: {df_missing.isnull().sum().sum()}")
print(f"Total non-missing: {df_missing.notnull().sum().sum()}")
```

```python
# Drop rows with missing values
print("Original:")
print(df_missing)

print("\nDrop any row with NaN:")
print(df_missing.dropna())

print("\nDrop rows where ALL values are NaN:")
print(df_missing.dropna(how='all'))

print("\nDrop rows where specific column has NaN:")
print(df_missing.dropna(subset=['A']))
```

```python
# Fill missing values
print("Fill with scalar:")
print(df_missing.fillna(0))

print("\nFill numeric with mean:")
print(df_missing.fillna(df_missing.mean(numeric_only=True)))

print("\nFill specific column:")
print(df_missing['C'].fillna(df_missing['C'].mean()))

print("\nForward fill:")
print(df_missing.ffill())

print("\nBackward fill:")
print(df_missing.bfill())
```

```python
# Replace specific values
df_replace = pd.DataFrame({
    'Status': ['Active', 'Inactive', 'Unknown', 'Active', 'Unknown'],
    'Value': [10, -1, 20, -1, 30]
})
print("Original:")
print(df_replace)

df_replace['Status'] = df_replace['Status'].replace('Unknown', 'Active')
df_replace['Value'] = df_replace['Value'].replace(-1, 0)
print("\nAfter replacement:")
print(df_replace)
```

## 11. Handling Duplicates

```python
# Create a DataFrame with duplicates
df_dup = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Alice', 'Charlie', 'Bob', 'Alice'],
    'Age': [25, 30, 25, 35, 30, 25],
    'City': ['NYC', 'LA', 'NYC', 'SF', 'LA', 'NYC']
})
print(df_dup)
print(f"\nShape: {df_dup.shape}")
```

```python
# Check for duplicates
print("Duplicated rows:")
print(df_dup.duplicated())

print(f"\nNumber of duplicates: {df_dup.duplicated().sum()}")

print("\nDuplicated based on specific column:")
print(df_dup.duplicated(subset=['Name']))

print("\nKeep last occurrence:")
print(df_dup.duplicated(subset=['Name'], keep='last'))
```

```python
# Remove duplicates
print("Drop all duplicates:")
df_no_dup = df_dup.drop_duplicates()
print(df_no_dup)
print(f"Shape: {df_no_dup.shape}")

print("\nDrop duplicates based on Name:")
df_no_dup_name = df_dup.drop_duplicates(subset=['Name'])
print(df_no_dup_name)

print("\nDrop duplicates, keep last:")
df_keep_last = df_dup.drop_duplicates(subset=['Name'], keep='last')
print(df_keep_last)
```

## 12. Basic Data Analysis

```python
# Create a sample dataset
np.random.seed(42)
df_analysis = pd.DataFrame({
    'Category': np.random.choice(['A', 'B', 'C'], size=20),
    'Value': np.random.randint(1, 100, size=20),
    'Score': np.random.uniform(0, 100, size=20).round(2)
})
print(df_analysis)
```

```python
# value_counts()
print("Category value counts:")
print(df_analysis['Category'].value_counts())

print("\nNormalized value counts:")
print(df_analysis['Category'].value_counts(normalize=True))

print("\nSorted by frequency:")
print(df_analysis['Category'].value_counts(sort=True))
```

```python
# describe()
print("Describe numeric columns:")
print(df_analysis.describe())

print("\nDescribe all columns:")
print(df_analysis.describe(include='all'))

print("\nDescribe specific column:")
print(df_analysis['Value'].describe())

print(f"\nMean: {df_analysis['Value'].mean():.2f}")
print(f"Median: {df_analysis['Value'].median():.2f}")
print(f"Std Dev: {df_analysis['Value'].std():.2f}")
print(f"Min: {df_analysis['Value'].min()}")
print(f"Max: {df_analysis['Value'].max()}")
print(f"Sum: {df_analysis['Value'].sum()}")
```

```python
# Other useful analysis methods
print(f"Unique categories: {df_analysis['Category'].unique()}")
print(f"Number of unique: {df_analysis['Category'].nunique()}")
print(f"Mode: {df_analysis['Value'].mode().tolist()}")
print(f"Quantiles: {df_analysis['Value'].quantile([0.25, 0.5, 0.75]).tolist()}")
print(f"Variance: {df_analysis['Value'].var():.2f}")
print(f"Skewness: {df_analysis['Value'].skew():.4f}")
print(f"Kurtosis: {df_analysis['Value'].kurt():.4f}")
```

## 13. GroupBy Basics

```python
# Create a sales dataset
sales_data = pd.DataFrame({
    'Product': ['A', 'B', 'A', 'C', 'B', 'A', 'C', 'B', 'A', 'C'],
    'Region': ['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South', 'East'],
    'Sales': [100, 200, 150, 300, 250, 120, 180, 220, 170, 290],
    'Units': [10, 20, 15, 30, 25, 12, 18, 22, 17, 29]
})
print(sales_data)
```

```python
# Basic GroupBy
grouped = sales_data.groupby('Product')
print("Group by Product:")
print(grouped['Sales'].sum())

print("\nGroup by Region:")
print(sales_data.groupby('Region')['Sales'].mean())

print("\nMultiple aggregations:")
print(grouped['Sales'].agg(['sum', 'mean', 'count']))

print("\nGroup by multiple columns:")
print(sales_data.groupby(['Product', 'Region'])['Sales'].sum())
```

```python
# Iterating over groups
for name, group in sales_data.groupby('Product'):
    print(f"\nProduct: {name}")
    print(group)
    print(f"Total Sales: {group['Sales'].sum()}")
```

```python
# GroupBy with transform
sales_data['Sales_Mean_By_Product'] = sales_data.groupby('Product')['Sales'].transform('mean')
sales_data['Sales_Pct_of_Product'] = sales_data['Sales'] / sales_data['Sales_Mean_By_Product'] * 100
print(sales_data)
```

## 14. Cross-tabulation

```python
# Create a survey dataset
survey_data = pd.DataFrame({
    'Gender': np.random.choice(['Male', 'Female'], size=50),
    'Preference': np.random.choice(['A', 'B', 'C'], size=50),
    'Age_Group': np.random.choice(['18-25', '26-35', '36-50'], size=50)
})

# Basic cross-tabulation
ct = pd.crosstab(survey_data['Gender'], survey_data['Preference'])
print("Cross-tabulation (Gender vs Preference):")
print(ct)

print("\nWith margins:")
ct_margins = pd.crosstab(survey_data['Gender'], survey_data['Preference'], margins=True)
print(ct_margins)

print("\nNormalized:")
ct_norm = pd.crosstab(survey_data['Gender'], survey_data['Preference'], normalize=True)
print(ct_norm)

print("\nRow-wise normalized:")
ct_row = pd.crosstab(survey_data['Gender'], survey_data['Preference'], normalize='index')
print(ct_row)
```

```python
# Three-way cross-tabulation
ct3 = pd.crosstab(
    [survey_data['Gender'], survey_data['Age_Group']],
    survey_data['Preference']
)
print("Three-way cross-tabulation:")
print(ct3)
```

## 15. Basic Visualization

```python
# Import matplotlib for visualization
import matplotlib.pyplot as plt
%matplotlib inline

# Create sample data for visualization
viz_data = pd.DataFrame({
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'Revenue': [100, 120, 140, 110, 150, 180],
    'Cost': [80, 90, 100, 85, 110, 120]
})

# Line plot
viz_data.set_index('Month')[['Revenue', 'Cost']].plot(kind='line', marker='o')
plt.title('Revenue and Cost Over Time')
plt.ylabel('Amount ($)')
plt.show()
```

```python
# Bar chart
viz_data.set_index('Month')['Revenue'].plot(kind='bar', color='steelblue')
plt.title('Revenue by Month')
plt.ylabel('Revenue ($)')
plt.xticks(rotation=0)
plt.show()
```

```python
# Histogram
np.random.seed(42)
hist_data = pd.Series(np.random.normal(50, 15, 1000))
hist_data.plot(kind='hist', bins=30, color='coral')
plt.title('Distribution of Values')
plt.xlabel('Value')
plt.show()
```

```python
# Scatter plot
scatter_data = pd.DataFrame({
    'X': np.random.rand(50) * 100,
    'Y': np.random.rand(50) * 100
})
scatter_data.plot(kind='scatter', x='X', y='Y', color='green', alpha=0.6)
plt.title('Scatter Plot')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

```python
# Box plot
box_data = pd.DataFrame({
    'Group A': np.random.normal(50, 10, 100),
    'Group B': np.random.normal(60, 15, 100),
    'Group C': np.random.normal(45, 8, 100)
})
box_data.plot(kind='box')
plt.title('Box Plot by Group')
plt.ylabel('Value')
plt.show()
```
