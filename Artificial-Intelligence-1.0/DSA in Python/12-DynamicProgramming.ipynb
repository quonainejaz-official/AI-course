# Dynamic Programming - DSA Patterns

Dynamic Programming optimizes recursive solutions by storing subproblem results.

## Topics Covered
1. Fibonacci and Climbing Stairs
2. Knapsack Problem
3. Longest Common Subsequence (LCS)
4. Longest Increasing Subsequence (LIS)

6. Coin Change

## 1. Fibonacci and Climbing Stairs

```python
def fibonacci_dp(n):
    """Fibonacci using DP.
    Time: O(n), Space: O(1)"""
    if n <= 1:
        return n
    
    a, b = 0, 1
    for _ in range(2, n + 1):
        a, b = b, a + b
    return b


def climb_stairs(n):
    """Climbing stairs problem (LeetCode 70).
    Time: O(n), Space: O(1)"""
    if n <= 2:
        return n
    
    a, b = 1, 2
    for _ in range(3, n + 1):
        a, b = b, a + b
    return b


# Examples
print(f"Fibonacci(10) = {fibonacci_dp(10)}")
print(f"Climb stairs 5: {climb_stairs(5)}")
```

## 2. Knapsack Problem

```python
def knapsack_01(weights, values, capacity):
    """0/1 Knapsack problem.
    Time: O(n * W)"""
    n = len(weights)
    dp = [[0] * (capacity + 1) for _ in range(n + 1)]
    
    for i in range(1, n + 1):
        for w in range(1, capacity + 1):
            if weights[i - 1] <= w:
                dp[i][w] = max(
                    dp[i - 1][w],
                    dp[i - 1][w - weights[i - 1]] + values[i - 1]
                )
            else:
                dp[i][w] = dp[i - 1][w]
    
    return dp[n][capacity]


def knapsack_unbounded(weights, values, capacity):
    """Unbounded Knapsack (items can be repeated)."""
    dp = [0] * (capacity + 1)
    
    for w in range(1, capacity + 1):
        for i in range(len(weights)):
            if weights[i] <= w:
                dp[w] = max(dp[w], dp[w - weights[i]] + values[i])
    
    return dp[capacity]


# Examples
weights = [2, 3, 4, 5]
values = [3, 4, 5, 6]
capacity = 8
print(f"0/1 Knapsack: {knapsack_01(weights, values, capacity)}")
print(f"Unbounded Knapsack: {knapsack_unbounded(weights, values, capacity)}")
```

## 3. Longest Common Subsequence (LCS)

```python
def longest_common_subsequence(text1, text2):
    """Find length of LCS (LeetCode 1143).
    Time: O(m * n)"""
    m, n = len(text1), len(text2)
    dp = [[0] * (n + 1) for _ in range(m + 1)]
    
    for i in range(1, m + 1):
        for j in range(1, n + 1):
            if text1[i - 1] == text2[j - 1]:
                dp[i][j] = dp[i - 1][j - 1] + 1
            else:
                dp[i][j] = max(dp[i - 1][j], dp[i][j - 1])
    
    return dp[m][n]


def longest_common_substring(s1, s2):
    """Find length of longest common substring."""
    m, n = len(s1), len(s2)
    dp = [[0] * (n + 1) for _ in range(m + 1)]
    max_len = 0
    
    for i in range(1, m + 1):
        for j in range(1, n + 1):
            if s1[i - 1] == s2[j - 1]:
                dp[i][j] = dp[i - 1][j - 1] + 1
                max_len = max(max_len, dp[i][j])
    
    return max_len


# Examples
print(f"LCS of 'abcde' and 'ace': {longest_common_subsequence('abcde', 'ace')}")
print(f"Longest common substring 'abcde' and 'abfce': {longest_common_substring('abcde', 'abfce')}")
```

## 4. Longest Increasing Subsequence (LIS)

```python
def length_of_lis(nums):
    """Find length of LIS (LeetCode 300).
    Time: O(n^2)"""
    n = len(nums)
    dp = [1] * n
    
    for i in range(1, n):
        for j in range(i):
            if nums[j] < nums[i]:
                dp[i] = max(dp[i], dp[j] + 1)
    
    return max(dp)


def length_of_lis_binary_search(nums):
    """LIS using binary search.
    Time: O(n log n)"""
    import bisect
    
    tails = []
    for num in nums:
        pos = bisect.bisect_left(tails, num)
        if pos == len(tails):
            tails.append(num)
        else:
            tails[pos] = num
    
    return len(tails)


# Examples
nums = [10, 9, 2, 5, 3, 7, 101, 18]
print(f"LIS of {nums}:")
print(f"  O(n^2) method: {length_of_lis(nums)}")
print(f"  O(n log n) method: {length_of_lis_binary_search(nums)}")
```

## 5. Coin Change

```python
def coin_change(coins, amount):
    """Minimum coins to make amount (LeetCode 322).
    Time: O(amount * len(coins))"""
    dp = [float('inf')] * (amount + 1)
    dp[0] = 0
    
    for i in range(1, amount + 1):
        for coin in coins:
            if coin <= i and dp[i - coin] != float('inf'):
                dp[i] = min(dp[i], dp[i - coin] + 1)
    
    return dp[amount] if dp[amount] != float('inf') else -1


def coin_change_ways(coins, amount):
    """Number of ways to make amount (LeetCode 518).
    Time: O(amount * len(coins))"""
    dp = [0] * (amount + 1)
    dp[0] = 1
    
    for coin in coins:
        for i in range(coin, amount + 1):
            dp[i] += dp[i - coin]
    
    return dp[amount]


# Examples
coins = [1, 5, 10, 25]
amount = 30
print(f"Minimum coins for {amount}: {coin_change(coins, amount)}")
print(f"Ways to make {amount}: {coin_change_ways(coins, amount)}")
```

## 6. Edit Distance

```python
def edit_distance(word1, word2):
    """Minimum operations to convert word1 to word2 (LeetCode 72).
    Time: O(m * n)"""
    m, n = len(word1), len(word2)
    dp = [[0] * (n + 1) for _ in range(m + 1)]
    
    for i in range(m + 1):
        dp[i][0] = i
    for j in range(n + 1):
        dp[0][j] = j
    
    for i in range(1, m + 1):
        for j in range(1, n + 1):
            if word1[i - 1] == word2[j - 1]:
                dp[i][j] = dp[i - 1][j - 1]
            else:
                dp[i][j] = 1 + min(
                    dp[i - 1][j],      # Delete
                    dp[i][j - 1],      # Insert
                    dp[i - 1][j - 1]   # Replace
                )
    
    return dp[m][n]


# Example
print(f"Edit distance 'horse' to 'ros': {edit_distance('horse', 'ros')}")
print(f"Edit distance 'intention' to 'execution': {edit_distance('intention', 'execution')}")
```
