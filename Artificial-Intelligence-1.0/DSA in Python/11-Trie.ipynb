# Trie (Prefix Tree) - DSA Patterns

A Trie is a tree-like data structure for storing strings, where each node represents a character.

## Topics Covered
1. Trie Implementation
2. Word Search
3. Prefix Operations
4. Autocomplete System

## 1. Trie Implementation

```python
class TrieNode:
    """Node in a Trie."""
    def __init__(self):
        self.children = {}
        self.is_end = False


class Trie:
    """Trie (Prefix Tree) implementation."""
    
    def __init__(self):
        self.root = TrieNode()
    
    def insert(self, word):
        """Insert a word into the trie.
        Time: O(m) where m is word length"""
        node = self.root
        for char in word:
            if char not in node.children:
                node.children[char] = TrieNode()
            node = node.children[char]
        node.is_end = True
    
    def search(self, word):
        """Search for a word in the trie.
        Time: O(m)"""
        node = self.root
        for char in word:
            if char not in node.children:
                return False
            node = node.children[char]
        return node.is_end
    
    def starts_with(self, prefix):
        """Check if any word starts with the prefix.
        Time: O(m)"""
        node = self.root
        for char in prefix:
            if char not in node.children:
                return False
            node = node.children[char]
        return True


# Example
trie = Trie()
words = ['apple', 'app', 'application', 'bat', 'ball']
for word in words:
    trie.insert(word)
print(f"Search 'apple': {trie.search('apple')}")
print(f"Search 'app': {trie.search('app')}")
print(f"Search 'ap': {trie.search('ap')}")
print(f"Starts with 'app': {trie.starts_with('app')}")
print(f"Starts with 'bat': {trie.starts_with('bat')}")
```

## 2. Word Search

```python
def word_search_2d(board, word):
    """Search for word in 2D board (LeetCode 79).
    Time: O(m * n * 4^l) where l is word length"""
    if not board or not word:
        return False
    
    rows, cols = len(board), len(board[0])
    
    def dfs(r, c, index):
        if index == len(word):
            return True
        if (r < 0 or r >= rows or c < 0 or c >= cols or
                board[r][c] != word[index]):
            return False
        
        temp = board[r][c]
        board[r][c] = '#'  # Mark as visited
        
        found = (dfs(r + 1, c, index + 1) or
                 dfs(r - 1, c, index + 1) or
                 dfs(r, c + 1, index + 1) or
                 dfs(r, c - 1, index + 1))
        
        board[r][c] = temp  # Restore
        return found
    
    for r in range(rows):
        for c in range(cols):
            if dfs(r, c, 0):
                return True
    
    return False


# Example
board = [
    ['A', 'B', 'C', 'E'],
    ['S', 'F', 'C', 'S'],
    ['A', 'D', 'E', 'E']
]
print(f"Search 'ABCCED': {word_search_2d(board, 'ABCCED')}")
print(f"Search 'SEE': {word_search_2d(board, 'SEE')}")
print(f"Search 'ABCB': {word_search_2d(board, 'ABCB')}")
```

## 3. Prefix Operations

```python
def find_words(board, words):
    """Find all words from dictionary in board (LeetCode 212).
    Time: O(m * n * 4 * 3^(l-1))"""
    trie = Trie()
    for word in words:
        trie.insert(word)
    
    rows, cols = len(board), len(board[0])
    result = set()
    
    def dfs(r, c, node, word):
        if node.is_end:
            result.add(word)
        
        if (r < 0 or r >= rows or c < 0 or c >= cols or
                board[r][c] not in node.children):
            return
        
        temp = board[r][c]
        board[r][c] = '#'
        
        for dr, dc in [(0, 1), (1, 0), (0, -1), (-1, 0)]:
            dfs(r + dr, c + dc, node.children[temp], word + temp)
        
        board[r][c] = temp
    
    for r in range(rows):
        for c in range(cols):
            dfs(r, c, trie.root, '')
    
    return list(result)


# Example
board = [
    ['o', 'a', 'a', 'n'],
    ['e', 't', 'a', 'e'],
    ['i', 'h', 'k', 'r'],
    ['i', 'f', 'l', 'v']
]
words = ['oath', 'pea', 'eat', 'rain']
print(f"Words found: {find_words(board, words)}")
```

## 4. Autocomplete System

```python
class AutocompleteSystem:
    """Autocomplete system using Trie."""
    
    def __init__(self, sentences, times):
        self.trie = Trie()
        self.counts = {}
        self.current = ''
        
        for sentence, time in zip(sentences, times):
            self.trie.insert(sentence)
            self.counts[sentence] = time
    
    def input(self, c):
        """Input a character and return top 3 suggestions."""
        if c == '#':
            self.trie.insert(self.current)
            self.counts[self.current] = self.counts.get(self.current, 0) + 1
            self.current = ''
            return []
        
        self.current += c
        suggestions = []
        node = self.trie.root
        
        for char in self.current:
            if char not in node.children:
                return suggestions
            node = node.children[char]
        
        self._collect_words(node, self.current, suggestions)
        suggestions.sort(key=lambda x: (-self.counts.get(x, 0), x))
        return suggestions[:3]
    
    def _collect_words(self, node, prefix, result):
        if node.is_end:
            result.append(prefix)
        for char, child in node.children.items():
            self._collect_words(child, prefix + char, result)


# Example
ac = AutocompleteSystem(['i love you', 'island', 'iron', 'i love leetcode'], [5, 3, 2, 2])
print(f"Input 'i': {ac.input('i')}")
print(f"Input ' ': {ac.input(' ')}")
print(f"Input 'a': {ac.input('a')}")
```
