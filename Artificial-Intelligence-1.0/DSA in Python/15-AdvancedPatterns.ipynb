## 1. Dutch National Flag (3-Way Partition)

```python
def sort_colors(nums):
    """Sort 0s, 1s, and 2s (LeetCode 75).
    Time: O(n), Space: O(1)"""
    low = mid = 0
    high = len(nums) - 1
    
    while mid <= high:
        if nums[mid] == 0:
            nums[low], nums[mid] = nums[mid], nums[low]
            low += 1
            mid += 1
        elif nums[mid] == 1:
            mid += 1
        else:
            nums[mid], nums[high] = nums[high], nums[mid]
            high -= 1
    
    return nums


def three_way_partition(arr, low_val, high_val):
    """Partition array around range [low_val, high_val]."""
    low = 0
    mid = 0
    high = len(arr) - 1
    
    while mid <= high:
        if arr[mid] < low_val:
            arr[low], arr[mid] = arr[mid], arr[low]
            low += 1
            mid += 1
        elif arr[mid] <= high_val:
            mid += 1
        else:
            arr[mid], arr[high] = arr[high], arr[mid]
            high -= 1
    
    return arr


# Examples
print(f"Sort colors [2,0,2,1,1,0]: {sort_colors([2, 0, 2, 1, 1, 0])}")
print(f"Three-way partition [3,1,4,1,5,9,2,6]: {three_way_partition([3, 1, 4, 1, 5, 9, 2, 6], 2, 5)}")
```

## 2. Two Pointers - Middle Finding

```python
def find_middle_of_linked_list(head):
    """Find middle of linked list using slow-fast pointers.
    Time: O(n)"""
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
    return slow


def find_peak_element(nums):
    """Find peak element (LeetCode 162).
    Time: O(log n) using binary search"""
    left, right = 0, len(nums) - 1
    
    while left < right:
        mid = (left + right) // 2
        if nums[mid] < nums[mid + 1]:
            left = mid + 1
        else:
            right = mid
    
    return left


def find_duplicate_number(nums):
    """Find duplicate number (LeetCode 287).
    Time: O(n) using Floyd's cycle detection"""
    slow = fast = nums[0]
    
    # Detect cycle
    while True:
        slow = nums[slow]
        fast = nums[nums[fast]]
        if slow == fast:
            break
    
    # Find cycle start
    slow = nums[0]
    while slow != fast:
        slow = nums[slow]
        fast = nums[fast]
    
    return slow


# Examples
print(f"Peak element [1,2,3,1]: {find_peak_element([1, 2, 3, 1])}")
print(f"Duplicate in [1,3,4,2,2]: {find_duplicate_number([1, 3, 4, 2, 2])}")
```

## 3. Cycle Detection in Arrays

```python
def find_cycle_in_array(arr):
    """Find cycle in array using Floyd's algorithm."""
    slow = fast = 0
    
    # Detect cycle
    while True:
        slow = arr[slow]
        fast = arr[arr[fast]]
        if slow == fast:
            break
    
    # Find cycle start
    slow = 0
    while slow != fast:
        slow = arr[slow]
        fast = arr[fast]
    
    # Find cycle length
    length = 1
    fast = arr[slow]
    while fast != slow:
        fast = arr[fast]
        length += 1
    
    return slow, length


def circular_array_loop(nums):
    """Check if array has circular loop (LeetCode 457)."""
    n = len(nums)
    
    def next_index(i):
        return (i + nums[i]) % n
    
    for i in range(n):
        if nums[i] == 0:
            continue
        
        slow = i
        fast = i
        
        while True:
            slow = next_index(slow)
            fast = next_index(next_index(fast))
            
            if slow == fast:
                # Check if cycle length > 1 and same direction
                if slow == next_index(slow):
                    break
                return True
            
            if nums[slow] * nums[next_index(slow)] < 0:
                break
    
    return False


# Examples
arr = [1, 2, 3, 4, 5, 2]
cycle_start, cycle_length = find_cycle_in_array(arr)
print(f"Cycle starts at: {cycle_start}, length: {cycle_length}")
print(f"Circular array loop [2,-1,1,2,2]: {circular_array_loop([2, -1, 1, 2, 2])}")
```

## 4. Meeting Point Problems

```python
def min_meeting_rooms(intervals):
    """Find minimum meeting rooms required (LeetCode 253).
    Time: O(n log n)"""
    if not intervals:
        return 0
    
    starts = sorted(i[0] for i in intervals)
    ends = sorted(i[1] for i in intervals)
    
    rooms = 0
    end_ptr = 0
    
    for start in starts:
        if start < ends[end_ptr]:
            rooms += 1
        else:
            end_ptr += 1
    
    return rooms


def max_meeting_in_one_room(meetings):
    """Find maximum meetings that can be attended in one room."""
    # Sort by end time
    meetings.sort(key=lambda x: x[1])
    
    count = 1
    last_end = meetings[0][1]
    
    for i in range(1, len(meetings)):
        if meetings[i][0] >= last_end:
            count += 1
            last_end = meetings[i][1]
    
    return count


# Examples
intervals = [[0, 30], [5, 10], [15, 20]]
print(f"Meeting rooms needed: {min_meeting_rooms(intervals)}")

meetings = [(1, 4), (3, 5), (0, 6), (5, 7), (3, 8)]
print(f"Max meetings: {max_meeting_in_one_room(meetings)}")
```

## 5. Advanced Two Pointer Problems

```python
def trap_rain_water(height):
    """Trapping rain water (LeetCode 42).
    Time: O(n)"""
    if not height:
        return 0
    
    left, right = 0, len(height) - 1
    left_max, right_max = height[left], height[right]
    water = 0
    
    while left < right:
        if left_max < right_max:
            left += 1
            left_max = max(left_max, height[left])
            water += left_max - height[left]
        else:
            right -= 1
            right_max = max(right_max, height[right])
            water += right_max - height[right]
    
    return water


def container_with_most_water(height):
    """Container with most water (LeetCode 11).
    Time: O(n)"""
    left, right = 0, len(height) - 1
    max_area = 0
    
    while left < right:
        area = min(height[left], height[right]) * (right - left)
        max_area = max(max_area, area)
        
        if height[left] < height[right]:
            left += 1
        else:
            right -= 1
    
    return max_area


# Examples
height = [0, 1, 0, 2, 1, 0, 1, 3, 2, 1, 2, 1]
print(f"Trapped water: {trap_rain_water(height)}")
print(f"Container with most water: {container_with_most_water(height)}")
```
