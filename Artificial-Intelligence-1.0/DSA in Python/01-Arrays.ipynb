## 1. Prefix Sum

Prefix sum allows O(1) range sum queries after O(n) preprocessing.

**Use Case:** Find sum of elements between indices i and j efficiently.

```python
def build_prefix_sum(arr):
    """Build prefix sum array."""
    n = len(arr)
    prefix = [0] * (n + 1)
    for i in range(n):
        prefix[i + 1] = prefix[i] + arr[i]
    return prefix


def range_sum(prefix, left, right):
    """Get sum of arr[left..right] in O(1)."""
    return prefix[right + 1] - prefix[left]


# Example
arr = [2, 4, 6, 8, 10]
prefix = build_prefix_sum(arr)
print(f"Array: {arr}")
print(f"Prefix sum: {prefix}")
print(f"Sum of arr[1..3] = {range_sum(prefix, 1, 3)}")  # 4+6+8 = 18
print(f"Sum of arr[0..4] = {range_sum(prefix, 0, 4)}")  # 2+4+6+8+10 = 30
```

### Prefix Sum 2D (Matrix)

```python
def build_prefix_sum_2d(matrix):
    """Build 2D prefix sum for matrix region sum queries."""
    rows = len(matrix)
    cols = len(matrix[0])
    prefix = [[0] * (cols + 1) for _ in range(rows + 1)]
    
    for i in range(rows):
        for j in range(cols):
            prefix[i + 1][j + 1] = (
                matrix[i][j]
                + prefix[i][j + 1]
                + prefix[i + 1][j]
                - prefix[i][j]
            )
    return prefix


def range_sum_2d(prefix, row1, col1, row2, col2):
    """Get sum of submatrix from (row1,col1) to (row2,col2)."""
    return (
        prefix[row2 + 1][col2 + 1]
        - prefix[row1][col2 + 1]
        - prefix[row2 + 1][col1]
        + prefix[row1][col1]
    )


# Example
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
prefix_2d = build_prefix_sum_2d(matrix)
print("Matrix:")
for row in matrix:
    print(row)
print(f"\nSum of submatrix (0,0) to (1,1) = {range_sum_2d(prefix_2d, 0, 0, 1, 1)}")  # 1+2+4+5 = 12
```

## 2. Two Pointers

Two pointers technique uses two indices to traverse the array, typically from opposite ends or same direction.

**Use Cases:** Pair sum, remove duplicates, merge sorted arrays.

```python
def two_sum_sorted(arr, target):
    """Find two numbers that sum to target in sorted array.
    Time: O(n), Space: O(1)"""
    left, right = 0, len(arr) - 1
    while left < right:
        current_sum = arr[left] + arr[right]
        if current_sum == target:
            return [left, right]
        elif current_sum < target:
            left += 1
        else:
            right -= 1
    return []


# Example
arr = [1, 2, 3, 4, 5, 6, 7]
target = 9
result = two_sum_sorted(arr, target)
print(f"Array: {arr}")
print(f"Target: {target}")
print(f"Indices: {result} -> Values: {arr[result[0]]} + {arr[result[1]]} = {target}")
```

```python
def remove_duplicates(arr):
    """Remove duplicates in-place from sorted array.
    Time: O(n), Space: O(1)"""
    if not arr:
        return 0
    
    write = 0
    for read in range(1, len(arr)):
        if arr[read] != arr[write]:
            write += 1
            arr[write] = arr[read]
    
    return write + 1


# Example
arr = [1, 1, 2, 2, 3, 4, 4, 5]
print(f"Original: {arr}")
new_length = remove_duplicates(arr)
print(f"After removing duplicates: {arr[:new_length]} (length: {new_length})")
```

```python
def merge_sorted(arr1, arr2):
    """Merge two sorted arrays.
    Time: O(n+m), Space: O(n+m)"""
    result = []
    i = j = 0
    
    while i < len(arr1) and j < len(arr2):
        if arr1[i] <= arr2[j]:
            result.append(arr1[i])
            i += 1
        else:
            result.append(arr2[j])
            j += 1
    
    result.extend(arr1[i:])
    result.extend(arr2[j:])
    return result


# Example
arr1 = [1, 3, 5, 7]
arr2 = [2, 4, 6, 8]
print(f"Array 1: {arr1}")
print(f"Array 2: {arr2}")
print(f"Merged: {merge_sorted(arr1, arr2)}")
```

## 3. Kadane's Algorithm (Maximum Subarray Sum)

Finds the contiguous subarray with the largest sum.

**Time:** O(n) | **Space:** O(1)

```python
def kadane(arr):
    """Find maximum subarray sum using Kadane's algorithm."""
    max_sum = arr[0]
    current_sum = arr[0]
    start = end = temp_start = 0
    
    for i in range(1, len(arr)):
        if current_sum + arr[i] < arr[i]:
            current_sum = arr[i]
            temp_start = i
        else:
            current_sum += arr[i]
        
        if current_sum > max_sum:
            max_sum = current_sum
            start = temp_start
            end = i
    
    return max_sum, arr[start:end + 1]


# Example
arr = [-2, 1, -3, 4, -1, 2, 1, -5, 4]
max_sum, subarray = kadane(arr)
print(f"Array: {arr}")
print(f"Maximum subarray sum: {max_sum}")
print(f"Subarray: {subarray}")
```

```python
def kadane_circular(arr):
    """Find maximum circular subarray sum."""
    # Case 1: Maximum subarray is not circular
    max_kadane = kadane(arr)[0]
    
    # Case 2: Maximum subarray is circular
    # Find minimum subarray sum and subtract from total
    total_sum = sum(arr)
    min_sum = 0
    current_min = 0
    
    for num in arr:
        current_min = min(num, current_min + num)
        min_sum = min(min_sum, current_min)
    
    max_circular = total_sum - min_sum
    
    # If all elements are negative, max_kadane is the answer
    if max_circular == 0:
        return max_kadane
    
    return max(max_kadane, max_circular)


# Example
arr = [5, -3, 5]
print(f"Array: {arr}")
print(f"Maximum circular subarray sum: {kadane_circular(arr)}")
```

## 4. Sliding Window

Sliding window technique maintains a window of elements and slides it across the array.

**Types:**
- Fixed size window
- Variable size window

```python
def max_sum_subarray_k(arr, k):
    """Find maximum sum of subarray of size k (Fixed Window).
    Time: O(n)"""
    window_sum = sum(arr[:k])
    max_sum = window_sum
    
    for i in range(k, len(arr)):
        window_sum += arr[i] - arr[i - k]
        max_sum = max(max_sum, window_sum)
    
    return max_sum


# Example
arr = [1, 4, 2, 10, 2, 3, 1, 0, 20]
k = 4
print(f"Array: {arr}")
print(f"Window size: {k}")
print(f"Maximum sum: {max_sum_subarray_k(arr, k)}")
```

```python
def longest_subarray_sum_k(arr, k):
    """Find length of longest subarray with sum = k.
    Time: O(n) using prefix sum + hash map"""
    prefix_sum = 0
    max_length = 0
    sum_map = {0: -1}  # prefix_sum -> index
    
    for i in range(len(arr)):
        prefix_sum += arr[i]
        
        if prefix_sum - k in sum_map:
            max_length = max(max_length, i - sum_map[prefix_sum - k])
        
        if prefix_sum not in sum_map:
            sum_map[prefix_sum] = i
    
    return max_length


# Example
arr = [1, -1, 5, -2, 3]
k = 3
print(f"Array: {arr}")
print(f"Target sum: {k}")
print(f"Length of longest subarray: {longest_subarray_sum_k(arr, k)}")
```

```python
def min_window_substring(s, t):
    """Find minimum window substring containing all characters of t.
    Time: O(n)"""
    from collections import Counter
    
    if not t or not s:
        return ""
    
    dict_t = Counter(t)
    required = len(dict_t)
    
    # Filter s to contain only characters in t
    l, r = 0, 0
    formed = 0
    window_counts = {}
    
    ans = float("inf"), None, None
    
    while r < len(s):
        character = s[r]
        window_counts[character] = window_counts.get(character, 0) + 1
        
        if character in dict_t and window_counts[character] == dict_t[character]:
            formed += 1
        
        while l <= r and formed == required:
            character = s[l]
            
            if r - l + 1 < ans[0]:
                ans = (r - l + 1, l, r)
            
            window_counts[character] -= 1
            if character in dict_t and window_counts[character] < dict_t[character]:
                formed -= 1
            
            l += 1
        
        r += 1
    
    return "" if ans[0] == float("inf") else s[ans[1]:ans[2] + 1]


# Example
s = "ADOBECODEBANC"
t = "ABC"
print(f"String: {s}")
print(f"Pattern: {t}")
print(f"Minimum window substring: {min_window_substring(s, t)}")
```

## 5. Dutch National Flag (3-way Partitioning)

Sort an array of 0s, 1s, and 2s in a single pass.

**Use Case:** Sort array with three distinct values.

```python
def dutch_national_flag(arr):
    """Sort array of 0s, 1s, and 2s in-place.
    Time: O(n), Space: O(1)"""
    low = 0
    mid = 0
    high = len(arr) - 1
    
    while mid <= high:
        if arr[mid] == 0:
            arr[low], arr[mid] = arr[mid], arr[low]
            low += 1
            mid += 1
        elif arr[mid] == 1:
            mid += 1
        else:  # arr[mid] == 2
            arr[mid], arr[high] = arr[high], arr[mid]
            high -= 1
    
    return arr


# Example
arr = [2, 0, 1, 2, 0, 1, 2, 0, 1]
print(f"Before sorting: {arr}")
print(f"After sorting:  {dutch_national_flag(arr.copy())}")
```

## 6. Sorting Algorithms

```python
def quicksort(arr):
    """Quick Sort implementation.
    Average Time: O(n log n), Space: O(log n)"""
    if len(arr) <= 1:
        return arr
    
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    
    return quicksort(left) + middle + quicksort(right)


def mergesort(arr):
    """Merge Sort implementation.
    Time: O(n log n), Space: O(n)"""
    if len(arr) <= 1:
        return arr
    
    mid = len(arr) // 2
    left = mergesort(arr[:mid])
    right = mergesort(arr[mid:])
    
    return merge(left, right)


def merge(left, right):
    """Merge two sorted arrays."""
    result = []
    i = j = 0
    
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    
    result.extend(left[i:])
    result.extend(right[j:])
    return result


# Example
arr = [38, 27, 43, 3, 9, 82, 10]
print(f"Original: {arr}")
print(f"Quick Sort: {quicksort(arr)}")
print(f"Merge Sort: {mergesort(arr)}")
```

## 7. Important Array Problems

```python
def find_missing_number(arr):
    """Find missing number from 0 to n.
    Time: O(n), Space: O(1)"""
    n = len(arr)
    expected_sum = n * (n + 1) // 2
    actual_sum = sum(arr)
    return expected_sum - actual_sum


def find_duplicate(arr):
    """Find duplicate number in array of n+1 integers (1 to n).
    Time: O(n), Space: O(1) - Floyd's Cycle Detection"""
    slow = fast = arr[0]
    
    # Detect cycle
    while True:
        slow = arr[slow]
        fast = arr[arr[fast]]
        if slow == fast:
            break
    
    # Find entrance to cycle
    slow = arr[0]
    while slow != fast:
        slow = arr[slow]
        fast = arr[fast]
    
    return slow


def move_zeroes(arr):
    """Move all zeroes to end of array while maintaining order."""
    write = 0
    for read in range(len(arr)):
        if arr[read] != 0:
            arr[write], arr[read] = arr[read], arr[write]
            write += 1
    return arr


# Examples
print(f"Missing number in [3,0,1]: {find_missing_number([3, 0, 1])}")
print(f"Duplicate in [1,3,4,2,2]: {find_duplicate([1, 3, 4, 2, 2])}")
print(f"Move zeroes [0,1,0,3,12]: {move_zeroes([0, 1, 0, 3, 12])}")
```
