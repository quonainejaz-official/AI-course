# Linked List - DSA Patterns

Linked List is a linear data structure where elements are stored in nodes, each pointing to the next node.

## Topics Covered

3. Cycle Detection
4. Merge Sorted Lists

## 1. Singly Linked List Implementation

```python
class ListNode:
    """Definition for singly-linked list node."""
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


class LinkedList:
    """Singly linked list implementation."""
    
    def __init__(self):
        self.head = None
    
    def append(self, val):
        """Add node at end."""
        new_node = ListNode(val)
        if not self.head:
            self.head = new_node
            return
        current = self.head
        while current.next:
            current = current.next
        current.next = new_node
    
    def prepend(self, val):
        """Add node at beginning."""
        new_node = ListNode(val)
        new_node.next = self.head
        self.head = new_node
    
    def display(self):
        """Print the linked list."""
        elements = []
        current = self.head
        while current:
            elements.append(str(current.val))
            current = current.next
        print(' -> '.join(elements))


# Example
ll = LinkedList()
for val in [1, 2, 3, 4, 5]:
    ll.append(val)
print("Linked List:")
ll.display()
```

## 2. Fast-Slow Pointers (Floyd's Algorithm)

```python
def has_cycle(head):
    """Detect cycle in linked list (LeetCode 141).
    Time: O(n), Space: O(1)"""
    if not head or not head.next:
        return False
    
    slow = head
    fast = head.next
    
    while slow != fast:
        if not fast or not fast.next:
            return False
        slow = slow.next
        fast = fast.next.next
    
    return True


def detect_cycle_start(head):
    """Find where cycle begins (LeetCode 142).
    Time: O(n), Space: O(1)"""
    if not head or not head.next:
        return None
    
    slow = fast = head
    
    # Detect cycle
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        if slow == fast:
            break
    else:
        return None
    
    # Find cycle start
    slow = head
    while slow != fast:
        slow = slow.next
        fast = fast.next
    
    return slow


def middle_of_linked_list(head):
    """Find middle node (LeetCode 876).
    Time: O(n)"""
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
    return slow


# Example
head = ListNode(1)
head.next = ListNode(2)
head.next.next = ListNode(3)
head.next.next.next = ListNode(4)
print(f"Has cycle: {has_cycle(head)}")
print(f"Middle node value: {middle_of_linked_list(head).val}")
```

## 3. Reverse Linked List

```python
def reverse_list(head):
    """Reverse linked list iteratively (LeetCode 206).
    Time: O(n), Space: O(1)"""
    prev = None
    current = head
    
    while current:
        next_temp = current.next
        current.next = prev
        prev = current
        current = next_temp
    
    return prev


def reverse_list_recursive(head):
    """Reverse linked list recursively.
    Time: O(n), Space: O(n) for recursion stack"""
    if not head or not head.next:
        return head
    
    new_head = reverse_list_recursive(head.next)
    head.next.next = head
    head.next = None
    
    return new_head


# Helper function to create linked list from array
def create_list(arr):
    if not arr:
        return None
    head = ListNode(arr[0])
    current = head
    for val in arr[1:]:
        current.next = ListNode(val)
        current = current.next
    return head


# Helper function to convert linked list to array
def to_array(head):
    result = []
    while head:
        result.append(head.val)
        head = head.next
    return result


# Example
head = create_list([1, 2, 3, 4, 5])
print(f"Original: {to_array(head)}")
reversed_head = reverse_list(head)
print(f"Reversed: {to_array(reversed_head)}")
```

## 4. Merge Sorted Lists

```python
def merge_two_sorted_lists(l1, l2):
    """Merge two sorted linked lists (LeetCode 21).
    Time: O(n + m)"""
    dummy = ListNode(0)
    current = dummy
    
    while l1 and l2:
        if l1.val <= l2.val:
            current.next = l1
            l1 = l1.next
        else:
            current.next = l2
            l2 = l2.next
        current = current.next
    
    current.next = l1 or l2
    return dummy.next


# Example
l1 = create_list([1, 2, 4])
l2 = create_list([1, 3, 4])
merged = merge_two_sorted_lists(l1, l2)
print(f"List 1: [1, 2, 4]")
print(f"List 2: [1, 3, 4]")
print(f"Merged: {to_array(merged)}")
```

## 5. Advanced Linked List Problems

```python
def remove_nth_from_end(head, n):
    """Remove nth node from end (LeetCode 19).
    Time: O(L) using two pointers"""
    dummy = ListNode(0)
    dummy.next = head
    fast = slow = dummy
    
    # Move fast pointer n steps ahead
    for _ in range(n + 1):
        fast = fast.next
    
    # Move both pointers until fast reaches end
    while fast:
        fast = fast.next
        slow = slow.next
    
    # Remove the node
    slow.next = slow.next.next
    return dummy.next


def add_two_numbers(l1, l2):
    """Add two numbers represented as linked lists (LeetCode 2)."""
    dummy = ListNode(0)
    current = dummy
    carry = 0
    
    while l1 or l2 or carry:
        val = carry
        if l1:
            val += l1.val
            l1 = l1.next
        if l2:
            val += l2.val
            l2 = l2.next
        carry, digit = divmod(val, 10)
        current.next = ListNode(digit)
        current = current.next
    
    return dummy.next


# Example
l1 = create_list([2, 4, 3])
l2 = create_list([5, 6, 4])
result = add_two_numbers(l1, l2)
print(f"342 + 465 = {to_array(result)}")
```
