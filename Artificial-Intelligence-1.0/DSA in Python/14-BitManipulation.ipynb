# Bit Manipulation - DSA Patterns

Bit manipulation involves directly manipulating bits of numbers for efficient computation.

## 1. Basic Bit Operations

```python
# AND (&) - both bits must be 1
# OR (|) - at least one bit must be 1
# XOR (^) - bits must be different
# NOT (~) - flip all bits
# Left Shift (<<) - multiply by 2
# Right Shift (>>) - divide by 2

a, b = 5, 3  # 101, 011

print(f"a = {a} ({bin(a)})")
print(f"b = {b} ({bin(b)})")
print(f"a & b = {a & b} ({bin(a & b)})")
print(f"a | b = {a | b} ({bin(a | b)})")
print(f"a ^ b = {a ^ b} ({bin(a ^ b)})")
print(f"~a = {~a}")
print(f"a << 2 = {a << 2} ({bin(a << 2)})")
print(f"a >> 1 = {a >> 1} ({bin(a >> 1)})")
```

## 2. Swap Two Bits

```python
def swap_bits(n):
    """Swap even and odd position bits.
    Example: 1010 -> 0101"""
    even_bits = n & 0xAAAAAAAA  # Extract even positions
    odd_bits = n & 0x55555555   # Extract odd positions
    
    even_bits >>= 1
    odd_bits <<= 1
    
    return even_bits | odd_bits


def swap_bits_at_positions(n, i, j):
    """Swap bits at positions i and j."""
    bit_i = (n >> i) & 1
    bit_j = (n >> j) & 1
    
    if bit_i != bit_j:
        n ^= (1 << i) | (1 << j)
    
    return n


# Examples
n = 0b1010  # 10
print(f"Original: {bin(n)}")
print(f"Swapped even/odd: {bin(swap_bits(n))}")
print(f"Swap bits 0 and 3 of {bin(n)}: {bin(swap_bits_at_positions(n, 0, 3))}")
```

## 3. Count Set Bits

```python
def count_set_bits(n):
    """Count number of 1 bits (Brian Kernighan's algorithm).
    Time: O(number of set bits)"""
    count = 0
    while n:
        n &= (n - 1)  # Clear lowest set bit
        count += 1
    return count


def count_set_bits_range(n):
    """Count set bits from 0 to n."""
    count = 0
    for i in range(n + 1):
        count += count_set_bits(i)
    return count


def is_power_of_two(n):
    """Check if n is power of 2 (LeetCode 231)."""
    return n > 0 and (n & (n - 1)) == 0


# Examples
print(f"Set bits in 13 ({bin(13)}): {count_set_bits(13)}")
print(f"Set bits from 0 to 13: {count_set_bits_range(13)}")
print(f"Is 16 power of 2? {is_power_of_two(16)}")
print(f"Is 18 power of 2? {is_power_of_two(18)}")
```

## 4. Single Number Problems

```python
def single_number(nums):
    """Find number appearing once (LeetCode 136).
    Time: O(n), Space: O(1)"""
    result = 0
    for num in nums:
        result ^= num
    return result


def single_number_ii(nums):
    """Find number appearing once when others appear three times (LeetCode 137).
    Time: O(n)"""
    ones = twos = 0
    
    for num in nums:
        ones = (ones ^ num) & ~twos
        twos = (twos ^ num) & ~ones
    
    return ones


def two_single_numbers(nums):
    """Find two numbers appearing once (LeetCode 260).
    Time: O(n)"""
    xor = 0
    for num in nums:
        xor ^= num
    
    # Find rightmost set bit
    diff_bit = xor & (-xor)
    
    a = 0
    for num in nums:
        if num & diff_bit:
            a ^= num
    
    return [a, xor ^ a]


# Examples
print(f"Single number [2,2,1]: {single_number([2, 2, 1])}")
print(f"Single number II [2,2,3,2]: {single_number_ii([2, 2, 3, 2])}")
print(f"Two single numbers [1,2,1,3,2,5]: {two_single_numbers([1, 2, 1, 3, 2, 5])}")
```

## 5. Bit Masking

```python
def subsets_bitmask(nums):
    """Generate all subsets using bit masking."""
    n = len(nums)
    result = []
    
    for mask in range(1 << n):  # 2^n subsets
        subset = []
        for i in range(n):
            if mask & (1 << i):
                subset.append(nums[i])
        result.append(subset)
    
    return result


def check_duplicate_exists(arr):
    """Check if array has duplicates using bit manipulation."""
    seen = 0
    for num in arr:
        if seen & (1 << num):
            return True
        seen |= (1 << num)
    return False


def power_set(nums):
    """Generate power set using bit manipulation."""
    n = len(nums)
    result = []
    
    for i in range(1 << n):
        subset = [nums[j] for j in range(n) if i & (1 << j)]
        result.append(subset)
    
    return result


# Examples
print(f"Subsets of [1,2,3] using bitmask: {subsets_bitmask([1, 2, 3])}")
print(f"Has duplicates [1,2,3]: {check_duplicate_exists([1, 2, 3])}")
print(f"Has duplicates [1,2,2]: {check_duplicate_exists([1, 2, 2])}")
```
