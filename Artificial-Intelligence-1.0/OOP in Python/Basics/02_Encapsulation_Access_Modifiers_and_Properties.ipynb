## 3. Name Mangling — Proof**What is:** When class inside `self.__x` write, Python compile time par it`self._ClassName__x` in convert do gives is.
````python
class Demo2:
    def __init__(self):
        self.__secret = 42  # Python isay inside hi inside '_Demo2__secret' bana dega

d2 = Demo2()
print(d2.__dict__)  # see: key '_Demo2__secret' is, '__secret' no

try:
    print(d2.__secret)  # this fail will be - is naam of attribute exist hi no does (mangled be chuka is)
except AttributeError as e:
    print("AttributeError pakड़ liya:", e)

# but mangled naam from access be goes - proof these 'real' private no, only naam badla isprint("Mangled naam from access:", d2._Demo2__secret)
assert d2._Demo2__secret == 42
````

## 5. Classic Getter/Setter (Java-style) — and These Pythonic why no**What is:** Java/C++ in tradition is of every private field for `getX()` and `setX(value)`methods write.**Syntax (Python in also write can be, but...):**

````python
class Employee:
    def __init__(self, salary):
        self.__salary = salary

    def get_salary(self):          # classic getter
        return self.__salary

    def set_salary(self, value):   # classic setter
        if value < 0:
            raise ValueError("salary negative no be can")
        self.__salary = value

e = Employee(50000)
print(e.get_salary())
e.set_salary(60000)
print(e.get_salary())
try:
    e.set_salary(-100)
except ValueError as ex:
    print("ValueError:", ex)
````

**These Pythonic why no:**- Caller to `e.salary` jaisi natural attribute syntax not milti, `e.get_salary()` then writehas then — verbose and "un-Pythonic".- if kal aap plain attribute from property in switch then do chahein, syntax largel jathesegaand sara calling code thendna padega.- Python of jawnow: **`@property`** — jisse method attribute like dikhta is, but validationalso can do. ## 6. `@property` — Getter**What is:** `@property` decorator ek method to "computed attribute" create gives is — calldo time parenthesis `()` not lagane padte.**Syntax:**```pythonclass C:@propertydef x(self):return self.__x```**where use is:** When chahte be of attribute like access be (`obj.x`), but peechelogic/validation/computation chal rahi be.

````python
class Circle:
    def __init__(self, radius):
        self.__radius = radius

    @property
    def radius(self):     # ab 'radius' ek property is, method no lagta bahar from
        return self.__radius

c = Circle(5)
print(c.radius)          # note: c.radius, NOT c.radius() - parenthesis no
assert c.radius == 5
````

## 7. `@x.setter` — Validation of Sath**What is:** `@property` with `@<name>.setter` decorator laga do hum assignment(`obj.x = value`) to also intercept can do and validation laga can are.**Syntax:**```python@x.setterdef x(self, value):# validationself.__x = value```**Gotcha:** Setter inside `self.<same_property_name> = value` mat then write — isse infiniterecursion be jathesegi (these hum Common Mistakes section in dikhathesenge). Hamesha mangled/privateattribute (`self.__x`) in store do.

````python
class Circle2:
    def __init__(self, radius):
        self.radius = radius  # yahan constructor me hi setter call be raha is (validation of sath)

    @property
    def radius(self):
        return self.__radius

    @radius.setter
    def radius(self, value):
        if value <= 0:
            raise ValueError("radius positive hona should")
        self.__radius = value

c2 = Circle2(3)
print(c2.radius)
c2.radius = 10           # normal attribute syntax, lekin peeche validation chal rahi is
print(c2.radius)
try:
    c2.radius = -1        # validation fail will be
except ValueError as e:
    print("ValueError:", e)
assert c2.radius == 10
````

## 8. `@x.deleter`**What is:** `del obj.x` chalane par custhenm logic run doing for `@<name>.deleter`decorator use is.**where use is:** Cleanup logic, logging, or attribute reset then do when `del` call be.

````python
class Circle3:
    def __init__(self, radius):
        self.radius = radius

    @property
    def radius(self):
        return self.__radius

    @radius.setter
    def radius(self, value):
        if value <= 0:
            raise ValueError("radius positive hona should")
        self.__radius = value

    @radius.deleter
    def radius(self):
        print("Radius delete be raha is...")   # cleanup / logging yahan kar can are
        del self.__radius

c3 = Circle3(4)
del c3.radius   # deleter call will be
try:
    print(c3.radius)   # ab attribute exist no does
except AttributeError as e:
    print("AttributeError:", e)
````

## 9. Computed / Derived Properties**What is:** property hamesha `self.__x` return not doti — these anyone also expression/computationof result also return can do, that every time access hone par (re)calculate is.**Examples below:** circle of `area`, person of `full_name`, and ek fixed reference date from`age` (note: hum `date.thenday()` use **no** doenge, warna asserts future in fail be increase —hamesha fixed date use do when deterministic output needed).

````python
class CircleFull:
    PI = 3.14159  # constant neeche class-level constants section me detail me dekhenge

    def __init__(self, radius):
        self.radius = radius

    @property
    def area(self):                              # computed property - every call par calculate is is
        return round(CircleFull.PI * self.radius ** 2, 2)

    @property
    def circumference(self):
        return round(2 * CircleFull.PI * self.radius, 2)

cf = CircleFull(10)
print("Area:", cf.area, "| Circumference:", cf.circumference)
assert cf.area == 314.16
assert cf.circumference == 62.83
````

````python
class Person:
    def __init__(self, first, last):
        self.first = first
        self.last = last

    @property
    def full_name(self):                          # do attributes to combine kar of derive did
        return f"{self.first} {self.last}"

p = Person("Ayesha", "Khan")
print(p.full_name)
assert p.full_name == "Ayesha Khan"

p.first = "Sana"       # first badalne from full_name automatically update be goes is
print(p.full_name)
assert p.full_name == "Sana Khan"
````

````python
from datetime import date

class PersonAge:
# FIXED reference date - deterministic output for asal 'aaj of date' no use do rahe    _FIXED_TODAY = date(2026, 1, 1)

    def __init__(self, name, dob):
        self.name = name
        self.dob = dob   # date object

    @property
    def age(self):
        ref = PersonAge._FIXED_TODAY
        years = ref.year - self.dob.year
# if birthday now is saal not aaya (month/day fixed date from aage is) then ek saal ghatao        if (ref.month, ref.day) < (self.dob.month, self.dob.day):
            years -= 1
        return years

pa = PersonAge("Bilal", date(2000, 5, 15))
print("Age (fixed reference 2026-01-01):", pa.age)
assert pa.age == 25   # 2000-05-15 from 2026-01-01 tak - birthday abhi no aaya is saal, is for 25
````

## 10. Read-Only property**What is:** if only `@property` (getter) define do and `@x.setter` absolutely not writeo, then`obj.x = value` doing par Python itself `AttributeError` give dega — without extra code writeeread-only attribute mil goes.**where use is:** IDs, timestamps, computed values that know also outside from set not honeneeded.

````python
class Employee2:
    def __init__(self, emp_id, name):
        self.__emp_id = emp_id     # this kabhi change no hona should
        self.name = name

    @property
    def emp_id(self):
        return self.__emp_id
# NOTE: setter define not done - therefore 'emp_id' authenmatically read-only ban gaya
emp = Employee2("E101", "Sara")
print(emp.emp_id)
try:
    emp.emp_id = "E999"     # setter no is, is for this fail will be
except AttributeError as e:
    print("AttributeError:", e)
assert emp.emp_id == "E101"
````

## 11. `property()` function Form (Purana Style)**What is:** `@property` decorator of aane from first (and aaj also valid is) `property()`built-in function directly use is thi — getter, setter, deleter, and docstring pass doke.**Syntax:**```pythonx = property(fget, fset, fdel, "docstring")```both style **functionally same** are — `@property` only syntactic sugar is.

````python
class Square:
    def __init__(self, side):
        self.__side = side

    def get_side(self):
        return self.__side

    def set_side(self, value):
        if value <= 0:
            raise ValueError("side positive hona should")
        self.__side = value

    def del_side(self):
        del self.__side

# purana style - decorator of bagair, seedha property() function call do of    side = property(get_side, set_side, del_side, "Square of side length")

sq = Square(5)
print(sq.side)
sq.side = 8
print(sq.side)
print("Docstring:", Square.side.__doc__)
assert sq.side == 8
````

## 12. Backwards-Compatible Refacthenr: Plain attribute → property**What is:** These `@property` of allse large practical faida is — if aapne first plainattribute (`self.width`) expose done tha and very sara user code `obj.width = 20` write chukais, then aap **without it code to thende** property in switch can be, whyki syntax absolutelysame rehta is (`obj.width`), only peeche validation add be goes.

````python
# V1: start in simple class, width plain public attribute isclass RectangleV1:
    def __init__(self, width):
        self.width = width

r1 = RectangleV1(10)
r1.width = 20              # purana user code - direct assignment, koi validation no
print("V1 width:", r1.width)

# V2: now humein validation needed (width know negative not be), LEKIN user code thendna notclass RectangleV2:
    def __init__(self, width):
        self.width = width   # this ab property setter call karega (neeche define is)

    @property
    def width(self):
        return self.__width

    @width.setter
    def width(self, value):
        if value <= 0:
            raise ValueError("width positive honi should")
        self.__width = value

r2 = RectangleV2(10)
r2.width = 20     # exactly WAHI syntax jo V1 me tha - some no tuta, validation transparently add be gayi
print("V2 width:", r2.width)
try:
    r2.width = -5
except ValueError as e:
    print("ValueError:", e)
assert r2.width == 20
````

## 13. Data Validation Patterns + Custhenm exception Classes**What is:** property setters in validation then write best practice is. if validationdomain-specific is (like "invalid age"), then generic `ValueError` of bajathese **custhenmexception class** createna behtar is — isse calling code specific `except` laga can is anderror messages also more meaningful are.**Syntax:**```pythonclass MyError(exception):pass```

````python
class InvalidAgeError(exception):
# Custhenm exception - when age invalid be (negative, or range from outside, or wrong type)    pass

class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age   # setter call will be

    @property
    def age(self):
        return self.__age

    @age.setter
    def age(self, value):
        if not isinstance(value, int):
            raise InvalidAgeError(f"Age integer honi should, mila: {type(value).__name__}")
        if value < 0 or value > 120:
            raise InvalidAgeError(f"Age valid range (0-120) me no is: {value}")
        self.__age = value

s = Student("Ahmed", 20)
print(s.age)
try:
    s.age = -5
except InvalidAgeError as e:
    print("InvalidAgeError:", e)
try:
    s.age = "twenty"
except InvalidAgeError as e:
    print("InvalidAgeError:", e)
assert s.age == 20   # invalid assignments ne asal age no badla
````

````python
class Circle4:
    PI = 3.14159   # class-level constant - all instances share do are, memory bhi bachti is

    def __init__(self, radius):
        self.radius = radius

    @property
    def area(self):
        return round(Circle4.PI * self.radius ** 2, 2)

print("class of through:", Circle4.PI)
c4a = Circle4(2)
c4b = Circle4(3)
print("Instance of through bhi wahi value:", c4a.PI, c4b.PI)
assert Circle4.PI == 3.14159
assert c4a.PI is c4b.PI   # both instances same object to refer kar rahe are (constant shared is)
````

## 15. `__slots__` — Memory Saving and Typo-Protection**What is:** `__slots__` class attribute define doke hum Python to batate are of instanceof paas **only thisi fixed attributes** honge — Python then every instance for `__dict__`not createta, seedha fixed slots allocate does.**why use is:**1. **Memory saving** — bohat sari instances createni be (lakhon objects) then `__dict__` ofoverhead bachta is.2. **Typo protection** — if galti from `obj.nmae = "x"` write do (typo), error mil jathesegainstead of silently naya attribute ban jane of.**Syntax:**```pythonclass C:__slots__ = ("x", "y")```

````python
class PointSlots:
    __slots__ = ("x", "y")   # sirf this do attributes allowed - koi __dict__ no banega

    def __init__(self, x, y):
        self.x = x
        self.y = y

ps = PointSlots(1, 2)
print(ps.x, ps.y)

try:
    ps.z = 100     # typo ya naya attribute add karne of koshish - slots isay rokta is
except AttributeError as e:
    print("AttributeError (typo-protection of fayda):", e)

try:
    print(ps.__dict__)    # slots those objects of __dict__ is hi no
except AttributeError as e:
    print("AttributeError:", e)
````

**Gotcha:** `__slots__` and properties of naam clash not be can — if `__slots__` in`"value"` writea is and class in `value` naam of property also define do, Python classcreatete time only error give dega. Below dikhate are (dynamic `exec` + `try/except` withso that notebook crash not be).

````python
# __slots__ + property same naam - class DEFINITION time par only error comestry:
    exec(
        "class BadSlots:\n"
        "    __slots__ = ('value',)\n"
        "    @property\n"
        "    def value(self):\n"
        "        return self.__value\n"
    )
except ValueError as e:
    print("ValueError (slots + property naam clash):", e)
````

Doosri gotchas:- Inheritance in if parent class `__slots__` use doti is but child no, then child tothen also `__dict__` mil goes (memory benefit end) — both jagah `__slots__` needed.- Default values seedha `__slots__` from not mil can — `__init__` in set doing have then.- Multiple inheritance in multiple non-empty `__slots__` that classes combine then do tricky(layout conflict) be can is.

## Common Mistakes / Gotchas1. **Setter inside itself to call then do (infinite recursion).** Setter inside`self.<same_name> = value` then write bhool from — hamesha private attribute(`self.__name`) in store do, property name in no.2. **Sochna of `__x` security is.** These only naming convention + name mangling is —`obj._ClassName__x` write do any also access/modify can do.3. **`__slots__` + property/class-variable same naam** — class definition time par error.4. **Getter/setter both not hona** when intention read-write be — only `@property` write dobhool jana of `@x.setter` also needed, then confthe sameon is "attribute set why not be raha".5. **property inside heavy computation** every time chalana when woh thing cache be can thi(performance gotcha — if expensive be then `functhenols.cached_property` see).6. **`_protected` to outside from seedha access then do** whyki "Python rokta no" — conventionfollow do, only therefore of chal goes ismeans these not of sahi is.Below do live demos — both safe ways from (`try/except`) dikhathese gathese are so that any crashnot be.

````python
import sys

# MISTAKE #1: setter authenmatically to only call do raha is -> infinite recursionclass BadSetter:
    def __init__(self, value):
        self.value = value   # this setter call karega

    @property
    def value(self):
        return self.__value

    @value.setter
    def value(self, v):
        self.value = v   # BUG: 'self.__value' hona should tha, yahan khud 'value' to hi call kar raha is

old_limit = sys.getrecursionlimit()
sys.setrecursionlimit(120)   # chhota kar dete are taake RecursionError jaldi (and cheap) mil jaye
try:
    bs = BadSetter(10)
except RecursionError:
    print("RecursionError pakड़ liya - common mistake: setter apne aap to dobara call kar raha is")
finally:
    sys.setrecursionlimit(old_limit)   # limit wapis normal kar do
````

````python
# MISTAKE #2: sochna of __x 'security' gives isclass Account:
    def __init__(self, balance):
        self.__balance = balance

    @property
    def balance(self):
        return self.__balance
# any setter not - balance only deposit/withdraw jaisi methods from change hona needed tha
acc2 = Account(100)
acc2._Account__balance = -500   # mangled naam from seedha access karke validation bypass kar diya!
print("Bypass of baad balance:", acc2.balance)   # -500 -> proof of __x security no, convention is
assert acc2.balance == -500
````

## Real-World ProblemsNow chaar realistic examples dekhte are that encapsulation + properties + validation to sahiways from combine do.

### 1. BankAccount — Balance Protection, Deposit/Withdraw with ValidationRequirement: `balance` know also seedha set not be sake (read-only property), only`deposit()`/`withdraw()` of zarithese change be, and invalid operations par custhenm exception aathese.

````python
class InsufficientFundsError(exception):
# Custhenm exception - when withdraw amount available balance from more be    pass

class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.__balance = 0
        if balance > 0:
            self.deposit(balance)   # constructor bhi wahi validated path use does is

    @property
    def balance(self):
        return self.__balance   # read-only - koi setter no

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit amount positive hona should")
        self.__balance += amount
        return self.__balance

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Withdraw amount positive hona should")
        if amount > self.__balance:
            raise InsufficientFundsError(f"Balance {self.__balance} is, {amount} withdraw no be can")
        self.__balance -= amount
        return self.__balance

acc = BankAccount("Imran", 1000)
print("Starting balance:", acc.balance)
acc.deposit(500)
print("Deposit of baad:", acc.balance)
acc.withdraw(200)
print("Withdraw of baad:", acc.balance)

try:
    acc.withdraw(999999)
except InsufficientFundsError as e:
    print("InsufficientFundsError:", e)

try:
    acc.balance = 999999    # property read-only is
except AttributeError as e:
    print("AttributeError:", e)

assert acc.balance == 1300
````

### 2. Temperature — Celsius/Fahrenheit SyncRequirement: `celsius` and `fahrenheit` both properties honi needed that hamesha a fewosre ofwith sync rahen — any also ek badlo, doosra authenmatically update be.

````python
class Temperature:
    def __init__(self, celsius=0):
        self.celsius = celsius   # setter call will be

    @property
    def celsius(self):
        return self.__celsius

    @celsius.setter
    def celsius(self, value):
        if value < -273.15:
            raise ValueError("Absolute zero from neeche temperature possible no")
        self.__celsius = value

    @property
    def fahrenheit(self):
        return round(self.__celsius * 9 / 5 + 32, 2)

    @fahrenheit.setter
    def fahrenheit(self, value):
        self.celsius = (value - 32) * 5 / 9   # dusre setter to call kar of both sync rehte are

t = Temperature(25)
print(t.celsius, "C =", t.fahrenheit, "F")

t.fahrenheit = 98.6
print("Fahrenheit set karne of baad celsius:", round(t.celsius, 2))

try:
    t.celsius = -300
except ValueError as e:
    print("ValueError:", e)

assert t.fahrenheit == 98.6
assert round(t.celsius, 2) == 37.0
````

### 3. User — Password Hashing (Salt + SHA-256)Requirement: Plaintext password **know store no** then do. every user of apna random salt be,and password verify doing for hash compare do, plaintext no.

````python
import hashlib
import secrets

class User:
    def __init__(self, username, password):
        self.username = username
        self.__salt = secrets.token_hex(8)              # every user of for random salt
        self.__password_hash = self.__hash_password(password)  # sirf hash store is is

    def __hash_password(self, password):
# salt + password combine do of sha256 hash - plaintext know save not is        return hashlib.sha256((self.__salt + password).encode()).hexdigest()

    def check_password(self, password):
        return self.__hash_password(password) == self.__password_hash

u = User("ahmed123", "MySecret@1")
print("Sahi password check:", u.check_password("MySecret@1"))
print("Ghalat password check:", u.check_password("wrongpass"))

try:
    print(u.__password_hash)    # bahar from seedha access no be can (name mangling)
except AttributeError as e:
    print("AttributeError:", e)

assert u.check_password("MySecret@1") is True
assert u.check_password("wrong") is False
````

### 4. Rectangle — Width/Height/Area/PerimeterRequirement: `width` and `height` positive honi needed (validated properties), `area` and`perimeter` computed (derived) properties hon.

````python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    @property
    def width(self):
        return self.__width

    @width.setter
    def width(self, value):
        if value <= 0:
            raise ValueError("width positive honi should")
        self.__width = value

    @property
    def height(self):
        return self.__height

    @height.setter
    def height(self, value):
        if value <= 0:
            raise ValueError("height positive honi should")
        self.__height = value

    @property
    def area(self):
        return self.__width * self.__height

    @property
    def perimeter(self):
        return 2 * (self.__width + self.__height)

r = Rectangle(4, 5)
print("Area:", r.area, "| Perimeter:", r.perimeter)
assert r.area == 20
assert r.perimeter == 18

try:
    r.width = -1
except ValueError as e:
    print("ValueError:", e)
````

### LC 705. Design HashSet — Easyhttps://leetcode.com/problems/design-hashset/**Bayan (Roman Urdu):** Bina built-in `set`/hash-thenle library use kithese, apna `MyHashSet`design do jisme `add(key)`, `remove(key)`, and `contains(key)` methods hon.**Approach (bucket-based, encapsulated internals):**1. Fixed number of buckets (like 1000) le lo, every bucket ek list is.2. `key % num_buckets` from index nikalo (simple hashing).3. `add`: if key it bucket in not is then append do (duplicate avoid).4. `remove`: if key bucket in is then list from nikal do.5. `contains`: bucket inside linear search do of tell.6. Buckets and helper `__index` method both **private** have — user to only`add`/`remove`/`contains` API dikhti is, internal sthenrage encapsulated is.**Time complexity:** Average O(1) per operation (if keys buckets in achi way distribute hon),worst case O(n/buckets) ek bucket inside linear scan of wajah from.**Space complexity:** O(n + buckets) where n = stored keys of tadaad.

````python
class MyHashSet:
    def __init__(self):
        self.__size = 1000
        self.__buckets = [[] for _ in range(self.__size)]   # encapsulated internal storage

    def __index(self, key):
        return key % self.__size   # private helper - bahar from call no is (name mangled)

    def add(self, key: int) -> None:
        idx = self.__index(key)
        if key not in self.__buckets[idx]:
            self.__buckets[idx].append(key)

    def remove(self, key: int) -> None:
        idx = self.__index(key)
        if key in self.__buckets[idx]:
            self.__buckets[idx].remove(key)

    def contains(self, key: int) -> bool:
        idx = self.__index(key)
        return key in self.__buckets[idx]

hs = MyHashSet()
hs.add(1)
hs.add(2)
assert hs.contains(1) == True
assert hs.contains(3) == False
hs.add(2)                  # duplicate add - koi asar no hona should
assert hs.contains(2) == True
hs.remove(2)
assert hs.contains(2) == False
print("LC 705 all test cases pass")
````

### LC 1470. Shuffle the Array — Easyhttps://leetcode.com/problems/shuffle-the-array/**Bayan (Roman Urdu):** Array `nums` di gayi is length `2n`, jisme `[x1,...,xn,y1,...,yn]`format in numbers are. Isay `[x1,y1,x2,y2,...,xn,yn]` format in shuffle do of return do.**Approach:** Direct simulation — first half and doosre half from ek-ek element interleave(alternately) do jao.**Time complexity:** O(n).**Space complexity:** O(n) — result array for.

````python
def shuffle(nums: list[int], n: int) -> list[int]:
    result = []
    for i in range(n):
        result.append(nums[i])        # pehle half from element
        result.append(nums[i + n])    # doosre half from corresponding element
    return result

assert shuffle([2, 5, 1, 3, 4, 7], 3) == [2, 3, 5, 4, 1, 7]
assert shuffle([1, 2, 3, 4, 4, 3, 2, 1], 4) == [1, 4, 2, 3, 3, 2, 4, 1]
assert shuffle([1, 1, 2, 2], 2) == [1, 2, 1, 2]
print("LC 1470 all test cases pass")
````

### LC 2160. Minimum sum of Four Digit Number After Splitting Digits — Easyhttps://leetcode.com/problems/minimum-sum-of-four-digit-number-after-splitting-digits/**Bayan (Roman Urdu):** Ek 4-digit number diya is. Uske digits to do nathese 2-digit numbers insplit do (any also comwithouttion), so that both of sum **minimum** be. sum return do.**Approach:** Digits to sort do do (small first). Snowse small 2 digits both numbers of**tens place** in daalo, and bache hue 2 (large) digits **ones place** in — isse sum minimumis (whyki tens place of value 10x more matter doti is, therefore there small digitshone needed).**Time complexity:** O(1) — hamesha only 4 digits are (sorting also O(4 log 4) = O(1)).**Space complexity:** O(1).

````python
def minimum_sum(num: int) -> int:
    digits = sorted(str(num))                   # chhote digits pehle aa jayenge
    n1 = int(digits[0] + digits[2])              # sabse chhota + teesra chhota -> tens+ones
    n2 = int(digits[1] + digits[3])              # doosra chhota + sabse bada -> tens+ones
    return n1 + n2

assert minimum_sum(2932) == 52     # digits sorted: 2,2,3,9 -> 22 + 39 ya 23+29, min sum = 52
assert minimum_sum(4009) == 13     # digits sorted: 0,0,4,9 -> 04 + 09 = 13
print("LC 2160 all test cases pass")
````
