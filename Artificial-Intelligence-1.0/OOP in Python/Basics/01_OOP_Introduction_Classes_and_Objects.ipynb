## 1. Procedural Programming vs OOP### What is- **Procedural programming**: Data (variables/dict/list) and it work on doing that**functions** separate separate are. every function to manually data pass then do parta is.- **OOP (object-Oriented Programming)**: Data and it work on doing that functions(methods) the same unit — **object** — inside bundle be jate are.### why / where use isWhen program small be (a few functions), procedural style easy feels. But likeonly things badhti are (multiple accounts, multiple students, multiple cars) — proceduralstyle in everywhere data explicitly pass then do has then, and bugs of chance increase goes.OOP this problem to solve does: **data + behavior ek jagah, object inside.**Below hum **bank account** of SAME problem both ways from solve doenge, so that differenceapni ankhon from dekh sako.

````python
# ---- PROCEDURAL STYLE ----# Data (accounts) and behavior (functions) absolutely separate are.# every function to "accounts" dictionary explicitly pass doni has then.
accounts = {}  # all accounts yahan store honge: {acc_no: {"owner":..., "balance":...}}

def create_account(accounts, acc_no, owner, balance=0):
# naya account ek nested dictionary of thenr par create rahe are    accounts[acc_no] = {"owner": owner, "balance": balance}

def deposit(accounts, acc_no, amount):
# deposit doing for accounts dict + acc_no + amount, teeno needed    accounts[acc_no]["balance"] += amount

def withdraw(accounts, acc_no, amount):
# withdraw from first balance check then do important is    if accounts[acc_no]["balance"] >= amount:
        accounts[acc_no]["balance"] -= amount
    else:
        print("Insufficient balance!")

def show_balance(accounts, acc_no):
    info = accounts[acc_no]
    print(f"[Procedural] Account {acc_no} ({info['owner']}) balance: {info['balance']}")

# Now use do -- every step par 'accounts' dict pass doni pad rahi iscreate_account(accounts, 101, "Ahmed", 5000)
deposit(accounts, 101, 2000)
withdraw(accounts, 101, 1000)
show_balance(accounts, 101)
````

**output means:** Account 101 of final balance 5000 + 2000 - 1000 = **6000** print hua.**Dikkat what is it?** `accounts` dictionary to every function call in manually pass then dopad raha is. if bhool gathese then bug. Data (`accounts`) and behavior (`deposit`,`withdraw`, ...) of any organic connection not — only convention from jura hua is.

````python
# ---- OOP STYLE (same problem) ----# Now data (attributes) and behavior (methods) the same "BankAccount" object inside are.
class BankAccount:
    def __init__(self, acc_no, owner, balance=0):
# this 3 attributes every object of apne separate honge        self.acc_no = acc_no
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
# now 'accounts' dict pass doing of need not -- 'self' itself object is        self.balance += amount

    def withdraw(self, amount):
        if self.balance >= amount:
            self.balance -= amount
        else:
            print("Insufficient balance!")

    def show_balance(self):
        print(f"[OOP] Account {self.acc_no} ({self.owner}) balance: {self.balance}")

# Now use do -- object itself apna data carry doesacc1 = BankAccount(101, "Ahmed", 5000)
acc1.deposit(2000)
acc1.withdraw(1000)
acc1.show_balance()
````

### Comparison| thing | Procedural | OOP ||---|---|---|| Data + behavior | Alag separate | Bundle (encapsulated) object inside || Data pass then do | every function call in manually | `self` itself carry does || Nathese accounts | Naya dict entry + wahi functions | Bas naya object: `BankAccount(...)` || Readnowility | `withdraw(accounts, 101, 1000)` | `acc1.withdraw(1000)` — more natural || Scale then do | Mushkil is goes | Objects naturally scale do |This difference small program in itna nazar not aata, but when project large be jathese(multiple banks, multiple students, multiple cars...) then OOP of faida saaf dikhta is.

## 6. Instance Attributes vs class Attributes### What is- **Instance attribute**: every object of apna separate copy is (like `self.name`).- **class attribute**: Saari instances of darmiyan **shared** is — class of levelpar define is, `self.` of bajathese seedha class body me.### why / where use isclass attribute then use is when any value saari instances in **common** honineeded — like school of naam (every student same school in is), or thental objects ofcounter.### Syntax```pythonclass ClassName:class_attr = "shared value" # class attribute -- all instances share dodef __init__(self, x):self.instance_attr = x # instance attribute -- every object of apna```

````python
class Student:
    school_name = "Bizee Grammar School"  # class attribute -- all students shared do are

    def __init__(self, name, roll_no, marks=None):
        self.name = name
        self.roll_no = roll_no
        self.marks = marks if marks is not None else []

    def display(self):
# self.school_name writene from first instance in dhoonda goes, not mile then class in        print(f"{self.name} (Roll {self.roll_no}) -> School: {self.school_name}")

s1 = Student("Ayesha", 1)
s2 = Student("Bilal", 2)

s1.display()
s2.display()

# class attribute to class of through change do -> SABHI instances par asar will beStudent.school_name = "Bizee Model School"
s1.display()
s2.display()
````

### Shared-State Trap (utable class attribute) — Bohat Common Bug!Here real khatra is: if class attribute **utable** be (like `list` or `dict`), thenall instances the same **ek** object to share do — and ek instance of change all parasar dalta is. This ek bohat common real-world bug is.

````python
# ---- BUGGY VERSION ----class TeamBuggy:
    members = []  # DANGER: this ek SHARED mutable list is (class attribute)

    def __init__(self, team_name):
        self.team_name = team_name

    def add_member(self, name):
        self.members.append(name)  # this 'self.members' likh rahe are, lekin list SHARED is

team_a = TeamBuggy("Alpha")
team_b = TeamBuggy("Beta")

team_a.add_member("Ahmed")
team_b.add_member("Sara")

# Expected: team_a only 'Ahmed' havee, team_b only 'Sara' -- but both of members SAME are!print("team_a.members:", team_a.members)
print("team_b.members:", team_b.members)
print("Same list object?", team_a.members is team_b.members)  # True -- yehi bug of jarh is
````

````python
# ---- FIXED VERSION ----class TeamFixed:
    def __init__(self, team_name):
        self.team_name = team_name
        self.members = []  # FIX: mutable data to hamesha __init__ of inside, self. of sath create

    def add_member(self, name):
        self.members.append(name)

team_a = TeamFixed("Alpha")
team_b = TeamFixed("Beta")

team_a.add_member("Ahmed")
team_b.add_member("Sara")

print("team_a.members:", team_a.members)   # sirf Ahmed
print("team_b.members:", team_b.members)   # sirf Sara
print("Same list object?", team_a.members is team_b.members)  # ab False
````

### Common mistakes / gotchas- **Know also utable default** (list, dict, set) class attribute of thenr par mat createif tum chahte be every instance apna separate copy havee. Hamesha `__init__` inside`self.attr = []` writeo.- Imutable class attributes (str, int, tuple) safe are why of unhe "modify" notdone ja can — only naya value assign is (jo instance-level attribute ban goesis, class-level to thenuch not does).- `self.x = value` writene from **naya instance attribute** ban goes (if alreadyclass attribute `x` present be then woh it instance for "shadow"/hide be goes).**Interview tip:** "utable default class attributes are shared across all instances" —this ek classic Python gotcha interview question is. ## 8.
````python
class Student:
    school_name = "Bizee Grammar School"

    def __init__(self, name, roll_no, marks=None):
        self.name = name
        self.roll_no = roll_no
        self.marks = marks if marks is not None else []

    def display(self):
        print(f"{self.name} (Roll {self.roll_no}) -> Marks: {self.marks}")

    def add_marks(self, value):
# instance method: object of apne 'marks' list to update does        self.marks.append(value)

# class of namespace -- 'display' and 'add_marks' here dikhengi (methods)class_level_names = [n for n in Student.__dict__ if not n.startswith("__")]
print("Student class of apne namespace me:", class_level_names)

s1 = Student("Ayesha", 1)
s1.add_marks(88)
s1.add_marks(92)
s1.display()

# 'add_marks' instance of __dict__ in no is -- ththis class of __dict__ in isprint("'add_marks' in s1.__dict__?", "add_marks" in s1.__dict__)
print("'add_marks' in Student.__dict__?", "add_marks" in Student.__dict__)
````

## 10. `del`, object Lifecycle, `__new__` and `__del__`### What isobject of poori zindagi (lifecycle) teen phases in is:1. **Creation**: `__new__` call is — this **memory allocate** does andkhaali (raw) object return does.2. **Initialization**: `__new__` of turant after `__init__` call is — ththis object tovalues from initialize does.3. **Destruction**: When object of any reference baaki not bachta (reference count 0 begoes), CPython it garbage collect do gives is and `__del__` call is.`del obj` itself object to delete **no** does — this only `obj` naam (reference) tohata gives is. if woh **last** reference thi, then object garbage collect be goesand `__del__` chal goes.### Syntax```pythonclass MyClass:def __new__(cls, *args, **kwargs):instance = super().__new__(cls) # asal memory allocationreturn instancedef __init__(self, ...):... # initializationdef __del__(self):... # cleanup when object destroy be raha be```

````python
class Resource:
    def __new__(cls, *args, **kwargs):
# __new__ first runs is -- ththis object of memory allocate does        print("[1] __new__ chal raha is -> memory allocate be rahi is")
        instance = super().__new__(cls)
        return instance

    def __init__(self, name):
# __new__ of after __init__ runs is -- ththis object to initialize does        print(f"[2] __init__ chal raha is -> '{name}' initialize be raha is")
        self.name = name

    def __del__(self):
# when object of any reference not bachta, this cleanup of time runs is        print(f"[3] __del__ chal raha is -> '{self.name}' object destroy be raha is")

r = Resource("TempConnection")
print("object ban gaya, ab hum 'del r' do are:")
del r   # this 'r' naam hata gives is -- why of this aakhri reference thi, __del__ chal goes is
print("del of baad 'r' variable ab exist no does")
````

**output means:** Order see — `__new__` -> `__init__` -> (work) -> `del r` of after`__del__`. Ththis object of poori lifecycle is.### Common mistakes / gotchas- `__del__` par **bharosa mat do** critical cleanup (like file close then do, dateaseconnection band then do) for — CPython in refcounting from this usually turant chalgoes, but **doosre Python implementations** (PyPy waghera) in timing guaranteednot is. Critical cleanup for `with` statement / context managers use do.- `del obj` only reference hatcomes — if object of any and reference (list, dict,doosra variable) present is, then `__del__` turant **no** will run.- `__new__` bohat less itself then write has then — more tar normal classes in only`__init__` only kaafi is. `__new__` more tar imutable types (`str`, `tuple`) ofsubclassing or metaclasses in use is.**Interview tip:** "`__new__` creates the object, `__init__` initializes it" — is differenceto clearly explain do sako then this mid-level+ interview in achha impression gives is.

## 11. Default vs Parameterized Constructor, and "Constructor Overloading"### What is- **Default constructor**: When `__init__` to any also extra arguments not dithese jathesein(ya all of default values be).- **Parameterized constructor**: When `__init__` explicit values takes is.### why / where use isJava/C++ in class of **multiple `__init__`-like constructors** (overloading) ban canare. **Python ismein separate is** — Python in ek class of only **ek** `__init__` be canis. its Pythonic solution do ways from is:1. **Default arguments** — `def __init__(self, name, marks=None)`2. **`@classmethod` facthenry methods** — alternate "constructors" that `cls(...)` calldoke object create, separate separate input formats for.

````python
class Student:
    school_name = "Bizee Grammar School"

    def __init__(self, name, roll_no, marks=None):
# Default constructor like istemal: Student() also chal can is (all defaults)# Parameterized: Student("Ali", 3, [90, 85])        self.name = name
        self.roll_no = roll_no
        self.marks = marks if marks is not None else []

    def add_marks(self, value):
        self.marks.append(value)

    @classmethod
    def from_string(cls, csv_line):
# this ek "alternate constructor" is -- CSV jaisi string from Student creates# e.g. "Bilal,3,90-85-77"        name, roll_no, marks_str = csv_line.split(",")
        marks = [int(m) for m in marks_str.split("-")] if marks_str else []
        return cls(name, int(roll_no), marks)  # cls(...) == Student(...) yahan

    @classmethod
    def new_blank(cls, name):
# another facthenry -- only naam from student create, roll_no later set will be        return cls(name, roll_no=0)

    def display(self):
        print(f"{self.name} (Roll {self.roll_no}) -> Marks: {self.marks}")

# 1) "default-ish" constructor -- marks not ditheses1 = Student("Sara", 4)

# 2) parameterized constructor -- all some explicits2 = Student("Ali", 3, [90, 85])

# 3) classmethod facthenry -- CSV string from createyas3 = Student.from_string("Bilal,7,90-85-77")

# 4) classmethod facthenry -- only naam from createyas4 = Student.new_blank("Zara")

for s in (s1, s2, s3, s4):
    s.display()

assert s3.name == "Bilal" and s3.roll_no == 7 and s3.marks == [90, 85, 77]
assert s4.roll_no == 0
print("\nSab factory methods ne sahi objects banaye -- asserts pass!")
````

**output means:** `Student.from_string(...)` and `Student.new_blank(...)` both`Student` of **nathese objects** only create rahe are, just separate input format from — thisi"Pythonic constructor overloading" is.### Common mistakes / gotchas- Python in `def __init__` do time then write **error no** dega, but doosri definitionpehli to **overwrite** do degi (Java jaisi overloading not is). ## 12. `__str__` vs `__repr__`### What is- `__str__` — **user-friendly** readnowle string (when `print(obj)` or `str(obj)` run).- `__repr__` — **developer-friendly** unambiguous string (when only `obj` type do REPLme, or `repr(obj)` call do). Ideally aisi honi needed of it `eval()` do of wapasobject recreate be sake.### why / where use isif `__str__` define not be, Python `__repr__` to fallback of thenr par use does. ifboth not hon, default `<__main__.Student object at 0x...>` like ugly output comes.

````python
class Student:
    def __init__(self, name, roll_no, marks=None):
        self.name = name
        self.roll_no = roll_no
        self.marks = marks if marks is not None else []

    def __str__(self):
# print(student) or str(student) call hone par this will run -- readnowle        return f"Student({self.name}, Roll {self.roll_no})"

    def __repr__(self):
# repr(student) or only 'student' writene par this will run -- unambiguous/debug-friendly        return f"Student(name={self.name!r}, roll_no={self.roll_no!r}, marks={self.marks!r})"

s1 = Student("Ayesha", 1, [90, 85])

print(s1)          # __str__ use is is
print(str(s1))     # __str__ explicitly
print(repr(s1))    # __repr__ explicitly

print([s1])        # list of inside print karne par __repr__ use is is, __str__ no!
````

## 13. Objects to `list` / `dict` in Store to do### What is / whyReal applications in hume **kayi objects** ek with manage doing are — like"saare students of class list".
````python
class Student:
    def __init__(self, name, roll_no, marks=None):
        self.name = name
        self.roll_no = roll_no
        self.marks = marks if marks is not None else []

    def __repr__(self):
        return f"Student({self.name!r}, roll_no={self.roll_no})"

# ---- list in store then do ----students_list = [
    Student("Ayesha", 1, [90, 85]),
    Student("Bilal", 2, [70, 75]),
    Student("Sara", 3, [95, 92]),
]

print("Poori list:", students_list)   # __repr__ use will be

for s in students_list:
    avg = sum(s.marks) / len(s.marks)
    print(f"{s.name}: average marks = {avg:.2f}")

# ---- dict in store then do (roll_no -> object, fast lookup for) ----students_by_roll = {s.roll_no: s for s in students_list}

# now O(1) in anyone also roll_no from student mil can is (list in O(n) lagta)lookup = students_by_roll[2]
print("\nRoll 2 of student:", lookup)

assert students_by_roll[1].name == "Ayesha"
assert len(students_by_roll) == 3
print("Asserts pass!")
````

````python
class Book:
    def __init__(self, title, author, pages, price):
        self.title = title
        self.author = author
        self.pages = pages
        self.price = price
        self.is_issued = False   # library me is ya issue be chuki is

    def issue(self):
        if self.is_issued:
            print(f"'{self.title}' pehle from hi issue is!")
        else:
            self.is_issued = True
            print(f"'{self.title}' issue kar di gayi.")

    def return_book(self):
        self.is_issued = False
        print(f"'{self.title}' wapas aa gayi.")

    def __str__(self):
        status = "Issued" if self.is_issued else "Available"
        return f"'{self.title}' by {self.author} ({self.pages}p, Rs.{self.price}) [{status}]"

book1 = Book("Python Crash Course", "Eric Matthes", 560, 2500)
print(book1)
book1.issue()
book1.issue()      # dobara issue karne of koshish -> already-issued message
book1.return_book()
print(book1)
````

````python
class Car:
# class attribute: thental kitni cars bani are, all objects share do    total_cars_made = 0

    def __init__(self, brand, model, year, fuel_level=0):
        self.brand = brand
        self.model = model
        self.year = year
        self.fuel_level = fuel_level   # percentage: 0 to 100
        Car.total_cars_made += 1       # every naye object par class-level counter barhta is

    def refuel(self, amount):
        self.fuel_level = min(100, self.fuel_level + amount)  # 100% from zyada no be can

    def drive(self, distance_km):
        fuel_needed = distance_km // 10  # every 10km of for 1% fuel (simple simulation)
        if fuel_needed > self.fuel_level:
            print(f"{self.brand} {self.model}: itna fuel no is is safar of for!")
        else:
            self.fuel_level -= fuel_needed
            print(f"{self.brand} {self.model} ne {distance_km}km chala. Fuel left: {self.fuel_level}%")

    def __str__(self):
        return f"{self.year} {self.brand} {self.model} (Fuel: {self.fuel_level}%)"

car1 = Car("Toyota", "Corolla", 2022, fuel_level=50)
car2 = Car("Honda", "Civic", 2023, fuel_level=20)

car1.refuel(30)
car1.drive(400)
car2.drive(400)   # kam fuel is, warning aayegi

print(car1)
print(car2)
print("Total cars made so far:", Car.total_cars_made)  # class attribute -- shared counter

assert Car.total_cars_made == 2
````

### Problem 1: Library Book TrackerEk chis library system create where books track be sakein — kitni copies availableare, issue/return of hisano.

````python
class LibraryBook:
    def __init__(self, title, total_copies):
        self.title = title
        self.total_copies = total_copies
        self.issued_copies = 0   # kitni copies is waqt issue are

    @property
    def available_copies(self):
# this ek computed value is -- humeshthental - issued        return self.total_copies - self.issued_copies

    def issue_copy(self):
        if self.available_copies > 0:
            self.issued_copies += 1
            print(f"'{self.title}' of ek copy issue hui. Available: {self.available_copies}")
        else:
            print(f"'{self.title}' of koi copy available no is!")

    def return_copy(self):
        if self.issued_copies > 0:
            self.issued_copies -= 1
            print(f"'{self.title}' of ek copy wapas aayi. Available: {self.available_copies}")

book = LibraryBook("Clean Code", total_copies=2)
book.issue_copy()
book.issue_copy()
book.issue_copy()   # ab koi copy available no
book.return_copy()

assert book.available_copies == 1
assert book.issued_copies == 1
print("\nSab asserts pass!")
````

### Problem 2: Bank Account with ValidationBank account that negative balance or invalid amounts not hone give.

````python
class SafeBankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount <= 0:
            print("Deposit amount positive hona should!")
            return
        self.balance += amount

    def withdraw(self, amount):
        if amount <= 0:
            print("Withdraw amount positive hona should!")
            return
        if amount > self.balance:
            print("Insufficient funds!")
            return
        self.balance -= amount

acc = SafeBankAccount("Nadia", 1000)
acc.deposit(500)
acc.withdraw(200)
acc.withdraw(-50)     # invalid -> reject
acc.deposit(0)        # invalid -> reject
acc.withdraw(5000)     # insufficient -> reject

print("Final balance:", acc.balance)
assert acc.balance == 1300
````

### Problem 3: Simple Inventry ManagementStore of products of sthenck track do, low-sthenck warning with.

````python
class Product:
    LOW_STOCK_THRESHOLD = 5   # class attribute -- all products of for same rule

    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def sell(self, units):
        if units > self.quantity:
            print(f"'{self.name}': itni quantity stock me no is!")
            return
        self.quantity -= units
        print(f"{units} '{self.name}' beche gaye. Baaki stock: {self.quantity}")
        if self.quantity < Product.LOW_STOCK_THRESHOLD:
            print(f"  WARNING: '{self.name}' of stock kam be raha is ({self.quantity} bache)!")

    def restock(self, units):
        self.quantity += units

inventory = [
    Product("Notebook", 50, 20),
    Product("Pen", 10, 100),
]

inventory[0].sell(17)   # stock 20 -> 3, low stock warning ayegi
inventory[1].sell(30)   # stock 100 -> 70, koi warning no

assert inventory[0].quantity == 3
assert inventory[1].quantity == 70
````

### LC 1929. Concatenation of Array — Easyhttps://leetcode.com/problems/concatenation-of-array/**Bayan:** Ek array `nums` (length `n`) diya is. Ek naya array `ans` (length `2n`)create jahan:- `ans[i] = nums[i]` for `0 <= i < n`- `ans[i + n] = nums[i]` for `0 <= i < n`Yani `nums` to authenmatically with jorna is (concatenate).**Approach:**- **Brute force**: naya array create, do loops (ya ek loop, both index fill do).- **Optimal**: Python in `nums + nums` seedha concatenation do gives is — O(n) me.**Time complexity:** O(n) both approaches in (output itself O(n) size of is).**Space complexity:** O(n) for the output array.Below hum LeetCode of expected `class Solution` format with **class-wrapper style**also dikha rahe are (object-oriented ways from ek "converter" object createna).

````python
# ---- Brute-force style (both index manually fill then do) ----class SolutionBruteForce:
    def getConcatenation(self, nums: list[int]) -> list[int]:
        n = len(nums)
        ans = [0] * (2 * n)   # pehle from size fix kar to take efficient is
        for i in range(n):
            ans[i] = nums[i]        # pehla hissa
            ans[i + n] = nums[i]    # dusra hissa (concatenation)
        return ans


# ---- Optimal (LeetCode standard "class Solution" format) ----class Solution:
    def getConcatenation(self, nums: list[int]) -> list[int]:
        return nums + nums   # Python list concatenation -- ek line me work


# ---- class-wrapper style (OOP theme of mutheniq: ek object that array to "wrap" does) ----class ArrayConcatenator:
    def __init__(self, nums: list[int]):
        self.nums = nums   # object apna data khud carry does is

    def concatenate(self) -> list[int]:
        return self.nums + self.nums


# ---- Test cases ----assert SolutionBruteForce().getConcatenation([1, 2, 1]) == [1, 2, 1, 1, 2, 1]
assert Solution().getConcatenation([1, 2, 1]) == [1, 2, 1, 1, 2, 1]
assert Solution().getConcatenation([1, 3, 2, 1]) == [1, 3, 2, 1, 1, 3, 2, 1]

converter = ArrayConcatenator([1, 2, 1])
assert converter.concatenate() == [1, 2, 1, 1, 2, 1]

print("LC 1929: all asserts pass!")
````

### LC 2469. Convert the Temperature — Easyhttps://leetcode.com/problems/convert-the-temperature/**Bayan:** Ek `celsius` (float) diya is. Ek array `[kelvin, fahrenheit]` return do,jahan:- `Kelvin = Celsius + 273.15`- `Fahrenheit = Celsius * 1.8 + 32.00`Answer 10^-5 tak accurate hona needed.**Approach:** Seedha formula apply do, `round(value, 5)` from precision handle do.**Time complexity:** O(1) — only arithmetic.**Space complexity:** O(1).Is problem to hum poori way **class-based** solve do rahe are (theme of mutheniq) —ek `TemperatureConverter` object create that apna `celsius` value carry does.

````python
class TemperatureConverter:
    def __init__(self, celsius: float):
        self.celsius = celsius   # object apna input value store does is

    def convert(self) -> list[float]:
        kelvin = round(self.celsius + 273.15, 5)     # Kelvin formula
        fahrenheit = round(self.celsius * 1.8 + 32.00, 5)  # Fahrenheit formula
        return [kelvin, fahrenheit]


# ---- Test cases (LeetCode official examples) ----assert TemperatureConverter(36.50).convert() == [309.65, 97.7]
assert TemperatureConverter(122.11).convert() == [395.26, 251.798]

print("LC 2469: all asserts pass!")
print("Example:", TemperatureConverter(36.50).convert())
````

### LC 1656. Design a Ordered Stream — Easyhttps://leetcode.com/problems/design-an-ordered-stream/**Bayan:** `n` values of ek stream is, id `1` from `n` tak. `OrderedStream` class create:- `OrderedStream(int n)` — stream initialize do `n` values for.- `insert(int idKey, String value)` — id `idKey` par `value` store do, and woh saari**consecutive** values return do that pointer from start be do milti are (chunk),then pointer to aage badhao.Ththis problem itself OOP is — absolutely `class`-based design expect does.**Approach:**- Ek array `stream` create size `n+1` of (1-indexed, index 0 unused).- Ek `pointer` haveo that batata is "next whosa id needed tha".- `insert` par: value store do, then when tak `pointer` par value present is, itresult in daalo and `pointer` aage badhao.**Time complexity:** every `insert` call worst case O(n) lag can is, but **thental**(saari calls milado) O(n) only is — every id only ek time "consume" is(amortized O(1) per call).**Space complexity:** O(n) for the internal stream array.

````python
class OrderedStream:
    def __init__(self, n: int):
        self.stream = [None] * (n + 1)   # 1-indexed; index 0 kabhi use no will be
        self.ptr = 1                      # agla id jo hume "expect" is

    def insert(self, idKey: int, value: str) -> list[str]:
        self.stream[idKey] = value   # is id par value store do

        result = []
# when tak pointer par value present is, it result in daalo and aage increaseo        while self.ptr < len(self.stream) and self.stream[self.ptr] is not None:
            result.append(self.stream[self.ptr])
            self.ptr += 1
        return result


# ---- Test case (LeetCode official example) ----os_stream = OrderedStream(5)
r1 = os_stream.insert(3, "ccccc")
r2 = os_stream.insert(1, "aaaaa")
r3 = os_stream.insert(2, "bbbbb")
r4 = os_stream.insert(5, "eeeee")
r5 = os_stream.insert(4, "ddddd")

assert r1 == []
assert r2 == ["aaaaa"]
assert r3 == ["bbbbb", "ccccc"]
assert r4 == []
assert r5 == ["ddddd", "eeeee"]

print("LC 1656: all asserts pass!")
print("Results:", r1, r2, r3, r4, r5)
````
