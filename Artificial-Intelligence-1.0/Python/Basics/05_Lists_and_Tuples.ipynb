## 1. List Basics

**what is:** `list` Python of **ordelayed, mutable** (badla ja can is) collection is. Elements kisi bhi type of is can are.

**why/kahan:** Jab bhi tumhein items of ek sequence should jo **grow/shrink** is, ya jisko after me modify to do is — list use do
(if modify not to do then tuple better is — us but ahead aayenge).

**Syntax:** `[]`, `list()`, `list(iterable)`.

```python
# List banane of kayi tareeqe - sab common are, interview me bhi poochte are
lst_literal = [1, 2, 3, 4, 5]                 # seedha literal syntax - sabse common
lst_empty = []                                 # empty list
lst_from_range = list(range(5))                # list() constructor + range() from
lst_from_str = list("abc")                     # string then characters of list banate are
lst_repeat = [0] * 5                           # ek value then n baar repeat to do
lst_hetero = [1, "two", 3.0, True, None]       # heterogeneous - kisi bhi type of mix runs is
lst_nested = [1, [2, 3], [4, [5, 6]]]          # nested lists - list of inside list

print("literal:      ", lst_literal)
print("from range:   ", lst_from_range)
print("from string:  ", lst_from_str)
print("repeat [0]*5: ", lst_repeat)
print("hetero:       ", lst_hetero)
print("nested:       ", lst_nested)
print("len(nested):  ", len(lst_nested))          # len() top-level elements ginta is, inside those not
print("empty truthy? ", bool(lst_empty))           # empty list -> False
print("literal truthy?", bool(lst_literal))        # non-empty -> True
```

**Output of matlab:** `len(lst_nested)` = 3 (because top-level but 3 items are: `1`, `[2,3]`, `[4,[5,6]]`) —
nested elements then count not does.

**Common mistakes / gotchas:**
- `lst_repeat = [0] * 5` sirf **immutable** elements (int, str) of for safe is. if inside mutable object is
  (like `[[0]*3]*3`), sab references **same object** of taraf point karti are — this trap ahead detail me dekhenge.
- Empty list `[]` **falsy** is is — `if my_list:` likhna `if len(my_list) > 0:` from zyada Pythonic is.

**Interview tip:** "Truthiness of empty containers" ek common interview question is — list, tuple, dict, str,
set sab empty hone but `False` are are.

## 2. Indexing and Negative Indexing

**what is:** Har element of ek position is is — `0` from shuru (positive) ya `-1` from end from (negative).

**why:** Kisi bhi element then O(1) me directly access to do of for.

**Syntax:** `lst[i]`

```python
nums = [10, 20, 30, 40, 50]
print(nums[0])    # pehla element -> 10
print(nums[-1])   # aakhri element -> 50 (negative indexing)
print(nums[2])    # teesra element -> 30
print(nums[-2])   # doosra last -> 40

try:
    print(nums[10])   # Indexerror - list of range from outside
except Indexerror as e:
    print("Indexerror pakda gaya:", e)
```

**Common mistakes:** `nums[len(nums)]` bhi `Indexerror` dega — last valid index `len(nums)-1` is, `len(nums)` not.

**Interview tip:** Negative index `-1` "last element" of for idiomatic is — `nums[len(nums)-1]` likhna beginner smell is.

## 3. Slicing `[start:stop:step]`

**what is:** Ek sath multiple elements nikalna, **new list** return does is (original untouched).

**Syntax:** `lst[start:stop:step]` — `stop` **exclusive** is is.

```python
nums = [10, 20, 30, 40, 50, 60]
print(nums[1:4])       # index 1,2,3 (4 exclusive) -> [20,30,40]
print(nums[:3])        # start from index 2 tak -> [10,20,30]
print(nums[3:])        # index 3 from end tak -> [40,50,60]
print(nums[::2])       # har doosra element -> [10,30,50]
print(nums[::-1])      # pura list reverse -> [60,50,40,30,20,10]
print(nums[-3:])       # last 3 elements -> [40,50,60]

# COPY via slice - [:] new (shallow) list banata is, alias not
copy_via_slice = nums[:]
copy_via_slice.append(999)
print("original:", nums)
print("copy:    ", copy_via_slice)
print("same object?", nums is copy_via_slice)   # False - alag objects
```

**Output of matlab:** `nums[:]` ek new list object banata is jiska content same is — is for `copy_via_slice`
me append to do from original `nums` **change not** is. this "shallow copy" of sabse simple tareeqa is.

**Common mistakes:** `nums[3:1]` (start > stop, positive step) empty list gives is, error not. Slicing kabhi
`Indexerror` not deti — out-of-range indices silently clamp is go are.

**Interview tip:** `nums[::-1]` reverse to do of sabse fast/idiomatic tareeqa is (vs `list(reversed(nums))`).

## 4. Slice Assignment — in depth

**what is:** Slice then sirf read hi not, **assign** bhi kar can is — list of size khud-ba-khud grow/shrink is jata is.

**why:** Bulk replace, insert, ya delete to do — without new list banaye, in-place.

```python
# 1) SAME LENGTH replace - straightforward
a = [1, 2, 3, 4, 5]
a[1:3] = [20, 30]              # 2 elements of jagah 2 naye elements
print("same length:", a)

# 2) DIFFERENT LENGTH - GROW (2 elements of jagah 4 daal diye)
b = [1, 2, 3, 4, 5]
b[1:3] = [20, 30, 40, 50]
print("grow:       ", b)

# 3) DIFFERENT LENGTH - SHRINK (3 elements of jagah 1 daala)
c = [1, 2, 3, 4, 5]
c[1:4] = [99]
print("shrink:     ", c)

# 4) INSERT without removing - EMPTY slice (koi element remove not is, bas ghusaya)
d = [1, 2, 3]
d[1:1] = [100, 101]            # index 1 but insert, some delete not
print("insert:     ", d)

# 5) DELETE via empty list assign
e = [1, 2, 3, 4, 5]
e[1:3] = []                     # index 1,2 delete
print("delete:     ", e)
```

```python
# 6) EXTENDED SLICE (step != 1) - length EXACTLY match honi should
f = [1, 2, 3, 4, 5, 6]
f[::2] = [100, 200, 300]        # 3 positions (0,2,4), 3 values diye - theek
print("extended slice ok:", f)

try:
    g = [1, 2, 3, 4, 5, 6]
    g[::2] = [1, 2]              # sirf 2 values diye, 3 positions are -> Valueerror
except Valueerror as e:
    print("Valueerror (extended slice length mismatch):", e)
```

**Output of matlab:** Normal slice (`step=1`) of target list **kisi bhi length** of is can is (grow/shrink allowed).
but **extended slice** (`step != 1`) of for assigned values of count **exactly** utni honi should jitni slice
me positions are — warna `Valueerror`.

**Common mistakes:** Beginners `lst[2:2] = [x]` and `lst.insert(2, x)` me confuse are are — dono same kaam do
are single element of for, but slice assignment multiple elements ek sath insert kar can is.

**Interview tip:** `lst[a:b] = []` and `del lst[a:b]` **same effect** dete are — dono from poocho, dono chalne
should demo me.

## 5. `lst[:] = other` vs `lst = other` — In-place vs Rebinding (BOHAT IMPORTANT)

**what is:** this Python of sabse confuse to do that concept is — ek **same object then modify** does is,
doosra **variable then naye object from re-point** does ha.

**why important is:** if kisi function of inside list then "replace" to do is and caller then bhi dikhna is,
then `[:] =` use to do hoga — plain `=` sirf local reference badalega.

```python
# Case A: [:] = ... -> IN-PLACE replace, saare aliases update are are
original = [1, 2, 3]
alias = original                # alias and original SAME object
original[:] = [9, 9, 9]         # original object of inside of data replace happened
print("Case A - alias after [:] replace:", alias)   # [9,9,9] - alias bhi badla!
print("Case A - same object?", alias is original)    # True - dono ab bhi same object

# Case B: = ... -> REBINDING, sirf 'original2' naye object of taraf point does is
original2 = [1, 2, 3]
alias2 = original2
original2 = [9, 9, 9]           # new object banew, sirf original2 usko point does is
print("Case B - alias2 after rebind:", alias2)       # [1,2,3] - alias2 old hi is
print("Case B - original2:", original2)               # [9,9,9]
print("Case B - same object?", alias2 is original2)   # False - ab alag objects
```

**Output of matlab:** `original[:] = [9,9,9]` list of inside of data **mutate** does is — object of `id()` same
rehta is. `original2 = [9,9,9]` variable then naye object from **rebind** does is — old object jahan `alias2`
abhi bhi point does is, waisa hi rehta is.

**Common mistakes:** Function of inside `def reset(lst): lst = []` caller of list **empty not karega** (rebind is,
local scope me). `def reset(lst): lst[:] = []` caller of list **empty kar dega** (in-place).

**Interview tip:** this Python me "pass by object reference" samjhane of best example is — interview me
surely poocha jata is "list then function of inside reset how do taake caller then dikhe".

## 6. List Methods — mukammal tour

Har method: what does is, return value what is (bohat from **`None`** return do are — in-place mutate),
and common gotcha.

```python
# append() vs + : append single element end me add does is, original then MUTATE does is
lst = [1, 2, 3]
lst.append(4)
print("append:", lst)

# + new list BANATA is, original not badalta
lst2 = [1, 2, 3] + [4, 5]
print("plus (+):", lst2)

# extend() vs append(list) - CLASSIC confusion
lst3 = [1, 2, 3]
lst3.extend([4, 5])             # extend elements then ek-ek karke add does is
print("extend([4,5]):", lst3)

lst4 = [1, 2, 3]
lst4.append([4, 5])             # append pura [4,5] then EK single element of tarah daal gives is!
print("append([4,5]) - GOTCHA:", lst4)

# extend() ek STRING bhi le can is - string ek iterable is, characters add are are
lst5 = [1, 2, 3]
lst5.extend("ab")
print("extend('ab'):", lst5)
```

```python
# insert(index, value) - O(n) because index of after of saare elements shift are are
lst6 = [10, 20, 30]
lst6.insert(1, 99)
print("insert:", lst6)

# remove(value) - sirf PEHLA match hatata is, missing but Valueerror
lst7 = [1, 2, 3, 2, 1]
lst7.remove(2)
print("remove (first match only):", lst7)
try:
    lst7.remove(100)
except Valueerror as e:
    print("remove Valueerror:", e)

# pop([index]) - default last element, index bhi give can is; empty but Indexerror
lst8 = [1, 2, 3]
popped_last = lst8.pop()        # default -1
popped_first = lst8.pop(0)      # explicit index
print("pop() results:", popped_last, popped_first, "| remaining:", lst8)
try:
    [].pop()
except Indexerror as e:
    print("pop empty list Indexerror:", e)

# clear() - sab elements hata gives is, list object wahi rehta is (empty)
lst9 = [1, 2, 3]
lst9.clear()
print("clear:", lst9)
```

```python
# index(value, [start], [stop]) - pehla occurrence dhoondta is, optional start/stop range
lst10 = [5, 3, 8, 3, 9, 3]
print("index of 3 (default):     ", lst10.index(3))         # 1
print("index of 3 (start=2):     ", lst10.index(3, 2))        # 3
print("index of 3 (start=2,stop=4):", lst10.index(3, 2, 4))    # 3
try:
    lst10.index(100)
except Valueerror as e:
    print("index() Valueerror:", e)

# count(value) - kitni baar value gets is
print("count of 3:", lst10.count(3))

# sort() - IN-PLACE sort, returns None -> #1 BEGINNER bug
lst11 = [5, 2, 9, 1]
return_value = lst11.sort()
print("sort() in place:", lst11, "| return value (None!):", return_value)
# GOTCHA: x = lst.sort() writege then x hamesha None hoga - lst khud sorted is jata is

# reverse() - IN-PLACE, returns None bhi
lst12 = [1, 2, 3]
lst12.reverse()
print("reverse:", lst12)

# copy() - SHALLOW copy, new top-level list object
lst13 = [1, 2, 3]
lst14 = lst13.copy()
lst14.append(4)
print("copy - original:", lst13, "| copy:", lst14)
```

**Common mistakes/gotchas (summary):**
- `x = my_list.sort()` → `x` **None** hoga, list khud sort is gaya. should then `sorted(my_list)` use do.
- `append([4,5])` vs `extend([4,5])` — pehla nested list banata is, doosra flatten does is.
- `remove()` **value** from dhoondta is, `pop()` **index** from — dono confuse are are beginners me.
- `insert(0, x)` and `pop(0)` dono **O(n)** are (sab elements shift to do padte are) — queue like use case is
  then `collections.deque` behtar is (ahead dekhenge).

**Interview tip:** "In-place methods return `None`" — this ek favourite trick question is
(`sort`, `reverse`, `append`, `extend`, `insert`, `remove`, `clear` sab `None` return do are).

## 7. `del` Statement — `pop`/`remove` from kaisa alag

**what is:** `del` ek **statement** is (method not), list of element(s) ya poore variable then delete kar can is.

**why:** Jab value **wapas not should** (sirf hatani is) — `pop()` value return does is, `del` some return not does.

```python
lst = [1, 2, 3, 4, 5]
del lst[0]              # single index delete
print("del lst[0]:", lst)

del lst[1:3]            # slice delete
print("del lst[1:3]:", lst)

lst_full = [1, 2, 3]
del lst_full             # variable hi khatam - object bhi reference-free is jata is
try:
    print(lst_full)
except Nameerror as e:
    print("Nameerror after del lst_full:", e)
```

**Comparison table:**

| | `pop(i)` | `remove(v)` | `del lst[i]` |
|---|---|---|---|
| Input | index | value | index/slice |
| Return | removed value | `None` | `None` (statement) |
| Missing case | `Indexerror` | `Valueerror` | `Indexerror` |
| Poora variable delete kar can? | not | not | yes (`del lst`) |

**Interview tip:** `del lst` variable then **unbind** does is — if object of koi and reference no is then
garbage collected is jata is. this `lst.clear()` from alag is (jo variable then rakhta is, sirf empty does is).

## 8. `sort()` vs `sorted()`

**what is:** `list.sort()` **in-place** method is (list hi of), `sorted()` **built-in function** is jo
**kisi bhi iterable** but runs is and **new list** return does is.

**why/kahan:** if original ordelay preserve to do is ya iterable list no is (like `set`, `dict`, `str`) — `sorted()`.
if list then permanently sort to do is, memory bachani is — `sort()`.

```python
nums = [5, 2, 9, 1]
new_sorted = sorted(nums)          # new list, original untouched
print("sorted():", new_sorted, "| original still:", nums)

nums.sort()                        # IN-PLACE
print("sort():", nums)

# sorted() KISI BHI iterable but kaam does is - list, tuple, set, string
print("sorted string:", sorted("dcba"))
print("sorted tuple: ", sorted((3, 1, 2)))
print("sorted set:   ", sorted({3, 1, 2}))
```

```python
import operator

words = ["banana", "kiwi", "apple", "fig"]
print("key=len:       ", sorted(words, key=len))                 # length of hisab from
print("key=str.lower: ", sorted(words, key=str.lower))            # case-insensitive
print("key=last char: ", sorted(words, key=lambda w: w[-1]))      # lambda key
print("reverse=True:  ", sorted(words, key=len, reverse=True))    # descending

# MULTI-KEY sort - tuple key: first marks descending, phir naam ascending
students = [("Ali", 85), ("Sara", 90), ("Bilal", 85), ("Ayesha", 90)]
print("multi-key (marks desc, naam asc):", sorted(students, key=lambda s: (-s[1], s[0])))

# operator.itemgetter - faster + readable alternative then lambda for simple field access
print("itemgetter(1): ", sorted(students, key=operator.itemgetter(1)))

# STABILITY - equal keys of RELATIVE ORdelay preserve is is (Timsort guarantee)
data = [("a", 2), ("b", 1), ("c", 2), ("d", 1)]
print("stable sort:   ", sorted(data, key=lambda x: x[1]))   # b,d (1) phir a,c (2) - insertion ordelay preserved
```

```python
# Sorting list of DICTS by a field
people = [{"name": "Ali", "age": 30}, {"name": "Sara", "age": 25}]
print("dicts by age:", sorted(people, key=lambda p: p["age"]))

# Sorting DICT ITEMS by value - top-N pattern
scores = {"Ali": 80, "Sara": 95, "Bilal": 70, "Ayesha": 88}
top2 = sorted(scores.items(), key=lambda kv: kv[1], reverse=True)[:2]
print("top 2 scorers:", top2)

# Typeerror - incomparable types then sort not kar can
try:
    sorted([1, "two", 3])
except Typeerror as e:
    print("Typeerror (mixed types):", e)

# functools.cmp_to_key - purane C-style comparator (a-b) then key function me convert does is
from functools import cmp_to_key
def compare(a, b):
    return a - b   # negative -> a first, positive -> b first, 0 -> barabar
print("cmp_to_key:", sorted([3, 1, 2], key=cmp_to_key(compare)))
```

**Output of matlab:** `key=` function har element but **ek dafa** apply is is (result cache is is internally) —
isliye `key=` `cmp=`-style comparator from zyada efficient is. Stability of matlab: if do elements of key
barabar is then unka relative ordelay **wahi rahega** jo input me tha.

**Common mistakes:**
- `sorted(my_list)` of after `my_list` khud **sort not is** (new list alag variable me store to do padega).
- `my_list.sort()` `None` return does is — `x = my_list.sort()` likh kar `x` use to do bug is.
- Incomparable types (`int` vs `str`) sort to do but `Typeerror` — mixed-type lists sort to do from first normalize do.

**Interview tip:** "Python of sort algorithm what is?" → **Timsort** (merge sort + insertion sort hybrid),
worst case O(n log n), **stable**, best case O(n) (already sorted data but).

## 9. List Comprehensions

**what is:** Ek line me new list banane of concise syntax — loop + condition then expression me compress does is.

**Syntax:** `[expr for item in iterable if condition]`

```python
# Basic comprehension
squares = [x**2 for x in range(6)]
print("basic:", squares)

# with if FILTER
evens = [x for x in range(20) if x % 2 == 0]
print("filter:", evens)

# MULTIPLE if conditions - dono AND of tarah kaam do are
multiples = [x for x in range(50) if x % 2 == 0 if x % 5 == 0]
print("multiple if:", multiples)

# if-else TERNARY in the expression part (note: yahan 'else' zaroori is, filter that 'if' alag is)
labels = ["even" if x % 2 == 0 else "odd" for x in range(6)]
print("ternary expr:", labels)
```

```python
# NESTED LOOPS in a comprehension - ordelay same is is like normal nested for loop
pairs = [(x, y) for x in range(3) for y in range(2)]
print("nested loops:", pairs)
# equivalent then:
#   for x in range(3):
#       for y in range(2):
#           pairs.append((x, y))

# comprehension over NESTED LIST - flatten to do
nested = [[1, 2, 3], [4, 5], [6]]
flat = [item for row in nested for item in row]   # outside that loop first (row), inside that after me (item)
print("flatten:", flat)
```

```python
import time

# Comprehension vs map/filter vs plain loop - readability + chhota perf comparison
N = 20000

start = time.perf_counter()
r1 = [x * 2 for x in range(N)]
t1 = time.perf_counter() - start

start = time.perf_counter()
r2 = list(map(lambda x: x * 2, range(N)))
t2 = time.perf_counter() - start

start = time.perf_counter()
r3 = []
for x in range(N):
    r3.append(x * 2)
t3 = time.perf_counter() - start

print("results match:", r1 == r2 == r3)
print(f"comprehension: {t1:.5f}s | map+lambda: {t2:.5f}s | plain loop: {t3:.5f}s")
# aam tor but: comprehension >= map (map me lambda call overhead is is), dono loop from fast are are
```

**Kab comprehension use not to do should:** Jab expression of result use hi not hona (sirf **side effect**
should, like printing ya kisi dict me insert to do) — comprehension wahan **anti-pattern** is, memory bhi
waste is is (ek throwaway list ban jati is) and readability bhi kharab is is.

```python
# BAD practice - comprehension sirf side-effect (print) of for use to do
throwaway = [print(x) for x in range(3)]   # this [None, None, None] banata is - result kabhi use not hoga
print("throwaway result (wasteful):", throwaway)

# GOOD - jab sirf side effect should, simple loop use do
for x in range(3):
    pass   # yahan asal me print(x) ya koi side-effect do, comprehension not
print("plain loop is the right tool for side-effects only")
```

```python
# LEAKING VARIABLE behaviour - Python 2 me list comprehension of loop variable
# enclosing scope me LEAK is jata tha. Python 3 me comprehension of apna scope is is.
x = "before comprehension"
squares_leak_test = [x for x in range(3)]
print("x after comprehension (Py3 - NO leak, comp has own scope):", x)
```

```python
import sys

# GENERATOR EXPRESSION vs LIST COMPREHENSION - memory difference
list_comp = [x for x in range(10000)]      # () of jagah [] - pura list memory me banta is
gen_expr = (x for x in range(10000))       # () - lazy, ek time but ek value

print("list comprehension size (bytes):", sys.getsizeof(list_comp))
print("generator expression size (bytes):", sys.getsizeof(gen_expr))   # bohat chhota - values store not hotin

print("sum via generator (consumes it once):", sum(gen_expr))
try:
    next(gen_expr)   # generator ek dafa consume hone of after khatam is jata is
except StopIteration:
    print("generator exhausted after use - can't reuse")
```

```python
# NESTED comprehension - matrix banana (row-major, r*3+c pattern)
matrix = [[r * 3 + c for c in range(3)] for r in range(3)]
print("matrix via nested comprehension:")
for row in matrix:
    print(row)
```

## 10. 2D Lists / Matrices

**what is:** List of inside lists — rows of collection. `grid[r][c]` from access.

**GOTCHA (bohat common bug):** `[[0]*3]*3` teenon rows **SAME inner list object** of taraf point karti are!

```python
grid = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print("grid[1][2] (row 1, col 2):", grid[1][2])   # 6

print("row-wise iteration:")
for row in grid:
    print(row)

print("index-based iteration:")
for r in range(len(grid)):
    for c in range(len(grid[0])):
        print(grid[r][c], end=" ")
print()
```

```python
# THE TRAP: [[0]*3]*3 - 3 references then the SAME inner list
trap = [[0] * 3] * 3
trap[0][0] = 99
print("trap - mutation LEAKS then all rows (WRONG):", trap)   # sab rows badal gayi!
print("are rows the same object?", trap[0] is trap[1])       # True - yehi bug of wajah is

# CORRECT way - comprehension har row of for new object banati is
correct = [[0] * 3 for _ in range(3)]
correct[0][0] = 99
print("correct - only row 0 changed:", correct)
print("are rows the same object?", correct[0] is correct[1])   # False - alag objects
```

```python
# Same trap ek list of lists but bhi - * operator SHALLOW repeat does is (references copy is are)
row_template = [1, 2]
bad_matrix = [row_template] * 3      # teenon entries same object
bad_matrix[0].append(999)
print("bad_matrix (all rows affected):", bad_matrix)

good_matrix = [row_template.copy() for _ in range(3)]   # copy() from alag objects
good_matrix[0].append(888)
print("good_matrix (only row 0 affected):", good_matrix)
```

## 11. Copy Semantics — Alias vs Shallow vs Deep Copy

**what is:**
- **Alias** (`b = a`) — dono naam **same object** of taraf point do are.
- **Shallow copy** — new top-level object banta is, but **nested/inner objects same reference** rehte are.
- **Deep copy** — new object + saare nested objects bhi **recursively naye** ban go are.

```python
import copy

original_list = [1, 2, [3, 4]]        # aakhri element ek nested list is
alias_list = original_list             # ALIAS - same object
shallow1 = original_list.copy()        # SHALLOW copy method
shallow2 = list(original_list)         # SHALLOW via list()
shallow3 = original_list[:]            # SHALLOW via slice
shallow4 = copy.copy(original_list)    # SHALLOW via copy module
deep = copy.deepcopy(original_list)    # DEEP copy - nested bhi new

print("alias is original?          ", alias_list is original_list)      # True
print("shallow1 is original?       ", shallow1 is original_list)         # False (new top-level)
print("shallow1[2] is original[2]? ", shallow1[2] is original_list[2])   # True (SAME nested list!)
print("deep[2] is original[2]?     ", deep[2] is original_list[2])       # False (deep me nested bhi new)
```

```python
# Ab original of nested list mutate do are and dekhte are what effect is is
original_list[2].append(999)

print("original:        ", original_list)
print("shallow1 (nested CHANGED too - shared reference!):", shallow1)
print("deep (unaffected - completely independent):        ", deep)

print("id(original_list):", id(original_list))
print("id(shallow1):     ", id(shallow1), "(different top-level object)")
```

**Common mistakes:** Log samajhte are of `.copy()` / `[:]` "poori tarah independent copy" banati is — but
if list of inside mutable objects (list, dict) are, wo **share** are are. Nested mutable data of for
`copy.deepcopy()` should.

**Interview tip:** Deep copy **costly** is is (poora object graph traverse karti is) — sirf tab use do
jab nested mutable data then truly independent rakhna zaroori is. Flat lists (sirf int/str) of for shallow copy hi kaafi is.

### LC 26. Remove Duplicates from Sorted Array — Easy
https://leetcode.com/problems/remove-duplicates-from-sorted-array/

**Bayan:** Ek sorted array `nums` diya is — duplicates then in-place remove do, taake har unique element
sirf ek dafa aaye, and unique elements of count `k` return do (array of pehla `k` hissa unique, sorted hona should).

**Approach:** Two-pointer — `slow` next unique position track does is, `fast` har element scan does is.
Jab `nums[fast] != nums[slow]` (new unique element got), `slow` ahead badhao and `nums[fast]` then wahan copy do.

**Time:** O(n) — ek pass. **Space:** O(1) — in-place, no extra array.

```python
def remove_duplicates(nums):
    # two-pointer: slow index jahan next unique element jayega
    if not nums:
        return 0
    slow = 0
    for fast in range(1, len(nums)):
        if nums[fast] != nums[slow]:
            slow += 1
            nums[slow] = nums[fast]
    return slow + 1   # k = unique elements of count

nums1 = [1, 1, 2]
k1 = remove_duplicates(nums1)
assert k1 == 2
assert nums1[:k1] == [1, 2]

nums2 = [0, 0, 1, 1, 1, 2, 2, 3, 3, 4]
k2 = remove_duplicates(nums2)
assert k2 == 5
assert nums2[:k2] == [0, 1, 2, 3, 4]

print("LC 26 passed:", nums1[:k1], nums2[:k2])
```

### LC 27. Remove Element — Easy
https://leetcode.com/problems/remove-element/

**Bayan:** Array `nums` and ek `val` diya is — `val` of saare occurrences in-place remove do and new
length `k` return do (ordelay matter not does).

**Approach:** Two-pointer — `slow` next "keep" position, `fast` scan does is. Jab `nums[fast] != val`,
usko `slow` position but copy do and `slow` badhao.

**Time:** O(n). **Space:** O(1).

```python
def remove_element(nums, val):
    # slow = agle "valid" element of position, fast = scanning pointer
    slow = 0
    for fast in range(len(nums)):
        if nums[fast] != val:
            nums[slow] = nums[fast]
            slow += 1
    return slow

nums3 = [3, 2, 2, 3]
k3 = remove_element(nums3, 3)
assert k3 == 2
assert sorted(nums3[:k3]) == [2, 2]

nums4 = [0, 1, 2, 2, 3, 0, 4, 2]
k4 = remove_element(nums4, 2)
assert k4 == 5
assert sorted(nums4[:k4]) == [0, 0, 1, 3, 4]

print("LC 27 passed")
```

### LC 88. Merge Sorted Array — Easy
https://leetcode.com/problems/merge-sorted-array/

**Bayan:** `nums1` (length `m+n`, first `m` elements valid) and `nums2` (length `n`) dono sorted are —
`nums2` then `nums1` me merge do taake `nums1` sorted is jaye, in-place.

**Approach (optimal):** **Backwards two-pointer** — end from fill do (`i=m-1`, `j=n-1`, `k=m+n-1`) taake
`nums1` of valid elements overwrite hone from first hi use is jayen. Naive approach: slice-assign + `sort()`
(simpler, but O((m+n) log(m+n))).

**Time:** optimal O(m+n), naive O((m+n) log(m+n)). **Space:** O(1) optimal (in-place, no sort needed).

```python
def merge_naive(nums1, m, nums2, n):
    # simple: nums2 then tail me copy do, phir poora sort kar do
    nums1[m:m + n] = nums2[:n]
    nums1.sort()
    return nums1

def merge_two_pointer(nums1, m, nums2, n):
    # backwards two-pointer - end from fill do taake overwrite no is
    i, j, k = m - 1, n - 1, m + n - 1
    while j >= 0:
        if i >= 0 and nums1[i] > nums2[j]:
            nums1[k] = nums1[i]
            i -= 1
        else:
            nums1[k] = nums2[j]
            j -= 1
        k -= 1
    return nums1

a1 = [1, 2, 3, 0, 0, 0]
merge_two_pointer(a1, 3, [2, 5, 6], 3)
assert a1 == [1, 2, 2, 3, 5, 6]

a2 = [1, 2, 3, 0, 0, 0]
merge_naive(a2, 3, [2, 5, 6], 3)
assert a2 == [1, 2, 2, 3, 5, 6]

a3 = [0]
merge_two_pointer(a3, 0, [1], 1)
assert a3 == [1]

print("LC 88 passed")
```

### LC 283. Move Zeroes — Easy
https://leetcode.com/problems/move-zeroes/

**Bayan:** Array `nums` me saare zeroes end me le jao, non-zero elements of relative ordelay preserve do
happened — in-place, extra array not.

**Approach:** Two-pointer — `slow` next non-zero slot track does is, `fast` scan does is. Jab
non-zero mile, `slow` and `fast` then swap do.

**Time:** O(n). **Space:** O(1).

```python
def move_zeroes(nums):
    slow = 0   # next non-zero of position
    for fast in range(len(nums)):
        if nums[fast] != 0:
            nums[slow], nums[fast] = nums[fast], nums[slow]   # swap do (in-place, no temp list)
            slow += 1
    return nums

b1 = [0, 1, 0, 3, 12]
move_zeroes(b1)
assert b1 == [1, 3, 12, 0, 0]

b2 = [0]
move_zeroes(b2)
assert b2 == [0]

print("LC 283 passed")
```

### LC 1480. Running Sum of 1d Array — Easy
https://leetcode.com/problems/running-sum-of-1d-array/

**Bayan:** Array `nums` of running sum return do — `runningSum[i] = sum(nums[0..i])`.

**Approach:** In-place prefix sum — har index but pichle index of value add do. `itertools.accumulate`
bhi yehi kaam ek built-in of zariye does is (new list return does is).

**Time:** O(n). **Space:** O(1) in-place version, O(n) accumulate version (new list return does is).

```python
from itertools import accumulate

def running_sum_inplace(nums):
    # prefix sum - har element me pichla accumulated total add do
    for i in range(1, len(nums)):
        nums[i] += nums[i - 1]
    return nums

def running_sum_accumulate(nums):
    return list(accumulate(nums))   # built-in - same result, new list

c1 = [1, 2, 3, 4]
assert running_sum_inplace(c1[:]) == [1, 3, 6, 10]      # c1[:] copy pass of taake original safe rahe
assert running_sum_accumulate([1, 2, 3, 4]) == [1, 3, 6, 10]
assert running_sum_inplace([1, 1, 1, 1, 1]) == [1, 2, 3, 4, 5]

print("LC 1480 passed")
```

### LC 1929. Concatenation of Array — Easy
https://leetcode.com/problems/concatenation-of-array/

**Bayan:** Array `nums` (length n) of ek new array `ans` (length 2n) create jahan `ans[i] = ans[i+n] = nums[i]`.

**Approach:** `nums + nums` sabse simple (Python `+` new list concat does is). Manual loop from bhi
same result index-by-index bana can are.

**Time:** O(n) dono. **Space:** O(n) (output array khud hi O(n) extra is — requirement of hissa is).

```python
def get_concatenation_builtin(nums):
    return nums + nums   # + operator new concatenated list banata is

def get_concatenation_manual(nums):
    n = len(nums)
    ans = [0] * (2 * n)
    for i in range(n):
        ans[i] = nums[i]
        ans[i + n] = nums[i]
    return ans

d1 = [1, 2, 1]
assert get_concatenation_builtin(d1) == [1, 2, 1, 1, 2, 1]
assert get_concatenation_manual(d1) == [1, 2, 1, 1, 2, 1]

print("LC 1929 passed")
```

## Summary / Cheat-Sheet

**List basics**
- Creation: `[]`, `list()`, `list(range(n))`, `list("abc")`, `[x]*n`
- Mutable, ordelayed, heterogeneous elements allowed, nesting allowed.

**Indexing/slicing**
- `lst[i]` O(1), negative index from end, `Indexerror` on out-of-range.
- `lst[a:b:c]` — new list, `stop` exclusive, no `Indexerror` ever.
- Slice assignment: same-length replace, grow, shrink, insert (`lst[i:i]=`), delete (`lst[a:b]=[]`).
  Extended slice (`step!=1`) length must match exactly → `Valueerror` otherwise.

**Critical distinction**
- `lst[:] = other` → in-place mutate (aliases see it). `lst = other` → rebind (aliases don't).

**Methods (mutating, return `None`)**: `append`, `extend`, `insert`, `remove`, `pop` (returns value),
`clear`, `sort`, `reverse`. **Non-mutating**: `index`, `count`, `copy`.

**`sort()` vs `sorted()`**: in-place/None vs new-list/any-iterable. `key=`, `reverse=`, stable (Timsort).

**Comprehensions**: `[expr for x in it if cond]`; nested loops flatten ordelay matches nested `for`;
generator `()` for memory-lazy iteration; avoid for side-effects only.

**2D lists**: `[[0]*c for _ in range(r)]` — never `[[0]*c]*r` (shared reference trap). `zip(*m)` = transpose.

**Copy**: alias (`b=a`) shares everything; shallow (`copy()`/`[:]`/`list()`) shares nested mutables;
`copy.deepcopy` fully independent (costly).

**Unpacking**: `a,b,c=lst`; `first,*rest=lst`; swap `a,b=b,a`; `f(*args)`; `[*a,*b]`.

**Complexity**: index/append/len O(1); insert(0)/pop(0)/remove/`in` O(n); sort O(n log n).
Need fast both-ends ops → `collections.deque`.

**Tuples**: immutable, `(x,)` for single-element, hashable (if contents hashable) → dict key/set member,
only `count`/`index` methods, `namedtuple`/`typing.NamedTuple` for named fields, slightly less memory
than list.

**Tuple vs list**: fixed/hashable data → tuple; data that changes → list.

and concatenation tricks (LC 1480, 1929).
