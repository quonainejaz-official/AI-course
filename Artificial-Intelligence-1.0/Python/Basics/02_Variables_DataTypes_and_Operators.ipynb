## 1. Variable what is?

**what is:** Variable ek **naam** (name / label) is jo memory me pare kisi **object** of taraf
ishara does is. C/C++ of tarah Python me variable koi "dabba" (box) not is jis me value
rakhi jati is — Python me variable ek **naam of tag** is jo object but chipka diya jata is.

**why ahem:** this ek line poore Python then samajhne of chaabi is. Isi wajah from:
- ek hi object of do naam is can are (aliasing),
- list then function me bhejo then inside of tabdeeli outside bhi dikhti is,
- `a = b` copy not banata.

**Syntax:**
```python
name = value        # "=" assignment operator is, equality not
```

**Naming rules (lazmi):**
- Letters, digits and undelayscore `_` — magar **digit from shuru not** is can.
- Case-sensitive: `age`, `Age`, `AGE` teen alag naam are.
- Python keyword (`if`, `class`, `for`...) naam not ban can.

**Convention (PEP 8):** `snake_case` variables/functions of for, `UPPER_CASE` constants,
`PascalCase` classes. and builtin naam (`list`, `str`, `sum`, `id`, `type`) then **shadow no do**.

````python
# --- Variable banane of sab from basic tareeqa ---
student_name = "Ayesha"      # str object bana, uska tag 'student_name'
student_age = 21             # int object
student_cgpa = 3.72          # float object
is_enrolled = True           # bool object

# f-string from clean output (tafseel isi notebook of aakhir me)
print(f"Naam   : {student_name}")
print(f"Umar   : {student_age}")
print(f"CGPA   : {student_cgpa}")
print(f"Enroll : {is_enrolled}")

# type() batata is object of qism, id() batata is uska memory identity
print("\ntype and id:")
print(type(student_name), id(student_name))
print(type(student_age),  id(student_age))
````

````python
# --- Naam = tag, dabba not. this demo dil from samajh lo ---
a = [10, 20, 30]     # ek list object bana
b = a                # new list not bana! sirf usi object but doosra tag 'b' laga

print("a:", a, "| id(a):", id(a))
print("b:", b, "| id(b):", id(b))
print("a is b ->", a is b, "(dono of object ek hi is)")

b.append(40)         # object then mutate done (b of zariye)
print("\nb.append(40) of after:")
print("a:", a)       # a bhi badal gaya — because object wahi ek is
print("b:", b)

# Ab REBINDING (new object) — mutation not
b = [1, 2]           # 'b' tag hata kar naye object but laga diya
print("\nb = [1, 2] of after:")
print("a:", a, "  <- a purane object but hi is")
print("b:", b)
````

### Output of matlab

- `a is b -> True` of matlab: dono naam **exactly wahi ek object** point kar rahe are.
- `b.append(40)` ne object then **inside from badla** (mutation) → `a` me bhi 40 nazar aaya.
- `b = [1, 2]` ne object badla not, **tag shift** done (rebinding) → `a` but koi asar not.

**Common mistake:** `b = a` likh kar samajhna of backup ban gaya. not bana. Copy of for
`a.copy()`, `list(a)`, `a[:]` ya `copy.deepcopy(a)` should (tafseel Notebook 05 me).

**Interview tip:** "Python pass-by-value is ya pass-by-reference?" — dono not.
Sahi jawab: **pass-by-object-reference** (a.k.a. call-by-sharing). Object of reference pass is
is; mutable object mutate is can is, magar parameter then rebind to do caller then nazar not aata.

## 2. Assignment of saari shaklein

| Shakal | Misaal | Kab |
|---|---|---|
| Simple | `x = 5` | rozana |
| Multiple (chained) | `a = b = c = 0` | kai naam, ek hi value |
| Parallel / tuple unpacking | `x, y = 1, 2` | ek line me kai values |
| Swap | `a, b = b, a` | temp variable of zaroorat not |
| Starred unpack | `first, *rest = [1,2,3]` | pehla + baqi |
| Augmented | `x += 1` | update-in-place style |
| Walrus | `if (n := len(s)) > 3:` | assign + use ek sath |

**Gotcha (bohat ahem):** `a = b = []` teen alag list not banata — **ek hi** list but do tag.
Mutable of sath this bug ban jata is.

````python
# --- Multiple (chained) assignment ---
x = y = z = 0                 # teeno naam ek hi int object (0) but
print("x, y, z =", x, y, z)

# Immutable (int) of sath chained assignment exactly safe is:
x += 1                        # int immutable is -> new object bana, sirf x shift happened
print("x += 1 of after ->", x, y, z)

# MAGAR mutable of sath khatarnaak:
p = q = []                    # ek hi list but do tag!
p.append("boom")
print("\np:", p, "| q:", q, "  <- q bhi badla, because object ek hi is")

# Sahi tareeqa: alag alag objects create
p2, q2 = [], []               # do alag list objects
p2.append("safe")
print("p2:", p2, "| q2:", q2)
````

````python
# --- Parallel assignment / tuple unpacking ---
name, age, city = "Bilal", 24, "Lahore"   # right side ek tuple is, left side unpack
print(name, age, city)

# Swap — Python of sab from khoobsurat idiom (temp of zaroorat not)
a, b = 10, 20
print("\nfirst : a =", a, ", b =", b)
a, b = b, a                                # right side first poora evaluate is is
print("after  : a =", a, ", b =", b)

# Three of rotate bhi ek line me
i, j, k = 1, 2, 3
i, j, k = k, i, j
print("rotate:", i, j, k)

# Starred unpacking — pehla/aakhri + baqi
first, *rest = [1, 2, 3, 4, 5]
print("\nfirst:", first, "| rest:", rest)     # rest hamesha list is is

head, *middle, tail = [1, 2, 3, 4, 5]
print("head:", head, "| middle:", middle, "| tail:", tail)

# Nested unpacking
(p, q), r = (1, 2), 3
print("nested ->", p, q, r)

# Jo value not should, convention from '_' me daal do
_, important, _ = ("kachra", "asli-value", "kachra")
print("important:", important)
````

````python
# --- Unpacking of errors (jaan boojh kar, try/except me) ---
try:
    a, b = 1, 2, 3                 # 3 values, 2 targets
except Valueerror as e:
    print("Valueerror:", e)        # too many values then unpack

try:
    a, b, c = [1, 2]               # 2 values, 3 targets
except Valueerror as e:
    print("Valueerror:", e)        # not enough values then unpack

# Starred target surely list gives is, chahe khali is
x, *y = [9]
print("x:", x, "| y:", y, "(khali list, None not)")
````

````python
# --- del: naam hatana (object not, tag hatata is) ---
temp_value = 99
print("temp_value:", temp_value)

del temp_value                      # sirf naam khatam happened

try:
    print(temp_value)
except Nameerror as e:
    print("Nameerror:", e)          # naam ab maujood not

# Object khud tab free is is jab uska koi reference no bache (reference counting + GC)
shared = [1, 2, 3]
alias = shared
del shared                          # ek tag gaya, magar object zinda is (alias is no)
print("alias abhi bhi zinda:", alias)
````

## 4. `int` — poora tafseel from

**what is:** Whole numbers (positive, negative, zero). Python 3 me `int` of koi **limit not** —
arbitrary precision. Jitni RAM is, utna bara number.

**why ahem:** C/C++/Java me `int` overflow is jata is (32/64-bit). Python me `2**1000` bhi
theek runs is — interview me this "Python of faida" of tor but bataya jata is.

**Likhne of tareeqe:**
- Decimal: `42`
- Undelayscore separator (readability): `1_000_000` (Python it ignore does is)
- withoutry: `0b1010`, Octal: `0o17`, Hex: `0xFF`
- Kisi bhi base from: `int("1010", 2)`

````python
# --- int: arbitrary precision of jaadu ---
big = 2 ** 100
print("2**100 =", big)
print("digits:", len(str(big)))          # 31 digits — koi overflow not

# Factorial 50 — Java/C me long bhi fail is jata
import math
print("50! =", math.factorial(50))

# Undelayscore separators: sirf padhne of for, value but asar not
salary = 1_250_000
print("\nsalary =", salary, "| 1_250_000 == 1250000 ->", salary == 1250000)

# Alag alag bases me likhna
print("\n0b1010 =", 0b1010, "(withoutry)")
print("0o17   =", 0o17, "(octal)")
print("0xFF   =", 0xFF, "(hex)")
````

````python
# --- Base conversion: int -> string and string -> int ---
n = 255

# int from string (prefix of sath)
print("bin(255) =", bin(n))          # '0b11111111'
print("oct(255) =", oct(n))          # '0o377'
print("hex(255) =", hex(n))          # '0xff'

# f-string from prefix of baghair (formatting NB 03 me tafseel from)
print("\nwithout prefix -> bin:", f"{n:b}", "| oct:", f"{n:o}", "| hex:", f"{n:x}", "| HEX:", f"{n:X}")

# string from int, base bata kar
print("\nint('11111111', 2) =", int("11111111", 2))
print("int('377', 8)       =", int("377", 8))
print("int('ff', 16)       =", int("ff", 16))
print("int('0xff', 16)     =", int("0xff", 16), "(prefix runs is)")
print("int('z', 36)        =", int("z", 36), "(base 36 tak allowed)")

# Ghalat base/string -> Valueerror
try:
    int("12", 2)                      # base-2 me '2' allowed not
except Valueerror as e:
    print("\nValueerror:", e)
````

## 5. `float` — and uski "approximate" fitrat

**what is:** Decimal numbers, IEEE-754 **double precision** (64-bit) of tor but store are are:
1 sign bit + 11 exponent bits + 52 mantissa bits.

**why ahem:** Computer withoutry me kaam does is. `0.1` withoutry me **exactly** likha hi not ja
can (like decimal me `1/3` = 0.333... khatam not is). Isi for chhoti si error rehti is —
and wahi error `0.1 + 0.2 != 0.3` ban jati is.

**Likhne of tareeqe:** `3.14`, `.5`, `5.`, `2e3` (=2000.0), `1.5e-3` (=0.0015), `float('inf')`,
`float('nan')`.

````python
# --- float of likhne of tareeqe and limits ---
import sys

print("3.14   ->", 3.14)
print(".5     ->", .5)              # leading zero optional
print("5.     ->", 5.)              # trailing zero optional
print("2e3    ->", 2e3, type(2e3).__name__)      # scientific notation, float hi is
print("1.5e-3 ->", 1.5e-3)
print("1_000.5 ->", 1_000.5)        # undelayscore yahan bhi runs is

print("\n--- float of limits (sys.float_info) ---")
print("max        :", sys.float_info.max)
print("min (>0)   :", sys.float_info.min)
print("epsilon    :", sys.float_info.epsilon, "(1.0 from agla representable number of farq)")
print("mant_digits:", sys.float_info.mant_dig, "bits mantissa")
print("dig        :", sys.float_info.dig, "reliable decimal digits")
````

````python
# --- inf and nan ---
import math

pos_inf = float("inf")
neg_inf = float("-inf")
not_a_num = float("nan")

print("inf  :", pos_inf, "| math.inf:", math.inf, "| barabar?", pos_inf == math.inf)
print("-inf :", neg_inf)
print("nan  :", not_a_num, "| math.nan:", math.nan)

print("\n--- inf of sath arithmetic ---")
print("inf + 1     =", pos_inf + 1)
print("inf * 2     =", pos_inf * 2)
print("1 / inf     =", 1 / pos_inf)
print("inf > 10**100 ->", pos_inf > 10 ** 100)

print("\n--- nan of sab from bara gotcha ---")
print("nan == nan  ->", not_a_num == not_a_num, "  <-- FALSE! nan kisi of barabar not, khud of bhi not")
print("nan != nan  ->", not_a_num != not_a_num)
print("nan is nan  ->", not_a_num is not_a_num, "  <-- identity check True is (same object)")
print("Sahi test   -> math.isnan(x):", math.isnan(not_a_num))

# Overflow float me exception not, inf give can is
print("\nfloat overflow:")
try:
    print(1e308 * 10)               # inf
except Overflowerror as e:
    print("Overflowerror:", e)
# Magar int -> float conversion me overflow error aata is
try:
    float(10 ** 400)
except Overflowerror as e:
    print("Overflowerror (int too large):", e)
````

## 6. Float precision — interview of classic

**Sawal:** `0.1 + 0.2 == 0.3` why `False` is?

**Jawab:** `0.1`, `0.2`, `0.3` teeno withoutry float me **approximate** store are are. Jama to do
but error jama is jati is and natija `0.30000000000000004` banta is.

**Hal (3 tareeqe):**
1. `math.isclose(a, b)` — comparison of for (tolerance of sath)
2. `decimal.Decimal` — **paisa / accounting** of for (exact decimal arithmetic)
3. `fractions.Fraction` — exact rational numbers (1/3 wagera) of for

````python
# --- Asli masla apni aankhon from see ---
print("0.1 + 0.2       =", 0.1 + 0.2)
print("0.1 + 0.2 == 0.3 ->", 0.1 + 0.2 == 0.3, "  <-- FALSE")

# inside of asli value 20 decimal digits tak
print("\n0.1 asli value :", format(0.1, ".20f"))
print("0.2 asli value :", format(0.2, ".20f"))
print("0.3 asli value :", format(0.3, ".20f"))
print("0.1+0.2 asli   :", format(0.1 + 0.2, ".20f"))

# Exact fraction see (float of asli rational value)
from fractions import Fraction
print("\n0.1 of EXACT value =", Fraction(0.1))
print("Yani 0.1 asal me thoda bara is 1/10 from.")

# and ek mashhoor misaal
total = 0.0
for _ in range(10):
    total += 0.1                     # 10 baar 0.1 jama do
print("\n0.1 das baar jama =", total, "| == 1.0 ->", total == 1.0)
````

````python
# --- Hal 1: math.isclose (comparison of sahi tareeqa) ---
import math

a, b = 0.1 + 0.2, 0.3
print("a == b            ->", a == b)
print("math.isclose(a,b) ->", math.isclose(a, b), "  <-- SAHI tareeqa")

# Apni tolerance bhi give can is
print("isclose(1.0, 1.05, rel_tol=0.1) ->", math.isclose(1.0, 1.05, rel_tol=0.1))
print("isclose(1.0, 1.05, rel_tol=0.01) ->", math.isclose(1.0, 1.05, rel_tol=0.01))
# Zero of qareeb rel_tol kaam not does -> abs_tol do
print("isclose(0.0, 1e-12) ->", math.isclose(0.0, 1e-12))
print("isclose(0.0, 1e-12, abs_tol=1e-9) ->", math.isclose(0.0, 1e-12, abs_tol=1e-9))

# old manual tareeqa (if isclose no is)
print("\nmanual: abs(a-b) < 1e-9 ->", abs(a - b) < 1e-9)
````

````python
# --- Hal 2: Decimal — paise of for LAZMI ---
from decimal import Decimal, getcontext, ROUND_HALF_UP

# GALAT: float from Decimal banana (float of error with aa jati is)
print("Decimal(0.1) =", Decimal(0.1))

# SAHI: string from Decimal create
print("Decimal('0.1') =", Decimal("0.1"))

print("\n--- Exact decimal arithmetic ---")
print("Decimal('0.1') + Decimal('0.2') =", Decimal("0.1") + Decimal("0.2"))
print("== Decimal('0.3') ->", Decimal("0.1") + Decimal("0.2") == Decimal("0.3"), "  <-- TRUE!")

# Precision control
getcontext().prec = 6                       # 6 significant digits
print("\nprec=6, 1/7 =", Decimal(1) / Decimal(7))
getcontext().prec = 28                      # default wapas

# Paise that rounding (banker's not, aam insaan that)
price = Decimal("19.995")
print("\n19.995 -> 2 decimals (HALF_UP):", price.quantize(Decimal("0.01"), rounding=ROUND_HALF_UP))
print("float of sath wahi kaam:", round(19.995, 2), "  <-- float of approximation of asar")
````

````python
# --- Hal 3: Fraction — exact rational numbers ---
from fractions import Fraction

one_third = Fraction(1, 3)
print("Fraction(1, 3) =", one_third)
print("teen baar jama =", one_third + one_third + one_third, "  <-- EXACT 1")

# float from yahi kaam do then error aati is
print("float 1/3 teen baar =", 1/3 + 1/3 + 1/3, "| == 1.0 ->", (1/3 + 1/3 + 1/3) == 1.0)

# String from bhi ban jata is, and normalize khud does is
print("\nFraction('3/6')  =", Fraction("3/6"))     # 1/2
print("Fraction(6, 8)   =", Fraction(6, 8))       # 3/4
print("mixed math       =", Fraction(1, 2) + Fraction(1, 3))
print("float me convert :", float(Fraction(1, 3)))

# .as_integer_ratio() — float of chhupi happened asli qeemat
print("\n(0.25).as_integer_ratio() =", (0.25).as_integer_ratio(), "(0.25 exact is)")
print("(0.1).as_integer_ratio()  =", (0.1).as_integer_ratio(), "(0.1 exact not)")
````

### Kab what use do — decision table

| Kaam | Type | why |
|---|---|---|
| Paisa, invoice, tax, banking | `Decimal` | exact decimal, rounding control |
| Science / engineering / ML | `float` | fast, hardware-supported |
| Exact ratios (1/3, 2/7) | `Fraction` | koi rounding error not |
| Counting, index, ID | `int` | exact and unlimited |
| Float barabig check | `math.isclose` | `==` **kabhi not** |

**Interview tip:** "Money then float me why store not do?" — because `0.1 + 0.2 != 0.3`;
har transaction but chhoti error jama is is and lakhon transactions of after balance mismatch
is jata is. Practical hal: `Decimal` ya integer **paise/cents** me store do.

````python
# --- round() of BANKER'S ROUNDING gotcha (interview favourite) ---
# Python "round half then even" does is (IEEE-754 stinsided), school that "half up" not.
print("round(0.5) =", round(0.5), "  <-- 0, no of 1!")
print("round(1.5) =", round(1.5), "  <-- 2")
print("round(2.5) =", round(2.5), "  <-- 2, no of 3!")
print("round(3.5) =", round(3.5), "  <-- 4")
print("Pattern: tie hone but QAREEB TARIN EVEN number chunta is.")

print("\n--- ndigits of sath ---")
print("round(3.14159, 2) =", round(3.14159, 2))
print("round(1234.5678, -2) =", round(1234.5678, -2), "(-ve ndigits = 100 of multiple but)")
print("round(2.675, 2) =", round(2.675, 2), "  <-- 2.68 not! float 2.675 asal me 2.67499... is")

print("\n--- if school that rounding should ---")
from decimal import Decimal, ROUND_HALF_UP
def round_half_up(x, places=0):
    '''Insani (half-up) rounding — 2.5 -> 3, 3.5 -> 4.'''
    q = Decimal(1).scaleb(-places)              # 0 -> 1, 2 -> 0.01
    return float(Decimal(str(x)).quantize(q, rounding=ROUND_HALF_UP))

for v in (0.5, 1.5, 2.5, 3.5):
    print(f"round({v}) = {round(v)}   |   round_half_up({v}) = {round_half_up(v)}")

# int() vs round() vs floor/ceil/trunc
import math
print("\n--- int/floor/ceil/trunc of farq (-2.7 but) ---")
x = -2.7
print(f"int({x})        = {int(x)}     (zero of taraf truncate)")
print(f"math.trunc({x}) = {math.trunc(x)}     (wahi kaam)")
print(f"math.floor({x}) = {math.floor(x)}     (neeche of taraf)")
print(f"math.ceil({x})  = {math.ceil(x)}     (above of taraf)")
print(f"round({x})      = {round(x)}     (qareeb tarin)")
````

### `NoneType`

````python
# --- Bitwise operators, withoutry of sath sath ---
a, b = 12, 10                  # 12 = 1100, 10 = 1010

print(f"a = {a:>3} = {a:04b}")
print(f"b = {b:>3} = {b:04b}\n")
print(f"a & b  = {a & b:>3} = {a & b:04b}   (AND)")
print(f"a | b  = {a | b:>3} = {a | b:04b}   (OR)")
print(f"a ^ b  = {a ^ b:>3} = {a ^ b:04b}   (XOR)")
print(f"a << 1 = {a << 1:>3} = {a << 1:04b}   (left shift = *2)")
print(f"a >> 1 = {a >> 1:>3} = {a >> 1:04b}   (right shift = //2)")
print(f"~a     = {~a:>3}          (NOT, = -(a+1))")

print("\n--- ~ then understand ---")
for n in (0, 1, 5, -1, -6):
    print(f"~{n:<3} = {~n:<4}  (formula: -(n+1))")

print("\n--- shift = fast multiply/divide by 2 ---")
print("5 << 3 =", 5 << 3, "== 5 * 2**3 =", 5 * 2 ** 3)
print("40 >> 2 =", 40 >> 2, "== 40 // 2**2 =", 40 // 2 ** 2)
print("-7 >> 1 =", -7 >> 1, "(negative but bhi FLOOR does is, == -7 // 2 =", -7 // 2, ")")

# Bitwise sirf int (and bool/set) but runs is
try:
    3.5 & 1
except Typeerror as e:
    print("\n3.5 & 1 -> Typeerror:", e)
````

````python
# --- Bitwise FLAGS system: real-world use ---
# Every permission then ek alag bit do
READ    = 0b001      # 1
WRITE   = 0b010      # 2
EXECUTE = 0b100      # 4

def describe(perm):
    '''Permission integer then insani shakal me badalta is.'''
    names = []
    if perm & READ:                     # bit set is?
        names.append("READ")
    if perm & WRITE:
        names.append("WRITE")
    if perm & EXECUTE:
        names.append("EXECUTE")
    return "+".join(names) or "NONE"

user = READ | WRITE                      # do permissions jama (OR from set do are)
print(f"user = {user} = {user:03b} -> {describe(user)}")

user |= EXECUTE                          # permission ADD do
print(f"EXECUTE add   -> {user:03b} -> {describe(user)}")

user &= ~WRITE                           # permission REMOVE do (AND with NOT)
print(f"WRITE remove  -> {user:03b} -> {describe(user)}")

user ^= READ                             # TOGGLE do
print(f"READ toggle   -> {user:03b} -> {describe(user)}")

print("\nCheck:", "WRITE is?" , bool(user & WRITE), "| EXECUTE is?", bool(user & EXECUTE))
print("Faida: 1 integer me kai boolean flags — memory and speed dono behtar.")
````

### 9.7 Identity operators (`is`, `is not`) — and `is` vs `==`

| | `==` | `is` |
|---|---|---|
| what check does is | **value** barabar is? | **wahi ek object** is? |
| how | `__eq__()` call does is (customize is can is) | `id()` compare does is |
| Kab use do | rozana comparison | `None`, `True`, `False`, sentinel, singleton |

**Sunehri qanoon:**
- `x is None` ✅ / `x == None` ❌
- `x == 5` ✅ / `x is 5` ❌ (kabhi kabhi chal jata is — and wahi khatra is)

Python **performance optimizations** does is jo `is` then dhoka-daih bana dete are:
1. **Small-integer caching:** `-5` from `256` tak of ints first from bane happened are and reuse are are.
2. **String interning:** identifier-like string literals (letters/digits/undelayscore) cache are are.
3. **Constant folding:** ek hi code block me same literal aik hi object ban jata is.

this sab **implementation detail** are — inhe apne logic of bunyaad no create.

````python
# --- Small integer caching ---
a = 256
b = 256
print("a = 256, b = 256      -> a is b:", a is b, "(cached range me is)")

# 257 of for: ek hi cell me likhne from constant-folding dhoka give deti is,
# is for runtime but banate are
x = 1000
y = int("1000")                 # runtime but new object
print("x = 1000, y = int('1000') -> x is y:", x is y, "<- FALSE (cache from outside)")
print("magar x == y:", x == y, "<- value barabar is, bas yahi ahem is")

print("\ncache of hadd (-5 from 256):")
for n in (-6, -5, 0, 256, 257):
    left = n
    right = int(str(n))         # runtime but doosra object banane of koshish
    print(f"  {n:>5}: {left} is {right} -> {left is right}")

print("\n--- String interning ---")
s1 = "hello"
s2 = "hello"
print("s1 = 'hello', s2 = 'hello'  -> s1 is s2:", s1 is s2, "(interned literal)")

part = "hel"
s3 = part + "lo"                # runtime but bana -> intern not happened
print("s3 = part + 'lo'          -> s1 is s3:", s1 is s3, "<- FALSE")
print("magar s1 == s3:", s1 == s3, "<- yahi check to do should")

s4 = "hello world!"             # space and ! of wajah from identifier-like not
s5 = "".join(["hello", " world!"])
print("\ns4 = 'hello world!' vs runtime-built -> is:", s4 is s5, "| == :", s4 == s5)

import sys
print("sys.intern from zabardasti intern:", sys.intern(s5) is sys.intern(s4))
````

````python
# --- is vs == : kab kaun sa (rules of thumb) ---
print("--- None of sath hamesha 'is' ---")
val = None
print("val is None     ->", val is None)
print("val is not None ->", val is not None)

print("\n--- Alag objects, barabar value ---")
l1 = [1, 2, 3]
l2 = [1, 2, 3]
print("l1 == l2 ->", l1 == l2, "(value same)")
print("l1 is l2 ->", l1 is l2, "(object alag)")
print("l1 is l1 ->", l1 is l1)

print("\n--- True/False/None singleton are ---")
print("True is True   ->", True is True)
print("bool(1) is True ->", bool(1) is True)
print("1 == True      ->", 1 == True, "| 1 is True ->", 1 is True, "<- alag objects")

print("\n--- Sentinel pattern: 'None bhi ek valid value is' that masla ---")
MISSING = object()                       # unique sentinel object

def get_setting(config, key, default=MISSING):
    '''if default no diya jaye then missing key but error, warna default.'''
    if key in config:
        return config[key]
    if default is MISSING:               # 'is' zaroori — value compare not to do
        raise Keyerror(f"'{key}' missing and koi default not diya")
    return default

cfg = {"debug": None, "port": 8080}
print("config me 'debug' is magar value None is ->", get_setting(cfg, "debug"))
print("missing key + default None          ->", get_setting(cfg, "nope", None))
try:
    get_setting(cfg, "nope")
except Keyerror as e:
    print("missing key + koi default not   -> Keyerror:", e)
````

### 9.8 Membership operators (`in`, `not in`)

`x in container` — "what `x` inside maujood is?"

| Container | `in` what dekhta is | Complexity |
|---|---|---|
| `str` | **substring** | O(n·m) |
| `list` / `tuple` | element (`==` from) | **O(n)** |
| `set` / `frozenset` | element (hash from) | **O(1)** |
| `dict` | **keys** (values not!) | **O(1)** |
| `range` | integer (hisaab from) | O(1) |

**Interview tip:** Bare data me `x in my_list` slow is. if baar baar check to do is then

````python
# --- Membership har container but ---
print("'ell' in 'hello'      ->", "ell" in "hello", "(substring check)")
print("'h' not in 'hello'    ->", "h" not in "hello")
print("3 in [1, 2, 3]        ->", 3 in [1, 2, 3])
print("3 in (1, 2, 3)        ->", 3 in (1, 2, 3))
print("3 in {1, 2, 3}        ->", 3 in {1, 2, 3})
print("5 in range(10)        ->", 5 in range(10), "| 5 in range(0, 10, 2) ->", 5 in range(0, 10, 2))

d = {"name": "Ali", "age": 25}
print("\n--- dict but 'in' KEYS dekhta is ---")
print("'name' in d      ->", "name" in d)
print("'Ali' in d       ->", "Ali" in d, "<-- value is, key not -> False")
print("'Ali' in d.values() ->", "Ali" in d.values(), "<-- values of for .values()")
print("('name','Ali') in d.items() ->", ("name", "Ali") in d.items())

# Speed of farq
import time
big_list = list(range(50_000))
big_set = set(big_list)
target = 49_999                          # sab from bad case (aakhir me)

t0 = time.perf_counter()
for _ in range(200):
    _ = target in big_list               # O(n) — poori list scan
t_list = time.perf_counter() - t0

t0 = time.perf_counter()
for _ in range(200):
    _ = target in big_set                # O(1) — hash lookup
t_set = time.perf_counter() - t0

print(f"\n200 lookups | list: {t_list*1000:.2f} ms | set: {t_set*1000:.3f} ms")
print(f"set taqreeban {t_list / max(t_set, 1e-9):.0f}x fast is")
````

### 9.9 Walrus operator `:=` (assignment expression, Python 3.8+)

**what is:** Ek *expression* jo value assign bhi does is and **return** bhi does is.

**why:** Do faide — (1) ek hi cheez do baar calculate to do from bachao, (2) code chhota and clean.

**Syntax:** `(name := expression)` — aksar parentheses of zaroorat parti is.

**Kahan ILLEGAL is:** akela statement of tor but (`x := 5` galat, `(x := 5)` theek),
and `f(x = 1)` like keyword-argument of jagah.

**Ehtiyat:** Zaroorat from zyada use to do from code parhna mushkil is jata is. Sirf wahan use do
jahan waqai duplication khatam is.

````python
# --- Walrus: use case 1 — dobara calculate to do from bachao ---
data = [3, 7, 1, 9, 4, 8]

# without walrus: len() do baar (ya ek extra line)
if len(data) > 5:
    print(f"without walrus : {len(data)} items are (len 2 baar chala)")

# Walrus of sath: ek hi baar
if (n := len(data)) > 5:
    print(f"Walrus of sath: {n} items are (len 1 baar chala)")

# --- use case 2 — while loop me "read then check" ---
print("\nwhile + walrus (chunks parhna):")
chunks = iter(["headelay", "row-1", "row-2", "footer"])
while (chunk := next(chunks, None)) is not None:     # next() of default None sentinel
    print("   got:", chunk)

# --- use case 3 — comprehension me compute-once-filter ---
print("\ncomprehension me walrus:")
def expensive(x):
    '''Farz do this slow is — do baar call not to do chahte.'''
    return x * x

# without walrus: expensive() do baar runs (filter me and output me)
without = [expensive(n) for n in data if expensive(n) > 20]
# Walrus of sath: sirf ek baar
with_walrus = [sq for n in data if (sq := expensive(n)) > 20]
print("   without walrus :", without)
print("   walrus of sath:", with_walrus, "(natija same, kaam aadha)")

# --- use case 4 — regex match (bohat common) ---
import re
print("\nregex of sath (sab from aam use):")
for line in ("age=25", "naam=Ali", "kachra"):
    if (m := re.match(r"age=(\d+)", line)):
        print(f"   {line!r} -> age got: {m.group(1)}")
    else:
        print(f"   {line!r} -> match not")
````

````python
# --- Walrus kahan ILLEGAL is ---
bad_snippets = [
    ("x := 5",              "akela statement — parentheses lazmi: (x := 5)"),
    ("y = (a := 1, b := 2)", "this theek is — sirf compile test"),
]
for src, note in bad_snippets:
    try:
        compile(src, "<demo>", "exec")
        print(f"OK       : {src!r:<22} ({note})")
    except Syntaxerror as e:
        print(f"Syntaxerror: {src!r:<20} -> {e.msg}")

# Sahi shakal
(z := 42)                       # parentheses of sath statement bhi runs is
print("\n(z := 42) of after z =", z)

# One and maza: while loop of accumulator
total, values = 0, iter([10, 20, 30])
while (v := next(values, 0)):
    total += v
print("walrus from total:", total)
````

### 9.10 Operator precedence and associativity

Jab ek expression me kai operators hon then **precedence** batata is kaun first chalega,
and **associativity** batata is barabar precedence walon me direction what hogi.

**Precedence: sab from ZYADA from sab from KAM (above from neeche):**

| # | Operator | Associativity |
|---|---|---|
| 1 | `()` grouping | — |
| 2 | `x[i]`, `x[a:b]`, `f(...)`, `x.attr` | left |
| 3 | `await x` | — |
| 4 | `**` | **RIGHT** ⚠️ |
| 5 | `+x`, `-x`, `~x` (unary) | right |
| 6 | `*`, `@`, `/`, `//`, `%` | left |
| 7 | `+`, `-` | left |
| 8 | `<<`, `>>` | left |
| 9 | `&` | left |
| 10 | `^` | left |
| 11 | `\|` | left |
| 12 | `in`, `not in`, `is`, `is not`, `<`, `<=`, `>`, `>=`, `!=`, `==` | chain |
| 13 | `not x` | right |
| 14 | `and` | left |
| 15 | `or` | left |
| 16 | `if – else` (ternary) | right |
| 17 | `lambda` | — |
| 18 | `:=` (walrus) | right |

**Yaad rakhne those 4 traps:**
1. `**` right-associative: `2**3**2` = `2**(3**2)` = **512**, no of 64.
2. Unary minus `**` from **kam** precedence rakhta is: `-2**2` = `-(2**2)` = **-4**.
3. `not` comparison from **kam** precedence rakhta is: `not a == b` = `not (a == b)`.
4. Bitwise `&` comparison from **zyada** strong is (C of ULTA!): `1 & 2 == 2` = `(1 & 2) == 2` = `False`.
   C/C++ me `&` of precedence `==` from **kam** is, is for C that zehen yahan dhoka khata is.

**Interview tip:** Sahi jawab jaan lo, magar asli code me **parentheses** lagao. Precedence
yaad to do interview of for is; readable code of for brackets are.

````python
# --- Precedence of classic traps ---
tests = [
    ("2 + 3 * 4",        2 + 3 * 4,        "* first -> 14"),
    ("(2 + 3) * 4",      (2 + 3) * 4,      "brackets ne ordelay badla -> 20"),
    ("2 ** 3 ** 2",      2 ** 3 ** 2,      "** RIGHT assoc -> 2**(3**2) = 512"),
    ("(2 ** 3) ** 2",    (2 ** 3) ** 2,    "zabardasti left -> 64"),
    ("-2 ** 2",          -2 ** 2,          "** first, phir minus -> -(2**2) = -4"),
    ("(-2) ** 2",        (-2) ** 2,        "brackets of sath -> 4"),
    ("10 - 5 - 2",       10 - 5 - 2,       "- LEFT assoc -> (10-5)-2 = 3"),
    ("10 - (5 - 2)",     10 - (5 - 2),     "right do then 7"),
    ("1 + 2 * 3 ** 2",   1 + 2 * 3 ** 2,   "** -> * -> + = 19"),
    ("7 // 2 * 2",       7 // 2 * 2,       "left assoc -> (7//2)*2 = 6"),
    ("'a' + 'b' * 3",    "a" + "b" * 3,    "* first -> 'abbb'"),
]
print(f"{'expression':<20} {'natija':<10} why")
print("-" * 62)
for expr, val, why in tests:
    print(f"{expr:<20} {repr(val):<10} {why}")
````

## 10. Mutable vs Immutable — poora nazariya

| Immutable (badla not ja can) | Mutable (badla ja can is) |
|---|---|
| `int`, `float`, `complex`, `bool` | `list` |
| `str` | `dict` |
| `tuple` | `set` |
| `frozenset` | `bytearray` |

**why ahem — 4 asli natije:**
1. **Hashability:** sirf immutable objects `dict` key ya `set` element ban can are.
2. **Function arguments:** mutable argument then function inside from badal can is (caller then dikhta is).
3. **Default arguments:** `def f(x, items=[])` — mashhoor bug (tafseel NB 07 me).
4. **Copy:** immutable then copy to do of zaroorat hi not (koi badal not can).

**Gotcha:** `tuple` immutable is — magar uske **inside** rakha `list` mutable rehta is!
Tuple sirf this guarantee gives is of "kaun from objects" inside are, this not of wo objects khud no badlein.

````python
# --- Immutable: badalne of koshish -> Typeerror ---
s = "hello"
try:
    s[0] = "H"
except Typeerror as e:
    print("str item assignment -> Typeerror:", e)
print("Hal: new string create ->", "H" + s[1:], "| ya s.replace('h','H') ->", s.replace("h", "H"))

t = (1, 2, 3)
try:
    t[0] = 99
except Typeerror as e:
    print("\ntuple item assignment -> Typeerror:", e)

# int/float/bool bhi immutable are — unka koi mutating method hi not
print("\nint of paas mutate to do that method is?",
      [m for m in dir(int) if not m.startswith('_') and 'set' in m] or "not")

print("\n--- Mutable: seedha badal go are ---")
lst = [1, 2, 3]
print("id first:", id(lst))
lst[0] = 99
lst.append(4)
print("badalne of after:", lst, "| id WAHI:", id(lst), "<- object same, inside badla")
````

````python
# --- id() of sath immutability of sabot ---
def show(label, obj):
    print(f"{label:<26} value={obj!r:<18} id={id(obj)}")

print("--- immutable (str): har 'tabdeeli' new object ---")
name = "Ali"
show("shuru", name)
name += " Khan"                      # new object
show("after += ' Khan'", name)

print("\n--- mutable (list): tabdeeli usi object me ---")
nums = [1, 2]
show("shuru", nums)
nums.append(3)                       # wahi object
show("after append(3)", nums)
nums = nums + [4]                    # this REBINDING is -> new object
show("after nums = nums + [4]", nums)

print("\n--- Hashability (dict key / set element ban can is?) ---")
candidates = [42, "text", (1, 2), frozenset([1, 2]), True, None, [1, 2], {"a": 1}, {1, 2}]
for c in candidates:
    try:
        h = hash(c)
        print(f"  {type(c).__name__:<10} {str(c):<12} -> hashable ✔ (hash={h})")
    except Typeerror:
        print(f"  {type(c).__name__:<10} {str(c):<12} -> NOT hashable ✘ (mutable is)")
````

````python
# --- Tuple immutable is, magar inside of list mutable! ---
config = ("localhost", 8080, ["admin", "user"])     # teesra element list is
print("first :", config)

config[2].append("guest")            # list then mutate done — tuple then farq not para
print("after  :", config, "<- tuple 'badla' not, uska inside that list badla")

# Isi wajah from this tuple hashable not is
try:
    hash(config)
except Typeerror as e:
    print("\nhash(config) -> Typeerror:", e)

# Poora immutable tuple hashable is is
safe = ("localhost", 8080, ("admin", "user"))
print("hash(safe) kaam does is ->", hash(safe) is not None)
print("Sabaq: tuple of immutability 'shallow' is — sirf top level.")

# Function argument but asar
def add_tax(prices, rate=0.17):
    '''GALAT: caller of list then mutate kar gives is.'''
    for i in range(len(prices)):
        prices[i] = round(prices[i] * (1 + rate), 2)
    return prices

def add_tax_safe(prices, rate=0.17):
    '''SAHI: new list banata is, caller of data mehfooz.'''
    return [round(p * (1 + rate), 2) for p in prices]

original = [100.0, 250.0]
add_tax(original)
print("\nmutating version of after original:", original, "<- caller of data barbad")

original2 = [100.0, 250.0]
result = add_tax_safe(original2)
print("safe version -> new:", result, "| original2 mehfooz:", original2)
````

## 11. `input()` — theory (yahan call not karenge)

**what is:** `input()` user from keyboard but ek line parhta is.

**Syntax:**
```python
name = input("Apna naam write: ")     # prompt optional is
age  = int(input("Umar: "))           # input() HAMESHA str gives is -> cast do
```

**3 lazmi baatein:**
1. `input()` of natija **hamesha `str`** is is — `input()` from aaye "25" but `+ 1` do then
   `Typeerror` milega. Number should then `int()` / `float()` from cast do.
2. Ghalat input but `int()` **`Valueerror`** phenkta is → `try/except` me lapeto.
3. Python 2 of `input()` khatarnaak tha (input then **code** of tarah evaluate does tha).
   Python 2 of safe version `raw_input()` tha. Python 3 of `input()` = Python 2 of `raw_input()`.

**Safe pattern (production style):**
```python
def read_int(prompt, lo=None, hi=None):
    while True:
        raw = input(prompt).strip()
        try:
            val = int(raw)
        except Valueerror:
            print("Sirf number write.");  continue
        if lo is not None and val < lo:  print(f"{lo} from bara write.");  continue
        if hi is not None and val > hi:  print(f"{hi} from chhota write."); continue
        return val
```

> **Is notebook (and poori series) me `input()` kabhi call not hoga.** Wajah: notebook
> automated tor but runs is; `input()` but wo hamesha of for ruk jayega (hang is jayega).
> Iske bajaye hum data **inline** (list/dict) me rakhte are — jo testing of for behtar bhi is.
> this asal me acha software design is: logic then I/O from alag rakho.

````python
# --- input() of jagah: list from "simulated input" (testable style) ---
# GALAT tareeqa (notebook hang kar dega):  age = int(input("Umar: "))
# SAHI tareeqa: logic then alag function me rakho, data outside from do.

def parse_age(raw):
    '''String then age me badalta is; masla is then (None, error) gives is.'''
    raw = raw.strip()                              # ahead behind of space hatao
    if not raw:
        return None, "khali input"
    try:
        age = int(raw)                             # cast — yahan Valueerror aa can is
    except Valueerror:
        return None, f"'{raw}' number not is"
    if not 0 < age < 130:                          # chained comparison
        return None, f"{age} ghair-haqeeqi umar is"
    return age, None

# "Farzi user inputs" — asli input() of bajaye
fake_inputs = ["25", " 30 ", "abc", "", "-5", "200", "42"]
for raw in fake_inputs:
    age, err = parse_age(raw)
    status = f"OK -> {age}" if err is None else f"REJECT -> {err}"
    print(f"input {raw!r:<7} : {status}")

print("\nFaida: this function testable is. if input() inside is, test likhna mushkil is.")
````

## 12. String formatting — teen tareeqe

````python
# =====================================================================
# problem 1: Invoice generator — Decimal of sath (paisa NEVER float me)
# =====================================================================
from decimal import Decimal, ROUND_HALF_UP

TAX_RATE = Decimal("0.17")            # 17% GST — string from banew (float from not)
PAISA = Decimal("0.01")               # rounding of quantum

def money(d):
    '''Decimal then 2 decimal places but half-up round does is.'''
    return d.quantize(PAISA, rounding=ROUND_HALF_UP)

# (item, qty, unit_price) — prices string me, taake exact rahen
cart = [
    ("Notebook A4",      3, "249.99"),
    ("Ball Pen (pack)",  2, "120.50"),
    ("Highlighter",      5,  "89.95"),
    ("Stapler",          2, "455.00"),
]

DISCOUNT_THRESHOLD = Decimal("2000")
DISCOUNT_RATE = Decimal("0.10")       # 2000 from above 10% off

print("=" * 54)
print(f"{'ITEM':<20}{'QTY':>4}{'RATE':>11}{'AMOUNT':>13}")
print("=" * 54)

subtotal = Decimal("0")
for item, qty, rate_str in cart:
    rate = Decimal(rate_str)
    amount = money(rate * qty)        # exact zarb, phir round
    subtotal += amount
    print(f"{item:<20}{qty:>4}{rate:>11}{amount:>13,}")

discount = money(subtotal * DISCOUNT_RATE) if subtotal > DISCOUNT_THRESHOLD else Decimal("0.00")
taxable = subtotal - discount
tax = money(taxable * TAX_RATE)
total = money(taxable + tax)

print("-" * 54)
print(f"{'Subtotal':<35}{subtotal:>19,}")
print(f"{'Discount (10%)' if discount else 'Discount':<35}{-discount:>19,}")
print(f"{'Tax (17%)':<35}{tax:>19,}")
print("=" * 54)
print(f"{'TOTAL (PKR)':<35}{total:>19,}")
print("=" * 54)

# Float of sath wahi hisaab — farq see
float_total = 0.0
for _, qty, rate_str in cart:
    float_total += float(rate_str) * qty
print(f"\nDecimal subtotal : {subtotal}")
print(f"float   subtotal : {float_total!r}  <- yahan chhoti error nazar aa rahi is")
assert money(total) == total
````

````python
# =====================================================================
# problem 2: Universal unit converter — casting + formatting of mazboot use
# =====================================================================
# Every conversion: (new unit, factor) — value * factor
CONVERSIONS = {
    ("km", "mile"):  ("mile", 0.621371),
    ("mile", "km"):  ("km",   1.609344),
    ("kg", "lb"):    ("lb",   2.2046226),
    ("lb", "kg"):    ("kg",   0.45359237),
    ("cm", "inch"):  ("inch", 0.3937008),
}

def convert(value, frm, then):
    '''Linear units convert does is; temperature alag from handle is is.'''
    if frm == then:
        return float(value)
    # Temperature linear-factor from not chalti (offset is)
    if (frm, then) == ("c", "f"):
        return value * 9 / 5 + 32
    if (frm, then) == ("f", "c"):
        return (value - 32) * 5 / 9
    if (frm, then) == ("c", "k"):
        return value + 273.15
    key = (frm, then)
    if key not in CONVERSIONS:                  # membership operator
        raise Valueerror(f"'{frm}' from '{then}' of conversion maloom not")
    _, factor = CONVERSIONS[key]
    return value * factor

jobs = [(5, "km", "mile"), (10, "mile", "km"), (70, "kg", "lb"),
        (30, "c", "f"), (98.6, "f", "c"), (25, "c", "k"), (12, "cm", "inch")]

print(f"{'INPUT':>12}  {'':2}  {'OUTPUT':>14}")
print("-" * 34)
for val, frm, then in jobs:
    out = convert(val, frm, then)
    print(f"{val:>8,.2f} {frm:<3} ->  {out:>10,.2f} {then}")

# error handling
try:
    convert(1, "km", "kg")
except Valueerror as e:
    print("\nValueerror:", e)

# Round-trip test (float of for isclose, == not)
import math
back = convert(convert(5, "km", "mile"), "mile", "km")
print(f"\nRound-trip 5 km -> mile -> km = {back!r}")
print("== 5 ?", back == 5, "| isclose ?", math.isclose(back, 5, rel_tol=1e-6), "<- SAHI check")
assert math.isclose(back, 5, rel_tol=1e-6)
assert math.isclose(convert(30, "c", "f"), 86.0)
````

````python
# =====================================================================
# problem 3: Human-readable file size — // % ** and formatting of got-jula use
# =====================================================================
UNITS = ["B", "KB", "MB", "GB", "TB", "PB"]

def human_size(num_bytes, withoutry=True):
    '''Bytes then padhne layak shakal me badalta is.
    withoutry=True -> 1 KB = 1024 B (Windows style), False -> 1000 B (SI style).'''
    if num_bytes < 0:
        raise Valueerror("bytes negative not is can")
    step = 1024 if withoutry else 1000
    if num_bytes < step:
        return f"{num_bytes} B"
    idx = 0
    size = float(num_bytes)
    while size >= step and idx < len(UNITS) - 1:   # while loop + comparison
        size /= step                               # true division
        idx += 1
    return f"{size:,.2f} {UNITS[idx]}"

samples = [0, 512, 1023, 1024, 1536, 1_048_576, 15_728_640, 3_221_225_472, 2 ** 50]
print(f"{'BYTES':>20}  {'withoutRY (1024)':>15}  {'SI (1000)':>15}")
print("-" * 56)
for b in samples:
    print(f"{b:>20,}  {human_size(b):>15}  {human_size(b, withoutry=False):>15}")

# Bitwise from bhi is can is (shift = divide by 1024 == >> 10)
print("\nBitwise shortcut: 15,728,640 >> 20 =", 15_728_640 >> 20, "MB (integer, fast magar precision kam)")

assert human_size(0) == "0 B"
assert human_size(1024) == "1.00 KB"
assert human_size(1_048_576) == "1.00 MB"
assert human_size(1000, withoutry=False) == "1.00 KB"
try:
    human_size(-1)
except Valueerror as e:
    print("negative -> Valueerror:", e)
print("\nSab tests pass ✔")
````

````python
# =====================================================================
# problem 4: Config loadelay — safe type coercion (casting + exceptions)
# =====================================================================
# Asli duniya me config (env vars, .ini, CLI args) hamesha STRING me aati is.
RAW_CONFIG = {
    "app_name":    "BizeeApp",
    "port":        "8080",
    "debug":       "true",
    "timeout":     "2.5",
    "max_retries": "3",
    "workers":     "abc",        # kharab value
    "rate_limit":  "",           # khali
    "verbose":     "0",
}

TRUE_WORDS = {"true", "yes", "y", "1", "on"}
FALSE_WORDS = {"false", "no", "n", "0", "off"}

def to_bool(text):
    '''String then bool me badalta is — bool(text) not use do (wo sab some True kehta is).'''
    t = text.strip().lower()
    if t in TRUE_WORDS:
        return True
    if t in FALSE_WORDS:
        return False
    raise Valueerror(f"'{text}' then bool not bana can")

def coerce(raw, key, caster, default=None):
    '''raw[key] then caster from convert does is; fail hone but default gives is.'''
    if key not in raw:
        return default
    text = raw[key]
    if not text.strip():                       # khali string -> default
        return default
    try:
        return caster(text)
    except (Valueerror, Typeerror) as e:       # ek hi except me kai types
        print(f"  ⚠ '{key}' = {text!r} invalid ({type(e).__name__}) -> default {default!r}")
        return default

print("Config parse is rahi is...")
config = {
    "app_name":    coerce(RAW_CONFIG, "app_name", str,     "MyApp"),
    "port":        coerce(RAW_CONFIG, "port",     int,     8000),
    "debug":       coerce(RAW_CONFIG, "debug",    to_bool, False),
    "timeout":     coerce(RAW_CONFIG, "timeout",  float,   5.0),
    "max_retries": coerce(RAW_CONFIG, "max_retries", int,  1),
    "workers":     coerce(RAW_CONFIG, "workers",  int,     4),      # 'abc' -> default
    "rate_limit":  coerce(RAW_CONFIG, "rate_limit", int,   100),    # khali -> default
    "verbose":     coerce(RAW_CONFIG, "verbose",  to_bool, True),   # '0' -> False
    "missing_key": coerce(RAW_CONFIG, "nope",     int,     -1),     # not is -> default
}

print(f"\n{'KEY':<14}{'VALUE':<12}{'TYPE'}")
print("-" * 34)
for k, v in config.items():
    print(f"{k:<14}{repr(v):<12}{type(v).__name__}")

assert config["port"] == 8080 and isinstance(config["port"], int)
assert config["debug"] is True
assert config["verbose"] is False, "'0' then False banna should tha — bool('0') of trap!"
assert config["workers"] == 4, "kharab value but default aana should"
print("\nSab assertions pass ✔  (note: bool('0') True is is — isi for to_bool likha)")
````

````python
# =====================================================================
# problem 5: Marks statistics report — operators + formatting + booleans
# =====================================================================
students = [
    ("Ali Raza",       78, 85, 92),
    ("Fatima Noor",    95, 88, 91),
    ("Bilal Ahmed",    45, 52, 38),
    ("Sana Iqbal",     67, 71, 80),
    ("Usman Tariq",    88, 94, 79),
]
SUBJECTS = ("Math", "Physics", "CS")
PASS_MARK = 50

def grade(pct):
    '''Percentage then grade me badalta is (chained comparison of khoobsurat use).'''
    if pct >= 90:  return "A+"
    if 80 <= pct < 90: return "A"
    if 70 <= pct < 80: return "B"
    if 60 <= pct < 70: return "C"
    if PASS_MARK <= pct < 60: return "D"
    return "F"

print(f"{'STUDENT':<15}{'MATH':>6}{'PHY':>6}{'CS':>6}{'TOTAL':>7}{'%':>8}{'GRADE':>7}{'STATUS':>9}")
print("=" * 64)

all_pct = []
for name, *marks in students:                      # starred unpacking!
    total = sum(marks)
    pct = total / (len(marks) * 100) * 100         # true division
    all_pct.append(pct)
    # sum of booleans idiom: kitne subjects pass happened
    passed_count = sum(m >= PASS_MARK for m in marks)
    status = "PASS" if passed_count == len(marks) else "FAIL"
    print(f"{name:<15}{marks[0]:>6}{marks[1]:>6}{marks[2]:>6}{total:>7}{pct:>7.2f}%{grade(pct):>7}{status:>9}")

print("-" * 64)
class_avg = sum(all_pct) / len(all_pct)
best = max(all_pct)
worst = min(all_pct)
print(f"{'CLASS AVERAGE':<15}{class_avg:>36.2f}%")
print(f"{'HIGHEST':<15}{best:>36.2f}%")
print(f"{'LOWEST':<15}{worst:>36.2f}%")
print(f"{'SPREAD':<15}{best - worst:>36.2f}%")
print(f"{'PASS RATE':<15}{sum(p >= PASS_MARK for p in all_pct) / len(all_pct):>36.1%}")

assert grade(95) == "A+" and grade(85) == "A" and grade(45) == "F"
assert isinstance(class_avg, float)
print("\nAssertions pass ✔")
````

### LC 412. Fizz Buzz — Easy

**Link:** https://leetcode.com/problems/fizz-buzz/

**problem (English):** Ek integer `n` diya is. `1` from `n` tak strings of ek list create jahan:
- number 3 and 5 **dono** from divisible is → `"FizzBuzz"`
- sirf 3 from divisible is → `"Fizz"`
- sirf 5 from divisible is → `"Buzz"`
- warna → number khud, string of shakal me (`str(i)`)

**Approach 1 — modulo + if/elif (seedha):**
1. `1` from `n` tak loop chalao (`range(1, n + 1)`).
2. **Sab from first** `i % 15 == 0` check do (ya `i % 3 == 0 and i % 5 == 0`).
   this tarteeb ahem is — if `i % 3` first check do then 15 but sirf `"Fizz"` aayega. **Yahi is
   problem of asli trap is.**
3. Phir `i % 3`, phir `i % 5`, warna `str(i)`.

**Approach 2 — string concatenation (zyada scalable):**
Khali string from shuru do, `3` but `"Fizz"` jodo, `5` but `"Buzz"` jodo. Aakhir me
`s or str(i)` — jo humne `or` operator those section me seekha (khali string falsy is).
if kal "7 but Bazz" of requirement aaye then sirf ek line barhani paregi — `elif` of jungle not.

**Time complexity:** O(n) — har number ek baar. (String banane of lagat then constant maan rahe are.)
**Space complexity:** O(n) — output list me `n` strings. Output then chhor kar auxiliary space O(1).

````python
# --- LC 412: Fizz Buzz ---

def fizz_buzz(n):
    '''Approach 1: modulo + if/elif. 15 of check SAB from first aana should.'''
    result = []
    for i in range(1, n + 1):                 # 1 from n tak (n bhi shamil)
        if i % 15 == 0:                       # 3 and 5 of LCM = 15 -> first check
            result.append("FizzBuzz")
        elif i % 3 == 0:
            result.append("Fizz")
        elif i % 5 == 0:
            result.append("Buzz")
        else:
            result.append(str(i))             # int then str me cast to do LAZMI is
    return result


def fizz_buzz_concat(n):
    '''Approach 2: string jodte jao. Naye rules add to do easy.'''
    result = []
    for i in range(1, n + 1):
        s = ""                                # khali string from shuru
        if i % 3 == 0:
            s += "Fizz"
        if i % 5 == 0:                        # 'elif' not -> 15 but dono jud jayenge
            s += "Buzz"
        result.append(s or str(i))            # khali string falsy -> str(i) chun lega
    return result


def fizz_buzz_rules(n, rules=((3, "Fizz"), (5, "Buzz"))):
    '''Approach 3: rules data-driven. Production code isi shakal me is is.'''
    out = []
    for i in range(1, n + 1):
        word = "".join(label for div, label in rules if i % div == 0)
        out.append(word or str(i))
    return out


# --- Tests ---
expected_15 = ["1", "2", "Fizz", "4", "Buzz", "Fizz", "7", "8", "Fizz", "Buzz",
               "11", "Fizz", "13", "14", "FizzBuzz"]

for fn in (fizz_buzz, fizz_buzz_concat, fizz_buzz_rules):
    assert fn(3) == ["1", "2", "Fizz"], fn.__name__
    assert fn(5) == ["1", "2", "Fizz", "4", "Buzz"], fn.__name__
    assert fn(15) == expected_15, fn.__name__
    assert fn(1) == ["1"], fn.__name__
    assert len(fn(100)) == 100, fn.__name__
    assert fn(15)[14] == "FizzBuzz", fn.__name__      # index 14 = number 15
    print(f"{fn.__name__:<18} ✔ saare tests pass")

print("\nn = 15 of natija:")
print(" ".join(fizz_buzz(15)))

# 15 of trap dikhao: if tarteeb ghalat is then what is is
def fizz_buzz_buggy(n):
    '''GHALAT version — 3 of check first, 15 of after me (ya exactly not).'''
    out = []
    for i in range(1, n + 1):
        if i % 3 == 0:
            out.append("Fizz")
        elif i % 5 == 0:
            out.append("Buzz")
        elif i % 15 == 0:      # yahan kabhi not pohnchega!
            out.append("FizzBuzz")
        else:
            out.append(str(i))
    return out

print("\nbuggy version, 15 but:", fizz_buzz_buggy(15)[14], "<- 'FizzBuzz' hona should tha")
assert fizz_buzz_buggy(15)[14] == "Fizz"      # bug of sabot
print("Sabaq: if/elif me KHAAS from AAM of tarteeb rakho.")
````

### LC 231. Power of Two — Easy

**Link:** https://leetcode.com/problems/power-of-two/

**problem (English):** Ek integer `n` diya is. Batao of `n` **2 of koi power** is ya not —
yani koi aisa `x` maujood is jahan `n == 2**x`? (`n` negative ya zero bhi is can is.)

**Approach 1 — Brute force (division loop):**
1. `n <= 0` is then foran `False` (2 of koi power negative ya zero not is).
2. Jab tak `n` even is, `n //= 2` do jao.
3. Aakhir me if `n == 1` bacha then yes, 2 of power thi.
   - Misaal: `16 → 8 → 4 → 2 → 1` ✔ ; `12 → 6 → 3` ✘ (3 bacha)

**Time:** O(log n) — har step me n aadha is is. **Space:** O(1).

**Approach 2 — Bitwise (optimal, one-liner):**
withoutry me 2 of har power me **theek ek** `1` bit is is:
```
1  = 0001      2  = 0010      4  = 0100      8  = 1000
```
and `n - 1` us akele `1` of neeche of saare bits then `1` kar gives is, us `1` then `0`:
```
   8 = 1000            12 = 1100
 8-1 = 0111          12-1 = 1011
 AND = 0000  ✔        AND = 1000  ✘ (zero not)
```
then formula: **`n > 0 and n & (n - 1) == 0`**

`n > 0` of shart lazmi is — warna `0` galat tor but `True` ban jata (`0 & -1 == 0`).

**Time:** O(1) — sirf ek subtraction and ek AND. **Space:** O(1).

**Approach 3 — `bit_count()` (Python 3.10+):** `n > 0 and n.bit_count() == 1` — sab from
padhne-layak, because seedha "ek hi 1 bit is" keh raha is.

````python
# --- LC 231: Power of Two ---

def is_power_of_two_loop(n):
    '''Approach 1: brute force. O(log n) time, O(1) space.'''
    if n <= 0:                       # 2 of power hamesha positive is is
        return False
    while n % 2 == 0:                # jab tak even is, aadha do jao
        n //= 2                      # floor division (/ not — warna float ban jayega)
    return n == 1                    # sirf 1 bacha then yes


def is_power_of_two(n):
    '''Approach 2: bitwise. O(1) time, O(1) space — interview of expected jawab.'''
    # n > 0 of shart zaroori: 0 & -1 == 0 is is, jo galat True give gives
    return n > 0 and n & (n - 1) == 0


def is_power_of_two_bitcount(n):
    '''Approach 3: 2 of power = withoutry me theek ek '1' bit.'''
    return n > 0 and n.bit_count() == 1


def is_power_of_two_recursive(n):
    '''Bonus: recursion from (NB 07 me tafseel).'''
    if n < 1:
        return False
    if n == 1:
        return True
    if n % 2:                        # odd is (and 1 not) -> not
        return False
    return is_power_of_two_recursive(n // 2)


# --- Bit-level tashreeh ---
print(f"{'n':>6} {'n (withoutry)':>12} {'n-1 (withoutry)':>14} {'n & (n-1)':>10}  jawab")
print("-" * 58)
for n in (1, 2, 3, 4, 8, 12, 16, 31, 32, 64):
    print(f"{n:>6} {n:>12b} {n - 1:>14b} {n & (n - 1):>10}  {is_power_of_two(n)}")

# --- Tests: chaaron approach ek like jawab dein ---
cases = [
    (1, True), (2, True), (4, True), (8, True), (16, True),
    (1024, True), (2 ** 30, True), (2 ** 62, True),
    (3, False), (5, False), (6, False), (12, False), (100, False),
    (0, False), (-1, False), (-2, False), (-16, False),
]
funcs = (is_power_of_two, is_power_of_two_loop, is_power_of_two_bitcount, is_power_of_two_recursive)
for fn in funcs:
    for n, expected in cases:
        got = fn(n)
        assert got == expected, f"{fn.__name__}({n}) ne {got} diya, {expected} should tha"
    print(f"{fn.__name__:<28} ✔ {len(cases)} tests pass")

print("\nEdge cases jo log bhool go are: n = 0, and negative n.")
print("  is_power_of_two(0)   =", is_power_of_two(0),  "(0 & -1 == 0 is — is for n > 0 lazmi)")
print("  is_power_of_two(-16) =", is_power_of_two(-16))
print("\nBonus: powers of two 1 from 1024 tak ->",
      [n for n in range(1, 1025) if is_power_of_two(n)])
````
