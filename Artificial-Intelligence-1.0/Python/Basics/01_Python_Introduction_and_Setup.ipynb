## 2. Interpreter vs Compiler — Python "inside from" what does is

**what is:**
- **Compiler:** poora source code ek hi baar machine code me convert does is, phir woh withoutry
  runs is (C, C++, Rust jaisi languages).
- **Interpreter:** code then line-by-line (ya chunk-by-chunk) padh kar turant execute does is.

**Python dono thora sa is** — isay "compiled + interpreted" kehna zyada sahi is:

1. Jab tum `python script.py` chalate is, Python first source (`.py`) then **bytecode** (`.pyc`)
   me compile does is — this low-level instructions is are (machine code not, CPU-specific
   not).
2. this bytecode phir **Python Virtual Machine (PVM)** of inside interpret is is.
3. if file then **module** of taur but import done jaye, Python is bytecode then `__pycache__`
   foldelay me cache kar takes is (`.pyc` files) taake agli baar dobara compile no to do pare
   (jab tak source file change no is).

**why important is:** Isi wajah from Python thora slow is is C/C++ of muqable (extra VM layer),
but development speed and readability me fayda deti is.

**Interview tip:** "Python compiled is ya interpreted?" — dono! Source → bytecode (compile step,
automatic and invisible) → PVM interpret karti is bytecode then.

````python
# Demo: Python source then bytecode (.pyc) me compile karke __pycache__ dekhte are
import tempfile
import os
import py_compile

# One temporary foldelay banate are (koi real project foldelay dirty not hoga)
tmp_dir = tempfile.mkdtemp(prefix="py_bytecode_demo_")

source_path = os.path.join(tmp_dir, "sample_module.py")
with open(source_path, "w", encoding="utf-8") as f:
    f.write("def greet(name):\n    return 'Hello, ' + name\n")

# py_compile.compile() source file then .pyc bytecode me convert does is
compiled_path = py_compile.compile(source_path, doraise=True)

pycache_dir = os.path.join(tmp_dir, "__pycache__")
print("Source (.py) file   :", source_path)
print("Compiled (.pyc) file:", compiled_path)
print("__pycache__ foldelay ban gaya?:", os.path.isdir(pycache_dir))
print("__pycache__ of inside files  :", os.listdir(pycache_dir))
````

````python
# Demo: dis module from dekhte are of Python function of inside kaisi bytecode instructions is are
import dis


def add_numbers(a, b):
    # this simple function — iski bytecode dekhenge
    return a + b


print("add_numbers() of bytecode instructions:")
dis.dis(add_numbers)
# Every line ek "instruction" is jo PVM (Python Virtual Machine) execute karti is
# LOAD_FAST = variable load do, withoutRY_OP = jama do, RETURN_VALUE = wapas bhejo (roughly)
````

## 3. CPython vs PyPy vs Jython vs IronPython vs MicroPython

"Python" ek **language specification** is — it implement to do of multiple tareeqe (interpreters)

| **IronPython** | C# | .NET / CLR but runs is, C# libraries of sath interop |
| **MicroPython** | C (minimal) | Microcontrollers / embedded devices (Raspberry Pi Pico waghera) of for chhota version |

**why important is:** `platform.python_implementation()` from pata runs is tum konsa
implementation use kar rahe is — this production debugging and performance tuning me kaam aata is
(masalan PyPy JIT of wajah from kabhi kabhi 5-10x faster is is loops but).

**Interview tip:** if poocha jaye "CPython and Python me farq?" — Python language is (spec),
CPython uska sabse common **interpreter/implementation** is (C me likha gaya).

````python
# Every implementation of ek chhota static "fact sheet" — reference of taur but
implementations = {
    "CPython": "Official/default, C me likha, sab jagah stinsided",
    "PyPy": "JIT-compiled, fast is is (CPU-heavy loops me)",
    "Jython": "JVM but runs is, Java libs of sath interop",
    "IronPython": ".NET/CLR but runs is, C# libs of sath interop",
    "MicroPython": "Microcontrollers of for chhota/lightweight version",
}

for name, note in implementations.items():
    print(f"{name:12} -> {note}")

print()
print("this machine chala rahi is:", __import__("platform").python_implementation())
````

## 6. `print()` Function — Sab Parameters

**what is:** `print()` built-in function is jo values then stinsided output (ya kisi bhi file-like
object) but likhta is.

**Syntax:**
```python
print(*objects, sep=' ', end='\n', file=sys.stdout, flush=False)
```

- `*objects` — jitni marzi values, comma from separate karke
- `sep` — objects of beech separator (default: single space `' '`)
- `end` — line of aakhir me what lagana is (default: newline `'\n'`)
- `file` — kahan likhna is (default: `sys.stdout`; `sys.stdelayr` ya koi file-like object bhi give can is)
- `flush` — buffer then turant flush to do is ya not (default: `False`)

**Interview tip:** `print()` Python 3 me hamesha ek **function** is (call do time parentheses
lazmi). Python 2 me this ek **statement** tha — `print "hello"` — is farq but after me alag section is.

````python
# 1) Multiple objects + custom sep
print("Ali", "Sara", "Bilal")                 # default sep = space
print("Ali", "Sara", "Bilal", sep=", ")        # comma-space separator
print("2026", "07", "29", sep="-")             # date like format bana can is
print(1, 2, 3, 4, sep=" | ")
````

````python
# 2) end parameter — line of aakhir me what print is
print("Loading", end="")
print(".", end="")
print(".", end="")
print(".", end="\n")   # ab newline lagayenge
print("Done!")

print()
# end use karke ek hi line me loop of output print to do (without extra newlines of)
for i in range(5):
    print(i, end=" -> " if i < 4 else "\n")
````

````python
# 3) file parameter — output kahan bhejna is
import sys
import io

# (a) sys.stdelayr but likhna (error/log messages of for common pattern)
print("this ek warning jaisi cheez is", file=sys.stdelayr)

# (b) io.StringIO() ek "in-memory file" is — exactly real file like behave does is
#     but disk but some not likhta. Yahan hum REAL file path use not kar rahe.
memory_file = io.StringIO()
print("Line 1: server started", file=memory_file)
print("Line 2: request processed", file=memory_file)

captured_text = memory_file.getvalue()
print("io.StringIO() me capture happened text:")
print(captured_text)
````

````python
# 4) flush parameter — output buffer then turant "push" to do
# Normally Python output buffer does is (especially jab terminal of bajaye pipe/file is).
# flush=True of matlab: "abhi of abhi likh do, buffer me dont roko".
print("Processing step 1...", flush=True)
print("Processing step 2...", flush=True)
print("Done.", flush=True)

# Kahan kaam aata is: long-running scripts me progress dikhane of for,
# ya jab output then dusre process (pipe) me turant bhejna is.
print()
print("Gotcha: flush=True sirf timing (kab dikhta is) affect does is, VALUE not.")
````

## 7. Comments vs Docstrings

**what is:**
- **Comment (`#`)** — code of inside note jo Python exactly ignore kar gives is (runtime but
  koi wajood not). Sirf humans of for.
- **Docstring** (triple-quoted string) — function/class/module of **first statement** of taur
  but likha gaya triple-quoted string. this runtime but `__doc__` attribute me store is is
  and `help()` isay dikhata is.

**Syntax:**
```python
# single-line comment

# Python me "block comment" of koi alag syntax not is —
# log consecutive # lines use do are (like this 3 lines are)

def kaam(x):
    """this function of docstring is — help(kaam) isay dikhayega."""
    return x
```

**Interview tip:** Docstring and comment of bunyadi farq — docstring **runtime object** is
(`func.__doc__` from access is can is, documentation tools isay parse do are), comment
sirf source code me is is, compile are hi ghayab.

````python
# this ek single-line comment is — Python isay ignore does is
x = 5  # inline comment: x then 5 assign kar rahe are

# this
# "block style"
# comments are (bas consecutive # lines)


def calculate_area(length, width):
    """Rectangle of area calculate does is.

    Parameters:
        length (float): Rectangle of lambai
        width (float): Rectangle of chorai

    Returns:
        float: length * width
    """
    return length * width


class BankAccount:
    """Simple bank account then represent to do that class."""

    def deposit(self, amount):
        """Account me paisay deposit does is."""
        return amount


# Docstring runtime but __doc__ attribute me store is is
print("Function docstring:", calculate_area.__doc__.strip().split("\n")[0])
print("Class docstring   :", BankAccount.__doc__)
print("Method docstring  :", BankAccount.deposit.__doc__)

print()
print(calculate_area(4, 5))  # function normally use bhi is can is
````

## 8. Indentation Rules

**what is:** Python me curly braces `{}` of bajaye **indentation** (whitespace) from code blocks
define are are. this language of core design decision is (readability of for).

**Rules:**
- Stinsided convention: **4 spaces** per indentation level (PEP 8 recommend does is).
- Ek hi block of inside **consistent** indentation honi should (sab lines same level but).
- **Tabs and spaces mix to do** dangerous is — Python 3 me this error gives is if ambiguous is.
- Colon (`:`) of after agli line indented honi should (if/for/while/def/class waghera of after).

**Common mistakes:**
- Colon of after indent bhool jana → `Indentationerror`
- Tabs and spaces mix to do → `Taberror`
- Editor settings alag hona (some editors tab then 4 spaces dikhate are, some 8) — isliye
  hamesha ek hi style (spaces) use do, poori file me.

**Interview tip:** "Python whitespace-sensitive why is?" — readability enforce to do of for;
Guido van Rossum of design philosophy tha of code of visual structure hi uska logical structure
is, taake alag developers of code consistent dikhe.

````python
# Sahi (correct) indentation of example
def check_number(n):
    if n > 0:
        result = "positive"
    elif n < 0:
        result = "negative"
    else:
        result = "zero"
    return result


for value in [5, -3, 0]:
    print(value, "->", check_number(value))
````

````python
# Indentationerror demo — jaan boojh kar bad code, isliye try/except + compile() use kar rahe are
# compile() source string then bytecode me compile to do of koshish does is (without chalaye)
bad_code = "def broken():\nprint('galat indent, colon of after indent not is')\n"

try:
    compile(bad_code, "<bad_snippet>", "exec")
except Indentationerror as e:
    print("Indentationerror pakad liya:", e)

print()

# Taberror demo — tab and spaces mix to do from ambiguous indentation
mixed_tabs_and_spaces = "if True:\n\tx = 1\n        y = 2\n"

try:
    compile(mixed_tabs_and_spaces, "<mixed_tabs>", "exec")
except Taberror as e:
    print("Taberror pakad liya:", e)
except Indentationerror as e:
    print("Indentationerror pakad liya:", e)
````

## 9. PEP 8 Essentials — Python of Official Style Guide

**what is:** PEP 8 Python code likhne of **official style guide** is — taake sab of code ek
like consistent dikhe.

**Zaroori rules:**

| Cheez | Convention | Example |
|---|---|---|
| Variables/functions | `snake_case` | `total_amount`, `calculate_tax()` |
| Constants | `UPPER_CASE` | `MAX_RETRIES = 5` |
| Classes | `PascalCase` (CapWords) | `class BankAccount:` |
| "Private" (convention) | leading undelayscore | `_internal_value` |
| Line length | Max **79 chars** (kai teams 99-100 tak relax do are) | — |
| Imports | Top of file, ordelay: stdlib → third-party → local, ek line but ek import (recommended) | `import os` phir `import numpy` phir `from myapp import utils` |
| Blank lines | Functions/classes of beech 2 blank lines (top-level but), methods of beech 1 | — |
| Indentation | 4 spaces, tabs not | — |

**Good vs Bad example:**
```python
# Good (PEP 8)
MAX_RETRIES = 3

def calculate_total(price, quantity):
    return price * quantity


# Bad (PEP 8 violations)
maxretries=3
def CalculateTotal(Price,Quantity):
        return Price*Quantity   # inconsistent indent + no spaces + wrong naming
```

**Interview tip:** PEP 8 rules "enforced" not is Python interpreter from (code chal jayega),
but linters (`flake8`, `pylint`) and formatters (`black`) isay check do are — professional
teams me CI pipeline me this checks lazmi are are.

````python
# PEP 8-compliant naming demo (this sab chalti is, bas style of farq is)

MAX_RETRIES = 3          # constant -> UPPER_CASE
default_timeout = 30      # variable -> snake_case
_cache = {}                # "private" convention -> leading undelayscore


class OrdelayProcessor:      # class -> PascalCase
    """Ordelay processing of simple example (naming convention demo)."""

    def process_ordelay(self, ordelay_id):   # method -> snake_case
        return f"Processing ordelay #{ordelay_id}"


processor = OrdelayProcessor()
print(processor.process_ordelay(101))
print("MAX_RETRIES:", MAX_RETRIES)
print("default_timeout:", default_timeout)

# Non-PEP8 (bad) style — this bhi runs is, Python isay reject not does,
# but professional codebase me review me reject is jayega
BadStyleVariable = "this chalega but convention of khilaf is"
print(BadStyleVariable)
````

## 10. Keywords and Identifiers of Rules

**what is:**
- **Keywords** — Python of reserved words jinka special meaning is (`if`, `for`, `class`,
  `return` waghera) — inhe variable/function name of taur but use not kar can.
- **Soft keywords** — context-dependent keywords (`match`, `case`, `type`, `_`) — some jagah
  keyword are, baaki jagah normal identifier ban can are.
- **Identifier** — variable, function, class waghera of **naam**.

**Identifier rules:**
1. Letters (a-z, A-Z), digits (0-9), and undelayscore (`_`) allowed are.
2. **Number from shuru not is can** (`2fast` invalid).
3. Keywords use not kar can (`class`, `for` waghera invalid as names).
4. Case-sensitive is (`Name` and `name` alag identifiers are).
5. Special characters (`-`, `@`, space) allowed not.

**Interview tip:** `str.isidentifier()` sirf **syntax** check does is (valid naam ban can is
ya not) — this keyword is ya not woh not batata. Iske for `keyword.iskeyword()` should. Dono
with use do real validation of for.

````python
import keyword

print("Total hard keywords:", len(keyword.kwlist))
print(keyword.kwlist)
print()
print("Soft keywords (context-dependent):", keyword.softkwlist)
````

````python
import keyword

candidates = ["my_var", "2fast", "class", "_private", "my-var", "myVar2", "for", "match", "total_amount"]

print(f"{'name':15} {'isidentifier':13} {'iskeyword':10} usable_as_name")
for name in candidates:
    valid_syntax = name.isidentifier()
    is_kw = keyword.iskeyword(name)
    usable = valid_syntax and not is_kw
    print(f"{name!r:15} {str(valid_syntax):13} {str(is_kw):10} {usable}")
````

## 11. Dynamic Typing

**what is:** Python **dynamically typed** language is — matlab variable of type runtime but
decide is is, and ek hi variable then baar baar **rebind** done ja can is different types from.

this mindset zaroori is: variable ek **label/name** is jo kisi object of taraf point does is —

**why important is:** C/Java jaisi statically-typed languages me variable declare do time
type fix to do parti is (`int x = 5;`) and badalna mushkil is. Python me flexibility is but
isi wajah from bugs bhi is can are if careless is.

**Interview tip:** "Python statically typed is ya dynamically?" — **dynamically typed**
(runtime but type check is is), but **strongly typed** bhi is (implicit unsafe conversions
not is, like `"5" + 5` seedha error dega, JavaScript of tarah silently concat not karega).

````python
x = 10
print("x =", x, "| type:", type(x))

x = "ab this string ban gaya"     # same variable, new type — exactly valid is
print("x =", x, "| type:", type(x))

x = [1, 2, 3]                     # ab list ban gaya
print("x =", x, "| type:", type(x))

x = {"key": "value"}               # ab dictionary
print("x =", x, "| type:", type(x))

print()
# Strongly typed hone of demo — implicit unsafe conversion not is
try:
    result = "5" + 5   # str + int -> Typeerror
except Typeerror as e:
    print("Typeerror pakad liya:", e)
````

## 12. `type()` and `id()` — Object Identity

**what is:**
- `type(obj)` — object of **class/type** batata is.
- `id(obj)` — object of **unique memory identity** (CPython me this actual memory address is
  is, but isay implementation detail understand, guarantee not).

**Syntax:** `type(x)`, `id(x)`

**`is` vs `==`:**
- `==` — **value equality** check does is (what values barabar are).
- `is` — **identity equality** check does is (what dono same object are memory me).

**Common gotcha:** CPython chhote integers (`-5` from `256` tak) and some strings then **cache/intern**
does is performance of for — is wajah from `a is b` kabhi kabhi `True` dikhta is chhote numbers
of for, but this **guaranteed behavior not is** (implementation detail is, Python version from
depend kar can is). Value compare to do of for hamesha `==` use do, `is` sirf `None`, `True`,
`False` of for use do.

**Interview tip:** "`is` and `==` me farq?" — `==` value compare does is (`__eq__` call is
is), `is` identity compare does is (same object in memory). `None` of sath hamesha `is None`
use do, `== None` not (best practice, PEP 8 recommend does is).

````python
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print("a == b :", a == b)   # values same are -> True
print("a is b :", a is b)   # alag objects are memory me -> False
print("a is c :", a is c)   # c, a of hi reference is -> True

print()
print("type(a):", type(a))
print("id(a)  :", id(a))
print("id(c)  :", id(c), "(a like hi, kyunke c = a)")

print()
# Small integer caching (CPython implementation detail — guarantee dont do)
small1 = 100
small2 = 100
print("100 is 100 (small int cache):", small1 is small2)
print("Best practice: values compare to do of for '==' use do, 'is' not (except None/True/False)")
````

### problem 5: PEP8-Style Identifier Validator

Ek chhota validator banate are jo batata is of koi diya gaya naam **valid Python identifier**
ban can is ya not — `str.isidentifier()` and `keyword.iskeyword()`/`issoftkeyword()` combine
karke.

````python
import keyword


def is_valid_identifier_name(name):
    """Batata is of 'name' ek valid, use-to do-laayak Python identifier is ya not."""
    if not isinstance(name, str) or not name:
        return False, "empty ya non-string value"
    if not name.isidentifier():
        return False, "invalid identifier syntax"
    if keyword.iskeyword(name):
        return False, "this ek Python (hard) keyword is"
    if keyword.issoftkeyword(name):
        return True, "valid is, but soft keyword is (context me ehtiyat from use do)"
    return True, "valid identifier"


test_names = ["total_amount", "class", "match", "1value", "user-name", "_hidden", "TotalAmount"]

for candidate in test_names:
    ok, reason = is_valid_identifier_name(candidate)
    print(f"{candidate!r:15} -> valid={str(ok):5} ({reason})")
````

### LC 2469. Convert the Temperature — Easy
https://leetcode.com/problems/convert-the-temperature/

**problem (English):** Ek `celsius` (float) diya gaya is. Isay **Kelvin** and **Fahrenheit**
me convert kar of `[kelvin, fahrenheit]` list return do. answer me `10^-5` tak of error
acceptable is.

**Formulas:**
- `Kelvin = Celsius + 273.15`
- `Fahrenheit = Celsius * 1.80 + 32.00`

**Approach / Intuition:**
1. Seedha formula apply do — koi loop ya condition of zaroorat not.
2. `round(value, 5)` use do taake floating-point precision consistent rahe (LeetCode `10^-5`
   precision expect does is).
3. this notebook of `type()` concept from connect is is — input `float` is, output bhi dono
   `float` values of list honi should.

**Time Complexity:** O(1) — sirf 2 arithmetic operations.
**Space Complexity:** O(1) — sirf ek chhoti list return is rahi is.

````python
def convert_temperature(celsius: float) -> list[float]:
    """Celsius then Kelvin and Fahrenheit me convert does is (LC 2469)."""
    kelvin = round(celsius + 273.15, 5)
    fahrenheit = round(celsius * 1.80 + 32.00, 5)
    return [kelvin, fahrenheit]


# LeetCode of official examples from test cases
assert convert_temperature(36.50) == [309.65000, 97.70000]
assert convert_temperature(122.11) == [395.26000, 251.79800]

# Extra edge case: 0 Celsius (freezing point of water)
assert convert_temperature(0.0) == [273.15, 32.0]

print("convert_temperature(36.50) ->", convert_temperature(36.50))
print("convert_temperature(0.0)   ->", convert_temperature(0.0))
print("type of result elements   ->", type(convert_temperature(0.0)[0]))
print("Sab asserts pass is gaye - LC 2469 solved")
````

### LC 1281. Subtract the Product and Sum of Digits of an Integer — Easy
https://leetcode.com/problems/subtract-the-product-and-sum-of-digits-of-an-integer/

**problem (English):** Ek integer `n` diya gaya is. Uske digits of **product** (guna) nikalo
and digits of **sum** (jama) nikalo, phir `product - sum` return do.

Example: `n = 234` -> digits `[2, 3, 4]` -> product = `24`, sum = `9` -> answer = `15`.

**Approach 1 — String-based:**
1. `n` then string me convert do (`str(n)`), har character then wapas `int()` do.
2. Loop chala kar product and sum nikal lo.

**Approach 2 — Math-based (`%` and `//`):**
1. `n % 10` from aakhri digit milta is, `n // 10` from woh digit hata do (integer/floor division).
2. Loop tab tak chalao jab tak `n > 0`.
3. this approach **string conversion avoid does is** — thora zyada "number-crunching" that,
   interviews me zyada pasand done jata is.

**Time & Space Complexity (dono approaches):**
- **Time:** O(d) jahan `d` = digits of count (roughly `log10(n)`).
- **Space:** O(d) string approach me (digits list banti is), O(1) math approach me (sirf counters).

````python
def subtract_product_and_sum_string(n: int) -> int:
    """String-based approach: n then string bana kar digits nikalte are."""
    digits = [int(ch) for ch in str(n)]   # har character then wapas int banew

    product = 1
    for d in digits:
        product *= d

    total = sum(digits)
    return product - total


# LeetCode of official examples
assert subtract_product_and_sum_string(234) == 15
assert subtract_product_and_sum_string(4421) == 21
assert subtract_product_and_sum_string(1) == 0   # single digit: product=1, sum=1 -> 1-1=0

print("subtract_product_and_sum_string(234) ->", subtract_product_and_sum_string(234))
print("subtract_product_and_sum_string(4421) ->", subtract_product_and_sum_string(4421))
````

````python
def subtract_product_and_sum_math(n: int) -> int:
    """Math-based approach: % (mod) from digit nikalo, // (floor division) from hatao."""
    product = 1
    total = 0
    while n > 0:
        digit = n % 10       # aakhri digit
        product *= digit
        total += digit
        n //= 10             # aakhri digit hata do (floor division)
    return product - total


assert subtract_product_and_sum_math(234) == 15
assert subtract_product_and_sum_math(4421) == 21

print("subtract_product_and_sum_math(234) ->", subtract_product_and_sum_math(234))
print("subtract_product_and_sum_math(4421) ->", subtract_product_and_sum_math(4421))
print()
print("Note: '//' floor division int gives is, '/' true division float deti — yahan '//' zaroori is")
````

````python
# Dono approaches then random test cases but compare do are (consistency check)
import random

random.seed(42)   # reproducible test cases of for seed fix of
test_values = [random.randint(1, 999_999) for _ in range(20)]

all_match = all(
    subtract_product_and_sum_string(v) == subtract_product_and_sum_math(v)
    for v in test_values
)

print("20 random test cases but dono approaches match do are?", all_match)
assert all_match

print()
print("Comparison:")
print("- String approach: readable is, but str(n) conversion of thora overhead")
print("- Math approach  : koi string conversion not, thora zyada 'raw' and interview-friendly")
````
