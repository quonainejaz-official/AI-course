## 1. Polymorphism — "Many Forms"

Polymorphism lets objects of **different types** respond to the **same interface**.

Forms of polymorphism in Python:
1. **Duck typing** — "If it walks like a duck and quacks like a duck..."
2. **Method overriding** — Child redefines parent method
3. **Polymorphic functions** — Same function works on different types
4. **Operator overloading** — `__add__`, `__mul__`, etc.

## 2. Duck Typing

Python doesn't check types at compile time. If an object has the required method, it works.

```python
class Duck:
    def quack(self) -> str:
        return "Quack!"

class Person:
    def quack(self) -> str:
        return "I'm pretending to be a duck!"

def make_it_quack(thing):
    # Duck typing — no type check, just call .quack()
    print(thing.quack())

make_it_quack(Duck())    # Quack!
make_it_quack(Person())  # I'm pretending to be a duck!
```

## 3. Method Overriding

A child class redefines a method from the parent. The child's version takes precedence.

```python
class Animal:
    def sound(self) -> str:
        return "Some sound"

class Cat(Animal):
    def sound(self) -> str:
        return "Meow"

class Dog(Animal):
    def sound(self) -> str:
        return "Woof"

animals = [Animal(), Cat(), Dog()]
for a in animals:
    print(a.sound())  # Some sound, Meow, Woof
```

## 4. Polymorphic Functions

A single function that works with objects of different types as long as they support the expected interface.

```python
def double(value):
    # Polymorphic — works with numbers, strings, lists
    return value * 2

print(double(5))        # 10
print(double("Hi "))    # Hi Hi
print(double([1, 2]))   # [1, 2, 1, 2]

def total_area(shapes: list) -> float:
    # Works with any shape that has .area()
    return sum(s.area() for s in shapes)
```

## 5. Abstract Base Classes (ABC)

ABCs define an **interface** that subclasses **must** implement. Use `abc.ABC` and `@abstractmethod`.

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self) -> float:
        pass

    @abstractmethod
    def perimeter(self) -> float:
        pass

    def describe(self) -> str:
        # Concrete method — available to all subclasses
        return f"Area: {self.area():.2f}, Perimeter: {self.perimeter():.2f}"

# s = Shape()  # TypeError! Can't instantiate abstract class

class Circle(Shape):
    def __init__(self, radius: float):
        self.radius = radius

    def area(self) -> float:
        import math
        return math.pi * self.radius ** 2

    def perimeter(self) -> float:
        import math
        return 2 * math.pi * self.radius

c = Circle(5)
print(c.describe())  # Area: 78.54, Perimeter: 31.42
```

## 6. `isinstance()` and `issubclass()`

- `isinstance(obj, Class)` — checks if `obj` is an instance of `Class` (or subclass)
- `issubclass(Child, Parent)` — checks inheritance relationship

```python
class Animal(ABC):
    pass

class Dog(Animal):
    pass

d = Dog()
print(isinstance(d, Dog))     # True
print(isinstance(d, Animal))  # True (Dog IS-A Animal)
print(issubclass(Dog, Animal))  # True
print(issubclass(Animal, Dog))  # False
```

## 7. Protocol Classes (Structural Subtyping)

Python 3.8+ supports **Protocols** — static duck typing via `typing.Protocol`. A class "conforms" to a protocol if it has the required methods.

```python
from typing import Protocol, runtime_checkable

@runtime_checkable
class Drawable(Protocol):
    def draw(self) -> str:
        ...  # Method signature only (ellipsis)

class Circle:
    def draw(self) -> str:
        return "Drawing a circle"

class Square:
    def draw(self) -> str:
        return "Drawing a square"

class NotDrawable:
    pass

def render(obj: Drawable):
    print(obj.draw())

render(Circle())  # Drawing a circle
render(Square())  # Drawing a square
print(isinstance(Circle(), Drawable))  # True (with @runtime_checkable)
print(isinstance(NotDrawable(), Drawable))  # False
```

## Problem 1: Payment System

**Difficulty: Medium**

Design a payment processing system with multiple payment methods using ABC.

- **Time complexity**: O(1) per payment
- **Space complexity**: O(1)

```python
from abc import ABC, abstractmethod

class PaymentMethod(ABC):
    """Abstract payment interface."""
    @abstractmethod
    def pay(self, amount: float) -> str:
        pass

    @abstractmethod
    def refund(self, amount: float) -> str:
        pass

class CreditCard(PaymentMethod):
    def __init__(self, card_number: str):
        self.card_number = card_number[-4:]  # Store last 4 digits

    def pay(self, amount: float) -> str:
        return f"Paid ${amount:.2f} via Credit Card ****{self.card_number}"

    def refund(self, amount: float) -> str:
        return f"Refunded ${amount:.2f} to Credit Card ****{self.card_number}"

class PayPal(PaymentMethod):
    def __init__(self, email: str):
        self.email = email

    def pay(self, amount: float) -> str:
        return f"Paid ${amount:.2f} via PayPal ({self.email})"

    def refund(self, amount: float) -> str:
        return f"Refunded ${amount:.2f} to PayPal ({self.email})"

class Crypto(PaymentMethod):
    def __init__(self, wallet_address: str):
        self.wallet = wallet_address[:8] + "..."

    def pay(self, amount: float) -> str:
        return f"Paid ${amount:.2f} via Crypto ({self.wallet})"

    def refund(self, amount: float) -> str:
        return f"Refunded ${amount:.2f} to Crypto ({self.wallet})"

# Polymorphic payment processing
def process_payment(method: PaymentMethod, amount: float):
    print(method.pay(amount))

methods = [
    CreditCard("1234567890123456"),
    PayPal("alice@example.com"),
    Crypto("0xABC123DEF456GHI789")
]
for m in methods:
    process_payment(m, 99.99)
```

## Problem 2: Animal Sound System

**Difficulty: Easy**

Create an animal sound system demonstrating polymorphism. Each animal makes its own sound.

```python
class Animal:
    def __init__(self, name: str):
        self.name = name

    def sound(self) -> str:
        raise NotImplementedError("Subclass must implement")

class Cow(Animal):
    def sound(self) -> str:
        return "Moo"

class Snake(Animal):
    def sound(self) -> str:
        return "Hiss"

class Owl(Animal):
    def sound(self) -> str:
        return "Hoot"

# Polymorphic function
def animal_chorus(animals: list[Animal]) -> str:
    return ' | '.join(f"{a.name}: {a.sound()}" for a in animals)

zoo = [Cow("Bessie"), Snake("Slyther"), Owl("Hedwig")]
print(animal_chorus(zoo))
# Bessie: Moo | Slyther: Hiss | Hedwig: Hoot
```

## Key Takeaways

- **Duck typing**: "If it quacks, it's a duck" — focus on behavior, not type.
- **Method overriding**: child redefines parent method for specialized behavior.
- **ABCs** (`abc.ABC`, `@abstractmethod`): enforce interface contracts.
- **Protocols** (`typing.Protocol`): structural subtyping — static duck typing.
- **Polymorphism**: write code that works with any object that supports the expected interface.
- Use `isinstance()`/`issubclass()` sparingly — prefer duck typing and ABCs.
