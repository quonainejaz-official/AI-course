## What is Inheritance?

Inheritance allows a **child class** to **reuse** and **extend** the behavior of a **parent class**.

- **Code reuse**: child gets all parent methods/attributes
- **Extensibility**: child can override or add new behavior
- **IS-A relationship**: `Dog IS-A Animal`

## 1. Single Inheritance

One child class inherits from one parent class.

```python
class Animal:
    def __init__(self, name: str):
        self.name = name

    def speak(self) -> str:
        return f"{self.name} makes a sound"

class Dog(Animal):  # Dog inherits from Animal
    def speak(self) -> str:
        # Override parent method
        return f"{self.name} says Woof!"

d = Dog("Buddy")
print(d.speak())  # Buddy says Woof!
print(isinstance(d, Animal))  # True
```

## 2. The `super()` Function

`super()` gives access to the **parent class** — used to call parent methods and avoid explicit parent class names.

```python
class Vehicle:
    def __init__(self, brand: str, model: str):
        self.brand = brand
        self.model = model

    def info(self) -> str:
        return f"{self.brand} {self.model}"

class Car(Vehicle):
    def __init__(self, brand: str, model: str, doors: int):
        super().__init__(brand, model)  # Call parent constructor
        self.doors = doors

    def info(self) -> str:
        # Extend parent info
        return f"{super().info()} — {self.doors} doors"

c = Car("Toyota", "Camry", 4)
print(c.info())  # Toyota Camry — 4 doors
```

## 3. Types of Inheritance

| Type | Description | Example |
|---|---|---|
| **Single** | One parent, one child | `A → B` |
| **Multilevel** | Chain of inheritance | `A → B → C` |
| **Multiple** | Child inherits from multiple parents | `A, B → C` |
| **Hierarchical** | One parent, multiple children | `A → B, A → C` |
| **Hybrid** | Mix of multiple and multilevel | (Diamond problem) |

### 3a. Multilevel Inheritance

```python
class Animal:
    def __init__(self, name: str):
        self.name = name

class Mammal(Animal):
    def warm_blooded(self) -> bool:
        return True

class Dog(Mammal):  # Dog → Mammal → Animal
    def speak(self) -> str:
        return f"{self.name} says Woof!"

d = Dog("Buddy")
print(f"Warm-blooded: {d.warm_blooded()}, Says: {d.speak()}")
```

### 3b. Hierarchical Inheritance

```python
class Shape:
    def __init__(self, color: str = "black"):
        self.color = color

class Circle(Shape):
    def __init__(self, radius: float, color: str = "black"):
        super().__init__(color)
        self.radius = radius

class Rectangle(Shape):
    def __init__(self, width: float, height: float, color: str = "black"):
        super().__init__(color)
        self.width = width
        self.height = height

c = Circle(5, "red")
r = Rectangle(3, 4, "blue")
print(f"Circle: radius={c.radius}, color={c.color}")  # 5, red
print(f"Rectangle: {r.width}x{r.height}, color={r.color}")  # 3x4, blue
```

## 4. Multiple Inheritance

A class can inherit from **multiple parent classes**.

```python
class Flyer:
    def fly(self) -> str:
        return "Flying..."

class Swimmer:
    def swim(self) -> str:
        return "Swimming..."

class Duck(Flyer, Swimmer):
    def __init__(self, name: str):
        self.name = name

d = Duck("Donald")
print(d.fly())   # Flying...
print(d.swim())  # Swimming...
```

## 5. Method Resolution Order (MRO)

Python uses the **C3 Linearization** algorithm to determine method lookup order.

- `ClassName.__mro__` shows the order
- `mro()` method returns a list
- Follows: child → parent → ... → `object`

```python
class A:
    def method(self):
        return "A"

class B(A):
    def method(self):
        return "B"

class C(A):
    def method(self):
        return "C"

class D(B, C):
    pass

d = D()
print(d.method())  # B — MRO: D → B → C → A → object
print(D.__mro__)   # (<class 'D'>, <class 'B'>, <class 'C'>, <class 'A'>, <class 'object'>)
```

## 6. The Diamond Problem

When a class inherits from two classes that share a common ancestor. Python resolves this via MRO — `super()` follows the MRO chain.

```python
class A:
    def __init__(self):
        print("A.__init__")

class B(A):
    def __init__(self):
        print("B.__init__")
        super().__init__()

class C(A):
    def __init__(self):
        print("C.__init__")
        super().__init__()

class D(B, C):
    def __init__(self):
        print("D.__init__")
        super().__init__()  # Follows MRO: D → B → C → A

print(D.__mro__)
d = D()
# Output:
# D.__init__
# B.__init__
# C.__init__
# A.__init__
# (A is initialized only ONCE thanks to cooperative MRO)
```

## Problem 1: Shape Hierarchy

**Difficulty: Medium**

Design a shape hierarchy:
- `Shape` (base): `color`, `area()` (abstract-like), `describe()`
- `Circle`: radius
- `Rectangle`: width, height
- `Triangle`: base, height

```python
import math

class Shape:
    def __init__(self, color: str = "black"):
        self.color = color

    def area(self) -> float:
        raise NotImplementedError("Subclasses must implement area()")

    def describe(self) -> str:
        return f"{self.color} {self.__class__.__name__}, area: {self.area():.2f}"

class Circle(Shape):
    def __init__(self, radius: float, color: str = "black"):
        super().__init__(color)
        self.radius = radius

    def area(self) -> float:
        return math.pi * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, width: float, height: float, color: str = "black"):
        super().__init__(color)
        self.width = width
        self.height = height

    def area(self) -> float:
        return self.width * self.height

class Triangle(Shape):
    def __init__(self, base: float, height: float, color: str = "black"):
        super().__init__(color)
        self.base = base
        self.height = height

    def area(self) -> float:
        return 0.5 * self.base * self.height

shapes = [
    Circle(5, "red"),
    Rectangle(3, 4, "blue"),
    Triangle(6, 4, "green")
]

for s in shapes:
    print(s.describe())
# red Circle, area: 78.54
# blue Rectangle, area: 12.00
# green Triangle, area: 12.00
```

## Problem 2: Vehicle Hierarchy

**Difficulty: Medium**

Build a vehicle hierarchy: `Vehicle → Car, Bike, Truck`. Each with `start()`, `stop()`, and `fuel_efficiency()`.

```python
class Vehicle:
    def __init__(self, brand: str, year: int, fuel_capacity: float):
        self.brand = brand
        self.year = year
        self.fuel_capacity = fuel_capacity  # liters
        self._is_running = False

    def start(self) -> str:
        self._is_running = True
        return f"{self.brand} {self.__class__.__name__} started"

    def stop(self) -> str:
        self._is_running = False
        return f"{self.brand} {self.__class__.__name__} stopped"

    def fuel_efficiency(self) -> float:
        # km per liter — to be overridden
        raise NotImplementedError

class Car(Vehicle):
    def __init__(self, brand: str, year: int, fuel_capacity: float, doors: int):
        super().__init__(brand, year, fuel_capacity)
        self.doors = doors

    def fuel_efficiency(self) -> float:
        return 15.0  # km/L average

class Bike(Vehicle):
    def fuel_efficiency(self) -> float:
        return 45.0  # km/L

class Truck(Vehicle):
    def __init__(self, brand: str, year: int, fuel_capacity: float, load_capacity: float):
        super().__init__(brand, year, fuel_capacity)
        self.load_capacity = load_capacity  # tons

    def fuel_efficiency(self) -> float:
        return 5.0  # km/L

vehicles = [
    Car("Toyota", 2022, 50, 4),
    Bike("Yamaha", 2023, 15),
    Truck("Volvo", 2021, 200, 20)
]

for v in vehicles:
    print(v.start())
    print(f"  Efficiency: {v.fuel_efficiency()} km/L")
    print(v.stop())
```

## Key Takeaways

- **Single inheritance**: one parent, one child — simplest form.
- **`super()`**: calls parent methods and follows MRO chain.
- **Multiple inheritance**: inherit from ≥2 parents — use with care.
- **MRO** (C3 Linearization): determines method lookup order; view with `ClassName.__mro__`.
- **Diamond problem**: Python handles it gracefully via cooperative `super()` and MRO.
- Prefer composition over inheritance when the relationship is HAS-A, not IS-A.
