# Methods in Python

**OOP-in-Python — Basics**  
Instance methods, class methods, static methods, and method overloading patterns.

## Overview of Method Types

| Method Type | Decorator | First Parameter | Access to Class | Access to Instance |
|---|---|---|---|---|
| Instance | None | `self` | Via `self.__class__` | Yes |
| Class | `@classmethod` | `cls` | Yes | No |
| Static | `@staticmethod` | None | No | No |

## 1. Instance Methods

- Operate on **instance** data.
- First parameter is `self`.
- Can access both instance and class attributes.

```python
class Rectangle:
    def __init__(self, width: float, height: float):
        self.width = width
        self.height = height

    # Instance method — works with self
    def area(self) -> float:
        return self.width * self.height

    def perimeter(self) -> float:
        return 2 * (self.width + self.height)

    def scale(self, factor: float):
        self.width *= factor
        self.height *= factor

r = Rectangle(3, 4)
print(f"Area: {r.area()}, Perimeter: {r.perimeter()}")
r.scale(2)
print(f"After scale — Area: {r.area()}")  # 6 * 8 = 48
```

## 2. Class Methods (`@classmethod`)

- Bound to the **class**, not the instance.
- First parameter is `cls`.
- Often used for **alternative constructors** or class-level state.

```python
class Pizza:
    def __init__(self, ingredients: list):
        self.ingredients = ingredients

    def __str__(self) -> str:
        return f"Pizza with {', '.join(self.ingredients)}"

    # Class method as an alternative constructor
    @classmethod
    def margherita(cls) -> "Pizza":
        return cls(["mozzarella", "tomatoes", "basil"])

    @classmethod
    def pepperoni(cls) -> "Pizza":
        return cls(["mozzarella", "pepperoni", "tomato sauce"])

# Usage — no need to pass ingredients
p1 = Pizza.margherita()
p2 = Pizza.pepperoni()
print(p1)
print(p2)
```

## 3. Static Methods (`@staticmethod`)

- Like a regular function but namespaced inside the class.
- No `self` or `cls` — receives no special first parameter.
- Used for **utility** functions related to the class.

```python
class MathUtils:
    @staticmethod
    def is_even(n: int) -> bool:
        return n % 2 == 0

    @staticmethod
    def is_prime(n: int) -> bool:
        if n < 2:
            return False
        for i in range(2, int(n ** 0.5) + 1):
            if n % i == 0:
                return False
        return True

    @staticmethod
    def gcd(a: int, b: int) -> int:
        while b:
            a, b = b, a % b
        return a

# Called on the class, not an instance
print(MathUtils.is_even(10))   # True
print(MathUtils.is_prime(17))  # True
print(MathUtils.gcd(48, 18))   # 6
```

## 4. When to Use Each

| Use Case | Method Type |
|---|---|
| Needs instance data (`self.xxx`) | Instance method |
| Needs class data or alternative constructors | Class method |
| Utility function related to class (no instance/class data needed) | Static method |
| You *could* make it a standalone function | Static method |

```python
class Demo:
    class_attr = "shared"

    def __init__(self, value: str):
        self.value = value  # Instance data

    def instance_method(self):
        # Needs instance data
        return f"Instance: {self.value}"

    @classmethod
    def class_method(cls):
        # Only needs class data
        return f"Class: {cls.class_attr}"

    @staticmethod
    def static_method(x: int) -> int:
        # No class or instance data needed
        return x * x

d = Demo("test")
print(d.instance_method())  # Instance: test
print(Demo.class_method())  # Class: shared
print(Demo.static_method(5))  # 25
```

## 5. Method Overloading in Python

Python does **not** support traditional method overloading (multiple methods with same name, different signatures). Instead, use:

1. **Default arguments**
2. **Variable-length arguments** (`*args`, `**kwargs`)
3. **Type checking** with `isinstance()`

```python
class Calculator:
    # 'Overloading' via default arguments
    def add(self, a: float, b: float = 0.0, c: float = 0.0) -> float:
        return a + b + c

    # 'Overloading' via *args
    def multiply(self, *args: float) -> float:
        result = 1
        for num in args:
            result *= num
        return result

    # 'Overloading' via type dispatch
    def describe(self, value) -> str:
        if isinstance(value, str):
            return f"String of length {len(value)}"
        elif isinstance(value, (int, float)):
            return f"Number: {value}"
        elif isinstance(value, list):
            return f"List with {len(value)} items"
        else:
            return f"Type: {type(value).__name__}"

calc = Calculator()
print(calc.add(5))            # 5
print(calc.add(5, 10))        # 15
print(calc.add(5, 10, 15))    # 30
print(calc.multiply(2, 3, 4)) # 24
print(calc.describe("hello")) # String of length 5
print(calc.describe(42))      # Number: 42
```

## Problem 1: Employee Class with Company Info

**Difficulty: Easy**

Build an `Employee` class that:
- Tracks employee count via class attribute
- Provides a `@classmethod` to display company info
- Provides a `@classmethod` to create from a comma-separated string

```python
class Employee:
    company = "InnoTech Solutions"
    _count = 0

    def __init__(self, name: str, position: str, salary: float):
        self.name = name
        self.position = position
        self.salary = salary
        Employee._count += 1

    @classmethod
    def company_info(cls) -> str:
        return f"{cls.company} — Total Employees: {cls._count}"

    @classmethod
    def from_string(cls, csv: str) -> "Employee":
        # Alternative constructor: "Alice,Engineer,75000"
        name, position, salary = csv.split(",")
        return cls(name, position, float(salary))

    def __repr__(self) -> str:
        return f"Employee('{self.name}', '{self.position}')"

# Test
e1 = Employee("Alice", "Engineer", 75000)
e2 = Employee.from_string("Bob,Manager,95000")

print(Employee.company_info())  # InnoTech Solutions — Total Employees: 2
print(e2.name, e2.salary)       # Bob 95000.0
```

## Problem 2: Utility Static Methods

**Difficulty: Easy**

Create a `StringUtils` class with static utility methods for string operations.

```python
class StringUtils:
    @staticmethod
    def is_palindrome(s: str) -> bool:
        cleaned = ''.join(c.lower() for c in s if c.isalnum())
        return cleaned == cleaned[::-1]

    @staticmethod
    def reverse_words(s: str) -> str:
        return ' '.join(word[::-1] for word in s.split())

    @staticmethod
    def count_vowels(s: str) -> int:
        return sum(1 for c in s.lower() if c in 'aeiou')

    @staticmethod
    def to_title_case(s: str) -> str:
        return ' '.join(word.capitalize() for word in s.split())

# Test
print(StringUtils.is_palindrome("A man, a plan, a canal: Panama"))  # True
print(StringUtils.reverse_words("hello world"))      # olleh dlrow
print(StringUtils.count_vowels("Beautiful"))          # 5
print(StringUtils.to_title_case("hello world"))      # Hello World
```

## Key Takeaways

- **Instance methods**: need `self`; operate on instance state.
- **Class methods** (`@classmethod`): need `cls`; operate on class state; great for alternative constructors.
- **Static methods** (`@staticmethod`): no `self`/`cls`; pure utility functions.
- Python lacks traditional overloading — use default args, `*args`, or type dispatch.
- Choose the right method type to make your intent clear and your code organized.
