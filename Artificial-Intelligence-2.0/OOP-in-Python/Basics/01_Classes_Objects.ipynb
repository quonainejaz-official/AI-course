## 1. The `class` Keyword

A **class** is a blueprint for creating objects. Use the `class` keyword followed by a name (PascalCase convention).

```python
class ClassName:
    pass
```

| Concept | Syntax | Purpose |
|---------|--------|---------|
| Class definition | `class MyClass:` | Create a blueprint |
| Constructor | `def __init__(self):` | Initialize new objects |
| Self | `self.attribute` | Reference to current instance |
| Object creation | `obj = MyClass()` | Instantiate from class |

````python
# Defining a simple empty class
class Dog:
    pass  # Placeholder — no attributes or methods yet

# Creating an object (instance)
my_dog = Dog()
print(type(my_dog))       # <class '__main__.Dog'>
print(isinstance(my_dog, Dog))  # True
````

## 2. The `__init__` Constructor

The `__init__` method is called **automatically** when an object is created. It initializes instance attributes.

````python
class Dog:
    # Constructor — runs once at object creation
    def __init__(self, name: str, age: int):
        self.name = name  # Instance attribute
        self.age = age    # Instance attribute

    def bark(self) -> str:
        return f"{self.name} says woof!"

# Instantiate with arguments
buddy = Dog("Buddy", 3)
print(buddy.bark())      # Buddy says woof!
print(buddy.age)         # 3
````

## 3. The `self` Parameter

- `self` refers to the **current instance** of the class.
- It must be the **first parameter** of every instance method.
- Python passes it automatically — you never pass it explicitly.

````python
class Counter:
    def __init__(self):
        self.count = 0  # self binds 'count' to this specific instance

    def increment(self):
        self.count += 1  # self ensures we modify THIS object's count

    def get_count(self):
        return self.count

c1 = Counter()
c2 = Counter()
c1.increment()
c1.increment()
c2.increment()
print(f"c1: {c1.get_count()}, c2: {c2.get_count()}")  # c1: 2, c2: 1
````

## 4. Instance Attributes vs Class Attributes

| Feature | Instance Attribute | Class Attribute |
|---------|-------------------|----------------|
| Defined | Inside `__init__` via `self` | Outside any method, directly in class |
| Scope | Per-object | Shared across all instances |
| Changed | One instance only | Affects all instances (if modified via class) |

````python
class Employee:
    # Class attribute — shared by all employees
    company = "TechCorp"
    emp_count = 0

    def __init__(self, name: str, salary: float):
        # Instance attributes — unique per employee
        self.name = name
        self.salary = salary
        Employee.emp_count += 1  # Access class attribute via class name

e1 = Employee("Alice", 70000)
e2 = Employee("Bob", 80000)

print(e1.company)              # TechCorp (inherited from class)
print(Employee.emp_count)      # 2
print(e2.name)                 # Bob (instance-specific)
````

## 5. `__str__` and `__repr__`

- `__str__`: **human-readable** string representation — used by `print()` and `str()`.
- `__repr__`: **unambiguous** representation — used by the REPL and `repr()`. Should ideally be valid Python code to recreate the object.

````python
class Book:
    def __init__(self, title: str, author: str, year: int):
        self.title = title
        self.author = author
        self.year = year

    def __repr__(self) -> str:
        # Should ideally be evaluatable: eval(repr(obj)) == obj
        return f"Book('{self.title}', '{self.author}', {self.year})"

    def __str__(self) -> str:
        # Friendly, readable description
        return f"\"{self.title}\" by {self.author} ({self.year})"

b = Book("1984", "George Orwell", 1949)
print(repr(b))  # Book('1984', 'George Orwell', 1949)
print(str(b))   # "1984" by George Orwell (1949)
print(b)        # (implicitly calls __str__)
````

## 6. `del` and Destructors

Python uses **garbage collection** — you rarely need explicit destructors. The `__del__` method runs when an object is about to be destroyed.

````python
import weakref

class Resource:
    def __init__(self, name: str):
        self.name = name
        print(f"{self.name} acquired")

    def __del__(self):
        # Destructor — called when reference count hits zero
        print(f"{self.name} released")

r = Resource("Database")
del r  # Explicit deletion triggers __del__ immediately
print("After deletion")
````

## Problem 1: BankAccount Class

**Difficulty: Easy**

Design a `BankAccount` class with:
- Owner name and initial balance (default 0)
- `deposit(amount)` — adds to balance
- `withdraw(amount)` — subtracts if sufficient funds
- `__str__` — display account info
- Class attribute `bank_name`

````python
class BankAccount:
    bank_name = "National Bank"  # Class attribute

    def __init__(self, owner: str, balance: float = 0.0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount: float) -> float:
        if amount <= 0:
            raise ValueError("Deposit amount must be positive")
        self.balance += amount
        return self.balance

    def withdraw(self, amount: float) -> float:
        if amount <= 0:
            raise ValueError("Withdrawal amount must be positive")
        if amount > self.balance:
            raise ValueError("Insufficient funds")
        self.balance -= amount
        return self.balance

    def __str__(self) -> str:
        return f"{self.bank_name} — {self.owner}: ${self.balance:.2f}"

# Test
acc = BankAccount("Alice", 1000)
print(acc)                    # National Bank — Alice: $1000.00
acc.deposit(500)
print(f"After deposit: ${acc.balance}")  # $1500
acc.withdraw(200)
print(f"After withdrawal: ${acc.balance}")  # $1300
print(f"Bank name: {BankAccount.bank_name}")
````

## Problem 2: Student Class with Roll Number

**Difficulty: Easy**

Create a `Student` class with:
- Roll number (auto-generated via class attribute counter)
- Name, marks in 3 subjects
- Methods to compute total, average, grade
- `__repr__` for clean debugging output

````python
class Student:
    _id_counter = 1000  # Class attribute for auto-generated roll numbers

    def __init__(self, name: str, marks: list):
        self.roll_no = Student._id_counter
        Student._id_counter += 1
        self.name = name
        self.marks = marks  # List of 3 marks

    def total(self) -> int:
        return sum(self.marks)

    def average(self) -> float:
        return round(self.total() / len(self.marks), 2)

    def grade(self) -> str:
        avg = self.average()
        if avg >= 90:
            return 'A'
        elif avg >= 75:
            return 'B'
        elif avg >= 60:
            return 'C'
        else:
            return 'D'

    def __repr__(self) -> str:
        return f"Student(roll={self.roll_no}, name='{self.name}')"

# Test
s1 = Student("Alice", [85, 92, 78])
s2 = Student("Bob", [72, 68, 74])

for s in (s1, s2):
    print(f"{s.name} (Roll: {s.roll_no}) — "
          f"Total: {s.total()}, Avg: {s.average()}, Grade: {s.grade()}")

# Output:
# Alice (Roll: 1000) — Total: 255, Avg: 85.0, Grade: B
# Bob (Roll: 1001) — Total: 214, Avg: 71.33, Grade: C
````

## Key Takeaways

- **Class**: blueprint; **Object**: instance of a class.
- `__init__` is the constructor — called on instantiation.
- `self` binds attributes and methods to the current instance.
- **Class attributes** are shared; **instance attributes** are per-object.
- `__str__` is for humans; `__repr__` is for developers (aim to be evaluatable).
- `__del__` is the destructor, but you rarely need it thanks to Python's GC.
