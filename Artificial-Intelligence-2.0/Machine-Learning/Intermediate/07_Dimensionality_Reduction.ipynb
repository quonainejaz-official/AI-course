# Dimensionality Reduction

Techniques for reducing the number of features while preserving information. Essential for visualization, noise reduction, and computational efficiency.

## The Curse of Dimensionality

As the number of features increases:
1. Data becomes **sparse** in high-dimensional space
2. Distance metrics become less meaningful (all points appear equally far)
3. Need exponentially more samples for reliable estimation
4. Models become more prone to overfitting
5. Computational cost increases

## Principal Component Analysis (PCA)

PCA finds directions (principal components) that maximize variance.

**How it works**:
1. Center the data (subtract mean)
2. Compute covariance matrix
3. Compute eigenvectors and eigenvalues (these are the PCs)
4. Project data onto top-K eigenvectors

**Key concepts**:
- **Explained variance ratio**: proportion of total variance captured by each component
- **Cumulative explained variance**: sum of variance ratios - used to choose K
- **Components**: the direction vectors (loadings) in original feature space

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_digits, load_iris, load_wine, load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
sns.set_style('whitegrid')
print("Libraries imported.")
```

```python
digits = load_digits()
X, y = digits.data, digits.target
print(f"Digits dataset: {X.shape[0]} samples, {X.shape[1]} features (64 pixels)")
print(f"10 digit classes (0-9)")
```

## PCA for Visualization: 2D Projection of Digits

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
print(f"Reduced shape: {X_pca.shape}")
print(f"Explained variance ratio (2 PCs): {pca.explained_variance_ratio_.round(4)}")
print(f"Total variance captured: {pca.explained_variance_ratio_.sum():.4f}")
```

```python
plt.figure(figsize=(10, 8))
scatter = plt.scatter(X_pca[:, 0], X_pca[:, 1], c=y, cmap='tab10', alpha=0.7, edgecolors='k', s=50)
plt.xlabel(f'PC1 ({pca.explained_variance_ratio_[0]:.2%} variance)')
plt.ylabel(f'PC2 ({pca.explained_variance_ratio_[1]:.2%} variance)')
plt.title('PCA: 2D Projection of Digits Dataset')
plt.colorbar(scatter, label='Digit')
plt.show()
```

```python
pca_full = PCA().fit(X_scaled)
cumulative_var = np.cumsum(pca_full.explained_variance_ratio_)
plt.figure(figsize=(8, 4))
plt.plot(range(1, len(cumulative_var)+1), cumulative_var, 'bo-', linewidth=2)
plt.axhline(y=0.95, color='red', linestyle='--', label='95% variance threshold')
plt.axhline(y=0.99, color='green', linestyle='--', label='99% variance threshold')
plt.xlabel('Number of Components')
plt.ylabel('Cumulative Explained Variance')
plt.title('Choosing Number of PCA Components')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
n_95 = np.argmax(cumulative_var >= 0.95) + 1
n_99 = np.argmax(cumulative_var >= 0.99) + 1
print(f"Components needed for 95% variance: {n_95}")
print(f"Components needed for 99% variance: {n_99}")
print(f"Original dimensions: {X.shape[1]}")
```

```python
n_components_range = [5, 10, 20, 40, 64]
for n in n_components_range:
    pca_temp = PCA(n_components=n)
    X_pca_temp = pca_temp.fit_transform(X_scaled)
    var_explained = pca_temp.explained_variance_ratio_.sum()
    print(f"  {n:2d} components -> {var_explained:.2%} variance retained")
```

## PCA Components as Images

Each principal component can be reshaped back into an 8x8 image to see what pattern it captures.

```python
fig, axes = plt.subplots(2, 5, figsize=(10, 4))
for i, ax in enumerate(axes.ravel()):
    component = pca_full.components_[i].reshape(8, 8)
    ax.imshow(component, cmap='RdBu')
    ax.set_title(f'PC {i+1}')
    ax.axis('off')
plt.suptitle('First 10 Principal Components (as images)')
plt.tight_layout()
plt.show()
```

## t-SNE: Non-linear Dimensionality Reduction

t-SNE (t-distributed Stochastic Neighbor Embedding):
- Preserves **local** structure (neighbor relationships)
- Non-linear, probabilistic technique
- Great for visualization, NOT for preprocessing
- Perplexity parameter balances local vs global structure

```python
tsne = TSNE(n_components=2, perplexity=30, random_state=42)
X_tsne = tsne.fit_transform(X_scaled)
print(f"t-SNE reduced shape: {X_tsne.shape}")
plt.figure(figsize=(10, 8))
scatter = plt.scatter(X_tsne[:, 0], X_tsne[:, 1], c=y, cmap='tab10', alpha=0.7, edgecolors='k', s=50)
plt.title('t-SNE: 2D Projection of Digits (perplexity=30)')
plt.colorbar(scatter, label='Digit')
plt.show()
print("Note: t-SNE separates classes much better than PCA for non-linear data.")
```

```python
perplexities = [5, 15, 30, 50]
fig, axes = plt.subplots(1, 4, figsize=(20, 4))
for ax, perp in zip(axes, perplexities):
    tsne_temp = TSNE(n_components=2, perplexity=perp, random_state=42)
    X_tsne_temp = tsne_temp.fit_transform(X_scaled)
    ax.scatter(X_tsne_temp[:, 0], X_tsne_temp[:, 1], c=y, cmap='tab10', alpha=0.7, s=10)
    ax.set_title(f'Perplexity={perp}')
    ax.axis('off')
plt.suptitle('t-SNE with Different Perplexity Values')
plt.tight_layout()
plt.show()
```

## PCA vs t-SNE: Comparison

| Aspect | PCA | t-SNE |
|--------|-----|-------|
| Type | Linear | Non-linear |
| Speed | Fast | Slow (O(n²)) |
| Deterministic | Yes | No (different runs differ) |
| Preserves | Global variance | Local neighbor structure |
| Use | Preprocessing, visualization | Visualization only |
| Interpretable | Yes (components) | No (axes have no meaning) |

```python
pca_wine = PCA(n_components=2)
wine = load_wine()
X_w = scaler.fit_transform(wine.data)
X_w_pca = pca_wine.fit_transform(X_w)
plt.figure(figsize=(8, 5))
plt.scatter(X_w_pca[:, 0], X_w_pca[:, 1], c=wine.target, cmap='viridis', alpha=0.7, edgecolors='k')
plt.xlabel(f'PC1 ({pca_wine.explained_variance_ratio_[0]:.1%})')
plt.ylabel(f'PC2 ({pca_wine.explained_variance_ratio_[1]:.1%})')
plt.title('PCA on Wine Dataset')
plt.colorbar(label='Wine cultivar')
plt.show()
print(f"Wine dataset: {X_w.shape[1]} features -> 2 PCs ({pca_wine.explained_variance_ratio_.sum():.1%} variance)")
```

```python
pca = PCA(n_components=30)
X_train_pca = pca.fit_transform(X_scaled)
X_train, X_test, y_train, y_test = train_test_split(X_train_pca, y, test_size=0.2, random_state=42)
rf_pca = RandomForestClassifier(n_estimators=100, random_state=42)
rf_pca.fit(X_train, y_train)
acc_pca = accuracy_score(y_test, rf_pca.predict(X_test))
X_train_f, X_test_f, y_train_f, y_test_f = train_test_split(X_scaled, y, test_size=0.2, random_state=42)
rf_full = RandomForestClassifier(n_estimators=100, random_state=42)
rf_full.fit(X_train_f, y_train_f)
acc_full = accuracy_score(y_test_f, rf_full.predict(X_test_f))
print(f"RF on 64 features:      {acc_full:.4f}")
print(f"RF on 30 PCA features:  {acc_pca:.4f}")
print(f"Time saved: ~{(1 - 30/64)*100:.0f}% fewer features")
```

## Feature Extraction vs Feature Selection

**Feature Extraction** (PCA, t-SNE):
- Creates NEW features from combinations of original features
- Reduces dimensionality but loses interpretability
- Good for: visualization, noise reduction, unsupervised learning

**Feature Selection** (SelectKBest, RFE):
- Selects a SUBSET of original features
- Preserves interpretability
- Good for: understanding important features, reducing overfitting

```python
X_iris, y_iris = load_iris(return_X_y=True)
pca_iris = PCA(n_components=2)
X_iris_pca = pca_iris.fit_transform(X_iris)
print("Iris: 4D -> 2D via PCA")
print(f"Explained variance: {pca_iris.explained_variance_ratio_.sum():.2%}")
print("\nPCA components (loadings):")
loadings = pd.DataFrame(pca_iris.components_.T, columns=['PC1', 'PC2'], index=load_iris().feature_names)
print(loadings.round(3))
```

```python
print("""
Interview Q: When to use PCA vs t-SNE?

PCA:
- As a preprocessing step before ML (noise reduction)
- When you need deterministic, reproducible results
- For large datasets (fast to compute)
- When interpretability of components matters

t-SNE:
- ONLY for visualization (exploratory data analysis)
- When you want to see clusters/hidden structure
- For small to medium datasets (n < 10,000)
- When linear methods fail to separate classes
""")
```
