## Problem 1: Movie Recommendation System

### Design
- **Collaborative Filtering**: users who liked similar movies will like this movie
- **Content-Based Filtering**: recommend movies with similar attributes (genre, cast)
- **Hybrid**: combine both approaches

```python
np.random.seed(42)
n_movies = 50
n_users = 200
movies = [f"Movie_{i}" for i in range(n_movies)]
genres = ['Action', 'Comedy', 'Drama', 'Horror', 'Sci-Fi', 'Romance']
movie_genres = {m: np.random.choice(genres) for m in movies}
ratings_data = []
for user in range(n_users):
    n_rated = np.random.randint(5, 20)
    rated_movies = np.random.choice(movies, n_rated, replace=False)
    for movie in rated_movies:
        rating = np.random.randint(1, 6)
        ratings_data.append({'user_id': user, 'movie': movie, 'rating': rating})
ratings_df = pd.DataFrame(ratings_data)
print(f"Ratings matrix: {ratings_df.shape}")
print(f"Users: {ratings_df['user_id'].nunique()}, Movies: {ratings_df['movie'].nunique()}")
print(ratings_df.head())
```

```python
# Build user-item matrix for collaborative filtering
user_item_matrix = ratings_df.pivot_table(index='user_id', columns='movie', values='rating').fillna(0)
print(f"User-Item matrix: {user_item_matrix.shape}")
user_similarity = cosine_similarity(user_item_matrix)
user_similarity_df = pd.DataFrame(user_similarity, index=user_item_matrix.index, columns=user_item_matrix.index)
print(f"User similarity matrix: {user_similarity_df.shape}")
def recommend_collaborative(user_id, n_recommendations=5):
    similar_users = user_similarity_df[user_id].sort_values(ascending=False)[1:11]
    user_ratings = user_item_matrix.loc[user_id]
    unrated_movies = user_ratings[user_ratings == 0].index
    scores = {}
    for movie in unrated_movies:
        sim_users_who_rated = similar_users.index[user_item_matrix.loc[similar_users.index, movie] > 0]
        if len(sim_users_who_rated) > 0:
            scores[movie] = user_item_matrix.loc[sim_users_who_rated, movie].mean()
    recommendations = sorted(scores.items(), key=lambda x: x[1], reverse=True)[:n_recommendations]
    return [movie for movie, _ in recommendations]
recs = recommend_collaborative(user_id=0)
print(f"Collaborative recommendations for User 0: {recs}")
```

```python
# Content-based filtering
movie_features = pd.get_dummies(pd.Series(movie_genres)).reindex(movies).fillna(0)
movie_similarity = cosine_similarity(movie_features)
movie_similarity_df = pd.DataFrame(movie_similarity, index=movies, columns=movies)
def recommend_content_based(movie_title, n_recommendations=5):
    similar = movie_similarity_df[movie_title].sort_values(ascending=False)[1:n_recommendations+1]
    return list(similar.index)
recs_content = recommend_content_based('Movie_0')
print(f"Users who liked Movie_0 also liked (content-based): {recs_content}")
print(f"Genre of Movie_0: {movie_genres['Movie_0']}")
print(f"Genres of recommendations: {[movie_genres[m] for m in recs_content]}")
```

## Problem 2: Customer Churn Prediction

### Design Considerations
- **Features**: usage patterns, demographics, support tickets, payment history
- **Target**: churned (left) vs retained
- **Imbalance**: typically 5-15% churn rate
- **Actionability**: predictions must lead to retention actions (discounts, outreach)

```python
np.random.seed(42)
n_customers = 5000
churn_data = pd.DataFrame({
    'tenure_months': np.random.randint(1, 72, n_customers),
    'monthly_charges': np.random.uniform(20, 120, n_customers),
    'total_charges': np.random.uniform(100, 8000, n_customers),
    'support_tickets': np.random.poisson(1, n_customers),
    'contract_type': np.random.choice(['Month-to-month', '1-year', '2-year'], n_customers),
    'payment_method': np.random.choice(['Electronic', 'Mailed', 'Bank', 'Credit'], n_customers)
})
churn_data['churn'] = (
    (churn_data['support_tickets'] > 2) |
    (churn_data['tenure_months'] < 6) |
    (churn_data['contract_type'] == 'Month-to-month')
).astype(int)
churn_data['churn'] = (churn_data['churn'] + np.random.binomial(1, 0.1, n_customers)) > 0
churn_data['churn'] = churn_data['churn'].astype(int)
print(f"Churn distribution: {churn_data['churn'].value_counts().to_dict()}")
print(f"Churn rate: {churn_data['churn'].mean()*100:.1f}%")
```

```python
X_churn = pd.get_dummies(churn_data.drop('churn', axis=1), drop_first=True)
y_churn = churn_data['churn']
X_c_train, X_c_test, y_c_train, y_c_test = train_test_split(X_churn, y_churn, test_size=0.2, random_state=42, stratify=y_churn)
scaler = StandardScaler()
X_c_train_s = scaler.fit_transform(X_c_train)
X_c_test_s = scaler.transform(X_c_test)
churn_model = GradientBoostingClassifier(n_estimators=100, learning_rate=0.1, max_depth=3, random_state=42)
churn_model.fit(X_c_train_s, y_c_train)
y_c_pred = churn_model.predict(X_c_test_s)
print("Customer Churn Prediction Results:")
print(f"  Accuracy: {accuracy_score(y_c_test, y_c_pred):.4f}")
print(f"  Precision: {precision_score(y_c_test, y_c_pred):.4f}")
print(f"  Recall: {recall_score(y_c_test, y_c_pred):.4f}")
print(f"  ROC-AUC: {roc_auc_score(y_c_test, churn_model.predict_proba(X_c_test_s)[:, 1]):.4f}")
print(classification_report(y_c_test, y_c_pred, target_names=['Retained', 'Churned']))
```

```python
# Feature importance for churn
churn_importance = pd.DataFrame({
    'Feature': X_churn.columns,
    'Importance': churn_model.feature_importances_
}).sort_values('Importance', ascending=False)
print("Top churn predictors:")
print(churn_importance.head(8).to_string(index=False))
print("\nAction: Target short-tenure, high-ticket customers with retention offers.")
```

## Problem 3: Real-time Fraud Detection Pipeline

### Architecture Considerations
- **Real-time scoring**: predict fraud within milliseconds
- **Feature store**: user historical patterns, device fingerprint, location
- **Model monitoring**: detect data drift (fraud patterns evolve)
- **Threshold tuning**: optimize false positive rate vs detection rate

```python
np.random.seed(42)
n_transactions = 10000
fraud_data = pd.DataFrame({
    'amount': np.random.exponential(100, n_transactions),
    'distance_from_home': np.random.exponential(50, n_transactions),
    'hour_of_day': np.random.randint(0, 24, n_transactions),
    'is_weekend': np.random.binomial(1, 0.3, n_transactions),
    'num_previous_transactions': np.random.poisson(5, n_transactions),
    'avg_transaction_amount_30d': np.random.exponential(80, n_transactions)
})
fraud_data['is_fraud'] = (
    (fraud_data['amount'] > 300) |
    (fraud_data['distance_from_home'] > 100) |
    ((fraud_data['hour_of_day'] >= 0) & (fraud_data['hour_of_day'] <= 5))
).astype(int) * np.random.binomial(1, 0.5, n_transactions)
print(f"Fraud rate: {fraud_data['is_fraud'].mean()*100:.2f}%")
print(fraud_data.head())
```

```python
X_f, y_f = fraud_data.drop('is_fraud', axis=1), fraud_data['is_fraud']
X_f_train, X_f_test, y_f_train, y_f_test = train_test_split(X_f, y_f, test_size=0.3, random_state=42, stratify=y_f)
scaler_f = StandardScaler()
X_f_train_s = scaler_f.fit_transform(X_f_train)
X_f_test_s = scaler_f.transform(X_f_test)
fraud_model = RandomForestClassifier(n_estimators=100, class_weight='balanced', random_state=42)
fraud_model.fit(X_f_train_s, y_f_train)
y_f_pred = fraud_model.predict(X_f_test_s)
y_f_prob = fraud_model.predict_proba(X_f_test_s)[:, 1]
print("Fraud Detection Results:")
print(f"  Precision: {precision_score(y_f_test, y_f_pred):.4f}")
print(f"  Recall: {recall_score(y_f_test, y_f_pred):.4f}")
print(f"  F1: {f1_score(y_f_test, y_f_pred):.4f}")
print(f"  ROC-AUC: {roc_auc_score(y_f_test, y_f_prob):.4f}")
```

```python
# Deployment pipeline
fraud_pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('classifier', RandomForestClassifier(n_estimators=100, class_weight='balanced', random_state=42))
])
fraud_pipeline.fit(X_f_train, y_f_train)
new_transaction = pd.DataFrame([{
    'amount': 450, 'distance_from_home': 200, 'hour_of_day': 3,
    'is_weekend': 1, 'num_previous_transactions': 2, 'avg_transaction_amount_30d': 75
}])
fraud_prob = fraud_pipeline.predict_proba(new_transaction)[0, 1]
fraud_pred = fraud_pipeline.predict(new_transaction)[0]
print("Real-time Fraud Detection:")
print(f"  New transaction: $450, 3AM, 200 miles from home")
print(f"  Fraud probability: {fraud_prob:.2%}")
print(f"  Prediction: {'FRAUD' if fraud_pred else 'LEGITIMATE'}")
```

## A/B Testing for ML Models

Framework: Old Model (control) vs New Model (treatment)

1. **Random assignment**: 50% users get old model, 50% get new
2. **Metrics**: precision, recall, business metric (revenue, retention)
3. **Statistical significance**: p-value, confidence intervals
4. **Guardrails**: monitor for regressions in other metrics

```python
print("""
A/B Testing for Churn Model:

1. Split customers into control (50%) and treatment (50%)
2. Control: existing rule-based system
3. Treatment: ML model (GradientBoosting)
4. Metrics: 
   - Churn reduction rate
   - Retention offer cost
   - Customer lifetime value (LTV)
5. Duration: 4-8 weeks (enough for churn cycle)
6. Statistical test: chi-squared or t-test
""")
```

## Model Monitoring: Data Drift & Concept Drift

- **Data Drift**: input distribution changes (e.g., users get older)
- **Concept Drift**: relationship between features and target changes (e.g., fraud patterns evolve)
- **Detection**: monitor feature distributions (PSI), prediction distribution, performance metrics

```python
print("""
Monitoring Dashboard Metrics:

| Metric | Method | Alert Trigger |
|--------|--------|---------------|
| Data Drift | PSI (Population Stability Index) | PSI > 0.1 |
| Prediction Drift | Mean prediction shift | > 2 std dev |
| Accuracy Drop | Rolling accuracy | > 5% drop |
| Feature Nulls | % missing values | > 1% |
| Latency | P99 response time | > 500ms |
""")
```

```python
def psi(expected, actual, bins=10):
    expected_hist, _ = np.histogram(expected, bins=bins, range=(0, 1))
    actual_hist, _ = np.histogram(actual, bins=bins, range=(0, 1))
    expected_pct = expected_hist / len(expected) + 1e-10
    actual_pct = actual_hist / len(actual) + 1e-10
    return np.sum((actual_pct - expected_pct) * np.log(actual_pct / expected_pct))
baseline_probs = pd.Series(np.random.beta(2, 5, 1000))
current_probs = pd.Series(np.random.beta(3, 4, 1000))
psi_value = psi(baseline_probs, current_probs)
print(f"Population Stability Index (PSI): {psi_value:.4f}")
if psi_value < 0.1:
    print("No significant drift detected.")
elif psi_value < 0.25:
    print("Moderate drift detected. Investigate.")
else:
    print("Severe drift! Model retraining needed.")
```
