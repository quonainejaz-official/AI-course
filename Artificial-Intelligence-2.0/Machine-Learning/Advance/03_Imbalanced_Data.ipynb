# Handling Imbalanced Data

Strategies for classification problems where class distribution is severely skewed. Covers resampling, SMOTE, weighted loss functions, and specialized ensemble methods.

## The Class Imbalance Problem

When one class (majority) vastly outnumbers another (minority), standard classifiers become biased toward the majority class.

### The Accuracy Paradox

A model that predicts **all samples as majority class** can achieve high accuracy but is useless. With 99% majority class, a dumb model scores 99% accuracy.

**Always use**: Precision, Recall, F1-Score, ROC-AUC, PR-AUC — NOT accuracy.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    confusion_matrix, classification_report, roc_auc_score,
    precision_recall_curve, average_precision_score
)
sns.set_style('whitegrid')
print("Libraries imported.")
```

```python
X, y = make_classification(
    n_samples=10000, n_features=20, n_informative=10, n_redundant=5,
    weights=[0.95, 0.05], flip_y=0.05, random_state=42
)
print(f"Class distribution:")
print(f"  Class 0 (majority): {np.sum(y==0)} ({np.mean(y==0)*100:.1f}%)")
print(f"  Class 1 (minority): {np.sum(y==1)} ({np.mean(y==1)*100:.1f}%)")
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
scaler = StandardScaler()
X_train_s = scaler.fit_transform(X_train)
X_test_s = scaler.transform(X_test)
```

## The Accuracy Paradox in Action

```python
lr = LogisticRegression(random_state=42)
lr.fit(X_train_s, y_train)
y_pred = lr.predict(X_test_s)
print("Logistic Regression on imbalanced data:")
print(f"  Accuracy:  {accuracy_score(y_test, y_pred):.4f}")
print(f"  Precision: {precision_score(y_test, y_pred):.4f}")
print(f"  Recall:    {recall_score(y_test, y_pred):.4f}")
print(f"  F1-Score:  {f1_score(y_test, y_pred):.4f}")
print(f"  ROC-AUC:   {roc_auc_score(y_test, lr.predict_proba(X_test_s)[:,1]):.4f}")
print(classification_report(y_test, y_pred, target_names=['Majority', 'Minority']))
```

## Random Undersampling

Randomly remove samples from the majority class to balance the distribution.
- **Pro**: Simple, fast
- **Con**: Loses potentially useful data

```python
try:
    from imblearn.under_sampling import RandomUnderSampler
    from imblearn.over_sampling import RandomOverSampler, SMOTE
    from imblearn.ensemble import EasyEnsembleClassifier
    rus = RandomUnderSampler(random_state=42)
    X_train_rus, y_train_rus = rus.fit_resample(X_train_s, y_train)
    print(f"After undersampling:")
    print(f"  Class 0: {np.sum(y_train_rus==0)}, Class 1: {np.sum(y_train_rus==1)}")
    lr_rus = LogisticRegression(random_state=42).fit(X_train_rus, y_train_rus)
    y_pred_rus = lr_rus.predict(X_test_s)
    print(f"  F1: {f1_score(y_test, y_pred_rus):.4f}, Recall: {recall_score(y_test, y_pred_rus):.4f}")
except ImportError:
    print("imbalanced-learn not installed. Install with: pip install imbalanced-learn")
```

## Random Oversampling

Randomly duplicate minority class samples.
- **Pro**: No data loss
- **Con**: Can cause overfitting (just copies of existing data)

```python
try:
    ros = RandomOverSampler(random_state=42)
    X_train_ros, y_train_ros = ros.fit_resample(X_train_s, y_train)
    print(f"After oversampling:")
    print(f"  Class 0: {np.sum(y_train_ros==0)}, Class 1: {np.sum(y_train_ros==1)}")
    lr_ros = LogisticRegression(random_state=42).fit(X_train_ros, y_train_ros)
    y_pred_ros = lr_ros.predict(X_test_s)
    print(f"  F1: {f1_score(y_test, y_pred_ros):.4f}, Recall: {recall_score(y_test, y_pred_ros):.4f}")
except ImportError:
    print("imbalanced-learn not installed.")
```

## SMOTE (Synthetic Minority Over-sampling Technique)

Creates **synthetic** samples by interpolating between minority class neighbors rather than duplicating.

**Algorithm**:
1. For each minority sample, find its k nearest neighbors
2. Randomly select one neighbor
3. Create a synthetic point along the line connecting them
4. Repeat until class balance is achieved

```python
try:
    smote = SMOTE(random_state=42)
    X_train_sm, y_train_sm = smote.fit_resample(X_train_s, y_train)
    print(f"After SMOTE:")
    print(f"  Class 0: {np.sum(y_train_sm==0)}, Class 1: {np.sum(y_train_sm==1)}")
    lr_sm = LogisticRegression(random_state=42).fit(X_train_sm, y_train_sm)
    y_pred_sm = lr_sm.predict(X_test_s)
    print(f"  F1: {f1_score(y_test, y_pred_sm):.4f}")
    print(f"  Recall: {recall_score(y_test, y_pred_sm):.4f}")
except ImportError:
    print("imbalanced-learn not installed.")
```

```python
try:
    models = {
        'No Resampling': LogisticRegression(random_state=42).fit(X_train_s, y_train),
        'Undersampling': LogisticRegression(random_state=42).fit(X_train_rus, y_train_rus),
        'Oversampling': LogisticRegression(random_state=42).fit(X_train_ros, y_train_ros),
        'SMOTE': LogisticRegression(random_state=42).fit(X_train_sm, y_train_sm)
    }
    results = []
    for name, model in models.items():
        y_p = model.predict(X_test_s)
        results.append({
            'Method': name,
            'Accuracy': accuracy_score(y_test, y_p),
            'Precision': precision_score(y_test, y_p),
            'Recall': recall_score(y_test, y_p),
            'F1': f1_score(y_test, y_p)
        })
    print("Comparison of Resampling Methods:")
    print(pd.DataFrame(results).round(4).to_string(index=False))
except:
    print("Run previous cells first.")
```

## Weighted Loss Functions

Most sklearn classifiers support `class_weight='balanced'` which automatically adjusts weights inversely proportional to class frequencies.

```python
lr_weighted = LogisticRegression(class_weight='balanced', random_state=42)
lr_weighted.fit(X_train_s, y_train)
y_pred_w = lr_weighted.predict(X_test_s)
print("Logistic Regression with class_weight='balanced':")
print(f"  F1: {f1_score(y_test, y_pred_w):.4f}")
print(f"  Recall: {recall_score(y_test, y_pred_w):.4f}")
print(f"  Weight: {lr_weighted.class_weight_}")
```

```python
rf_weighted = RandomForestClassifier(n_estimators=100, class_weight='balanced', random_state=42)
rf_weighted.fit(X_train_s, y_train)
y_pred_rf = rf_weighted.predict(X_test_s)
print("Random Forest with class_weight='balanced':")
print(f"  F1: {f1_score(y_test, y_pred_rf):.4f}")
print(f"  Recall: {recall_score(y_test, y_pred_rf):.4f}")
print(f"  ROC-AUC: {roc_auc_score(y_test, rf_weighted.predict_proba(X_test_s)[:,1]):.4f}")
```

## Ensemble Methods for Imbalanced Data

**EasyEnsemble**: builds multiple balanced subsets via undersampling, trains a classifier on each, and combines them.

```python
try:
    eec = EasyEnsembleClassifier(n_estimators=10, random_state=42)
    eec.fit(X_train_s, y_train)
    y_pred_ee = eec.predict(X_test_s)
    print("EasyEnsembleClassifier:")
    print(f"  F1: {f1_score(y_test, y_pred_ee):.4f}")
    print(f"  Recall: {recall_score(y_test, y_pred_ee):.4f}")
except ImportError:
    print("imbalanced-learn not installed.")
```
