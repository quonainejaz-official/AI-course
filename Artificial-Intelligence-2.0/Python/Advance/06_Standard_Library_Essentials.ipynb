## collections — Specialized Data Structures

The `collections` module provides alternatives to built-in data types.

```python
from collections import Counter, defaultdict, OrderedDict, namedtuple, deque

# Counter — count hashable objects
cnt = Counter('abracadabra')
print(f"Counter: {dict(cnt)}")
print(f"Most common 2: {cnt.most_common(2)}")

# defaultdict — factory default for missing keys
dd = defaultdict(list)
dd['group1'].append(1)
dd['group1'].append(2)
dd['group2'].append(3)
print(f"\ndefaultdict: {dict(dd)}")

# OrderedDict — remembers insertion order (dict already does this in 3.7+)
od = OrderedDict()
od['a'] = 1
od['b'] = 2
od['c'] = 3
print(f"\nOrderedDict: {od}")

# namedtuple — lightweight immutable data containers
Point = namedtuple('Point', ['x', 'y', 'z'])
p = Point(1, 2, 3)
print(f"\nnamedtuple: {p}, x={p.x}, y={p.y}")

# deque — double-ended queue
dq = deque([1, 2, 3])
dq.appendleft(0)
dq.append(4)
dq.rotate(1)  # move all elements right by 1
print(f"\ndeque after rotate(1): {list(dq)}")
```

## datetime Module

Provides classes for manipulating dates, times, and intervals.

```python
from datetime import date, time, datetime, timedelta, timezone

# Current date and time
today = date.today()
now = datetime.now()
utc_now = datetime.now(timezone.utc)

print(f"Today: {today}")
print(f"Now:   {now}")
print(f"UTC:   {utc_now}")

# Creating specific dates/times
independence = date(1947, 8, 15)
meeting = time(14, 30, 0)
event = datetime(2024, 12, 25, 10, 0, 0)

print(f"\nIndependence day: {independence} ({independence.strftime('%A')})")
print(f"Meeting at: {meeting}")
print(f"Christmas: {event}")

# Date arithmetic
delta = timedelta(days=7, hours=3)
future = now + delta
past = now - delta
print(f"\n7 days + 3 hours from now: {future}")
print(f"7 days + 3 hours ago: {past}")

# Difference between dates
d1 = date(2024, 1, 1)
d2 = date(2025, 1, 1)
diff = d2 - d1
print(f"\nDays until 2025: {diff.days}")
```

## random Module

Generates pseudo-random numbers with various distributions.

```python
import random

# Seeding for reproducibility
random.seed(42)

# Basic random functions
print(f"Random float [0, 1): {random.random():.4f}")
print(f"Random int [1, 10]: {random.randint(1, 10)}")
print(f"Uniform [2.5, 7.5]: {random.uniform(2.5, 7.5):.2f}")

# Sequence operations
items = ['red', 'green', 'blue', 'yellow', 'black']
print(f"\nChoice: {random.choice(items)}")
print(f"Sample (3): {random.sample(items, 3)}")

# Shuffle
deck = list(range(1, 53))
random.shuffle(deck)
print(f"\nFirst 5 shuffled: {deck[:5]}")

# Distributions
print(f"\nGaussian(0, 1): {random.gauss(0, 1):.4f}")
print(f"Exponential(1): {random.expovariate(1):.4f}")
```

## itertools — Iterator Building Blocks

Provides iterator functions for efficient looping.

```python
import itertools

# Infinite iterators
counter = itertools.count(start=1, step=2)
print(f"First 5 from count(1, 2): {[next(counter) for _ in range(5)]}")

cycler = itertools.cycle(["A", "B", "C"])
print(f"First 7 from cycle: {[next(cycler) for _ in range(7)]}")

# Combinatoric iterators
items = [1, 2, 3]
print(f"\nPermutations: {list(itertools.permutations(items, 2))}")
print(f"Combinations: {list(itertools.combinations(items, 2))}")
print(f"Product: {list(itertools.product(items, repeat=2))}")
print(f"Combinations with replacement: {list(itertools.combinations_with_replacement(items, 2))}")

# Grouping
data = [("A", 1), ("A", 2), ("B", 3), ("B", 4), ("B", 5)]
for key, group in itertools.groupby(data, key=lambda x: x[0]):
    print(f"Group {key}: {list(group)}")

# accumulate — running reduction
print(f"\nAccumulate (sum): {list(itertools.accumulate([1, 2, 3, 4, 5]))}")
import operator
print(f"Accumulate (mul): {list(itertools.accumulate([1, 2, 3, 4, 5], operator.mul))}")
```

## json Module

JSON encoding/decoding for data interchange.

```python
import json

# Python -> JSON
data = {
    "name": "Alice",
    "age": 30,
    "is_student": False,
    "courses": ["Python", "ML"],
    "address": None
}

json_str = json.dumps(data, indent=2, sort_keys=True)
print("Python -> JSON:")
print(json_str)

# JSON -> Python
parsed = json.loads(json_str)
print(f"\nJSON -> Python: {parsed}")
print(f"Type: {type(parsed)}")

# Custom serialization
from datetime import datetime

class DateTimeEncoder(json.JSONEncoder):
    def default(self, obj):
        if isinstance(obj, datetime):
            return obj.isoformat()
        return super().default(obj)

data_with_date = {"event": "meeting", "time": datetime(2024, 6, 15, 14, 30)}
print(f"\nCustom serialization: {json.dumps(data_with_date, cls=DateTimeEncoder, indent=2)}")
```

## argparse — Command-Line Argument Parsing

Build command-line interfaces with argument parsing, help, and error messages.

```python
import argparse

def create_parser():
    """Create an argument parser for a sample CLI tool."""
    parser = argparse.ArgumentParser(
        description="Sample CLI tool for processing files",
        epilog="Example: python script.py input.txt -o output.txt --verbose"
    )
    
    # Positional argument
    parser.add_argument(
        "input",
        help="Input file path"
    )
    
    # Optional arguments
    parser.add_argument(
        "-o", "--output",
        help="Output file path (default: output.txt)",
        default="output.txt"
    )
    
    parser.add_argument(
        "-v", "--verbose",
        help="Increase output verbosity",
        action="store_true"
    )
    
    parser.add_argument(
        "--limit",
        type=int,
        default=100,
        help="Number of lines to process (default: 100)"
    )
    
    return parser

# Simulate command-line arguments
import sys
sys.argv = ["script.py", "data.txt", "-v", "--limit", "50", "-o", "result.txt"]

parser = create_parser()
args = parser.parse_args()

print("Parsed arguments:")
print(f"  Input:    {args.input}")
print(f"  Output:   {args.output}")
print(f"  Verbose:  {args.verbose}")
print(f"  Limit:    {args.limit}")
```

## logging Module

Flexible logging system with multiple severity levels.

```python
import logging

# Configure logging
logging.basicConfig(
    level=logging.DEBUG,
    format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',
    datefmt='%Y-%m-%d %H:%M:%S',
    filename='app.log',
    filemode='w'
)

# Also log to console
console = logging.StreamHandler()
console.setLevel(logging.INFO)
console.setFormatter(logging.Formatter('%(levelname)s: %(message)s'))
logging.getLogger('').addHandler(console)

# Log messages at different levels
logger = logging.getLogger(__name__)
logger.debug("Detailed debug information")
logger.info("Application started successfully")
logger.warning("Configuration file not found, using defaults")
logger.error("Failed to connect to database")

try:
    1 / 0
except ZeroDivisionError:
    logger.exception("An error occurred")  # logs with traceback

# Log levels: DEBUG < INFO < WARNING < ERROR < CRITICAL
print("Logging demo complete. Check 'app.log' for full output.")

# Show log file content
with open('app.log') as f:
    print(f"\nFirst 3 lines of log file:")
    for i, line in enumerate(f):
        if i >= 3:
            break
        print(f"  {line.rstrip()}")
```

## pickle — Object Serialization

Serialize Python objects to a binary format for storage/transmission.

```python
import pickle

# Complex Python object
data = {
    "users": [
        {"name": "Alice", "scores": [85, 92, 78]},
        {"name": "Bob", "scores": [90, 87, 95]},
    ],
    "metadata": {"version": 1, "created": "2024-01-15"}
}

# Serialize to bytes
pickled_bytes = pickle.dumps(data)
print(f"Pickled size: {len(pickled_bytes)} bytes")

# Deserialize back
unpickled = pickle.loads(pickled_bytes)
print(f"Unpickled: {unpickled}")

# Save to file
with open("data.pkl", "wb") as f:
    pickle.dump(data, f)

# Load from file
with open("data.pkl", "rb") as f:
    loaded = pickle.load(f)

print(f"\nLoaded from file: {loaded}")
print(f"Equals original: {data == loaded}")

import os
os.remove("data.pkl")
os.remove("app.log")
```

### Problem 1: Build an LRU Cache with OrderedDict
Implement a Least Recently Used (LRU) cache.

```python
from collections import OrderedDict

class LRUCache:
    """LRU Cache using OrderedDict."""
    def __init__(self, capacity: int):
        self.capacity = capacity
        self.cache = OrderedDict()
    
    def get(self, key):
        """Get value for key; move to end (most recently used)."""
        if key not in self.cache:
            return -1
        self.cache.move_to_end(key)  # mark as recently used
        return self.cache[key]
    
    def put(self, key, value):
        """Add/update key-value; evict LRU if over capacity."""
        if key in self.cache:
            self.cache.move_to_end(key)
        self.cache[key] = value
        if len(self.cache) > self.capacity:
            self.cache.popitem(last=False)  # remove LRU (first item)

# Test
cache = LRUCache(3)
cache.put(1, "one")
cache.put(2, "two")
cache.put(3, "three")
print(f"After 3 puts: {dict(cache.cache)}")

cache.get(1)  # access 1, making 2 the LRU
print(f"After get(1): {dict(cache.cache)}")

cache.put(4, "four")  # should evict 2
print(f"After put(4):  {dict(cache.cache)}")

print(f"get(2): {cache.get(2)}")  # -1 (evicted)
print(f"get(1): {cache.get(1)}")  # 'one'

# Time: O(1) for both get and put
# Space: O(capacity)
```

### Problem 2: Log Analyzer with Counter
Analyze server log entries to find the most active IP addresses.

```python
from collections import Counter
import re

def analyze_server_log(log_lines: list, top_n: int = 5) -> list:
    """Analyze log lines and return top N most frequent IPs."""
    ip_pattern = r'\b(?:\d{1,3}\.){3}\d{1,3}\b'
    ips = []
    
    for line in log_lines:
        match = re.search(ip_pattern, line)
        if match:
            ips.append(match.group())
    
    return Counter(ips).most_common(top_n)

# Sample log data
logs = [
    "192.168.1.1 - - [15/Jan/2024:10:00:00] GET /index.html",
    "10.0.0.2 - - [15/Jan/2024:10:00:01] POST /login",
    "192.168.1.1 - - [15/Jan/2024:10:00:02] GET /about",
    "10.0.0.3 - - [15/Jan/2024:10:00:03] GET /contact",
    "192.168.1.1 - - [15/Jan/2024:10:00:04] GET /products",
    "10.0.0.2 - - [15/Jan/2024:10:00:05] POST /checkout",
    "10.0.0.2 - - [15/Jan/2024:10:00:06] GET /cart",
    "172.16.0.1 - - [15/Jan/2024:10:00:07] GET /images/logo.png",
]

top_ips = analyze_server_log(logs)
print("Top IP addresses:")
for ip, count in top_ips:
    print(f"  {ip:15} {count} requests")

# Time: O(n) for counting, O(k log k) for most_common
# Space: O(k) where k = unique IPs
```

### Problem 3: CLI Tool with argparse
Create a word count utility (like wc).

```python
import argparse
import sys

def word_count(text: str) -> dict:
    """Count lines, words, and characters in text."""
    lines = len(text.splitlines())
    words = len(text.split())
    chars = len(text)
    return {"lines": lines, "words": words, "chars": chars}

def create_wc_parser():
    """Create argument parser for wc-like tool."""
    parser = argparse.ArgumentParser(description="Word count utility")
    parser.add_argument("file", nargs="?", help="Input file")
    parser.add_argument("-l", "--lines", action="store_true", help="Count lines")
    parser.add_argument("-w", "--words", action="store_true", help="Count words")
    parser.add_argument("-c", "--chars", action="store_true", help="Count characters")
    return parser

# Simulate CLI
test_content = "Hello World\nThis is a test\nLine three\n"
with open("wc_test.txt", "w") as f:
    f.write(test_content)

# Simulate: python wc.py wc_test.txt -l -w -c
sys.argv = ["wc.py", "wc_test.txt", "-l", "-w", "-c"]
parser = create_wc_parser()
args = parser.parse_args()

# Process file
if args.file:
    with open(args.file, "r") as f:
        text = f.read()
    counts = word_count(text)
    
    # If no flags set, show all
    show_all = not (args.lines or args.words or args.chars)
    
    parts = []
    if show_all or args.lines:
        parts.append(f"{counts['lines']} lines")
    if show_all or args.words:
        parts.append(f"{counts['words']} words")
    if show_all or args.chars:
        parts.append(f"{counts['chars']} chars")
    
    print(f"{' '.join(parts)} in {args.file}")

import os
os.remove("wc_test.txt")
```
