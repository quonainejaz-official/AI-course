# Regular Expressions in Python

Regular expressions (**regex**) are patterns used to match character combinations in strings. Python's `re` module provides full regex support.

### Use Cases
- **Validation**: Email, phone, password, URL
- **Extraction**: Pull data from text (logs, HTML)
- **Search/Replace**: Find and replace patterns
- **Parsing**: Tokenize and parse structured text

> **Prerequisite**: Strings, file handling.

## The `re` Module — Core Functions

| Function | Description | Returns |
|----------|-------------|---------|
| `search(pattern, text)` | Find first match anywhere | `Match` or `None` |
| `match(pattern, text)` | Match at beginning of string | `Match` or `None` |
| `fullmatch(pattern, text)` | Match entire string | `Match` or `None` |
| `findall(pattern, text)` | Find all non-overlapping matches | `list` of strings |
| `finditer(pattern, text)` | Find all matches lazily | iterator of `Match` |
| `sub(pattern, repl, text)` | Replace matches | `str` |
| `split(pattern, text)` | Split on pattern | `list` |
| `compile(pattern)` | Compile pattern for reuse | `Pattern` object |

```python
import re

# search() — find first match anywhere
text = "My email is alice@email.com and bob@test.org"
match = re.search(r'\w+@\w+\.\w+', text)
if match:
    print(f"search result: {match.group()} (span: {match.span()})")

# match() — only matches at start of string
m1 = re.match(r'\w+', "hello world")
m2 = re.match(r'\w+', "  hello world")
print(f"match at start: {m1.group() if m1 else None}")
print(f"match not at start: {m2}")

# fullmatch() — matches entire string
m3 = re.fullmatch(r'\d{5}', "12345")
m4 = re.fullmatch(r'\d{5}', "12345-6789")
print(f"fullmatch (exact): {m3.group() if m3 else None}")
print(f"fullmatch (longer): {m4}")
```

```python
# findall() and finditer()
text = "Contact: alice@email.com, bob@test.org, or charlie@work.net"

# findall — returns list of matches
emails = re.findall(r'\w+@\w+\.\w+', text)
print(f"findall: {emails}")

# finditer — returns iterator of Match objects
for match in re.finditer(r'\w+@\w+\.\w+', text):
    print(f"  Email: {match.group():20} at {match.span()}")

# sub() — replace matches
censored = re.sub(r'\w+@\w+\.\w+', '[EMAIL]', text)
print(f"\nsub: {censored}")

# split() — split on pattern
csv_line = "apple,banana;cherry|date"
parts = re.split(r'[,;|]', csv_line)
print(f"split: {parts}")
```

## Metacharacters and Character Classes

### Metacharacters
Characters with special meaning: `. ^ $ * + ? { } [ ] \ | ( )`

### Character Classes

| Pattern | Matches |
|---------|---------|
| `\d` | A digit (0-9) |
| `\D` | A non-digit |
| `\w` | Word character (a-z, A-Z, 0-9, _) |
| `\W` | Non-word character |
| `\s` | Whitespace (space, tab, newline) |
| `\S` | Non-whitespace |
| `.` | Any character except newline |
| `[abc]` | Any of a, b, or c |
| `[a-z]` | Range from a to z |
| `[^abc]` | Not a, b, or c |

```python
# Character classes
text = "My phone: (555) 123-4567, ZIP: 90210, Price: $49.99"

# \d — digits
digits = re.findall(r'\d+', text)
print(f"Digits: {digits}")

# \w — word characters
words = re.findall(r'\w+', text)
print(f"Words: {words}")

# Custom character class — vowels
vowels = re.findall(r'[aeiouAEIOU]+', text)
print(f"Vowels: {vowels}")

# Negated character class — non-vowels
non_vowels = re.findall(r'[^aeiouAEIOU\s\d\W]+', text)
print(f"Non-vowels: {non_vowels}")

# Range in character class — hex digits
hex_matches = re.findall(r'[0-9A-Fa-f]+', "Colors: FF5733, 00FF00, 0000FF")
print(f"Hex colors: {hex_matches}")
```

## Quantifiers and Anchors

### Quantifiers

| Quantifier | Meaning | Example |
|------------|---------|---------|
| `*` | 0 or more | `\d*` — empty or digits |
| `+` | 1 or more | `\d+` — at least one digit |
| `?` | 0 or 1 | `colou?r` — color or colour |
| `{n}` | Exactly n | `\d{5}` — ZIP code |
| `{n,}` | n or more | `\d{3,}` — 3+ digits |
| `{n,m}` | n to m | `\d{2,4}` — 2 to 4 digits |

### Anchors

| Anchor | Matches |
|--------|---------|
| `^` | Start of string |
| `$` | End of string |
| `\b` | Word boundary |
| `\B` | Non-word boundary |

```python
# Quantifiers
text = "I have 2 cats, 15 dogs, and 100 fish. My ZIP is 90210."

print(f"1+ digits: {re.findall(r'\d+', text)}")
print(f"Exactly 5 digits: {re.findall(r'\b\d{5}\b', text)}")
print(f"2-3 digits: {re.findall(r'\b\d{2,3}\b', text)}")

# Greedy vs lazy quantifiers
html = "<div><p>Hello</p><p>World</p></div>"
greedy = re.findall(r'<.*>', html)  # greedy: matches as much as possible
lazy = re.findall(r'<.*?>', html)   # lazy: matches as little as possible
print(f"\nGreedy: {greedy}")
print(f"Lazy:   {lazy}")

# Anchors
lines = ["hello world", "hello", "world hello"]
for line in lines:
    start = bool(re.match(r'^hello', line))
    end = bool(re.search(r'world$', line))
    print(f"'{line:15}' ^hello:{start}, world$:{end}")

# Word boundary \b
text2 = "cat category catalog concatenate cat"
print(f"\nWhole word 'cat': {re.findall(r'\bcat\b', text2)}")
print(f"Words starting with 'cat': {re.findall(r'\bcat\w*', text2)}")
```

## Groups and Capturing

**Groups** allow you to extract specific parts of a match. Use parentheses `()` to create groups.

```python
# Capturing groups
text = "Alice: (555) 123-4567, Bob: (888) 987-6543"

pattern = r'(\w+): \((\d{3})\) (\d{3})-(\d{4})'
for match in re.finditer(pattern, text):
    name, area, prefix, line = match.groups()
    print(f"Name: {name:6} | Area: {area} | Phone: {prefix}-{line}")

# Named groups — (?P<name>...)
pattern2 = r'(?P<name>\w+): \((?P<area>\d{3})\) (?P<prefix>\d{3})-(?P<line>\d{4})'
match = re.search(pattern2, text)
if match:
    print(f"\nNamed groups: {match.groupdict()}")

# Non-capturing groups — (?:...)
text2 = "color or colour"
matches = re.findall(r'col(?:o|ou)r', text2)
print(f"Non-capturing: {matches}")

# Backreferences — \1, \2, etc.
dup_words = "the the book is is great"
fixed = re.sub(r'\b(\w+)\s+\1\b', r'\1', dup_words)
print(f"\nBefore: {dup_words}")
print(f"After:  {fixed}")
```

## Lookahead and Lookbehind

**Lookaround** assertions match a position without consuming characters.

| Assertion | Syntax | Matches |
|-----------|--------|---------|
| Positive lookahead | `(?=...)` | Followed by |
| Negative lookahead | `(?!...)` | Not followed by |
| Positive lookbehind | `(?<=...)` | Preceded by |
| Negative lookbehind | `(?<!...)` | Not preceded by |

```python
# Lookahead — match only if followed by
text = "1 apple, 2 apples, 3 apple pies"
matches = re.findall(r'\d+(?=\s+apple)', text)  # digits before 'apple'
print(f"Lookahead (digits before apple): {matches}")

# Negative lookahead — match if NOT followed by
text2 = "apple, application, apple pie, apple"
matches2 = re.findall(r'\bapple(?!\s+pie)\b', text2)  # 'apple' not followed by ' pie'
print(f"Negative lookahead: {matches2}")

# Lookbehind — match only if preceded by
prices = "Price: $10.50, Cost: $5.00, $20.00"
dollar_amounts = re.findall(r'(?<=\$)\d+\.\d{2}', prices)
print(f"Lookbehind (amounts after $): {dollar_amounts}")

# Negative lookbehind — match if NOT preceded by
text3 = "Python is fun. I love Python."
matches3 = re.findall(r'(?<!\.\s)\bPython\b', text3)
print(f"Negative lookbehind: {matches3}")

# Practical: match numbers not followed by %
data = "Growth: 5%, Increase: 100 units, Rate: 3.5%"
plain_numbers = re.findall(r'\d+\.?\d*(?!%)', data)
print(f"Numbers not followed by %: {plain_numbers}")
```

## Regex Flags

| Flag | Abbrev | Effect |
|------|--------|--------|
| `re.IGNORECASE` | `re.I` | Case-insensitive matching |
| `re.MULTILINE` | `re.M` | `^`/`$` match start/end of each line |
| `re.DOTALL` | `re.S` | `.` matches newline too |
| `re.VERBOSE` | `re.X` | Allow whitespace and comments in pattern |
| `re.ASCII` | `re.A` | Make `\w`, `\d`, `\s` ASCII-only |

```python
import re

# re.IGNORECASE
text = "Python PYTHON python"
print(f"Case insensitive: {re.findall(r'python', text, re.I)}")

# re.MULTILINE — ^ and $ apply to each line
multiline = "Line 1\nLine 2\nLine 3"
print(f"Without MULTILINE, ^ matches: {re.findall(r'^\w+', multiline)}")
print(f"With MULTILINE, ^ matches:    {re.findall(r'^\w+', multiline, re.M)}")

# re.DOTALL — . matches newlines
text2 = "Hello\nWorld"
print(f"Without DOTALL: {re.findall(r'Hello.World', text2)}")
print(f"With DOTALL:    {re.findall(r'Hello.World', text2, re.S)}")

# re.VERBOSE — readable patterns with comments
pattern = re.compile(r'''
    \b          # word boundary
    [A-Z][a-z]*  # capitalized word
    \s+         # whitespace
    \d{3}       # 3 digits
    \b          # word boundary
''', re.VERBOSE)

result = pattern.findall("Alice 123, Bob 456, Charlie 789")
print(f"VERBOSE pattern: {result}")
```

## Compiling Patterns

Pre-compile regex patterns for **performance** when reusing them.

```python
# Compiling patterns
email_pattern = re.compile(r'\b[\w.-]+@[\w.-]+\.\w{2,}\b')
phone_pattern = re.compile(r'\(?\d{3}\)?[-.\s]?\d{3}[-.]\d{4}')

text = "Contact us: support@company.com or (555) 123-4567"

emails = email_pattern.findall(text)
phones = phone_pattern.findall(text)

print(f"Emails: {emails}")
print(f"Phones: {phones}")

# Performance comparison
import time

text_big = "alice@email.com " * 10000

# Uncompiled
start = time.perf_counter()
re.findall(r'\w+@\w+\.\w+', text_big)
t1 = time.perf_counter() - start

# Compiled
pattern = re.compile(r'\w+@\w+\.\w+')
start = time.perf_counter()
pattern.findall(text_big)
t2 = time.perf_counter() - start

print(f"\nUncompiled: {t1:.4f}s")
print(f"Compiled:   {t2:.4f}s")
```

## Interview Problems

### Problem 1: Email Validator
Validate email addresses according to common rules.

```python
import re

def validate_email(email: str) -> bool:
    """Validate an email address format."""
    pattern = r'^[a-zA-Z0-9][\w.-]*@[a-zA-Z0-9][\w.-]*\.[a-zA-Z]{2,}$'
    return bool(re.fullmatch(pattern, email))

test_emails = [
    "alice@email.com",          # valid
    "alice@email",              # invalid (no TLD)
    "@email.com",               # invalid (no local part)
    "alice@.com",               # invalid (no domain)
    "alice@sub.domain.com",     # valid
    "alice@email.co.uk",        # valid
    "alice name@email.com",     # invalid (space)
]

print("Email Validator:")
for email in test_emails:
    print(f"  {'✓' if validate_email(email) else '✗'} {email}")
```

### Problem 2: Phone Number Extractor
Extract phone numbers in various formats.

```python
def extract_phone_numbers(text: str) -> list:
    """Extract US phone numbers in various formats."""
    # Patterns: (555) 123-4567, 555-123-4567, 555.123.4567
    pattern = r'\(?\d{3}\)?[-.\s]?\d{3}[-.]\d{4}'
    return re.findall(pattern, text)

def normalize_phone(phone: str) -> str:
    """Normalize extracted phone to (XXX) XXX-XXXX format."""
    digits = re.sub(r'\D', '', phone)
    return f"({digits[:3]}) {digits[3:6]}-{digits[6:]}"

# Test
sample = "Call us at (555) 123-4567 or 888-987-6543 or 415.555.1212"
phones = extract_phone_numbers(sample)

print("Extracted phone numbers:")
for phone in phones:
    normalized = normalize_phone(phone)
    print(f"  {phone:20} -> {normalized}")
```

### Problem 3: Password Strength Checker
Check password strength using regex lookaheads.

```python
def password_strength(password: str) -> dict:
    """Check password strength using multiple criteria."""
    checks = {
        "length_8+": len(password) >= 8,
        "has_uppercase": bool(re.search(r'[A-Z]', password)),
        "has_lowercase": bool(re.search(r'[a-z]', password)),
        "has_digit": bool(re.search(r'\d', password)),
        "has_special": bool(re.search(r'[!@#$%^&*(),.?":{}|<>]', password)),
        "no_whitespace": not bool(re.search(r'\s', password)),
    }
    
    score = sum(checks.values())
    if score <= 2:
        strength = "Weak"
    elif score <= 4:
        strength = "Moderate"
    elif score == 5:
        strength = "Strong"
    else:
        strength = "Very Strong"
    
    return {"score": score, "strength": strength, "checks": checks}

# Test
passwords = ["abc", "Password", "Pass1234", "P@ssw0rd", "Str0ng!P@ss"]
print(f"{'Password':<20} {'Score':<6} {'Strength'}")
print("-" * 42)
for pwd in passwords:
    result = password_strength(pwd)
    print(f"{pwd:<20} {result['score']:<6} {result['strength']}")
```

### Problem 4: URL Parser
Parse URLs into components (protocol, domain, path, query, fragment).

```python
def parse_url(url: str) -> dict:
    """Parse a URL into its components."""
    pattern = r'^(?P<protocol>https?)://(?P<domain>[^/?#]+)(?P<path>/[^?#]*)?(?:\?(?P<query>[^#]*))?(?:#(?P<fragment>.*))?$'
    match = re.search(pattern, url)
    if match:
        return {k: v for k, v in match.groupdict().items() if v is not None}
    return {}

# Test
urls = [
    "https://www.example.com",
    "https://example.com/path/to/page",
    "https://api.example.com/v1/users?page=1&limit=10",
    "http://blog.example.com/post#comments",
    "https://shop.example.com/products?id=123&sort=price#reviews",
]

print("URL Parsing:")
for url in urls:
    parsed = parse_url(url)
    print(f"\n  URL: {url}")
    for k, v in parsed.items():
        print(f"    {k:10}: {v}")
```
