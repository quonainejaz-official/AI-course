# String Handling in Python

Strings are **sequences of characters** — one of Python's most versatile data types.

### Key Properties:
- **Immutable**: Strings cannot be modified in place
- **Iterable**: Can loop over characters
- **Ordered**: Characters have a defined position
- **Indexable**: Access characters by position (0-indexed)

> **Prerequisite**: Variables, data types, basic operators.

## String Creation and Fundamentals

Python supports single quotes, double quotes, triple quotes (multiline), and raw strings.

```python
# Different ways to create strings
single = 'Hello'
double = "World"
multiline = '''This is a
multiline string'''
raw = r"C:\Users\Name"  # raw string — backslashes are literal

print(f"Single: {single}")
print(f"Double: {double}")
print(f"Multiline:\n{multiline}")
print(f"Raw: {raw}")

# String immutability demonstration
s = "hello"
# s[0] = 'H'  # TypeError! Strings are immutable
s = "Hello"  # This creates a NEW string and reassigns
print(f"Reassigned: {s}")

# String concatenation and repetition
concat = "Py" + "thon"  # concatenation
repeated = "Ha" * 3     # repetition
print(f"Concat: {concat}, Repeated: {repeated}")
```

## Escape Sequences

| Sequence | Meaning |
|----------|---------|
| `\\` | Backslash |
| `\'` | Single quote |
| `\"` | Double quote |
| `\n` | Newline |
| `\t` | Tab |
| `\r` | Carriage return |
| `\b` | Backspace |
| `\xhh` | Hex value |
| `\uhhhh` | Unicode (16-bit) |

```python
# Escape sequences in action
print("Line1\nLine2")             # newline
print("Column1\tColumn2")         # tab
print("She said, \"Hello!\"")    # escaped quotes
print("Backslash: \\")             # backslash
print("\u03A9 is omega")          # unicode Ω
print("\x41 is A")                # hex value for 'A'
```

## String Indexing and Slicing

Strings support the same indexing and slicing as lists:
- `s[i]` — character at index `i`
- `s[i:j]` — substring from `i` (inclusive) to `j` (exclusive)
- `s[i:j:k]` — with step `k`
- Negative indices count from the end

```python
# String slicing
s = "Python Programming"

print(f"String: '{s}'")
print(f"Length: {len(s)}")
print(f"First char: '{s[0]}'")
print(f"Last char: '{s[-1]}'")
print(f"First 6: '{s[:6]}'")         # 'Python'
print(f"From index 7: '{s[7:]}'")    # 'Programming'
print(f"Middle: '{s[7:14]}'")        # 'Program'
print(f"Every 2nd char: '{s[::2]}'")
print(f"Reversed: '{s[::-1]}'")

# Practical: string reversal
word = "racecar"
if word == word[::-1]:
    print(f"'{word}' is a palindrome")
```

## Essential String Methods

Python strings have a rich set of built-in methods. Here are the most important ones:

### Case Conversion
`lower()`, `upper()`, `title()`, `capitalize()`, `swapcase()`

```python
# Case conversion
s = "hello WORLD"
print(f"lower():    '{s.lower()}'")
print(f"upper():    '{s.upper()}'")
print(f"title():    '{s.title()}'")
print(f"capitalize(): '{s.capitalize()}'")
print(f"swapcase(): '{s.swapcase()}'")

# Checking case
print(f"\nislower():  {s.islower()}")
print(f"isupper():  {s.isupper()}")
print(f"istitle():  {s.istitle()}")
```

### Search and Replace

`find()`, `index()`, `rfind()`, `count()`, `replace()`, `startswith()`, `endswith()`

```python
# Search and replace methods
s = "the quick brown fox jumps over the lazy dog"

print(f"String: '{s}'")
print(f"find('fox'): {s.find('fox')}")            # 16
print(f"find('cat'): {s.find('cat')}")            # -1 (not found)
print(f"index('fox'): {s.index('fox')}")          # 16
print(f"rfind('the'): {s.rfind('the')}")           # 31 (rightmost)
print(f"count('the'): {s.count('the')}")           # 2
print(f"startswith('the'): {s.startswith('the')}") # True
print(f"endswith('dog'): {s.endswith('dog')}")     # True

replaced = s.replace('fox', 'cat')
print(f"replace: '{replaced}'")
```

### Split, Join, and Strip

`split()`, `rsplit()`, `splitlines()`, `join()`, `strip()`, `lstrip()`, `rstrip()`

```python
# Split, join, strip
csv_data = "apple,banana,cherry,date"
fruits = csv_data.split(',')
print(f"split: {fruits}")

# join — the inverse of split
rejoined = ' | '.join(fruits)
print(f"join: '{rejoined}'")

# rsplit with limit
print(f"rsplit (max=2): {csv_data.rsplit(',', 2)}")

# strip — removes whitespace (or specified characters)
messy = "   Hello, World!   \n"
print(f"strip:    '{messy.strip()}'")    # 'Hello, World!'
print(f"lstrip:   '{messy.lstrip()}'")   # 'Hello, World!   \n'
print(f"rstrip:   '{messy.rstrip()}'")   # '   Hello, World!'

# Strip specific characters
url = "https://example.com/"
print(f"strip /: '{url.strip('/')}'")
```

### Validation Methods

`isalpha()`, `isdigit()`, `isalnum()`, `isspace()`, `isdecimal()`

```python
# String validation
tests = ["Hello", "123", "Hello123", "   ", "2024-01-15"]

print(f"{'String':<15} {'isalpha':<8} {'isdigit':<8} {'isalnum':<8} {'isspace':<8}")
print("-" * 47)
for s in tests:
    print(f"{s:<15} {str(s.isalpha()):<8} {str(s.isdigit()):<8} "
          f"{str(s.isalnum()):<8} {str(s.isspace()):<8}")
```

## String Formatting

Three approaches, in order of preference:
1. **f-strings** (Python 3.6+) — `f"{variable}"`
2. **`str.format()`** — `"{}".format(variable)`
3. **`%`-formatting** — `"%s" % variable` (old style)

```python
# f-strings (modern, preferred)
name, age, score = "Alice", 30, 92.5678
print(f"Name: {name}, Age: {age}, Score: {score:.1f}%")
print(f"Binary: {42:b}, Hex: {255:x}, Octal: {42:o}")
print(f"Percentage: {0.125:.2%}")
print(f"Padding: |{'hi':>10}| |{'hi':<10}| |{'hi':^10}|")

# str.format()
print("\nName: {}, Age: {}".format(name, age))
print("Name: {n}, Age: {a}".format(n=name, a=age))

# % formatting (legacy)
print("\nName: %s, Age: %d, Score: %.2f" % (name, age, score))

# Format specifiers table
print("\n--- Format Specifiers ---")
print(f"Integer:           {42:d}")
print(f"Float:             {3.14159:.3f}")
print(f"Scientific:        {3.14159:.2e}")
print(f"Percentage:        {0.75:.1%}")
print(f"Comma separator:   {1234567:,d}")
print(f"Left aligned:      |{'hello':10}|")
print(f"Right aligned:     |{'hello':>10}|")
print(f"Center aligned:    |{'hello':^10}|")
```

## Interview Problems

### Problem 1: Reverse a String
Reverse a string in three different ways.

```python
def reverse_slicing(s: str) -> str:
    """Reverse using slicing (most Pythonic)."""
    return s[::-1]

def reverse_loop(s: str) -> str:
    """Reverse using a loop."""
    result = ""
    for ch in s:
        result = ch + result  # prepend each character
    return result

def reverse_recurse(s: str) -> str:
    """Reverse using recursion."""
    if len(s) <= 1:
        return s
    return reverse_recurse(s[1:]) + s[0]

# Test
test = "Python"
print(f"Original: {test}")
print(f"Slicing:  {reverse_slicing(test)}")
print(f"Loop:     {reverse_loop(test)}")
print(f"Recurse:  {reverse_recurse(test)}")

# Time: O(n) for all, Space: O(n) for all
```

### Problem 2: Count Vowels and Consonants
Count the number of vowels and consonants in a string.

```python
def count_vowels_consonants(s: str):
    """Return (vowel_count, consonant_count) ignoring non-alphabetic chars."""
    vowels = set("aeiouAEIOU")
    v_count = c_count = 0
    for ch in s:
        if ch.isalpha():
            if ch in vowels:
                v_count += 1
            else:
                c_count += 1
    return v_count, c_count

# Test
test_str = "Python Programming 101!"
v, c = count_vowels_consonants(test_str)
print(f"String: '{test_str}'")
print(f"Vowels: {v}, Consonants: {c}")

# Time: O(n), Space: O(1)
```

### Problem 3: Remove Punctuation
Remove all punctuation from a string, leaving only letters, numbers, and spaces.

```python
import string

def remove_punctuation(s: str) -> str:
    """Remove punctuation characters from a string."""
    # string.punctuation contains: !"#$%&'()*+,-./:;<=>?@[\]^_`{|}~
    translator = str.maketrans('', '', string.punctuation)
    return s.translate(translator)

def remove_punctuation_comprehension(s: str) -> str:
    """Alternative using list comprehension."""
    return ''.join(ch for ch in s if ch not in string.punctuation)

# Test
test = "Hello, World! How's it going? (Python 3.10)"
print(f"Original: {test}")
print(f"Cleaned:  {remove_punctuation(test)}")

# Practical: word frequency counter
text = "It's a beautiful, beautiful day! Isn't it?"
cleaned = remove_punctuation(text)
words = cleaned.lower().split()
from collections import Counter
freq = Counter(words)
print(f"\nWord frequencies: {dict(freq)}")
```

### Problem 4: Check if String is a Valid Password
A valid password must have at least 8 characters, 1 uppercase, 1 lowercase, 1 digit, and 1 special character.

```python
import string

def is_valid_password(pwd: str) -> tuple:
    """Validate password strength. Returns (is_valid, reason)."""
    if len(pwd) < 8:
        return False, "Too short (min 8 chars)"
    if not any(ch.isupper() for ch in pwd):
        return False, "Needs uppercase letter"
    if not any(ch.islower() for ch in pwd):
        return False, "Needs lowercase letter"
    if not any(ch.isdigit() for ch in pwd):
        return False, "Needs at least one digit"
    if not any(ch in string.punctuation for ch in pwd):
        return False, "Needs at least one special character"
    return True, "Valid password"

# Test
passwords = ["abc", "Password", "Pass1234", "Pass1234!", "Strong!Pass#1"]
for pwd in passwords:
    valid, reason = is_valid_password(pwd)
    print(f"'{pwd:20}' -> {'✓' if valid else '✗'} {reason}")
```
