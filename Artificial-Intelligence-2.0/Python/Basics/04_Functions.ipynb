# Functions in Python

Functions are **reusable blocks of code** that perform a specific task. They help with:
- **Modularity**: Breaking code into logical units
- **Reusability**: Write once, use many times
- **Abstraction**: Hide implementation details

> **Prerequisite**: Variables, operators, data structures, control flow.

## Defining Functions

Syntax:
```python
def function_name(parameters):
    """Docstring: describes what the function does."""
    # function body
    return value
```

### Parameters vs Arguments
- **Parameters**: Variables listed in the function definition
- **Arguments**: Values passed to the function when calling it

````python
# Basic function definition
def greet(name):
    """Return a greeting message for the given name."""
    return f"Hello, {name}!"

# Calling the function
message = greet("Alice")
print(message)

# Function with multiple parameters and return value
def rectangle_area(length, width):
    """Calculate the area of a rectangle."""
    return length * width

area = rectangle_area(5, 3)
print(f"Area: {area}")
````

## Scope and LEGB Rule

Python resolves variable names using the **LEGB** rule:
1. **L**ocal — inside the current function
2. **E**nclosing — in enclosing functions (nested functions)
3. **G**lobal — at module level
4. **B**uilt-in — Python's built-in names

````python
# LEGB scope demonstration
x = "global x"  # global scope

def outer():
    x = "outer x"  # enclosing scope
    
    def inner():
        x = "inner x"  # local scope
        print(f"Inner: {x}")
    
    inner()
    print(f"Outer: {x}")

outer()
print(f"Global: {x}")

# Using global keyword
counter = 0

def increment():
    global counter  # declare intent to modify global
    counter += 1

increment()
increment()
print(f"Counter after 2 increments: {counter}")
````

## Default Arguments

Parameters can have **default values**. Default arguments are evaluated **once** at function definition time.

````python
# Default arguments
def power(base, exp=2):
    """Raise base to the given exponent (default: square)."""
    return base ** exp

print(f"3^2 = {power(3)}")        # uses default exp=2
print(f"3^4 = {power(3, 4)}")     # overrides default

# WARNING: Mutable default arguments are shared across calls!
def bad_append(item, lst=[]):  # [] is created once
    lst.append(item)
    return lst

print(f"First call: {bad_append(1)}")   # [1]
print(f"Second call: {bad_append(2)}")  # [1, 2]  -- BUG!

# Fix: use None as default
def good_append(item, lst=None):
    if lst is None:
        lst = []
    lst.append(item)
    return lst

print(f"First call: {good_append(1)}")   # [1]
print(f"Second call: {good_append(2)}")  # [2]
````

## *args and **kwargs

- `*args` — captures **positional** arguments as a **tuple**
- `**kwargs` — captures **keyword** arguments as a **dict**

Useful for wrapping functions, decorators, and flexible APIs.

````python
# *args — variable positional arguments
def sum_all(*args):
    """Sum any number of arguments."""
    total = 0
    for num in args:
        total += num
    return total

print(f"sum_all(1, 2, 3) = {sum_all(1, 2, 3)}")
print(f"sum_all(1, 2, 3, 4, 5) = {sum_all(1, 2, 3, 4, 5)}")

# **kwargs — variable keyword arguments
def print_details(**kwargs):
    """Print key-value pairs from keyword arguments."""
    for key, value in kwargs.items():
        print(f"  {key}: {value}")

print("\nDetails:")
print_details(name="Alice", age=30, city="NYC")

# Combined usage
def func(a, b, *args, **kwargs):
    print(f"a={a}, b={b}")
    print(f"args={args}")
    print(f"kwargs={kwargs}")

func(1, 2, 3, 4, 5, x=10, y=20)
````

## Keyword-Only and Positional-Only Arguments

Python 3 supports more precise parameter control:
- Arguments after `*` are **keyword-only**
- Arguments before `/` are **positional-only**

````python
# Keyword-only arguments (after *)
def greet_kw(name, *, greeting="Hello", punctuation="!"):
    """greeting and punctuation must be passed as keywords."""
    return f"{greeting}, {name}{punctuation}"

print(greet_kw("Alice"))                          # works
print(greet_kw("Bob", greeting="Hi"))            # works
# print(greet_kw("Bob", "Hi"))                   # TypeError!

# Positional-only arguments (before /)
def divide(a, b, /):
    """a and b must be passed positionally."""
    return a / b

print(f"divide(10, 3) = {divide(10, 3)}")
# print(divide(a=10, b=3))  # TypeError!
````

## Lambda Functions

Lambda functions are **anonymous**, **single-expression** functions.

Syntax: `lambda parameters: expression`

Use them for short, throwaway functions — especially with `map()`, `filter()`, `sorted()`.

````python
# Basic lambda
square = lambda x: x ** 2
print(f"Square of 5: {square(5)}")

# Lambda with sorted() — custom sort key
people = [("Alice", 30), ("Bob", 25), ("Charlie", 35)]
sorted_by_age = sorted(people, key=lambda p: p[1])
print(f"Sorted by age: {sorted_by_age}")

# Lambda with map()
numbers = [1, 2, 3, 4, 5]
doubled = list(map(lambda x: x * 2, numbers))
print(f"Doubled: {doubled}")

# Lambda with filter()
evens = list(filter(lambda x: x % 2 == 0, numbers))
print(f"Evens: {evens}")
````

## Docstrings and Type Hints

**Docstrings** document what a function does. **Type hints** indicate expected parameter and return types.

````python
# Function with complete documentation
def calculate_bmi(weight_kg: float, height_m: float) -> float:
    """
    Calculate Body Mass Index (BMI).
    
    Args:
        weight_kg: Weight in kilograms
        height_m: Height in meters
    
    Returns:
        BMI value rounded to 2 decimal places
    
    Raises:
        ValueError: If height_m is 0 or negative
    """
    if height_m <= 0:
        raise ValueError("Height must be positive")
    bmi = weight_kg / (height_m ** 2)
    return round(bmi, 2)

# Test
bmi = calculate_bmi(70, 1.75)
print(f"BMI: {bmi}")

# Viewing docstring
print(f"\nDocstring:\n{calculate_bmi.__doc__}")

# help(calculate_bmi)  # would show full help in interactive mode
````

## Recursion

A recursive function calls itself. Every recursive function needs:
1. **Base case** — stops the recursion
2. **Recursive case** — calls itself with modified arguments

````python
# Recursive factorial
def factorial(n: int) -> int:
    """Calculate n! recursively."""
    if n <= 1:  # base case
        return 1
    return n * factorial(n - 1)  # recursive case

print(f"Factorial 5: {factorial(5)}")

# Recursive Fibonacci
def fibonacci(n: int) -> int:
    """Return the nth Fibonacci number (0-indexed) recursively."""
    if n <= 1:
        return n  # base: F(0)=0, F(1)=1
    return fibonacci(n - 1) + fibonacci(n - 2)

print(f"Fibonacci F(10): {fibonacci(10)}")

# Time: O(2^n) for naive recursive Fibonacci, O(n) for factorial
# Space: O(n) call stack depth for both
````

## Interview Problems

### Problem 1: Palindrome Checker
Check if a string is a palindrome (reads same forwards and backwards).

````python
def is_palindrome(s: str) -> bool:
    """Check if s is a palindrome (case-insensitive, ignores non-alphanumeric)."""
    # Filter and normalize
    cleaned = ''.join(ch.lower() for ch in s if ch.isalnum())
    return cleaned == cleaned[::-1]

# Test cases
tests = [
    ("racecar", True),
    ("hello", False),
    ("A man, a plan, a canal: Panama", True),
    ("Was it a car or a cat I saw?", True),
    ("", True),
]

for s, expected in tests:
    result = is_palindrome(s)
    status = "✓" if result == expected else "✗"
    print(f"{status} '{s[:40]}...' -> {result}")

# Time: O(n), Space: O(n)
````

### Problem 2: Anagram Checker
Check if two strings are anagrams of each other (same characters, different order).

````python
def are_anagrams(s1: str, s2: str) -> bool:
    """Check if two strings are anagrams, ignoring case and spaces."""
    # Normalize: lowercase, remove spaces, sort characters
    return sorted(s1.lower().replace(' ', '')) == sorted(s2.lower().replace(' ', ''))

# Alternative: using Counter (O(n) instead of O(n log n))
from collections import Counter

def are_anagrams_fast(s1: str, s2: str) -> bool:
    """Check anagrams using character frequency count."""
    def normalize(s):
        return Counter(ch.lower() for ch in s if ch != ' ')
    return normalize(s1) == normalize(s2)

# Test
pairs = [
    ("listen", "silent"),
    ("hello", "world"),
    ("debit card", "bad credit"),
    ("Astronomer", "Moon starer"),
]

for s1, s2 in pairs:
    print(f"'{s1}' vs '{s2}' -> {are_anagrams_fast(s1, s2)}")

# Time: O(n log n) for sorted approach, O(n) for Counter
# Space: O(n)
````

### Problem 3: Recursive Sum of Digits
Write a recursive function that sums the digits of a number.

````python
def sum_digits(n: int) -> int:
    """Recursively sum the digits of a non-negative integer."""
    if n < 10:  # base case: single digit
        return n
    return n % 10 + sum_digits(n // 10)  # last digit + recursive call

# Test
test_nums = [123, 4567, 9, 0, 1000000]
for n in test_nums:
    print(f"sum_digits({n}) = {sum_digits(n)}")

# Time: O(log n) — one recursive call per digit
# Space: O(log n) — call stack depth equals number of digits
````
