# Python Basics

## What is Python?
Python is a **high-level**, **interpreted**, **general-purpose** programming language created by Guido van Rossum in 1991. It emphasizes code readability with its notable use of significant indentation.

### Key Features:
- **Interpreted**: Executes code line by line, no compilation needed
- **Dynamically typed**: No need to declare variable types explicitly
- **Object-oriented**: Supports classes, inheritance, polymorphism
- **Large standard library**: "Batteries included" philosophy
- **Cross-platform**: Runs on Windows, macOS, Linux

### Python vs Other Languages

| Feature | Python | Java | C++ |
|---------|--------|------|-----|
| Typing | Dynamic | Static | Static |
| Syntax | Minimal | Verbose | Complex |
| Speed | Moderate | Fast | Very Fast |
| Learning Curve | Easy | Moderate | Steep |
| Use Cases | AI/ML, Web, Scripting | Enterprise, Android | Systems, Games |

## Variables and Data Types

Variables are containers for storing data values. Python is **dynamically typed** — the interpreter infers the type at runtime.

### Core Data Types:
- `int`: Integer numbers (e.g., `42`, `-7`, `0`)
- `float`: Floating-point numbers (e.g., `3.14`, `-0.5`, `2.0`)
- `str`: String of characters (e.g., `"hello"`, `'world'`)
- `bool`: Boolean values (`True`, `False`)
- `NoneType`: Represents absence of value (`None`)

```python
# Variables and data types demonstration
name = "Alice"        # str
age = 25              # int
height = 5.6          # float
is_student = True     # bool
middle_name = None    # NoneType

# type() function tells us the data type
print(f"{name} is of type {type(name)}")
print(f"{age} is of type {type(age)}")
print(f"{height} is of type {type(height)}")
print(f"{is_student} is of type {type(is_student)}")
print(f"{middle_name} is of type {type(middle_name)}")
```

## Basic Operators

### Arithmetic Operators
| Operator | Description | Example |
|----------|-------------|---------|
| `+` | Addition | `5 + 3 = 8` |
| `-` | Subtraction | `5 - 3 = 2` |
| `*` | Multiplication | `5 * 3 = 15` |
| `/` | Division (float) | `5 / 2 = 2.5` |
| `//` | Floor Division | `5 // 2 = 2` |
| `%` | Modulus (remainder) | `5 % 2 = 1` |
| `**` | Exponentiation | `5 ** 2 = 25` |

### Comparison Operators
`==`, `!=`, `>`, `<`, `>=`, `<=` — return `bool`.

### Logical Operators
`and`, `or`, `not` — combine boolean expressions.

### Assignment Operators
`=`, `+=`, `-=`, `*=`, `/=` etc.

```python
# Arithmetic operators
a, b = 10, 3
print(f"{a} + {b} = {a + b}")     # 13
print(f"{a} - {b} = {a - b}")     # 7
print(f"{a} * {b} = {a * b}")     # 30
print(f"{a} / {b} = {a / b}")     # 3.333...
print(f"{a} // {b} = {a // b}")   # 3 (floor division)
print(f"{a} % {b} = {a % b}")     # 1 (modulus)
print(f"{a} ** {b} = {a ** b}")   # 1000 (exponentiation)

# Comparison operators
print(f"{a} == {b}: {a == b}")    # False
print(f"{a} != {b}: {a != b}")    # True
print(f"{a} > {b}: {a > b}")      # True
print(f"{a} <= {b}: {a <= b}")    # False

# Logical operators
x, y = True, False
print(f"{x} and {y}: {x and y}")  # False
print(f"{x} or {y}: {x or y}")    # True
print(f"not {x}: {not x}")        # False

# Assignment operators
c = 5
c += 3  # same as: c = c + 3
print(f"c after += 3: {c}")       # 8
```

## Bitwise Operators

Bitwise operators work on bits of integer operands.

| Operator | Description | Example (`a=5, b=3`) |
|----------|-------------|----------------------|
| `&` | AND | `5 & 3 = 1` (101 & 011 = 001) |
| `\|` | OR | `5 \| 3 = 7` (101 | 011 = 111) |
| `^` | XOR | `5 ^ 3 = 6` (101 ^ 011 = 110) |
| `~` | NOT | `~5 = -6` (inverts all bits) |
| `<<` | Left shift | `5 << 1 = 10` (101 -> 1010) |
| `>>` | Right shift | `5 >> 1 = 2` (101 -> 10) |

```python
# Bitwise operators
a, b = 5, 3  # 5 = 0b101, 3 = 0b011
print(f"{a} & {b} = {a & b}")    # 1 (001)
print(f"{a} | {b} = {a | b}")    # 7 (111)
print(f"{a} ^ {b} = {a ^ b}")    # 6 (110)
print(f"~{a} = {~a}")            # -6
print(f"{a} << 1 = {a << 1}")    # 10 (1010)
print(f"{a} >> 1 = {a >> 1}")    # 2 (10)

# Practical use: checking if a number is odd/even
num = 7
if num & 1:
    print(f"{num} is odd")       # 7 & 1 = 1 (True)
else:
    print(f"{num} is even")
```

## Type Conversion (Type Casting)

Python provides functions to convert between types:
- `int()` — convert to integer
- `float()` — convert to float
- `str()` — convert to string
- `bool()` — convert to boolean

```python
# Type conversion examples
# Implicit conversion (coercion)
result = 5 + 2.5  # int + float -> float
print(f"5 + 2.5 = {result} (type: {type(result)})")

# Explicit conversion
pi_str = "3.14159"
pi_float = float(pi_str)  # string to float
print(f"Float from string: {pi_float}")

age_str = "25"
age_int = int(age_str)  # string to int
print(f"Int from string: {age_int}")

num = 42
num_str = str(num)  # int to string
print(f"String from int: '{num_str}'")

# bool() conversion rules
print(f"bool(0) = {bool(0)}")        # False
print(f"bool(1) = {bool(1)}")        # True
print(f"bool('') = {bool('')}")      # False (empty string)
print(f"bool('abc') = {bool('abc')}") # True
print(f"bool(None) = {bool(None)}")   # False
```

## Print Formatting

Three main ways to format strings:
1. **f-strings** (Python 3.6+) — `f"value is {var}"`
2. **`str.format()`** — `"value is {}".format(var)`
3. **`%` formatting** — `"value is %d" % var` (old style)

```python
# Print formatting techniques
name, age, salary = "Alice", 30, 75000.4567

# 1. f-strings (preferred)
print(f"Name: {name}, Age: {age}, Salary: ${salary:.2f}")

# 2. str.format()
print("Name: {}, Age: {}, Salary: ${:.2f}".format(name, age, salary))
print("Name: {n}, Age: {a}".format(n=name, a=age))

# 3. % formatting
print("Name: %s, Age: %d, Salary: $%.2f" % (name, age, salary))

# Format specifiers
print(f"Left aligned:  |{'hello':<10}|")
print(f"Right aligned: |{'hello':>10}|")
print(f"Centered:      |{'hello':^10}|")
print(f"Binary:        {42:b}")        # 101010
print(f"Hex:           {255:x}")        # ff
print(f"Percentage:    {0.125:.1%}")    # 12.5%
```

## Interview Problems

### Problem 1: Swap Two Variables
Swap the values of two variables **without using a temporary variable**.

```python
# Solution 1: Tuple unpacking (Pythonic way)
a, b = 5, 10
print(f"Before swap: a={a}, b={b}")
a, b = b, a  # Pythonic swap
print(f"After swap:  a={a}, b={b}")

# Solution 2: Using arithmetic (works for numbers)
x, y = 5, 10
x = x + y  # x = 15
y = x - y  # y = 5
x = x - y  # x = 10
print(f"Arithmetic swap: x={x}, y={y}")

# Solution 3: Using XOR (works for integers)
m, n = 5, 10
m ^= n
n ^= m
m ^= n
print(f"XOR swap: m={m}, n={n}")
```

### Problem 2: Check Even or Odd
Write a function that checks if a number is even or odd.

```python
def check_even_odd(num):
    """Return 'Even' if num is even, 'Odd' otherwise."""
    if num % 2 == 0:
        return "Even"
    else:
        return "Odd"

# Test cases
test_nums = [0, 1, 7, 42, -3, -8]
for n in test_nums:
    print(f"{n:3} -> {check_even_odd(n)}")

# Alternative: using bitwise AND (more efficient)
def is_odd(num):
    return num & 1 == 1

print(f"\nUsing bitwise: 7 is odd: {is_odd(7)}")
print(f"Using bitwise: 8 is odd: {is_odd(8)}")
```

### Problem 3: Basic Calculator
Write a simple calculator that takes two numbers and an operator and performs the operation.

```python
def calculator(a, b, operator):
    """Perform basic arithmetic operations."""
    if operator == '+':
        return a + b
    elif operator == '-':
        return a - b
    elif operator == '*':
        return a * b
    elif operator == '/':
        if b == 0:
            return "Error: Division by zero"
        return a / b
    elif operator == '//':
        if b == 0:
            return "Error: Division by zero"
        return a // b
    elif operator == '%':
        if b == 0:
            return "Error: Division by zero"
        return a % b
    elif operator == '**':
        return a ** b
    else:
        return "Error: Invalid operator"

# Test the calculator
tests = [(10, 3, '+'), (10, 3, '-'), (10, 3, '*'), (10, 3, '/'),
         (10, 3, '//'), (10, 3, '%'), (2, 8, '**'), (10, 0, '/')]

for a, b, op in tests:
    result = calculator(a, b, op)
    print(f"{a} {op} {b} = {result}")
```

### Problem 4: FizzBuzz (Classic Interview Question)
Print numbers 1 to n, but for multiples of 3 print "Fizz", for multiples of 5 print "Buzz", and for multiples of both print "FizzBuzz".

```python
def fizzbuzz(n):
    """Classic FizzBuzz implementation."""
    result = []
    for i in range(1, n + 1):
        if i % 3 == 0 and i % 5 == 0:
            result.append("FizzBuzz")
        elif i % 3 == 0:
            result.append("Fizz")
        elif i % 5 == 0:
            result.append("Buzz")
        else:
            result.append(str(i))
    return result

# Test FizzBuzz for 1-20
output = fizzbuzz(20)
for i, val in enumerate(output, 1):
    print(f"{i:2}: {val}")
```
