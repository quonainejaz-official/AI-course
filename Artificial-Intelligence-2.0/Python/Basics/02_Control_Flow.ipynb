# Control Flow in Python

Control flow determines the **order** in which code statements are executed. Python provides:
- **Conditional statements**: `if`, `elif`, `else`
- **Looping constructs**: `for`, `while`
- **Loop control**: `break`, `continue`, `pass`

> **Prerequisite**: Basic Python syntax, variables, and operators (covered in `01_Python_Basics.ipynb`).

## if/elif/else Statements

Syntax:
```python
if condition1:
    # code block
elif condition2:
    # code block
else:
    # code block
```

Python uses **indentation** (4 spaces by convention) to define code blocks. Unlike other languages that use braces `{}`.

````python
# if/elif/else basics
score = 85

if score >= 90:
    grade = 'A'
elif score >= 80:
    grade = 'B'
elif score >= 70:
    grade = 'C'
elif score >= 60:
    grade = 'D'
else:
    grade = 'F'

print(f"Score: {score}, Grade: {grade}")

# Ternary conditional (one-liner)
age = 20
status = "Adult" if age >= 18 else "Minor"
print(f"Age {age}: {status}")
````

## Nested Conditionals

Conditions can be nested inside other conditions. Use **sparingly** — deep nesting reduces readability.

````python
# Nested conditionals example
year = 2024

if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print(f"{year} is a leap year")
        else:
            print(f"{year} is not a leap year")
    else:
        print(f"{year} is a leap year")
else:
    print(f"{year} is not a leap year")

# Better: combine conditions with logical operators
if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(f"{year} is a leap year (simplified check)")
else:
    print(f"{year} is not a leap year")
````

## for Loops

The `for` loop iterates over any **iterable** (sequences like lists, strings, tuples, or range objects).

### range() Function
| Form | Description | Example |
|------|-------------|---------|
| `range(stop)` | 0 to stop-1 | `range(5)` → 0,1,2,3,4 |
| `range(start, stop)` | start to stop-1 | `range(2,7)` → 2,3,4,5,6 |
| `range(start, stop, step)` | start to stop-1 by step | `range(1,10,2)` → 1,3,5,7,9 |

````python
# for loop with range
print("Counting 0 to 4:")
for i in range(5):
    print(f"  {i}", end=" ")
print()  # newline

# Iterating over sequences
fruits = ["apple", "banana", "cherry"]
print("\nFruits:")
for fruit in fruits:
    print(f"  - {fruit}")

# enumerate() — get index and value
print("\nEnumerated fruits:")
for idx, fruit in enumerate(fruits):
    print(f"  {idx}: {fruit}")

# Iterating over string
print("\nCharacters in 'Python':")
for ch in "Python":
    print(f"  '{ch}'", end=" ")
print()

# Iterating over dictionary
person = {"name": "Alice", "age": 30, "city": "NYC"}
print("\nPerson details:")
for key, value in person.items():
    print(f"  {key}: {value}")
````

## while Loops

A `while` loop executes as long as a condition remains `True`. **Be careful of infinite loops!**

````python
# while loop — countdown
count = 5
print("Countdown:")
while count > 0:
    print(f"  {count}...")
    count -= 1
print("  Liftoff! 🚀")

# while loop with user input simulation
secret = "python"
guess = ""
attempts = 0
while guess != secret and attempts < 3:
    # In real code: guess = input("Enter password: ")
    guess = "python"  # simulated correct guess
    attempts += 1
    print(f"Attempt {attempts}: {'Correct!' if guess == secret else 'Wrong'}")
````

## break, continue, pass

| Statement | Effect |
|-----------|--------|
| `break` | Exits the loop immediately |
| `continue` | Skips the rest of current iteration, moves to next |
| `pass` | Does nothing (placeholder/no-op) |

### Loop else Clause
The `else` block after a loop runs **only if the loop completed normally** (i.e., no `break` was hit).

````python
# break example — find first even number
numbers = [3, 7, 9, 12, 15, 18]
for num in numbers:
    if num % 2 == 0:
        print(f"First even number found: {num}")
        break  # exit loop immediately

# continue example — print odd numbers only
print("\nOdd numbers:")
for i in range(10):
    if i % 2 == 0:
        continue  # skip even numbers
    print(f"  {i}", end=" ")
print()

# pass example — placeholder for unimplemented code
def future_function():
    pass  # TODO: implement later

# else clause with loop
print("\nSearching for 7 in [1, 2, 3]:")
for item in [1, 2, 3]:
    if item == 7:
        print("Found 7!")
        break
else:
    print("7 not found — loop completed without break")

# else clause with break
print("\nSearching for 2 in [1, 2, 3]:")
for item in [1, 2, 3]:
    if item == 2:
        print("Found 2!")
        break
else:
    print("2 not found")  # This will NOT execute because break was hit
````

## Nested Loops

Loops can be nested inside other loops. Common use cases: matrix operations, multiplication tables, pattern printing.

````python
# Multiplication table (nested for loops)
print("Multiplication Table (1-5):\n")
for i in range(1, 6):
    for j in range(1, 6):
        print(f"{i*j:4}", end=" ")
    print()  # newline after each row

# Pattern printing — right triangle
print("\nRight Triangle Pattern:")
n = 5
for i in range(1, n + 1):
    print('*' * i)
````

### Problem 1: FizzBuzz (Classic)
Print numbers from 1 to n. For multiples of 3 print "Fizz", multiples of 5 print "Buzz", multiples of both print "FizzBuzz".

````python
def fizzbuzz(n):
    """Print FizzBuzz from 1 to n."""
    result = []
    for i in range(1, n + 1):
        if i % 15 == 0:          # divisible by both 3 and 5
            result.append("FizzBuzz")
        elif i % 3 == 0:
            result.append("Fizz")
        elif i % 5 == 0:
            result.append("Buzz")
        else:
            result.append(str(i))
    return result

# Test
fb = fizzbuzz(30)
for i in range(0, len(fb), 5):
    print(' '.join(fb[i:i+5]))

# Time complexity: O(n) — single pass
# Space complexity: O(n) — storing n results
````

### Problem 2: Factorial Calculation
Calculate the factorial of a number `n` (n! = n × (n-1) × ... × 1). Implement both iterative and recursive approaches.

````python
def factorial_iterative(n):
    """Iterative factorial calculation."""
    if n < 0:
        raise ValueError("Factorial not defined for negative numbers")
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

def factorial_recursive(n):
    """Recursive factorial calculation."""
    if n < 0:
        raise ValueError("Factorial not defined for negative numbers")
    if n <= 1:  # base case: 0! = 1, 1! = 1
        return 1
    return n * factorial_recursive(n - 1)

# Test both implementations
test_values = [0, 1, 5, 10]
for n in test_values:
    iter_res = factorial_iterative(n)
    rec_res = factorial_recursive(n)
    print(f"{n:2}! = {iter_res:10} (iter) = {rec_res:10} (rec)")

# Time complexity: O(n) for both
# Space complexity: O(1) iterative, O(n) recursive (call stack)
````

### Problem 3: Prime Number Check
Determine whether a given number is prime. A prime number is greater than 1 and divisible only by 1 and itself.

````python
def is_prime(n):
    """Check if n is a prime number."""
    if n <= 1:
        return False
    if n <= 3:
        return True  # 2 and 3 are prime
    if n % 2 == 0 or n % 3 == 0:
        return False
    # Check only up to sqrt(n), stepping by 6 (6k ± 1 optimization)
    i = 5
    while i * i <= n:
        if n % i == 0 or n % (i + 2) == 0:
            return False
        i += 6
    return True

# Test
test_nums = [1, 2, 3, 4, 17, 19, 21, 97, 100]
for num in test_nums:
    print(f"{num:3} -> {'Prime' if is_prime(num) else 'Not prime'}")

# Time complexity: O(√n) — checks up to sqrt(n)
# Space complexity: O(1)
````

### Problem 4: Fibonacci Sequence
Generate the first `n` numbers in the Fibonacci sequence (0, 1, 1, 2, 3, 5, 8, 13, ...).

````python
def fibonacci_iterative(n):
    """Generate first n Fibonacci numbers iteratively."""
    if n <= 0:
        return []
    if n == 1:
        return [0]
    fib = [0, 1]
    for i in range(2, n):
        fib.append(fib[i-1] + fib[i-2])
    return fib

def fibonacci_nth(n):
    """Return the nth Fibonacci number (0-indexed)."""
    if n <= 0:
        return 0
    a, b = 0, 1
    for _ in range(n - 1):
        a, b = b, a + b  # tuple unpacking
    return b if n > 0 else 0

# Test
print("First 12 Fibonacci numbers:")
print(fibonacci_iterative(12))

print("\nIndividual nth Fibonacci numbers:")
for n in [0, 1, 5, 10, 20]:
    print(f"F({n}) = {fibonacci_nth(n)}")

# Time complexity: O(n) for both
# Space complexity: O(n) for sequence, O(1) for nth number
````

### Problem 5: Greatest Common Divisor (GCD)
Find the GCD of two numbers using **Euclid's algorithm**.

````python
def gcd_euclidean(a, b):
    """Find GCD using Euclid's algorithm (iterative)."""
    while b != 0:
        a, b = b, a % b
    return a

def gcd_recursive(a, b):
    """Find GCD using Euclid's algorithm (recursive)."""
    if b == 0:
        return a
    return gcd_recursive(b, a % b)

# Test
pairs = [(48, 18), (56, 98), (101, 103), (0, 5)]
for a, b in pairs:
    print(f"gcd({a}, {b}) = {gcd_euclidean(a, b)}")

# Time complexity: O(log(min(a, b))) — Euclid's algorithm
# Space complexity: O(1) iterative, O(log(min(a,b))) recursive
````
