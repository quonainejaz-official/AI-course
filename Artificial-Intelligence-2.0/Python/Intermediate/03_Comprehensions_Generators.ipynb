# Comprehensions and Generators

Python provides **concise syntax** for creating collections and lazy sequences:

| Feature | Syntax | Produces |
|---------|--------|----------|
| List comprehension | `[expr for x in iterable]` | `list` |
| Dict comprehension | `{k: v for x in iterable}` | `dict` |
| Set comprehension | `{expr for x in iterable}` | `set` |
| Generator expression | `(expr for x in iterable)` | `generator` |

> **Prerequisite**: Lists, dictionaries, sets, functions.

## List Comprehensions

A **list comprehension** provides a concise way to create lists.

Syntax: `[expression for item in iterable if condition]`

```python
# Basic list comprehension vs for loop
# Traditional way
squares_loop = []
for i in range(10):
    squares_loop.append(i ** 2)

# List comprehension
squares_lc = [i ** 2 for i in range(10)]

print(f"Loop:      {squares_loop}")
print(f"List comp: {squares_lc}")

# With condition
evens = [x for x in range(20) if x % 2 == 0]
print(f"Evens:     {evens}")

# With if-else expression
labels = ["even" if x % 2 == 0 else "odd" for x in range(10)]
print(f"Labels:    {labels}")
```

### Nested List Comprehensions

Useful for flattening nested structures or working with matrices.

```python
# Flatten a matrix (list of lists)
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flattened = [num for row in matrix for num in row]
# Order: outer loop -> inner loop
print(f"Matrix:    {matrix}")
print(f"Flattened: {flattened}")

# Transpose a matrix
transposed = [[row[i] for row in matrix] for i in range(len(matrix[0]))]
print(f"Transposed: {transposed}")

# Cartesian product
colors = ["red", "green"]
items = ["shirt", "hat"]
combos = [(c, i) for c in colors for i in items]
print(f"Combinations: {combos}")
```

## Dictionary Comprehensions

Create dictionaries with a concise syntax: `{key_expr: value_expr for item in iterable}`

```python
# Basic dict comprehension
squares = {x: x**2 for x in range(6)}
print(f"Squares: {squares}")

# From two lists
names = ["Alice", "Bob", "Charlie"]
ages = [30, 25, 35]
people = {name: age for name, age in zip(names, ages)}
print(f"People: {people}")

# Filtering
even_squares = {x: x**2 for x in range(10) if x % 2 == 0}
print(f"Even squares: {even_squares}")

# Transform keys/values
original = {"a": 1, "b": 2, "c": 3}
inverted = {v: k.upper() for k, v in original.items()}
print(f"Inverted: {inverted}")
```

## Set Comprehensions

Create sets: `{expression for item in iterable if condition}` — automatically removes duplicates.

```python
# Set comprehension — automatic deduplication
nums = [1, 2, 2, 3, 4, 3, 5, 1, 6]
unique_squares = {x**2 for x in nums}
print(f"Unique squares: {unique_squares}")

# Extract unique characters
sentence = "hello world"
unique_chars = {ch for ch in sentence if ch != ' '}
print(f"Unique chars: {sorted(unique_chars)}")

# Set comprehension with condition
vowels = {ch for ch in "python programming" if ch in "aeiou"}
print(f"Vowels found: {sorted(vowels)}")
```

## Generator Functions and `yield`

A **generator function** uses `yield` instead of `return`. It produces a sequence of values **lazily** — one at a time, as requested.

### Generator vs Iterator
- **Iterator**: Any object with `__iter__` and `__next__`
- **Generator**: An iterator created with `yield` (simpler to write)

```python
# Generator function
def count_up_to(n):
    """Generator that yields numbers from 1 to n."""
    count = 1
    while count <= n:
        yield count
        count += 1

# Using the generator
counter = count_up_to(5)
print(f"Type: {type(counter)}")
for num in counter:
    print(f"  {num}", end=" ")
print()

# Generators are single-use iterators
counter2 = count_up_to(3)
print(f"list: {list(counter2)}")
print(f"After exhaustion: {list(counter2)}")  # empty!
```

## Generator Expressions

Similar to list comprehensions but with **parentheses** — produces a generator lazily.

Syntax: `(expression for item in iterable if condition)`

```python
# Generator expression vs list comprehension
import sys

# List comprehension — stores all in memory
lc = [x ** 2 for x in range(10000)]
print(f"List comp size: {sys.getsizeof(lc)} bytes")

# Generator expression — lazy, only one item at a time
ge = (x ** 2 for x in range(10000))
print(f"Gen expr size:  {sys.getsizeof(ge)} bytes")

# Using generator expression
gen = (x * 2 for x in range(10))
print(f"First 5: {[next(gen) for _ in range(5)]}")
print(f"Remaining: {list(gen)}")

# Generator expression with condition (memory efficient filter)
evens_gen = (x for x in range(100) if x % 2 == 0)
print(f"First 10 even numbers: {[next(evens_gen) for _ in range(10)]}")

# Sum of squares (no intermediate list)
total = sum(x ** 2 for x in range(10))  # note: extra () not needed here
print(f"Sum of squares 0-9: {total}")
```

## yield from

`yield from` delegates part of the generator's operation to another iterable or generator.

```python
# yield from — delegating to sub-generators
def flatten(nested):
    """Flatten a nested list using yield from."""
    for item in nested:
        if isinstance(item, list):
            yield from flatten(item)  # delegate to recursive call
        else:
            yield item

# Test
nested = [1, [2, [3, 4], 5], 6]
print(f"Nested: {nested}")
print(f"Flattened: {list(flatten(nested))}")

# yield from with range
def chain(*iterables):
    """Chain multiple iterables together."""
    for it in iterables:
        yield from it

result = list(chain([1, 2, 3], "AB", range(4, 7)))
print(f"Chained: {result}")
```

## itertools Basics

The `itertools` module provides **iterator building blocks** that work with generator-style lazy evaluation.

```python
import itertools

# count() — infinite counter
counter = itertools.count(start=1, step=2)
first_5 = [next(counter) for _ in range(5)]
print(f"Count (1,3,5,...): {first_5}")

# cycle() — infinite cycle
cycler = itertools.cycle(["red", "green", "blue"])
first_7 = [next(cycler) for _ in range(7)]
print(f"Cycle: {first_7}")

# chain() — chain iterables
print(f"Chain: {list(itertools.chain([1,2], [3,4], "ab"))}")

# permutations and combinations
items = ['A', 'B', 'C']
print(f"Permutations (2): {list(itertools.permutations(items, 2))}")
print(f"Combinations (2): {list(itertools.combinations(items, 2))}")
print(f"Product: {list(itertools.product(items, repeat=2))}")

# takewhile / dropwhile
nums = [1, 3, 5, 7, 2, 4, 6]
taken = list(itertools.takewhile(lambda x: x < 5, nums))
print(f"Take while < 5: {taken}")
```

## Interview Problems

### Problem 1: Flatten a Nested List
Flatten an arbitrarily nested list of integers.

```python
def flatten_list(nested):
    """Flatten arbitrarily nested lists using recursion and yield from."""
    for item in nested:
        if isinstance(item, list):
            yield from flatten_list(item)
        else:
            yield item

# Test
nested_list = [1, [2, [3, [4, 5]], 6], 7, [8, 9]]
flat = list(flatten_list(nested_list))
print(f"Original: {nested_list}")
print(f"Flattened: {flat}")

# Time: O(n) where n = total number of elements
# Space: O(d) where d = max depth of nesting (call stack)
```

### Problem 2: Generate Fibonacci with Generators
Create a generator that yields Fibonacci numbers indefinitely.

```python
def fibonacci_generator():
    """Infinite generator of Fibonacci numbers."""
    a, b = 0, 1
    while True:
        yield a
        a, b = b, a + b

# Get first 15 Fibonacci numbers
fib = fibonacci_generator()
first_15 = [next(fib) for _ in range(15)]
print(f"First 15 Fibonacci: {first_15}")

# Memory efficient: we can generate billions without storing
# Skip to the 100th Fibonacci number
fib2 = fibonacci_generator()
for _ in range(100):
    hundredth = next(fib2)
print(f"100th Fibonacci: {hundredth}")

# Time: O(n) per n numbers yielded
# Space: O(1) — constant memory!
```

### Problem 3: Prime Number Generator
Generate prime numbers lazily using the Sieve of Eratosthenes.

```python
def prime_generator():
    """Infinite generator of prime numbers using trial division."""
    yield 2
    primes = [2]
    candidate = 3
    while True:
        is_prime = True
        for p in primes:
            if p * p > candidate:
                break
            if candidate % p == 0:
                is_prime = False
                break
        if is_prime:
            primes.append(candidate)
            yield candidate
        candidate += 2  # skip even numbers

# Get first 20 primes
primes = prime_generator()
first_20 = [next(primes) for _ in range(20)]
print(f"First 20 primes: {first_20}")

# Get the 1000th prime
primes2 = prime_generator()
for _ in range(1000):
    thousandth = next(primes2)
print(f"1000th prime: {thousandth}")

# Time: approximately O(n log log n) per n primes
# Space: O(k) where k = number of primes found
```

### Problem 4: Generator Pipeline (Data Processing)
Chain multiple generators for efficient data processing.

```python
# Generator pipeline — each stage is a generator
def read_numbers(n):
    """Stage 1: Generate numbers from 0 to n."""
    for i in range(n):
        yield i

def filter_even(nums):
    """Stage 2: Keep only even numbers."""
    for num in nums:
        if num % 2 == 0:
            yield num

def square(nums):
    """Stage 3: Square each number."""
    for num in nums:
        yield num ** 2

def take(nums, count):
    """Stage 4: Take only first 'count' items."""
    for i, num in enumerate(nums):
        if i >= count:
            break
        yield num

# Build the pipeline
pipeline = take(square(filter_even(read_numbers(100))), 10)

# Execute lazily
print(f"Pipeline result: {list(pipeline)}")

# Memory: O(1) per stage — processes one item at a time!
# This scales to billions of items that wouldn't fit in memory
```
