# Modules and Packages in Python

Modules and packages help organize Python code into **reusable, maintainable** units.

### Key Concepts
- **Module**: A single `.py` file
- **Package**: A directory with `__init__.py` (can contain sub-packages)
- **Standard Library**: Python's built-in modules ("batteries included")

> **Prerequisite**: Functions, file structure.

## The `import` Statement

Python provides several ways to import modules:

```python
# Different import styles

# Style 1: import the whole module
import math
print(f"math.sqrt(16) = {math.sqrt(16)}")
print(f"math.pi = {math.pi}")

# Style 2: import specific names
from math import sqrt, pi
print(f"sqrt(25) = {sqrt(25)}")
print(f"pi = {pi}")

# Style 3: import all names (AVOID in production)
# from math import *  # pollutes namespace

# Style 4: alias
import datetime as dt
print(f"Today: {dt.date.today()}")

# Style 5: import with alias for specific function
from collections import Counter as CounterClass
print(f"Counter: {CounterClass('mississippi')}")
```

## The `__name__ == "__main__"` Pattern

This pattern allows a script to be both **imported as a module** and **run as a standalone script**.

```python
%%writefile demo_module.py
"""A demo module that can be imported or run directly."""

def greet(name):
    """Return a greeting."""
    return f"Hello, {name}!"

def add(a, b):
    """Add two numbers."""
    return a + b

if __name__ == "__main__":
    # This block runs ONLY when script is executed directly
    print("Running demo_module.py directly...")
    print(greet("World"))
    print(f"5 + 3 = {add(5, 3)}")
```

```python
# Importing the module we just created
import demo_module

# When imported, __name__ is 'demo_module', so the if-block doesn't run
print(f"Using imported module:")
print(demo_module.greet("Alice"))
print(f"10 + 20 = {demo_module.add(10, 20)}")
```

## Creating a Package

A **package** is a directory containing modules and an `__init__.py` file.

```python
%%writefile mypackage/__init__.py
"""MyPackage - A demonstration package."""

VERSION = "1.0.0"

def get_version():
    return VERSION
```

```python
%%writefile mypackage/math_utils.py
"""Mathematical utility functions."""

def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n - 1)

def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

def fibonacci(n):
    """Return first n Fibonacci numbers."""
    result = [0, 1]
    for _ in range(2, n):
        result.append(result[-1] + result[-2])
    return result[:n]
```

```python
%%writefile mypackage/string_utils.py
"""String utility functions."""

def reverse(s):
    return s[::-1]

def count_vowels(s):
    return sum(1 for ch in s.lower() if ch in "aeiou")

def is_palindrome(s):
    cleaned = ''.join(ch.lower() for ch in s if ch.isalnum())
    return cleaned == cleaned[::-1]
```

```python
# Using the package
import mypackage
from mypackage import math_utils, string_utils

print(f"Package version: {mypackage.get_version()}")
print(f"Factorial(5): {math_utils.factorial(5)}")
print(f"Is 17 prime? {math_utils.is_prime(17)}")
print(f"Fibonacci(10): {math_utils.fibonacci(10)}")
print(f"Reverse('hello'): {string_utils.reverse('hello')}")
print(f"Vowels in 'Python': {string_utils.count_vowels('Python')}")
print(f"Is 'racecar' palindrome? {string_utils.is_palindrome('racecar')}")
```

```python
# collections — specialized data structures
from collections import Counter, defaultdict, namedtuple, deque, OrderedDict

print("--- collections ---")

# Counter
cnt = Counter("mississippi")
print(f"Counter: {dict(cnt)}")
print(f"Most common: {cnt.most_common(2)}")

# defaultdict — never get KeyError
dd = defaultdict(list)
dd['fruits'].append('apple')
dd['fruits'].append('banana')
print(f"defaultdict: {dict(dd)}")

# namedtuple
Point = namedtuple('Point', ['x', 'y'])
p = Point(3, 4)
print(f"Point: x={p.x}, y={p.y}, dist={p[0]}")

# deque — fast appends/pops from both ends
dq = deque([1, 2, 3])
dq.appendleft(0)
dq.append(4)
print(f"Deque: {list(dq)}")
```

### Problem 1: Age Calculator using datetime
Calculate a person's exact age in years, months, and days.

```python
from datetime import date

def calculate_age(birth_date: date) -> dict:
    """Calculate exact age from birth date."""
    today = date.today()
    
    years = today.year - birth_date.year
    months = today.month - birth_date.month
    days = today.day - birth_date.day
    
    # Adjust for negative days/months
    if days < 0:
        months -= 1
        # Get days in previous month
        prev_month = today.month - 1
        prev_month_date = date(today.year, prev_month, 1) if prev_month > 0 else date(today.year - 1, 12, 1)
        import calendar
        days_in_prev_month = calendar.monthrange(prev_month_date.year, prev_month_date.month)[1]
        days += days_in_prev_month
    
    if months < 0:
        years -= 1
        months += 12
    
    return {"years": years, "months": months, "days": days}

# Test
birth = date(1990, 6, 15)
age = calculate_age(birth)
print(f"Birth date: {birth}")
print(f"Age: {age['years']} years, {age['months']} months, {age['days']} days")
```

### Problem 2: Word Frequency Analyzer with Counter
Read a text and return the most common words.

```python
from collections import Counter
import re

def word_frequency(text: str, top_n: int = 5) -> list:
    """Get the top N most common words from text."""
    # Normalize: lowercase, split by non-alphanumeric
    words = re.findall(r'\b\w+\b', text.lower())
    counter = Counter(words)
    return counter.most_common(top_n)

# Test
sample_text = """
Python is amazing. Python is powerful. Python is easy to learn.
Many developers love Python because Python is versatile.
Python can be used for web development, data science, and AI.
"""

top_words = word_frequency(sample_text, 5)
print("Top 5 words:")
for word, count in top_words:
    print(f"  {word:10}: {count}")
```

### Problem 3: Build a Simple Calculator Module
Create and use a custom calculator module.

```python
%%writefile calculator.py
"""Simple calculator module."""

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        raise ValueError("Division by zero")
    return a / b

def power(a, b):
    return a ** b

def calculator():
    """Interactive calculator."""
    ops = {'+': add, '-': subtract, '*': multiply, '/': divide, '^': power}
    
    while True:
        expr = input("Enter (a op b) or 'q' to quit: ")
        if expr.lower() == 'q':
            break
        parts = expr.split()
        if len(parts) != 3:
            print("Format: a op b (e.g., 5 + 3)")
            continue
        a, op, b = parts
        if op not in ops:
            print(f"Unknown operator: {op}")
            continue
        try:
            result = ops[op](float(a), float(b))
            print(f"Result: {result}")
        except (ValueError, ZeroDivisionError) as e:
            print(f"Error: {e}")

if __name__ == "__main__":
    calculator()
```

```python
# Use the calculator module
import calculator

print("Using calculator module:")
print(f"15 + 7 = {calculator.add(15, 7)}")
print(f"15 - 7 = {calculator.subtract(15, 7)}")
print(f"15 * 7 = {calculator.multiply(15, 7)}")
print(f"15 / 7 = {calculator.divide(15, 7):.4f}")
print(f"2 ^ 10 = {calculator.power(2, 10)}")
```
