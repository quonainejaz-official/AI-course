# File Handling in Python

File handling allows you to **read from** and **write to** files on disk. Python's built-in `open()` function and `with` statement make file operations clean and safe.

### File Modes

| Mode | Description |
|------|-------------|
| `'r'` | Read (default) — file must exist |
| `'w'` | Write — creates file, overwrites if exists |
| `'a'` | Append — writes to end, creates if not exists |
| `'x'` | Exclusive creation — fails if file exists |
| `'rb'` | Read binary |
| `'wb'` | Write binary |
| `'r+'` | Read and write |

> **Prerequisite**: Functions, strings, basic Python.

## The `open()` Function and `with` Statement

Always use the **`with` statement** (context manager) — it automatically closes the file, even if an exception occurs.

```python
# Writing to a file
with open("sample.txt", "w") as f:
    f.write("Hello, World!\n")
    f.write("This is line 2.\n")
    f.write("Line 3: Python file handling.\n")

print("File 'sample.txt' created successfully.")

# Reading the entire file
with open("sample.txt", "r") as f:
    content = f.read()

print("\n--- File Content ---")
print(content)
```

## Reading Files: Different Methods

| Method | Returns | Use Case |
|--------|---------|----------|
| `read()` | Single string | Small files, entire content |
| `readline()` | Single string (one line) | Line-by-line processing |
| `readlines()` | List of strings | Small files, need all lines |
| Iteration | Lines one at a time | Large files (memory efficient) |

```python
# Different reading approaches
print("1. Using read():")
with open("sample.txt", "r") as f:
    content = f.read()
    print(repr(content[:50]) + "..." if len(content) > 50 else content)

print("\n2. Using readline():")
with open("sample.txt", "r") as f:
    line1 = f.readline()
    line2 = f.readline()
    print(f"Line 1: {line1!r}")
    print(f"Line 2: {line2!r}")

print("\n3. Using readlines():")
with open("sample.txt", "r") as f:
    lines = f.readlines()
    print(f"Number of lines: {len(lines)}")
    for i, line in enumerate(lines, 1):
        print(f"  Line {i}: {line.rstrip()}")

print("\n4. Iterating (most efficient for large files):")
with open("sample.txt", "r") as f:
    for i, line in enumerate(f, 1):
        print(f"  Line {i}: {line.rstrip()}")
```

## Appending to Files

Use `'a'` mode to add content without overwriting.

```python
# Appending to file
with open("sample.txt", "a") as f:
    f.write("This line was appended.\n")
    f.write("And another appended line.\n")

# Verify the append
with open("sample.txt", "r") as f:
    print(f.read())
```

## Working with Binary Files

Binary mode is used for images, audio, serialized objects, etc.

```python
# Writing binary data
data = bytes(range(256))  # 0-255
with open("binary_sample.bin", "wb") as f:
    f.write(data)

# Reading binary data
with open("binary_sample.bin", "rb") as f:
    read_data = f.read()

print(f"Written {len(data)} bytes, read {len(read_data)} bytes")
print(f"First 10 bytes: {list(read_data[:10])}")
print(f"Last 10 bytes: {list(read_data[-10:])}")
```

## Working with CSV Files

The **`csv` module** provides functionality for reading/writing CSV files.

```python
import csv

# Writing CSV
data = [
    ["Name", "Age", "City"],
    ["Alice", 30, "New York"],
    ["Bob", 25, "San Francisco"],
    ["Charlie", 35, "Chicago"],
]

with open("people.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(data)

print("Created people.csv")

# Reading CSV
print("\n--- CSV Content ---")
with open("people.csv", "r") as f:
    reader = csv.reader(f)
    for row in reader:
        print(f"  {row}")

# Reading CSV as dictionaries
print("\n--- CSV as Dicts ---")
with open("people.csv", "r") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(f"  {row['Name']:10} | {row['Age']:3} | {row['City']}")
```

## File and Path Operations with `os.path`

The `os` module provides functions for interacting with the operating system.

```python
import os

# File checks
filename = "sample.txt"
print(f"File exists: {os.path.exists(filename)}")
print(f"Is file: {os.path.isfile(filename)}")
print(f"File size: {os.path.getsize(filename)} bytes")
print(f"File name: {os.path.basename(filename)}")
print(f"Directory: {os.path.dirname(filename)}")
print(f"Absolute path: {os.path.abspath(filename)}")

# Directory listing
print("\nCurrent directory contents:")
for item in os.listdir('.'):
    print(f"  {'[DIR]' if os.path.isdir(item) else '[FILE]'} {item}")

# Join paths safely
data_dir = "data"
os.makedirs(data_dir, exist_ok=True)
new_path = os.path.join(data_dir, "output.txt")
print(f"\nJoined path: {new_path}")
```

## Pathlib (Modern Path Handling)

`pathlib` provides an **object-oriented** approach to filesystem paths (Python 3.4+).

```python
from pathlib import Path

# Create Path objects
current = Path(".")
file_path = Path("sample.txt")

print(f"Current dir: {current.absolute()}")
print(f"File exists: {file_path.exists()}")
print(f"File name: {file_path.stem}")       # 'sample' (no extension)
print(f"Extension: {file_path.suffix}")      # '.txt'
print(f"Parent dir: {file_path.parent}")

# Reading/writing with pathlib
path = Path("pathlib_sample.txt")
path.write_text("Hello from pathlib!\nSecond line.")
content = path.read_text()
print(f"\nRead via pathlib:\n{content}")

# Glob patterns
print("\nAll .txt files:")
for txt_file in Path(".").glob("*.txt"):
    print(f"  {txt_file.name}")
```

### Problem 1: Read and Summarize a Text File
Read a file and return: line count, word count, character count.

```python
def summarize_file(filename: str) -> dict:
    """Return a summary of a text file."""
    with open(filename, "r") as f:
        content = f.read()
    
    lines = content.splitlines()
    words = content.split()
    
    return {
        "filename": filename,
        "line_count": len(lines),
        "word_count": len(words),
        "char_count": len(content),
        "avg_word_length": round(len("".join(words)) / len(words), 2) if words else 0
    }

# Write a test file
with open("summary_test.txt", "w") as f:
    f.write("Python is an amazing programming language.\n")
    f.write("It is widely used in AI and data science.\n")
    f.write("Learning Python opens many career opportunities.\n")

# Summarize it
summary = summarize_file("summary_test.txt")
for key, value in summary.items():
    print(f"{key:20}: {value}")
```

### Problem 2: CSV Parser — Average Age
Parse the people.csv file and compute the average age.

```python
import csv

def average_age_from_csv(filename: str) -> float:
    """Read CSV and compute average age from 'Age' column."""
    ages = []
    with open(filename, "r") as f:
        reader = csv.DictReader(f)
        for row in reader:
            ages.append(int(row["Age"]))
    return sum(ages) / len(ages) if ages else 0.0

avg = average_age_from_csv("people.csv")
print(f"Average age: {avg:.1f} years")

# Time: O(n), Space: O(n)
```

### Problem 3: Log File Analyzer
Parse a log file and count occurrences of each log level (INFO, WARNING, ERROR).

```python
import re
from collections import Counter

def analyze_log(filename: str) -> dict:
    """Analyze a log file and count log levels."""
    log_levels = Counter()
    pattern = r'\b(INFO|WARNING|ERROR|DEBUG|CRITICAL)\b'
    
    with open(filename, "r") as f:
        for line in f:
            matches = re.findall(pattern, line)
            log_levels.update(matches)
    
    return dict(log_levels)

# Create a sample log file
log_content = """
2024-01-15 10:00:00 INFO Starting application
2024-01-15 10:00:01 INFO Loading configuration
2024-01-15 10:00:02 WARNING Configuration file not found, using defaults
2024-01-15 10:00:03 INFO Database connected
2024-01-15 10:00:04 ERROR Failed to query users table
2024-01-15 10:00:05 INFO Retrying connection
2024-01-15 10:00:06 ERROR Timeout exceeded
2024-01-15 10:00:07 CRITICAL Application shutting down
"""

with open("sample.log", "w") as f:
    f.write(log_content)

# Analyze
analysis = analyze_log("sample.log")
print("Log Level Analysis:")
for level, count in sorted(analysis.items(), key=lambda x: -x[1]):
    print(f"  {level:10}: {count}")

# Time: O(n), Space: O(k) where k = number of unique log levels
```
