# Greedy Algorithms

Greedy algorithms make the locally optimal choice at each step, hoping to reach the global optimum.

## Topics Covered

- Optimal Substructure

## 1. Greedy Choice Property & Optimal Substructure

A problem is greedy-solvable if:
- **Greedy Choice Property**: A global optimum can be reached by making locally optimal choices
- **Optimal Substructure**: Optimal solution contains optimal solutions to subproblems

**When to use greedy:**
- Activity selection, interval scheduling
- Minimum spanning tree (Prim's, Kruskal's)
- Shortest path (Dijkstra's)
- Huffman coding

## 2. Activity Selection (Interval Scheduling)

Select the maximum number of non-overlapping intervals. Greedy: pick the one that finishes earliest.

```python
def activity_selection(intervals):
    """
    Select maximum non-overlapping intervals.
    intervals: list of (start, end)
    """
    intervals.sort(key=lambda x: x[1])  # Sort by end time
    count = 0
    last_end = float('-inf')
    selected = []
    for start, end in intervals:
        if start >= last_end:  # Non-overlapping
            count += 1
            selected.append((start, end))
            last_end = end
    return count, selected

intervals = [(1, 4), (3, 5), (0, 6), (5, 7), (3, 8), (5, 9), (6, 10), (8, 11)]
count, selected = activity_selection(intervals)
print(f"Max activities: {count}")
print(f"Selected: {selected}")
# Time: O(n log n), Space: O(n)
```

## 3. Minimum Number of Platforms

Given arrival and departure times of trains, find minimum platforms needed. Greedy: sort arrivals and departures, use two pointers.

```python
def min_platforms(arrivals, departures):
    """
    Minimum platforms needed for trains.
    arrivals: list of arrival times
    departures: list of departure times
    """
    arrivals.sort()
    departures.sort()
    n = len(arrivals)
    platforms = 0
    max_platforms = 0
    i = j = 0
    while i < n and j < n:
        if arrivals[i] <= departures[j]:
            platforms += 1  # Train arrives
            i += 1
            max_platforms = max(max_platforms, platforms)
        else:
            platforms -= 1  # Train departs
            j += 1
    return max_platforms

arr = [900, 940, 950, 1100, 1500, 1800]
dep = [910, 1200, 1120, 1130, 1900, 2000]
print(f"Min platforms: {min_platforms(arr, dep)}")  # 3
# Time: O(n log n), Space: O(1)
```

## 4. Fractional Knapsack

Unlike 0/1 knapsack, we can take fractions of items. Greedy: take items with highest value/weight ratio first.

```python
def fractional_knapsack(weights, values, capacity):
    """Max value by taking fractions of items."""
    # Items as (value, weight, ratio)
    items = [(v, w, v/w) for v, w in zip(values, weights)]
    items.sort(key=lambda x: x[2], reverse=True)  # Sort by ratio
    total_value = 0
    for v, w, _ in items:
        if capacity >= w:
            total_value += v
            capacity -= w
        else:
            total_value += v * (capacity / w)
            break
    return total_value

weights = [10, 20, 30]
values = [60, 100, 120]
capacity = 50
print(f"Fractional knapsack: {fractional_knapsack(weights, values, capacity)}")  # 240.0
# Time: O(n log n), Space: O(n)
```

## 5. Gas Station

Given gas available at each station and cost to reach next station, find starting station to complete circuit.

```python
def can_complete_circuit(gas, cost):
    """Return starting station index if circuit is possible."""
    total_gas = total_cost = 0
    start = 0
    tank = 0
    for i in range(len(gas)):
        total_gas += gas[i]
        total_cost += cost[i]
        tank += gas[i] - cost[i]
        if tank < 0:
            start = i + 1
            tank = 0
    return start if total_gas >= total_cost else -1

print(f"Gas station: {can_complete_circuit([1,2,3,4,5], [3,4,5,1,2])}")  # 3
print(f"No circuit: {can_complete_circuit([2,3,4], [3,4,3])}")  # -1
# Time: O(n), Space: O(1)
```

## 6. Jump Game

Can you reach the last index? Each element is max jump length from that position.

```python
def can_jump(nums):
    """Check if you can reach the last index."""
    max_reach = 0
    for i, jump in enumerate(nums):
        if i > max_reach:
            return False
        max_reach = max(max_reach, i + jump)
        if max_reach >= len(nums) - 1:
            return True
    return True

def jump_min(nums):
    """Minimum jumps to reach last index."""
    jumps = 0
    curr_end = 0
    farthest = 0
    for i in range(len(nums) - 1):
        farthest = max(farthest, i + nums[i])
        if i == curr_end:
            jumps += 1
            curr_end = farthest
    return jumps

print(f"Can jump [2,3,1,1,4]: {can_jump([2,3,1,1,4])}")  # True
print(f"Can jump [3,2,1,0,4]: {can_jump([3,2,1,0,4])}")  # False
print(f"Min jumps: {jump_min([2,3,1,1,4])}")  # 2
# Time: O(n), Space: O(1)
```

## 7. Candy Distribution

Distribute candies to children such that:
- Each child gets at least 1 candy
- Children with higher rating get more than neighbors

```python
def candy(ratings):
    """Minimum candies to distribute."""
    n = len(ratings)
    candies = [1] * n
    # Left to right
    for i in range(1, n):
        if ratings[i] > ratings[i-1]:
            candies[i] = candies[i-1] + 1
    # Right to left
    for i in range(n-2, -1, -1):
        if ratings[i] > ratings[i+1]:
            candies[i] = max(candies[i], candies[i+1] + 1)
    return sum(candies)

print(f"Candy [1,0,2]: {candy([1,0,2])}")  # 5
print(f"Candy [1,2,2]: {candy([1,2,2])}")  # 4
# Time: O(n), Space: O(n)
```

## 8. Non-overlapping Intervals

Find minimum number of intervals to remove to make the rest non-overlapping.

```python
def erase_overlap_intervals(intervals):
    """Minimum intervals to remove for non-overlapping set."""
    if not intervals:
        return 0
    intervals.sort(key=lambda x: x[1])  # Sort by end
    count = 0
    last_end = intervals[0][1]
    for i in range(1, len(intervals)):
        if intervals[i][0] < last_end:
            count += 1  # Overlap, remove this
        else:
            last_end = intervals[i][1]
    return count

print(f"Remove count: {erase_overlap_intervals([[1,2],[2,3],[3,4],[1,3]])}")  # 1
print(f"Remove count: {erase_overlap_intervals([[1,2],[1,2],[1,2]])}")  # 2
# Time: O(n log n), Space: O(1)
```

## 9. Task Scheduler

CPU tasks with cooldown n between same tasks. Find minimum time to complete all tasks.

```python
from collections import Counter

def task_scheduler(tasks, n):
    """Minimum time to complete tasks with cooldown."""
    freq = Counter(tasks)
    max_freq = max(freq.values())
    max_count = sum(1 for v in freq.values() if v == max_freq)
    # Formula: (max_freq - 1) * (n + 1) + max_count
    time = (max_freq - 1) * (n + 1) + max_count
    return max(time, len(tasks))

print(f"Task scheduler: {task_scheduler(['A','A','A','B','B','B'], 2)}")  # 8
print(f"Task scheduler (more tasks): {task_scheduler(['A','A','A','B','B','B','C','C','C'], 2)}")  # 9
# Time: O(n), Space: O(1)
```

## Summary

| Problem | Greedy Strategy | Time | Key Insight |
|---|---|---|---|
| Activity Selection | Earliest finish first | O(n log n) | Maximize count |
| Min Platforms | Sort + two pointers | O(n log n) | Arrival/departure pointers |
| Fractional Knapsack | Highest value/weight first | O(n log n) | Can take fractions |
| Gas Station | Cumulative tank, reset when negative | O(n) | Total must be >= 0 |
| Jump Game | Max reachable index | O(n) | Track farthest |
| Candy | Two passes | O(n) | Left then right |
| Task Scheduler | Fill idle slots with most frequent | O(n) | (max-1)*(n+1)+max_count |
