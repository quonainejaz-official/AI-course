# Dynamic Programming

Dynamic Programming (DP) solves problems by breaking them into overlapping subproblems and storing results to avoid recomputation.

## Topics Covered
- DP Fundamentals (Optimal Substructure, Overlapping Subproblems)
- Top-Down (Memoization) vs Bottom-Up (Tabulation)
- 1D DP (Fibonacci, Climbing Stairs, House Robber)
- 2D DP (Unique Paths, Longest Common Subsequence)
- Knapsack DP (0/1 Knapsack, Unbounded Knapsack)
- DP on Trees (Diameter, Max Path Sum)
- DP Patterns (Coin Change, Edit Distance, LIS)

## 1. DP Fundamentals

Two key properties:
- **Optimal Substructure**: Optimal solution depends on optimal solutions to subproblems
- **Overlapping Subproblems**: Same subproblems are solved repeatedly

Two approaches:
- **Top-Down (Memoization)**: Recursive + caching
- **Bottom-Up (Tabulation)**: Iterative, fill table

```python
# Fibonacci: Top-Down (Memoization)
def fib_memo(n, memo={}):
    if n in memo:
        return memo[n]
    if n <= 1:
        return n
    memo[n] = fib_memo(n-1, memo) + fib_memo(n-2, memo)
    return memo[n]

# Fibonacci: Bottom-Up (Tabulation)
def fib_tab(n):
    if n <= 1:
        return n
    dp = [0] * (n + 1)
    dp[1] = 1
    for i in range(2, n + 1):
        dp[i] = dp[i-1] + dp[i-2]
    return dp[n]

# Fibonacci: Space-optimized
def fib_opt(n):
    if n <= 1:
        return n
    a, b = 0, 1
    for _ in range(2, n + 1):
        a, b = b, a + b
    return b

n = 40
print(f"Fib({n}) memo: {fib_memo(n)}")
print(f"Fib({n}) tab: {fib_tab(n)}")
print(f"Fib({n}) opt: {fib_opt(n)}")
# Time: O(n), Space: O(n) memo/tab, O(1) optimized
```

## 2. 1D DP Pattern

Problems where dp[i] depends on dp[i-1] and/or dp[i-2].

```python
# Climbing Stairs: dp[i] = dp[i-1] + dp[i-2]
def climb_stairs(n):
    if n <= 2:
        return n
    a, b = 1, 2
    for _ in range(3, n + 1):
        a, b = b, a + b
    return b

# House Robber: max(dp[i-1], dp[i-2] + nums[i])
def rob(nums):
    if not nums:
        return 0
    if len(nums) <= 2:
        return max(nums)
    dp = [0] * len(nums)
    dp[0] = nums[0]
    dp[1] = max(nums[0], nums[1])
    for i in range(2, len(nums)):
        dp[i] = max(dp[i-1], dp[i-2] + nums[i])
    return dp[-1]

# House Robber: space optimized
def rob_opt(nums):
    prev2 = prev1 = 0
    for num in nums:
        prev2, prev1 = prev1, max(prev1, prev2 + num)
    return prev1

print(f"Climb 5 stairs: {climb_stairs(5)}")  # 8
print(f"Rob [2,7,9,3,1]: {rob([2,7,9,3,1])}")  # 12
print(f"Rob [2,1,1,2]: {rob_opt([2,1,1,2])}")  # 4
```

## 3. 2D DP Pattern

Matrix-based DP where dp[i][j] depends on neighbors.

```python
# Unique Paths: dp[i][j] = dp[i-1][j] + dp[i][j-1]
def unique_paths(m, n):
    """Number of ways to reach bottom-right from top-left."""
    dp = [[1] * n for _ in range(m)]
    for i in range(1, m):
        for j in range(1, n):
            dp[i][j] = dp[i-1][j] + dp[i][j-1]
    return dp[m-1][n-1]

# Unique Paths with obstacles
def unique_paths_with_obstacles(grid):
    if not grid or grid[0][0] == 1:
        return 0
    m, n = len(grid), len(grid[0])
    dp = [[0] * n for _ in range(m)]
    dp[0][0] = 1
    for i in range(m):
        for j in range(n):
            if grid[i][j] == 1:
                dp[i][j] = 0
            else:
                if i > 0:
                    dp[i][j] += dp[i-1][j]
                if j > 0:
                    dp[i][j] += dp[i][j-1]
    return dp[m-1][n-1]

print(f"Unique paths 3x7: {unique_paths(3, 7)}")  # 28
print(f"Unique paths with obstacles: {unique_paths_with_obstacles([[0,0,0],[0,1,0],[0,0,0]])}")  # 2
# Time: O(m*n), Space: O(m*n)
```

## 4. Longest Common Subsequence (LCS)

Classic 2D DP. dp[i][j] = LCS of text1[:i] and text2[:j].

```python
def longest_common_subsequence(text1, text2):
    m, n = len(text1), len(text2)
    dp = [[0] * (n + 1) for _ in range(m + 1)]
    for i in range(1, m + 1):
        for j in range(1, n + 1):
            if text1[i-1] == text2[j-1]:
                dp[i][j] = dp[i-1][j-1] + 1
            else:
                dp[i][j] = max(dp[i-1][j], dp[i][j-1])
    return dp[m][n]

def lcs_with_string(text1, text2):
    """Return LCS string itself."""
    m, n = len(text1), len(text2)
    dp = [[0] * (n + 1) for _ in range(m + 1)]
    for i in range(1, m + 1):
        for j in range(1, n + 1):
            if text1[i-1] == text2[j-1]:
                dp[i][j] = dp[i-1][j-1] + 1
            else:
                dp[i][j] = max(dp[i-1][j], dp[i][j-1])
    # Backtrack
    i, j = m, n
    result = []
    while i > 0 and j > 0:
        if text1[i-1] == text2[j-1]:
            result.append(text1[i-1])
            i -= 1; j -= 1
        elif dp[i-1][j] >= dp[i][j-1]:
            i -= 1
        else:
            j -= 1
    return ''.join(reversed(result))

print(f"LCS of 'abcde' and 'ace': {longest_common_subsequence('abcde', 'ace')}")  # 3
print(f"LCS string: {lcs_with_string('abcde', 'ace')}")  # 'ace'
# Time: O(m*n), Space: O(m*n)
```

## 5. 0/1 Knapsack

Given weights and values, maximize value with capacity constraint. Each item used at most once.

```python
def knapSack_01(weights, values, capacity):
    """0/1 Knapsack - max value with capacity constraint."""
    n = len(weights)
    dp = [[0] * (capacity + 1) for _ in range(n + 1)]
    for i in range(1, n + 1):
        w, v = weights[i-1], values[i-1]
        for c in range(capacity + 1):
            if w > c:
                dp[i][c] = dp[i-1][c]
            else:
                dp[i][c] = max(dp[i-1][c], dp[i-1][c-w] + v)
    return dp[n][capacity]

def knapSack_01_optimized(weights, values, capacity):
    """Space-optimized 0/1 Knapsack."""
    dp = [0] * (capacity + 1)
    for w, v in zip(weights, values):
        for c in range(capacity, w - 1, -1):  # Reverse!
            dp[c] = max(dp[c], dp[c - w] + v)
    return dp[capacity]

# Unbounded Knapsack (items can be reused)
def unbounded_knapsack(weights, values, capacity):
    dp = [0] * (capacity + 1)
    for c in range(1, capacity + 1):
        for w, v in zip(weights, values):
            if w <= c:
                dp[c] = max(dp[c], dp[c - w] + v)
    return dp[capacity]

weights = [2, 3, 4, 5]
values = [3, 4, 5, 6]
capacity = 5
print(f"0/1 Knapsack: {knapSack_01(weights, values, capacity)}")  # 7
print(f"0/1 Knapsack (opt): {knapSack_01_optimized(weights, values, capacity)}")  # 7
print(f"Unbounded: {unbounded_knapsack(weights, values, capacity)}")  # 8?
# Time: O(n*cap), Space: O(cap) optimized
```

## 6. Coin Change (Unbounded Knapsack)

Minimum number of coins to make a given amount (unlimited supply).

```python
def coin_change(coins, amount):
    """Minimum coins to make amount."""
    dp = [float('inf')] * (amount + 1)
    dp[0] = 0
    for coin in coins:
        for i in range(coin, amount + 1):
            dp[i] = min(dp[i], dp[i - coin] + 1)
    return dp[amount] if dp[amount] != float('inf') else -1

def coin_change_ways(coins, amount):
    """Number of ways to make amount with coins."""
    dp = [0] * (amount + 1)
    dp[0] = 1
    for coin in coins:
        for i in range(coin, amount + 1):
            dp[i] += dp[i - coin]
    return dp[amount]

print(f"Coin change [1,2,5] -> 11: {coin_change([1,2,5], 11)}")  # 3
print(f"Coin change [2] -> 3: {coin_change([2], 3)}")  # -1
print(f"Ways to make 5 with [1,2,5]: {coin_change_ways([1,2,5], 5)}")  # 4
# Time: O(n*amount), Space: O(amount)
```

## 7. Longest Increasing Subsequence

LIS: dp[i] = longest increasing subsequence ending at index i.

```python
def length_of_LIS(nums):
    """Longest Increasing Subsequence - O(n²)."""
    if not nums:
        return 0
    n = len(nums)
    dp = [1] * n
    for i in range(n):
        for j in range(i):
            if nums[j] < nums[i]:
                dp[i] = max(dp[i], dp[j] + 1)
    return max(dp)

def length_of_LIS_optimized(nums):
    """LIS with patience sorting - O(n log n)."""
    import bisect
    piles = []
    for num in nums:
        i = bisect.bisect_left(piles, num)
        if i == len(piles):
            piles.append(num)
        else:
            piles[i] = num
    return len(piles)

print(f"LIS [10,9,2,5,3,7,101,18]: {length_of_LIS([10,9,2,5,3,7,101,18])}")  # 4
print(f"LIS optimized: {length_of_LIS_optimized([10,9,2,5,3,7,101,18])}")  # 4
# Time: O(n²) for DP, O(n log n) for optimized
```

## 8. Edit Distance (Levenshtein Distance)

Minimum number of operations (insert, delete, replace) to convert word1 to word2.

```python
def min_distance(word1, word2):
    """Edit distance between two strings."""
    m, n = len(word1), len(word2)
    dp = [[0] * (n + 1) for _ in range(m + 1)]
    for i in range(m + 1):
        dp[i][0] = i
    for j in range(n + 1):
        dp[0][j] = j
    for i in range(1, m + 1):
        for j in range(1, n + 1):
            if word1[i-1] == word2[j-1]:
                dp[i][j] = dp[i-1][j-1]
            else:
                dp[i][j] = 1 + min(
                    dp[i-1][j],    # delete
                    dp[i][j-1],    # insert
                    dp[i-1][j-1]   # replace
                )
    return dp[m][n]

print(f"Edit distance 'horse' -> 'ros': {min_distance('horse', 'ros')}")  # 3
print(f"Edit distance 'intention' -> 'execution': {min_distance('intention', 'execution')}")  # 5
# Time: O(m*n), Space: O(m*n)
```

## 9. Partition Equal Subset Sum

Can the array be partitioned into two subsets with equal sum? (0/1 Knapsack variation)

```python
def can_partition(nums):
    """Check if array can be partitioned into two equal-sum subsets."""
    total = sum(nums)
    if total % 2 != 0:
        return False
    target = total // 2
    dp = [False] * (target + 1)
    dp[0] = True
    for num in nums:
        for i in range(target, num - 1, -1):
            dp[i] = dp[i] or dp[i - num]
    return dp[target]

print(f"Can partition [1,5,11,5]: {can_partition([1,5,11,5])}")  # True
print(f"Can partition [1,2,3,5]: {can_partition([1,2,3,5])}")  # False
# Time: O(n * target), Space: O(target)
```

## 10. DP on Trees

Tree DP: compute values bottom-up using postorder traversal.

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def diameter_of_tree(root):
    """Diameter = longest path between any two nodes."""
    diameter = 0
    def height(node):
        nonlocal diameter
        if not node:
            return 0
        left = height(node.left)
        right = height(node.right)
        diameter = max(diameter, left + right)
        return 1 + max(left, right)
    height(root)
    return diameter

def max_path_sum(root):
    """Maximum path sum (any node to any node)."""
    max_sum = float('-inf')
    def gain(node):
        nonlocal max_sum
        if not node:
            return 0
        left = max(gain(node.left), 0)  # Only positive contribution
        right = max(gain(node.right), 0)
        max_sum = max(max_sum, node.val + left + right)
        return node.val + max(left, right)
    gain(root)
    return max_sum

# Test
root = TreeNode(-10,
    TreeNode(9),
    TreeNode(20, TreeNode(15), TreeNode(7))
)
print(f"Max path sum: {max_path_sum(root)}")  # 42
# Time: O(n), Space: O(h) where h = height
```

## LeetCode Problems

### LeetCode 70 - Climbing Stairs

**Problem:** Number of ways to climb n stairs taking 1 or 2 steps.

```python
def climbStairs(n):
    if n <= 2:
        return n
    a, b = 1, 2
    for _ in range(3, n + 1):
        a, b = b, a + b
    return b
# Time: O(n), Space: O(1)
```

### LeetCode 198 - House Robber

**Problem:** Max money without robbing adjacent houses.

```python
def rob(nums):
    prev2 = prev1 = 0
    for num in nums:
        prev2, prev1 = prev1, max(prev1, prev2 + num)
    return prev1
# Time: O(n), Space: O(1)
```

### LeetCode 322 - Coin Change

**Problem:** Minimum coins to make amount (unlimited supply).

```python
def coinChange(coins, amount):
    dp = [amount + 1] * (amount + 1)
    dp[0] = 0
    for coin in coins:
        for i in range(coin, amount + 1):
            dp[i] = min(dp[i], dp[i - coin] + 1)
    return dp[amount] if dp[amount] != amount + 1 else -1
# Time: O(n * amount), Space: O(amount)
```

### LeetCode 1143 - Longest Common Subsequence

**Problem:** Longest common subsequence of two strings.

```python
def longestCommonSubsequence(text1, text2):
    m, n = len(text1), len(text2)
    dp = [[0] * (n + 1) for _ in range(m + 1)]
    for i in range(1, m + 1):
        for j in range(1, n + 1):
            if text1[i-1] == text2[j-1]:
                dp[i][j] = dp[i-1][j-1] + 1
            else:
                dp[i][j] = max(dp[i-1][j], dp[i][j-1])
    return dp[m][n]
# Time: O(m*n), Space: O(m*n)
```

## Summary

| Pattern | Recurrence | Time | Space |
|---|---|---|---|
| 1D DP | dp[i] = f(dp[i-1], dp[i-2]) | O(n) | O(1) or O(n) |
| 2D DP | dp[i][j] = f(dp[i-1][j], dp[i][j-1]) | O(mn) | O(mn) |
| 0/1 Knapsack | dp[c] = max(dp[c], dp[c-w] + v) | O(n*C) | O(C) |
| Unbounded | dp[c] = max(dp[c], dp[c-w] + v) | O(n*C) | O(C) |
| LCS | dp[i][j] = match + max(left, up) | O(mn) | O(mn) |
| LIS | dp[i] = max(dp[j] + 1) | O(n²) | O(n) |
| Edit Distance | min(insert, delete, replace) + 1 | O(mn) | O(mn) |
| Tree DP | Postorder + global max | O(n) | O(h) |
