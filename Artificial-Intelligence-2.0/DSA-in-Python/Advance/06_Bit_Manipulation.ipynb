## 1. Bitwise Operators

| Operator | Name | Description |
|---|---|---|
| `&` | AND | 1 if both bits are 1 |
| `|` | OR | 1 if at least one bit is 1 |
| `^` | XOR | 1 if bits are different |
| `~` | NOT | Invert bits |
| `<<` | Left Shift | Shift left, fill with 0 (multiply by 2) |
| `>>` | Right Shift | Shift right (divide by 2) |

```python
a = 5  # 101
b = 3  # 011

print(f"a = {a:04b}, b = {b:04b}")
print(f"a & b = {a & b:04b} = {a & b}")   # 001 = 1
print(f"a | b = {a | b:04b} = {a | b}")   # 111 = 7
print(f"a ^ b = {a ^ b:04b} = {a ^ b}")   # 110 = 6
print(f"~a = {~a & 0b1111:04b} = {~a}")
print(f"a << 1 = {a << 1:04b} = {a << 1}")  # 1010 = 10
print(f"a >> 1 = {a >> 1:04b} = {a >> 1}")  # 010 = 2
```

## 2. Binary Representation & Useful Operations

Python's `bin()` and `format()` for binary representation.

```python
num = 42
print(f"Binary of {num}: {bin(num)}")        # 0b101010
print(f"Binary (8-bit): {num:08b}")          # 00101010

# Check if i-th bit is set
def get_bit(num, i):
    return (num >> i) & 1

# Set i-th bit to 1
def set_bit(num, i):
    return num | (1 << i)

# Clear i-th bit (set to 0)
def clear_bit(num, i):
    return num & ~(1 << i)

# Toggle i-th bit
def toggle_bit(num, i):
    return num ^ (1 << i)

n = 10  # 1010
print(f"n = {n:04b}")
print(f"Bit at position 1: {get_bit(n, 1)}")  # 1
print(f"Set bit 0: {set_bit(n, 0):04b}")      # 1011
print(f"Clear bit 1: {clear_bit(n, 1):04b}")  # 1000
print(f"Toggle bit 3: {toggle_bit(n, 3):04b}")  # 0010
```

## 3. Counting Set Bits

**Brian Kernighan's Algorithm**: repeatedly clear the lowest set bit. Runs in O(number of set bits).

```python
def count_set_bits(n):
    """Count number of 1 bits using Brian Kernighan's algorithm."""
    count = 0
    while n:
        n = n & (n - 1)  # Clear the lowest set bit
        count += 1
    return count

def count_set_bits_naive(n):
    """Naive O(number of bits) approach."""
    count = 0
    while n:
        count += n & 1
        n >>= 1
    return count

n = 29  # 11101
print(f"Binary: {n:08b}")
print(f"Set bits (Kernighan): {count_set_bits(n)}")  # 4
print(f"Set bits (naive): {count_set_bits_naive(n)}")  # 4
# Time: O(k) where k = number of set bits
```

## 4. XOR Tricks

XOR properties:
- a ^ a = 0 (self-cancellation)
- a ^ 0 = a
- a ^ b ^ a = b (commutative & associative)

```python
# Find the unique number (all others appear twice)
def find_unique(nums):
    """Find element that appears once, others appear twice."""
    result = 0
    for num in nums:
        result ^= num
    return result

# Find missing number in [0..n]
def find_missing(nums):
    """Find missing number in range 0..n."""
    n = len(nums)
    result = n
    for i, num in enumerate(nums):
        result ^= i ^ num
    return result

# Swap two numbers without temp variable
def swap(a, b):
    a ^= b
    b ^= a
    a ^= b
    return a, b

print(f"Unique in [2,3,4,2,3]: {find_unique([2,3,4,2,3])}")  # 4
print(f"Missing in [0,1,3,4]: {find_missing([0,1,3,4])}")  # 2
print(f"Swap 5,7: {swap(5, 7)}")  # (7, 5)
```

## 5. Power of Two Check

A number is a power of two if it has exactly one set bit.

```python
def is_power_of_two(n):
    """Check if n is a power of two."""
    return n > 0 and (n & (n - 1)) == 0

# Find the next power of two >= n
def next_power_of_two(n):
    if n <= 0:
        return 1
    n -= 1
    while n & (n - 1):
        n &= n - 1
    return n << 1

for n in [1, 2, 3, 4, 16, 18]:
    print(f"{n}: power of two? {is_power_of_two(n)}, next power: {next_power_of_two(n)}")
```

## 6. Subsets Using Bitmask

Each subset corresponds to a bitmask where bit i = 1 means element i is included.

```python
def subsets_bitmask(nums):
    """Generate all subsets using bitmasks."""
    n = len(nums)
    result = []
    for mask in range(1 << n):  # 2^n iterations
        subset = []
        for i in range(n):
            if mask & (1 << i):  # Check if i-th bit is set
                subset.append(nums[i])
        result.append(subset)
    return result

def subset_sum_exists(nums, target):
    """Check if any subset sums to target using bitmask."""
    n = len(nums)
    for mask in range(1 << n):
        total = 0
        for i in range(n):
            if mask & (1 << i):
                total += nums[i]
        if total == target:
            return True
    return False

print(f"Subsets of [1,2]: {subsets_bitmask([1,2])}")
print(f"Subset sum 3 from [1,2,3]: {subset_sum_exists([1,2,3], 3)}")  # True
print(f"Subset sum 7 from [1,2,3]: {subset_sum_exists([1,2,3], 7)}")  # False
# Time: O(n * 2^n), Space: O(n * 2^n)
```

## 7. Bit Reversal

Reverse the bits of a number.

```python
def reverse_bits(n, bits=8):
    """Reverse the bits of an integer (within given number of bits)."""
    result = 0
    for _ in range(bits):
        result = (result << 1) | (n & 1)
        n >>= 1
    return result

def reverse_bits_32(n):
    """Reverse 32-bit integer (LeetCode style)."""
    result = 0
    for _ in range(32):
        result = (result << 1) | (n & 1)
        n >>= 1
    return result

n = 0b11001010  # 202
rev = reverse_bits(n)
print(f"Original:  {n:08b} = {n}")
print(f"Reversed:  {rev:08b} = {rev}")
```

## Additional Useful Bit Operations

```python
# Check if two numbers have opposite signs
def opposite_signs(a, b):
    return (a ^ b) < 0

# Get the lowest set bit
def lowest_set_bit(n):
    return n & -n

# Clear all but the lowest set bit
def isolate_lowest_set_bit(n):
    return n & -n

# Divide by 2 (integer division)
def divide_by_2(n):
    return n >> 1

# Multiply by 2
def multiply_by_2(n):
    return n << 1

# Check if numbers are equal
def is_equal(a, b):
    return (a ^ b) == 0

print(f"Opposite signs (5, -3): {opposite_signs(5, -3)}")  # True
print(f"Lowest set bit of 12 (1100): {lowest_set_bit(12):b}")  # 100 (4)
print(f"42 * 2 = {multiply_by_2(42)}")  # 84
print(f"42 // 2 = {divide_by_2(42)}")  # 21
```

## LeetCode Problems

### LeetCode 136 - Single Number

**Problem:** Find the element that appears once when all others appear twice.

```python
def singleNumber(nums):
    result = 0
    for num in nums:
        result ^= num
    return result
# Time: O(n), Space: O(1)
```

### LeetCode 191 - Number of 1 Bits

**Problem:** Count the number of set bits in an integer.

```python
def hammingWeight(n):
    count = 0
    while n:
        n &= n - 1
        count += 1
    return count
# Time: O(k) where k = number of 1 bits, Space: O(1)
```

### LeetCode 338 - Counting Bits

**Problem:** Return an array of size n+1 where ans[i] = number of 1 bits in i.

```python
def countBits(n):
    """DP approach: ans[i] = ans[i >> 1] + (i & 1)"""
    ans = [0] * (n + 1)
    for i in range(1, n + 1):
        ans[i] = ans[i >> 1] + (i & 1)
    return ans

print(f"Count bits up to 5: {countBits(5)}")  # [0, 1, 1, 2, 1, 2]
# Time: O(n), Space: O(n)
```

### LeetCode 78 - Subsets (Bitmask Approach)

**Problem:** Generate all possible subsets.

```python
def subsets(nums):
    n = len(nums)
    result = []
    for mask in range(1 << n):
        subset = [nums[i] for i in range(n) if mask & (1 << i)]
        result.append(subset)
    return result
# Time: O(n * 2^n), Space: O(n * 2^n)
```
