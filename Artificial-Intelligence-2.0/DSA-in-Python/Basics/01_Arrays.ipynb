# Arrays in Python

Arrays are fundamental data structures that store elements in contiguous memory locations. In Python, we use **lists** as dynamic arrays that can grow and shrink at runtime.

## 1. Array/List Basics

Python lists are dynamic arrays - they automatically resize. Under the hood, CPython allocates extra space to amortize the cost of appends.

```python
# Creating lists
empty = []                     # Empty list
numbers = [1, 2, 3, 4, 5]      # Literal initialization
zeros = [0] * 5                # [0, 0, 0, 0, 0]
mixed = [1, "hello", 3.14]    # Heterogeneous (not typical in practice)
from_list = list(range(5))     # [0, 1, 2, 3, 4]

print(f"numbers: {numbers}")
print(f"zeros: {zeros}")
print(f"from_list(range(5)): {from_list}")
print(f"Length: {len(numbers)}")
```

## 2. Indexing & Slicing

Python supports zero-based indexing, negative indexing (from end), and slicing with `[start:stop:step]`.

```python
arr = [10, 20, 30, 40, 50, 60]

# Indexing
print(f"arr[0] = {arr[0]}")      # First element
print(f"arr[-1] = {arr[-1]}")    # Last element
print(f"arr[-2] = {arr[-2]}")    # Second last

# Slicing [start:stop:step]
print(f"arr[1:4] = {arr[1:4]}")       # [20, 30, 40]
print(f"arr[:3] = {arr[:3]}")         # First 3: [10, 20, 30]
print(f"arr[3:] = {arr[3:]}")         # From index 3: [40, 50, 60]
print(f"arr[::2] = {arr[::2]}")       # Every 2nd: [10, 30, 50]
print(f"arr[::-1] = {arr[::-1]}")     # Reversed: [60, 50, 40, 30, 20, 10]
print(f"arr[-3:] = {arr[-3:]}")       # Last 3: [40, 50, 60]
print(f"arr[:-3] = {arr[:-3]}")       # All except last 3: [10, 20, 30]
```

## 3. Iteration Techniques

Multiple ways to iterate over arrays depending on whether you need indices, values, or both.

```python
arr = [10, 20, 30, 40]

# By value
for val in arr:
    print(val, end=" ")
print()

# By index
for i in range(len(arr)):
    print(arr[i], end=" ")
print()

# Both index and value (enumerate)
for i, val in enumerate(arr):
    print(f"{i}:{val}", end=" ")
print()

# Reverse iteration
for val in reversed(arr):
    print(val, end=" ")
print()

# Multiple lists (zip)
a = [1, 2, 3]
b = ["a", "b", "c"]
for x, y in zip(a, b):
    print(f"{x}{y}", end=" ")
```

## 4. Insertion & Deletion

Inserting or deleting elements in a list is O(n) because elements must shift.

```python
arr = [1, 2, 3, 4, 5]

# Insert at end - O(1) amortized
arr.append(6)
print(f"After append(6): {arr}")

# Insert at position - O(n)
arr.insert(0, 0)  # Insert 0 at start
print(f"After insert(0, 0): {arr}")

# Remove by value - O(n)
arr.remove(3)
print(f"After remove(3): {arr}")

# Pop from end - O(1)
last = arr.pop()
print(f"Popped {last}, arr: {arr}")

# Pop from index - O(n)
first = arr.pop(0)
print(f"Popped {first} from start, arr: {arr}")

# Delete slice
del arr[1:3]
print(f"After del arr[1:3]: {arr}")

# Extend (merge another list)
arr.extend([10, 11, 12])
print(f"After extend([10, 11, 12]): {arr}")
```

## 5. 2D Arrays (Matrix)

A 2D array (matrix) is a list of lists. Access with `matrix[row][col]`.

```python
# Creating a 2D array
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print(f"matrix[0][1] = {matrix[0][1]}")  # Row 0, Col 1 = 2

# Row-wise iteration
for row in matrix:
    for val in row:
        print(val, end=" ")
print()

# Column-wise iteration
cols = len(matrix[0])
rows = len(matrix)
for c in range(cols):
    for r in range(rows):
        print(matrix[r][c], end=" ")
print()

# Creating a 3x4 matrix of zeros
rows, cols = 3, 4
zeros = [[0] * cols for _ in range(rows)]  # Correct way
print(f"Zero matrix: {zeros}")

# WRONG way - creates references to same row!
wrong = [[0] * cols] * rows  # All rows are the same list object
wrong[0][0] = 99
print(f"Wrong matrix: {wrong}")  # All rows affected!
```

## 6. Prefix Sum Technique

Precomputes cumulative sums so that range sum queries are O(1).

- **Prefix sum array**: `prefix[i] = sum of arr[0..i]`
- **Range sum [l..r]**: `prefix[r] - prefix[l-1]` (handle l=0 edge case)

```python
def build_prefix_sum(arr):
    """Build prefix sum array where prefix[i] = sum(arr[0..i])."""
    n = len(arr)
    prefix = [0] * n
    prefix[0] = arr[0]
    for i in range(1, n):
        prefix[i] = prefix[i-1] + arr[i]
    return prefix

def range_sum(prefix, l, r):
    """Return sum(arr[l..r]) in O(1) using prefix sum."""
    if l == 0:
        return prefix[r]
    return prefix[r] - prefix[l-1]

# Example
arr = [3, 1, 4, 1, 5, 9, 2, 6]
prefix = build_prefix_sum(arr)
print(f"Array: {arr}")
print(f"Prefix sum: {prefix}")
print(f"Sum(2..5) = {range_sum(prefix, 2, 5)}")  # 4+1+5+9 = 19
print(f"Sum(0..3) = {range_sum(prefix, 0, 3)}")  # 3+1+4+1 = 9
print(f"Sum(3..6) = {range_sum(prefix, 3, 6)}")  # 1+5+9+2 = 17
```

## 7. Sliding Window Technique

Used for subarray/substring problems with a fixed or variable-size window. Reduces O(n²) brute force to O(n).

**Key patterns:**
- Fixed window size: maintain window, slide by 1
- Variable window size: expand right, shrink left when condition breaks

```python
# Fixed Window: Maximum sum of any subarray of size k
def max_sum_fixed_window(arr, k):
    """Return max sum of any subarray of size k."""
    if len(arr) < k:
        return None
    # Compute sum of first window
    window_sum = sum(arr[:k])
    max_sum = window_sum
    # Slide window
    for i in range(k, len(arr)):
        window_sum += arr[i] - arr[i - k]  # Add new, remove old
        max_sum = max(max_sum, window_sum)
    return max_sum

# Variable Window: Smallest subarray with sum >= target
def min_subarray_len(target, arr):
    """Return length of smallest subarray with sum >= target."""
    left = 0
    curr_sum = 0
    min_len = float('inf')
    for right in range(len(arr)):
        curr_sum += arr[right]
        while curr_sum >= target:
            min_len = min(min_len, right - left + 1)
            curr_sum -= arr[left]
            left += 1
    return min_len if min_len != float('inf') else 0

arr = [2, 1, 5, 2, 3, 2]
print(f"Max sum subarray of size 3: {max_sum_fixed_window(arr, 3)}")  # 10
print(f"Smallest subarray len sum>=7: {min_subarray_len(7, arr)}")    # 2
```

## 8. Kadane's Algorithm

Finds the maximum sum of a contiguous subarray in O(n) time. Core idea: at each position, decide whether to extend the current subarray or start fresh.

```python
def kadane(arr):
    """Return maximum subarray sum using Kadane's algorithm."""
    max_ending_here = arr[0]
    max_so_far = arr[0]
    for i in range(1, len(arr)):
        # Extend or start new
        max_ending_here = max(arr[i], max_ending_here + arr[i])
        max_so_far = max(max_so_far, max_ending_here)
    return max_so_far

# With indices for subarray reconstruction
def kadane_with_indices(arr):
    max_ending_here = arr[0]
    max_so_far = arr[0]
    start = end = 0
    temp_start = 0
    for i in range(1, len(arr)):
        if arr[i] > max_ending_here + arr[i]:
            max_ending_here = arr[i]
            temp_start = i
        else:
            max_ending_here += arr[i]
        if max_ending_here > max_so_far:
            max_so_far = max_ending_here
            start = temp_start
            end = i
    return max_so_far, arr[start:end+1]

arr = [-2, 1, -3, 4, -1, 2, 1, -5, 4]
print(f"Max subarray sum: {kadane(arr)}")
max_sum, subarray = kadane_with_indices(arr)
print(f"Maximum sum: {max_sum}, Subarray: {subarray}")
```

## 9. Binary Search

Efficient search on sorted arrays. O(log n) time. Three templates: basic, left-bound, right-bound.

```python
def binary_search(arr, target):
    """Standard binary search. Returns index or -1."""
    left, right = 0, len(arr) - 1
    while left <= right:
        mid = left + (right - left) // 2  # Avoid overflow
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1

def first_occurrence(arr, target):
    """Find first index where target appears."""
    left, right = 0, len(arr) - 1
    result = -1
    while left <= right:
        mid = left + (right - left) // 2
        if arr[mid] == target:
            result = mid
            right = mid - 1  # Search left
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return result

def last_occurrence(arr, target):
    """Find last index where target appears."""
    left, right = 0, len(arr) - 1
    result = -1
    while left <= right:
        mid = left + (right - left) // 2
        if arr[mid] == target:
            result = mid
            left = mid + 1  # Search right
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return result

arr = [1, 2, 3, 4, 4, 4, 5, 6, 7]
print(f"Index of 4: {binary_search(arr, 4)}")
print(f"First 4: {first_occurrence(arr, 4)}")
print(f"Last 4: {last_occurrence(arr, 4)}")
print(f"Index of 8: {binary_search(arr, 8)}")
```

### Binary Search in Rotated Sorted Array

A sorted array rotated at some pivot. At least one half (left or right of mid) is always fully sorted.

```python
def search_rotated(nums, target):
    """Search in rotated sorted array. O(log n)."""
    left, right = 0, len(nums) - 1
    while left <= right:
        mid = left + (right - left) // 2
        if nums[mid] == target:
            return mid
        # Left half is sorted
        if nums[left] <= nums[mid]:
            if nums[left] <= target < nums[mid]:
                right = mid - 1
            else:
                left = mid + 1
        # Right half is sorted
        else:
            if nums[mid] < target <= nums[right]:
                left = mid + 1
            else:
                right = mid - 1
    return -1

# Test
rotated = [4, 5, 6, 7, 0, 1, 2]
for t in [0, 3, 7, 5]:
    print(f"Index of {t}: {search_rotated(rotated, t)}")
```

### LeetCode 53 - Maximum Subarray

**Problem:** Given an integer array nums, find the contiguous subarray (containing at least one number) which has the largest sum and return its sum.

**Approach:** Kadane's Algorithm - O(n) time, O(1) space.

```python
def maxSubArray(nums):
    max_ending_here = nums[0]
    max_so_far = nums[0]
    for i in range(1, len(nums)):
        max_ending_here = max(nums[i], max_ending_here + nums[i])
        max_so_far = max(max_so_far, max_ending_here)
    return max_so_far

# Test
print(maxSubArray([-2,1,-3,4,-1,2,1,-5,4]))  # 6
print(maxSubArray([1]))                       # 1
print(maxSubArray([5,4,-1,7,8]))             # 23
# Time: O(n), Space: O(1)
```

### LeetCode 121 - Best Time to Buy and Sell Stock

**Problem:** You are given an array prices where prices[i] is the price of a given stock on day i. You want to maximize profit by choosing a single day to buy and a different day in the future to sell.

**Approach:** Track minimum price so far, compute max profit at each step. O(n) time, O(1) space.

```python
def maxProfit(prices):
    min_price = float('inf')
    max_profit = 0
    for price in prices:
        if price < min_price:
            min_price = price
        elif price - min_price > max_profit:
            max_profit = price - min_price
    return max_profit

# Test
print(maxProfit([7,1,5,3,6,4]))  # 5
print(maxProfit([7,6,4,3,1]))    # 0
# Time: O(n), Space: O(1)
```

### LeetCode 1 - Two Sum

**Problem:** Given an array of integers nums and an integer target, return indices of the two numbers that add up to target.

**Approach:** Use hash map to store complement. O(n) time, O(n) space.

```python
def twoSum(nums, target):
    seen = {}  # value -> index
    for i, num in enumerate(nums):
        complement = target - num
        if complement in seen:
            return [seen[complement], i]
        seen[num] = i
    return []

# Test
print(twoSum([2,7,11,15], 9))  # [0, 1]
print(twoSum([3,2,4], 6))      # [1, 2]
print(twoSum([3,3], 6))        # [0, 1]
# Time: O(n), Space: O(n)
```

### LeetCode 33 - Search in Rotated Sorted Array

**Problem:** Given a rotated sorted array and a target, return the index. If not found, return -1. Must be O(log n).

**Approach:** Modified binary search - determine which half is sorted and check if target is in that half.

```python
def search(nums, target):
    left, right = 0, len(nums) - 1
    while left <= right:
        mid = left + (right - left) // 2
        if nums[mid] == target:
            return mid
        if nums[left] <= nums[mid]:
            if nums[left] <= target < nums[mid]:
                right = mid - 1
            else:
                left = mid + 1
        else:
            if nums[mid] < target <= nums[right]:
                left = mid + 1
            else:
                right = mid - 1
    return -1

# Test
print(search([4,5,6,7,0,1,2], 0))  # 4
print(search([4,5,6,7,0,1,2], 3))  # -1
print(search([1], 0))              # -1
# Time: O(log n), Space: O(1)
```
