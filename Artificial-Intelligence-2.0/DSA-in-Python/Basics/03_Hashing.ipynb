## 1. Hash Map / Dict Fundamentals

Python dicts are hash tables with O(1) average operations. Keys must be hashable (immutable).

```python
# Creating dicts
d1 = {}  # Empty
d2 = {"name": "Alice", "age": 30}
d3 = dict([("a", 1), ("b", 2)])  # From pairs
d4 = dict.fromkeys(["x", "y", "z"], 0)  # Default values

# Access
print(f"d2['name']: {d2['name']}")
print(f"d2.get('age'): {d2.get('age')}")
print(f"d2.get('city', 'Not found'): {d2.get('city', 'Not found')}")

# Insert/Update
d2["city"] = "NYC"  # Insert
d2["age"] = 31       # Update

# Check existence
print(f"'name' in d2: {'name' in d2}")

# Delete
del d2["city"]       # Remove key
popped = d2.pop("age")  # Remove and return value
print(f"Popped: {popped}, d2: {d2}")

# Iteration
d = {"a": 1, "b": 2, "c": 3}
print("Keys:", list(d))
print("Values:", list(d.values()))
print("Items:", list(d.items()))
for k, v in d.items():
    print(f"{k}:{v}", end=" ")
print()
```

## 2. Set Operations

Sets store unique elements. They support mathematical set operations.

```python
s1 = {1, 2, 3, 4, 5}
s2 = {4, 5, 6, 7, 8}

# Basic operations
s1.add(6)          # Add element
s1.discard(1)      # Remove if exists (no error)
s1.remove(2)       # Remove (raises KeyError if missing)

# Set operations
print(f"Union: {s1 | s2}")              # {3, 4, 5, 6, 7, 8}
print(f"Intersection: {s1 & s2}")       # {4, 5, 6}
print(f"Difference: {s1 - s2}")         # {3}
print(f"Symmetric diff: {s1 ^ s2}")    # {3, 7, 8}
print(f"Subset: {s1 <= s2}")           # False
print(f"Superset: {s2 >= s1}")         # False
print(f"Disjoint: {s1.isdisjoint(s2)}") # False

# Set comprehension
squares = {x**2 for x in range(10)}
print(f"Squares: {squares}")
```

## 3. Frequency Map Pattern

Counting element frequencies is one of the most common hash map use cases.

```python
from collections import Counter, defaultdict

# Using Counter
arr = [1, 2, 3, 2, 1, 2, 4, 5, 4, 4, 4]
freq = Counter(arr)
print(f"Counter: {dict(freq)}")
print(f"Most common(2): {freq.most_common(2)}")

# Using defaultdict
freq2 = defaultdict(int)
for x in arr:
    freq2[x] += 1
print(f"DefaultDict: {dict(freq2)}")

# Manual dict
freq3 = {}
for x in arr:
    freq3[x] = freq3.get(x, 0) + 1
print(f"Manual dict: {freq3}")

# Character frequency
text = "hello world"
char_freq = Counter(text)
print(f"Char freq: {dict(char_freq)}")
```

## 4. Count Distinct Elements

Using set or dict to count unique elements.

```python
def count_distinct(arr):
    return len(set(arr))

def count_distinct_window(arr, k):
    """Count distinct elements in each window of size k."""
    if k > len(arr):
        return []
    freq = Counter(arr[:k])
    result = [len(freq)]
    for i in range(k, len(arr)):
        # Remove element going out
        freq[arr[i-k]] -= 1
        if freq[arr[i-k]] == 0:
            del freq[arr[i-k]]
        # Add new element
        freq[arr[i]] = freq.get(arr[i], 0) + 1
        result.append(len(freq))
    return result

arr = [1, 2, 1, 3, 4, 2, 3]
print(f"Distinct count: {count_distinct(arr)}")
print(f"Distinct in windows of size 4: {count_distinct_window(arr, 4)}")
```

## 5. Group Anagrams

Anagrams share the same sorted string or frequency signature. Use a dict to group them.

```python
from collections import defaultdict

def group_anagrams(words):
    """Group anagrams together."""
    groups = defaultdict(list)
    for w in words:
        # Key: sorted characters
        key = ''.join(sorted(w))
        groups[key].append(w)
    return list(groups.values())

def group_anagrams_count(words):
    """Group anagrams using character count tuple as key."""
    groups = defaultdict(list)
    for w in words:
        # Key: tuple of 26 character counts
        count = [0] * 26
        for c in w:
            count[ord(c) - ord('a')] += 1
        groups[tuple(count)].append(w)
    return list(groups.values())

words = ["eat", "tea", "tan", "ate", "nat", "bat"]
print(f"Using sorted key: {group_anagrams(words)}")
print(f"Using count key: {group_anagrams_count(words)}")
# Time: O(n * k log k) for sort, O(n * k) for count method
# Where n = number of words, k = max word length
```

## 6. Subarray Sum Equals K

Use a hash map to store prefix sums and their frequencies. If `prefix_sum - k` has been seen, it means a subarray with sum k exists.

```python
def subarray_sum_equals_k(nums, k):
    """Count subarrays whose sum equals k."""
    prefix_sum = 0
    count = 0
    # Map: prefix sum -> frequency
    sum_freq = {0: 1}  # Empty subarray
    for num in nums:
        prefix_sum += num
        # If (prefix_sum - k) exists, we found subarrays
        count += sum_freq.get(prefix_sum - k, 0)
        sum_freq[prefix_sum] = sum_freq.get(prefix_sum, 0) + 1
    return count

# Test
print(subarray_sum_equals_k([1, 1, 1], 2))       # 2
print(subarray_sum_equals_k([1, 2, 3], 3))       # 2
print(subarray_sum_equals_k([-1, -1, 1], 0))     # 1
# Time: O(n), Space: O(n)
```

## 7. Union/Intersection of Arrays

Using set operations for efficient union and intersection.

```python
def union_intersection(arr1, arr2):
    """Return union and intersection of two arrays."""
    s1, s2 = set(arr1), set(arr2)
    union = list(s1 | s2)
    intersection = list(s1 & s2)
    return union, intersection

def intersect_with_duplicates(nums1, nums2):
    """Return intersection including duplicates."""
    freq = Counter(nums1)
    result = []
    for x in nums2:
        if freq.get(x, 0) > 0:
            result.append(x)
            freq[x] -= 1
    return result

a1 = [1, 2, 2, 1, 3]
a2 = [2, 2, 3, 4]
u, i = union_intersection(a1, a2)
print(f"Union: {u}")
print(f"Intersection: {i}")
print(f"Intersection with duplicates: {intersect_with_duplicates(a1, a2)}")
```

## LeetCode Problems

### LeetCode 242 - Valid Anagram (Hash Map Approach)

**Problem:** Given two strings s and t, return true if t is an anagram of s.

**Approach:** Count character frequencies and compare.

```python
from collections import Counter

def isAnagram(s: str, t: str) -> bool:
    if len(s) != len(t):
        return False
    count = [0] * 26
    for c in s:
        count[ord(c) - ord('a')] += 1
    for c in t:
        count[ord(c) - ord('a')] -= 1
    return all(x == 0 for x in count)

# Test
print(isAnagram("anagram", "nagaram"))  # True
print(isAnagram("rat", "car"))          # False
# Time: O(n), Space: O(1)
```

### LeetCode 49 - Group Anagrams

**Problem:** Given an array of strings, group the anagrams together.

**Approach:** Use a hash map with sorted strings as keys.

```python
from collections import defaultdict

def groupAnagrams(strs):
    groups = defaultdict(list)
    for s in strs:
        key = ''.join(sorted(s))
        groups[key].append(s)
    return list(groups.values())

# Test
print(groupAnagrams(["eat","tea","tan","ate","nat","bat"]))
# Time: O(n * k log k), Space: O(n * k)
# where n = number of strings, k = max string length
```

### LeetCode 560 - Subarray Sum Equals K

**Problem:** Given an array of integers and an integer k, return the total number of subarrays whose sum equals k.

**Approach:** Hash map of prefix sum frequencies.

```python
def subarraySum(nums, k):
    prefix_sum = 0
    count = 0
    sum_freq = {0: 1}
    for num in nums:
        prefix_sum += num
        count += sum_freq.get(prefix_sum - k, 0)
        sum_freq[prefix_sum] = sum_freq.get(prefix_sum, 0) + 1
    return count

# Test
print(subarraySum([1,1,1], 2))       # 2
print(subarraySum([1,2,3], 3))       # 2
print(subarraySum([1,-1,0], 0))      # 3
# Time: O(n), Space: O(n)
```

### LeetCode 128 - Longest Consecutive Sequence

**Problem:** Given an unsorted array of integers, return the length of the longest consecutive elements sequence. Must run in O(n).

**Approach:** Use a set for O(1) lookups. Only count sequences starting from the smallest element.

```python
def longestConsecutive(nums):
    num_set = set(nums)
    longest = 0
    for num in num_set:
        # Only start counting if num-1 is not in set (start of sequence)
        if num - 1 not in num_set:
            current = num
            streak = 1
            while current + 1 in num_set:
                current += 1
                streak += 1
            longest = max(longest, streak)
    return longest

# Test
print(longestConsecutive([100,4,200,1,3,2]))  # 4 (1,2,3,4)
print(longestConsecutive([0,3,7,2,5,8,4,6,0,1]))  # 9
# Time: O(n), Space: O(n)
```
