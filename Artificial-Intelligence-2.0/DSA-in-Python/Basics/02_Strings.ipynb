## 1. String Basics & Immutability

Strings in Python are **immutable** - once created, they cannot be modified. Any operation that changes a string creates a new string object.

```python
# String creation
s1 = 'single quotes'
s2 = "double quotes"
s3 = '''multi line
string'''
s4 = """also multi
line"""

# Immutability demo
s = "hello"
# s[0] = 'H'  # TypeError!
s = "H" + s[1:]  # Creates new string
print(s)

# Strings are sequences
print(f"Length: {len(s)}")
print(f"First char: {s[0]}")
print(f"Last char: {s[-1]}")
print(f"Substring: {s[1:4]}")
```

## 2. Common String Methods

Python provides a rich set of string methods.

```python
text = "  Hello, World!  "

# Case manipulation
print(f"lower: '{text.lower()}'")       # '  hello, world!  '
print(f"upper: '{text.upper()}'")       # '  HELLO, WORLD!  '
print(f"title: '{text.title()}'")       # '  Hello, World!  '
print(f"swapcase: '{text.swapcase()}'") # '  hELLO, wORLD!  '

# Trimming
print(f"strip: '{text.strip()}'")        # 'Hello, World!'
print(f"lstrip: '{text.lstrip()}'")      # 'Hello, World!  '
print(f"rstrip: '{text.rstrip()}'")      # '  Hello, World!'

# Searching
print(f"find 'World': {text.find('World')}")      # 8
print(f"find 'Python': {text.find('Python')}")    # -1
print(f"index 'World': {text.index('World')}")    # 8 (raises if not found)
print(f"count 'l': {text.count('l')}")            # 3

# Checking
print(f"startswith '  H': {text.startswith('  H')}")  # True
print(f"endswith '!  ': {text.endswith('!  ')}")      # True
print(f"isdigit '123': {'123'.isdigit()}")            # True
print(f"isalpha 'abc': {'abc'.isalpha()}")            # True
print(f"isalnum 'abc123': {'abc123'.isalnum()}")      # True

# Splitting and joining
print(f"split: {text.split(',')}")        # ['  Hello', ' World!  ']
print(f"split with default: {'a b c'.split()}")  # ['a', 'b', 'c']

# Replace
print(f"replace: {text.replace('World', 'Python')}")
```

## 3. Two Pointers Technique

Two pointers (one at start, one at end) is a common pattern for string problems like palindrome checking, reversing, and more.

```python
def reverse_string(s):
    """Reverse a string using two pointers (in-place via list)."""
    chars = list(s)
    left, right = 0, len(chars) - 1
    while left < right:
        chars[left], chars[right] = chars[right], chars[left]
        left += 1
        right -= 1
    return ''.join(chars)

def is_palindrome(s):
    """Two-pointer palindrome check (ignoring non-alphanumeric)."""
    left, right = 0, len(s) - 1
    while left < right:
        while left < right and not s[left].isalnum():
            left += 1
        while left < right and not s[right].isalnum():
            right -= 1
        if s[left].lower() != s[right].lower():
            return False
        left += 1
        right -= 1
    return True

print(f"Reverse 'hello': {reverse_string('hello')}")
print(f"Palindrome 'racecar': {is_palindrome('racecar')}")
print(f"Palindrome 'A man, a plan, a canal: Panama': {is_palindrome('A man, a plan, a canal: Panama')}")
```

## 4. Pattern Matching - KMP Algorithm

Knuth-Morris-Pratt algorithm finds occurrences of a pattern in a text in O(n + m) time by avoiding redundant comparisons using a **LPS (Longest Proper Prefix which is also Suffix)** array.

```python
def build_lps(pattern):
    """Build Longest Prefix Suffix array for KMP."""
    m = len(pattern)
    lps = [0] * m
    length = 0  # length of previous longest prefix suffix
    i = 1
    while i < m:
        if pattern[i] == pattern[length]:
            length += 1
            lps[i] = length
            i += 1
        else:
            if length != 0:
                length = lps[length - 1]
            else:
                lps[i] = 0
                i += 1
    return lps

def kmp_search(text, pattern):
    """Return starting indices of pattern in text using KMP."""
    n, m = len(text), len(pattern)
    if m == 0:
        return []
    lps = build_lps(pattern)
    result = []
    i = j = 0  # i for text, j for pattern
    while i < n:
        if text[i] == pattern[j]:
            i += 1
            j += 1
            if j == m:
                result.append(i - j)
                j = lps[j - 1]
        else:
            if j != 0:
                j = lps[j - 1]
            else:
                i += 1
    return result

# Example
text = "ABABDABACDABABCABAB"
pattern = "ABABCABAB"
print(f"Pattern found at indices: {kmp_search(text, pattern)}")
print(f"LPS array for '{pattern}': {build_lps(pattern)}")
# Time: O(n + m), Space: O(m)
```

## 5. Anagram/Frequency Count

An anagram is formed by rearranging letters. Two strings are anagrams if they have the same character frequencies.

**Approaches:**
- Sort both strings and compare: O(n log n)
- Count character frequencies: O(n) using Counter or array

```python
from collections import Counter

def is_anagram_sort(s1, s2):
    """Anagram check by sorting."""
    return sorted(s1) == sorted(s2)

def is_anagram_counter(s1, s2):
    """Anagram check with Counter."""
    return Counter(s1) == Counter(s2)

def is_anagram_array(s1, s2):
    """Anagram check with array (for lowercase letters)."""
    if len(s1) != len(s2):
        return False
    count = [0] * 26
    a_ord = ord('a')
    for c in s1:
        count[ord(c) - a_ord] += 1
    for c in s2:
        count[ord(c) - a_ord] -= 1
    return all(x == 0 for x in count)

print(f"Anagram (sort): {is_anagram_sort('listen', 'silent')}")
print(f"Anagram (counter): {is_anagram_counter('listen', 'silent')}")
print(f"Anagram (array): {is_anagram_array('listen', 'silent')}")
print(f"Not anagram: {is_anagram_array('hello', 'world')}")
# Time: O(n) with counter/array, Space: O(1) for array (fixed 26)
```

## 6. Palindrome Problems

A palindrome reads the same forwards and backward. There are variations: valid palindrome, longest palindromic substring, palindrome partitioning.

```python
def valid_palindrome(s):
    """Check if string is a valid palindrome (ignore case & non-alnum)."""
    left, right = 0, len(s) - 1
    while left < right:
        while left < right and not s[left].isalnum():
            left += 1
        while left < right and not s[right].isalnum():
            right -= 1
        if s[left].lower() != s[right].lower():
            return False
        left += 1
        right -= 1
    return True

def expand_around_center(s, left, right):
    """Expand from center and return palindrome bounds."""
    while left >= 0 and right < len(s) and s[left] == s[right]:
        left -= 1
        right += 1
    return left + 1, right - 1

def longest_palindrome(s):
    """Find longest palindromic substring using expand around center."""
    if not s:
        return ""
    start, end = 0, 0
    for i in range(len(s)):
        # Odd length palindrome
        l1, r1 = expand_around_center(s, i, i)
        # Even length palindrome
        l2, r2 = expand_around_center(s, i, i + 1)
        if r1 - l1 > end - start:
            start, end = l1, r1
        if r2 - l2 > end - start:
            start, end = l2, r2
    return s[start:end+1]

print(f"Valid palindrome 'racecar': {valid_palindrome('racecar')}")
print(f"Longest palindrome in 'babad': {longest_palindrome('babad')}")
print(f"Longest palindrome in 'cbbd': {longest_palindrome('cbbd')}")
# Time: O(n²), Space: O(1)
```

## 7. String Building: join vs +

Using `+` in a loop creates many intermediate string objects. `join()` is O(n) and much more efficient.

```python
# Inefficient - O(n²)
def build_with_plus(words):
    result = ""
    for w in words:
        result += w  # Creates new string each time
    return result

# Efficient - O(n)
def build_with_join(words):
    return "".join(words)

# Efficient for conditional building: use list + join
def build_conditional(words):
    parts = []
    for w in words:
        if len(w) > 2:
            parts.append(w.upper())
    return ", ".join(parts)

words = ["hello", "world", "from", "Python"]
print(f"Plus: {build_with_plus(words)}")
print(f"Join: {build_with_join(words)}")
print(f"Conditional: {build_conditional(words)}")
```

### LeetCode 125 - Valid Palindrome

**Problem:** A phrase is a palindrome if it reads the same forward and backward after converting all uppercase letters to lowercase and removing all non-alphanumeric characters.

```python
def isPalindrome(s: str) -> bool:
    left, right = 0, len(s) - 1
    while left < right:
        while left < right and not s[left].isalnum():
            left += 1
        while left < right and not s[right].isalnum():
            right -= 1
        if s[left].lower() != s[right].lower():
            return False
        left += 1
        right -= 1
    return True

# Test
print(isPalindrome("A man, a plan, a canal: Panama"))  # True
print(isPalindrome("race a car"))                      # False
print(isPalindrome(" "))                               # True
# Time: O(n), Space: O(1)
```

### LeetCode 242 - Valid Anagram

**Problem:** Given two strings s and t, return true if t is an anagram of s.

```python
from collections import Counter

def isAnagram(s: str, t: str) -> bool:
    return Counter(s) == Counter(t)

# Test
print(isAnagram("anagram", "nagaram"))  # True
print(isAnagram("rat", "car"))          # False
# Time: O(n), Space: O(1) - fixed 26 chars
```

### LeetCode 5 - Longest Palindromic Substring

**Problem:** Given a string s, return the longest palindromic substring in s.

```python
def longestPalindrome(s: str) -> str:
    if not s:
        return ""
    start, end = 0, 0
    for i in range(len(s)):
        l1, r1 = i, i
        while l1 >= 0 and r1 < len(s) and s[l1] == s[r1]:
            l1 -= 1
            r1 += 1
        if r1 - l1 - 2 > end - start:
            start, end = l1 + 1, r1 - 1
        l2, r2 = i, i + 1
        while l2 >= 0 and r2 < len(s) and s[l2] == s[r2]:
            l2 -= 1
            r2 += 1
        if r2 - l2 - 2 > end - start:
            start, end = l2 + 1, r2 - 1
    return s[start:end+1]

# Test
print(longestPalindrome("babad"))  # "bab" or "aba"
print(longestPalindrome("cbbd"))   # "bb"
print(longestPalindrome("a"))      # "a"
# Time: O(n²), Space: O(1)
```

### LeetCode 28 - Find the Index of the First Occurrence in a String (KMP)

**Problem:** Given two strings needle and haystack, return the index of the first occurrence of needle in haystack, or -1 if not found.

```python
def build_lps(pattern):
    m = len(pattern)
    lps = [0] * m
    length = 0
    i = 1
    while i < m:
        if pattern[i] == pattern[length]:
            length += 1
            lps[i] = length
            i += 1
        else:
            if length != 0:
                length = lps[length - 1]
            else:
                lps[i] = 0
                i += 1
    return lps

def strStr(haystack: str, needle: str) -> int:
    n, m = len(haystack), len(needle)
    if m == 0:
        return 0
    lps = build_lps(needle)
    i = j = 0
    while i < n:
        if haystack[i] == needle[j]:
            i += 1
            j += 1
            if j == m:
                return i - j
        else:
            if j != 0:
                j = lps[j - 1]
            else:
                i += 1
    return -1

# Test
print(strStr("sadbutsad", "sad"))   # 0
print(strStr("leetcode", "leeto"))   # -1
print(strStr("hello", "ll"))         # 2
# Time: O(n + m), Space: O(m)
```
