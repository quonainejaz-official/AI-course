# Recursion & Backtracking

Recursion solves a problem by solving smaller instances of the same problem. Backtracking is a refinement where we explore partial solutions and abandon them when they cannot lead to a valid solution.

## 1. Recursion Fundamentals

Every recursive function needs:
- **Base case**: condition to stop recursion
- **Recursive case**: call to itself with smaller input
- **Call stack**: each call pushes a new frame

```python
def factorial(n):
    """n! = n * (n-1) * ... * 1"""
    if n <= 1:  # Base case
        return 1
    return n * factorial(n - 1)  # Recursive case

def fibonacci(n):
    """Fibonacci: F(n) = F(n-1) + F(n-2)"""
    if n <= 1:  # Base case
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)  # Recursive case

print(f"Factorial 5: {factorial(5)}")
print(f"Fibonacci 6: {fibonacci(6)}")
print(f"Fibonacci 10: {fibonacci(10)}")

# Time: O(n) for factorial, O(2^n) for naive fibonacci
# Space: O(n) call stack depth
```

## 2. Backtracking Framework

General backtracking follows the **choose-explore-unchoose** pattern:
1. **Choose**: Make a decision
2. **Explore**: Recursively solve the subproblem
3. **Unchoose**: Undo the decision to try alternatives

```python
def subsets(nums):
    """Generate all subsets (power set)."""
    result = []
    def backtrack(start, path):
        result.append(path[:])  # Include current subset
        for i in range(start, len(nums)):
            path.append(nums[i])  # Choose
            backtrack(i + 1, path)  # Explore
            path.pop()  # Unchoose
    backtrack(0, [])
    return result

def permutations(nums):
    """Generate all permutations."""
    result = []
    def backtrack(path, used):
        if len(path) == len(nums):
            result.append(path[:])
            return
        for i, num in enumerate(nums):
            if not used[i]:
                path.append(num)  # Choose
                used[i] = True
                backtrack(path, used)  # Explore
                path.pop()  # Unchoose
                used[i] = False
    backtrack([], [False] * len(nums))
    return result

def combinations(n, k):
    """Generate all combinations of size k from 1..n."""
    result = []
    def backtrack(start, path):
        if len(path) == k:
            result.append(path[:])
            return
        for i in range(start, n + 1):
            path.append(i)
            backtrack(i + 1, path)
            path.pop()
    backtrack(1, [])
    return result

print(f"Subsets of [1,2]: {subsets([1, 2])}")
print(f"Permutations of [1,2,3]: {permutations([1, 2, 3])}")
print(f"Combinations C(4,2): {combinations(4, 2)}")
```

## 3. Combinatorial Generation Patterns

Three core patterns: subsets, permutations, combinations. Understanding their differences is critical for interviews.

```python
def combination_sum(candidates, target):
    """Find all combinations that sum to target (unlimited use)."""
    result = []
    def backtrack(start, path, remaining):
        if remaining == 0:
            result.append(path[:])
            return
        if remaining < 0:
            return
        for i in range(start, len(candidates)):
            path.append(candidates[i])
            backtrack(i, path, remaining - candidates[i])  # same index
            path.pop()
    backtrack(0, [], target)
    return result

def combination_sum_no_repeat(candidates, target):
    """Each number used at most once."""
    result = []
    candidates.sort()
    def backtrack(start, path, remaining):
        if remaining == 0:
            result.append(path[:])
            return
        for i in range(start, len(candidates)):
            if i > start and candidates[i] == candidates[i-1]:
                continue  # Skip duplicates
            if candidates[i] > remaining:
                break
            path.append(candidates[i])
            backtrack(i + 1, path, remaining - candidates[i])
            path.pop()
    backtrack(0, [], target)
    return result

print(f"Combination sum to 7 from [2,3,6,7]: {combination_sum([2,3,6,7], 7)}")
print(f"Sum to 8 from [10,1,2,7,6,1,5] no repeat: {combination_sum_no_repeat([10,1,2,7,6,1,5], 8)}")
```

## 4. N-Queens

Place N queens on an NxN board so no two attack each other. Use backtracking with column, diagonal, and anti-diagonal tracking.

```python
def solve_n_queens(n):
    """Return all distinct solutions to N-Queens."""
    cols = set()
    diag1 = set()  # r + c
    diag2 = set()  # r - c
    result = []
    board = [["."] * n for _ in range(n)]

    def backtrack(r):
        if r == n:
            result.append(["".join(row) for row in board])
            return
        for c in range(n):
            if c in cols or (r + c) in diag1 or (r - c) in diag2:
                continue
            board[r][c] = "Q"
            cols.add(c)
            diag1.add(r + c)
            diag2.add(r - c)
            backtrack(r + 1)
            board[r][c] = "."
            cols.remove(c)
            diag1.remove(r + c)
            diag2.remove(r - c)

    backtrack(0)
    return result

solutions = solve_n_queens(4)
print(f"Number of solutions for 4-Queens: {len(solutions)}")
for sol in solutions:
    for row in sol:
        print(row)
    print()
# Time: O(n!), Space: O(n²)
```

## 5. Sudoku Solver

Fill empty cells with digits 1-9 respecting row, column, and 3x3 box constraints.

```python
def solve_sudoku(board):
    """Solve Sudoku in-place using backtracking."""
    def is_valid(r, c, ch):
        for i in range(9):
            if board[i][c] == ch:
                return False
            if board[r][i] == ch:
                return False
            if board[3 * (r // 3) + i // 3][3 * (c // 3) + i % 3] == ch:
                return False
        return True

    def backtrack():
        for r in range(9):
            for c in range(9):
                if board[r][c] == ".":
                    for ch in "123456789":
                        if is_valid(r, c, ch):
                            board[r][c] = ch
                            if backtrack():
                                return True
                            board[r][c] = "."
                    return False
        return True

    backtrack()
    return board

# Example
board = [
    ["5","3",".",".","7",".",".",".","."],
    ["6",".",".","1","9","5",".",".","."],
    [".","9","8",".",".",".",".","6","."],
    ["8",".",".",".","6",".",".",".","3"],
    ["4",".",".","8",".","3",".",".","1"],
    ["7",".",".",".","2",".",".",".","6"],
    [".","6",".",".",".",".","2","8","."],
    [".",".",".","4","1","9",".",".","5"],
    [".",".",".",".","8",".",".","7","9"]
]
solve_sudoku(board)
for row in board:
    print(" ".join(row))
```

## 6. Word Search

Find if a word exists in a 2D grid by moving up/down/left/right, without revisiting cells.

```python
def exist(board, word):
    """Check if word exists in grid by adjacent moves."""
    rows, cols = len(board), len(board[0])
    def dfs(r, c, i):
        if i == len(word):
            return True
        if r < 0 or r >= rows or c < 0 or c >= cols or board[r][c] != word[i]:
            return False
        temp = board[r][c]
        board[r][c] = "#"  # Mark visited
        found = dfs(r+1, c, i+1) or dfs(r-1, c, i+1) or \
                dfs(r, c+1, i+1) or dfs(r, c-1, i+1)
        board[r][c] = temp  # Unmark
        return found

    for r in range(rows):
        for c in range(cols):
            if board[r][c] == word[0] and dfs(r, c, 0):
                return True
    return False

grid = [
    ["A","B","C","E"],
    ["S","F","C","S"],
    ["A","D","E","E"]
]
print(f"'ABCCED' exists: {exist(grid, 'ABCCED')}")  # True
print(f"'SEE' exists: {exist(grid, 'SEE')}")        # True
print(f"'ABCB' exists: {exist(grid, 'ABCB')}")      # False
# Time: O(m * n * 4^L) where L = word length, Space: O(L)
```

## LeetCode Problems

### LeetCode 46 - Permutations

**Problem:** Given an array of distinct integers, return all possible permutations.

```python
def permute(nums):
    result = []
    def backtrack(path, used):
        if len(path) == len(nums):
            result.append(path[:])
            return
        for i, num in enumerate(nums):
            if not used[i]:
                path.append(num)
                used[i] = True
                backtrack(path, used)
                path.pop()
                used[i] = False
    backtrack([], [False] * len(nums))
    return result
# Time: O(n * n!), Space: O(n)
```

### LeetCode 78 - Subsets

**Problem:** Given an array of distinct integers, return all possible subsets.

```python
def subsets(nums):
    result = []
    def backtrack(start, path):
        result.append(path[:])
        for i in range(start, len(nums)):
            path.append(nums[i])
            backtrack(i + 1, path)
            path.pop()
    backtrack(0, [])
    return result
# Time: O(n * 2^n), Space: O(n)
```

### LeetCode 39 - Combination Sum

**Problem:** Find all unique combinations of candidates that sum to target (can reuse candidates).

```python
def combinationSum(candidates, target):
    result = []
    def backtrack(start, path, remaining):
        if remaining == 0:
            result.append(path[:])
            return
        for i in range(start, len(candidates)):
            if candidates[i] > remaining:
                continue
            path.append(candidates[i])
            backtrack(i, path, remaining - candidates[i])
            path.pop()
    backtrack(0, [], target)
    return result
# Time: O(n^(t/m)) where t=target, m=min candidate, Space: O(t/m)
```

### LeetCode 51 - N-Queens

**Problem:** Place N queens on NxN board without them attacking each other.

```python
def solveNQueens(n):
    cols = set()
    diag1 = set()
    diag2 = set()
    result = []
    board = [["."] * n for _ in range(n)]
    def backtrack(r):
        if r == n:
            result.append(["".join(row) for row in board])
            return
        for c in range(n):
            if c in cols or (r + c) in diag1 or (r - c) in diag2:
                continue
            board[r][c] = "Q"
            cols.add(c)
            diag1.add(r + c)
            diag2.add(r - c)
            backtrack(r + 1)
            board[r][c] = "."
            cols.remove(c)
            diag1.remove(r + c)
            diag2.remove(r - c)
    backtrack(0)
    return result
# Time: O(n!), Space: O(n²)
```

## Summary

| Problem | Pattern | Time | Space |
|---|---|---|---|
| Subsets | Inclusion/Exclusion | O(n * 2^n) | O(n) |
| Permutations | Used array + backtrack | O(n * n!) | O(n) |
| Combinations | Start index + backtrack | O(C(n,k) * k) | O(k) |
| N-Queens | Row by row, track attacks | O(n!) | O(n²) |
| Sudoku | Row-col iteration | O(9^(empty)) | O(1) |
| Word Search | DFS + backtracking | O(mn * 4^L) | O(L) |
