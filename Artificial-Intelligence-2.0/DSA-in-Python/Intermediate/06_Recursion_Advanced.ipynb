## 1. Divide & Conquer

**Divide**: Break problem into smaller subproblems
**Conquer**: Solve subproblems recursively
**Combine**: Merge solutions into final answer

```python
def merge_sort(arr):
    """Merge sort using divide and conquer."""
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    return merge(left, right)

def merge(left, right):
    """Merge two sorted arrays."""
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result

def quick_sort(arr):
    """Quick sort using divide and conquer."""
    if len(arr) <= 1:
        return arr
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quick_sort(left) + middle + quick_sort(right)

# In-place quicksort (interview-friendly)
def quick_sort_inplace(arr, low=0, high=None):
    if high is None:
        high = len(arr) - 1
    if low < high:
        pi = partition(arr, low, high)
        quick_sort_inplace(arr, low, pi - 1)
        quick_sort_inplace(arr, pi + 1, high)

def partition(arr, low, high):
    pivot = arr[high]
    i = low - 1
    for j in range(low, high):
        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]
    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return i + 1

arr = [38, 27, 43, 3, 9, 82, 10]
print(f"Original: {arr}")
print(f"Merge sort: {merge_sort(arr)}")
print(f"Quick sort: {quick_sort(arr)}")
# Merge sort: O(n log n) time, O(n) space
# Quick sort: O(n log n) avg, O(n²) worst, O(log n) space
```

## 2. Divide & Conquer - Max Subarray

Maximum subarray sum using divide and conquer: split array, find max in left, right, and crossing.

```python
def max_subarray_dc(arr, low, high):
    """Maximum subarray using divide and conquer."""
    if low == high:
        return arr[low]
    mid = (low + high) // 2
    left_max = max_subarray_dc(arr, low, mid)
    right_max = max_subarray_dc(arr, mid + 1, high)
    cross_max = max_crossing(arr, low, mid, high)
    return max(left_max, right_max, cross_max)

def max_crossing(arr, low, mid, high):
    left_sum = float('-inf')
    total = 0
    for i in range(mid, low - 1, -1):
        total += arr[i]
        left_sum = max(left_sum, total)
    right_sum = float('-inf')
    total = 0
    for i in range(mid + 1, high + 1):
        total += arr[i]
        right_sum = max(right_sum, total)
    return left_sum + right_sum

arr = [-2, 1, -3, 4, -1, 2, 1, -5, 4]
print(f"Max subarray (DC): {max_subarray_dc(arr, 0, len(arr)-1)}")
# Time: O(n log n), Space: O(log n)
```

## 3. Memoization (Top-Down DP)

Cache results of expensive function calls to avoid redundant computation. Transforms exponential to polynomial.

```python
from functools import lru_cache

# Fibonacci with memoization
def fib_memo(n, memo={}):
    if n in memo:
        return memo[n]
    if n <= 1:
        return n
    memo[n] = fib_memo(n-1, memo) + fib_memo(n-2, memo)
    return memo[n]

# Using lru_cache decorator
@lru_cache(maxsize=None)
def fib_cached(n):
    if n <= 1:
        return n
    return fib_cached(n-1) + fib_cached(n-2)

print(f"Fibonacci 40 (memo): {fib_memo(40)}")
print(f"Fibonacci 40 (cached): {fib_cached(40)}")
# Time: O(n), Space: O(n)
# Compare to naive O(2^n) without memoization
```

## 4. Tail Recursion

Tail recursion is when the recursive call is the **last operation** in the function. Python doesn't optimize tail calls, but it's a useful concept.

```python
# Non-tail recursive factorial
def fact_non_tail(n):
    if n <= 1:
        return 1
    return n * fact_non_tail(n - 1)  # Multiplication after recursion

# Tail recursive factorial (accumulator pattern)
def fact_tail(n, acc=1):
    if n <= 1:
        return acc
    return fact_tail(n - 1, n * acc)  # Only recursive call, no further work

# Non-tail recursive: sum
def sum_non_tail(arr):
    if not arr:
        return 0
    return arr[0] + sum_non_tail(arr[1:])

# Tail recursive sum
def sum_tail(arr, acc=0):
    if not arr:
        return acc
    return sum_tail(arr[1:], acc + arr[0])

print(f"Factorial 5 (non-tail): {fact_non_tail(5)}")
print(f"Factorial 5 (tail): {fact_tail(5)}")
print(f"Sum [1,2,3,4,5] (tail): {sum_tail([1,2,3,4,5])}")
```

## 5. Recursive vs Iterative Tradeoffs

| Aspect | Recursive | Iterative |
|---|---|---|
| Readability | Natural for divide-conquer | May be verbose |
| Memory | O(n) call stack | O(1) extra |
| Risk | Stack overflow for deep recursion | None |
| Performance | Function call overhead | Faster |

```python
# Recursive vs iterative comparison

# Recursive binary search
def binary_search_recursive(arr, target, left, right):
    if left > right:
        return -1
    mid = left + (right - left) // 2
    if arr[mid] == target:
        return mid
    elif arr[mid] < target:
        return binary_search_recursive(arr, target, mid + 1, right)
    else:
        return binary_search_recursive(arr, target, left, mid - 1)

# Iterative binary search
def binary_search_iterative(arr, target):
    left, right = 0, len(arr) - 1
    while left <= right:
        mid = left + (right - left) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1

# Recursive tree traversal (natural fit)
def tree_sum(root):
    if not root:
        return 0
    return root.val + tree_sum(root.left) + tree_sum(root.right)

arr = [1, 3, 5, 7, 9, 11]
print(f"Recursive BS: {binary_search_recursive(arr, 7, 0, len(arr)-1)}")
print(f"Iterative BS: {binary_search_iterative(arr, 7)}")
print(f"Recursive BS: {binary_search_recursive(arr, 8, 0, len(arr)-1)}")
```

## 6. Tree/Graph Recursion

Trees and graphs are naturally recursive structures.

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def max_depth(root):
    """Maximum depth of binary tree (tree recursion)."""
    if not root:
        return 0
    return 1 + max(max_depth(root.left), max_depth(root.right))

def count_paths(root, target):
    """Count paths in tree that sum to target (Path Sum III pattern)."""
    def dfs(node, curr_sum):
        if not node:
            return 0
        curr_sum += node.val
        count = 1 if curr_sum == target else 0
        return count + dfs(node.left, curr_sum) + dfs(node.right, curr_sum)

    if not root:
        return 0
    return dfs(root, 0) + count_paths(root.left, target) + count_paths(root.right, target)

# Build sample
root = TreeNode(10,
    TreeNode(5, TreeNode(3), TreeNode(2)),
    TreeNode(-3, None, TreeNode(11))
)
print(f"Max depth: {max_depth(root)}")
print(f"Paths summing to 8: {count_paths(root, 8)}")
```

### LeetCode 50 - Pow(x, n)

**Problem:** Implement pow(x, n) which calculates x raised to n. Fast exponentiation using divide & conquer.

```python
def myPow(x: float, n: int) -> float:
    if n == 0:
        return 1.0
    if n < 0:
        x = 1 / x
        n = -n
    half = myPow(x, n // 2)
    if n % 2 == 0:
        return half * half
    else:
        return half * half * x

# Test
print(myPow(2.0, 10))  # 1024.0
print(myPow(2.0, -2))  # 0.25
print(myPow(2.0, 0))   # 1.0
# Time: O(log n), Space: O(log n)
```

### LeetCode 169 - Majority Element (Divide & Conquer)

**Problem:** Find the element that appears more than n/2 times.

```python
def majorityElement(nums):
    """Divide & conquer approach."""
    def majority(l, r):
        if l == r:
            return nums[l]
        mid = l + (r - l) // 2
        left_majority = majority(l, mid)
        right_majority = majority(mid + 1, r)
        if left_majority == right_majority:
            return left_majority
        left_count = sum(1 for i in range(l, r + 1) if nums[i] == left_majority)
        right_count = sum(1 for i in range(l, r + 1) if nums[i] == right_majority)
        return left_majority if left_count > right_count else right_majority
    return majority(0, len(nums) - 1)

# Test
print(majorityElement([3, 2, 3]))              # 3
print(majorityElement([2, 2, 1, 1, 1, 2, 2]))  # 2
# Time: O(n log n), Space: O(log n)
# Note: O(n) Boyer-Moore voting is more efficient
```

### LeetCode 241 - Different Ways to Add Parentheses

**Problem:** Given a string expression of numbers and operators, return all possible results from different parenthesizations.

```python
def diffWaysToCompute(expression):
    if expression.isdigit():
        return [int(expression)]
    result = []
    for i, c in enumerate(expression):
        if c in '+-*':
            left = diffWaysToCompute(expression[:i])
            right = diffWaysToCompute(expression[i+1:])
            for l in left:
                for r in right:
                    if c == '+':
                        result.append(l + r)
                    elif c == '-':
                        result.append(l - r)
                    else:
                        result.append(l * r)
    return result

# Test
print(diffWaysToCompute("2-1-1"))  # [0, 2]
print(diffWaysToCompute("2*3-4*5"))  # [-34, -14, -10, -10, 10]
# Time: O(2^n), Space: O(2^n) - Catalan number growth
```

## Summary

| Pattern | Approach | Time | Space | Best For |
|---|---|---|---|---|
| Merge Sort | D&C - split, sort, merge | O(n log n) | O(n) | Stable sorting |
| Quick Sort | D&C - partition | O(n log n) avg | O(log n) | In-place sorting |
| Fast Exponentiation | D&C - half | O(log n) | O(log n) | Math |
| Memoization | Cache results | O(n) | O(n) | Overlapping subproblems |
| Tree Recursion | DFS on children | O(n) | O(h) | Tree problems |
| Tail Recursion | Accumulator pattern | O(n) | O(1)* | Linear recursion (*if TCO) |
