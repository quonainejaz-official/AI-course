## 1. Stack Implementation

Python's list is the most common stack. `collections.deque` also works as a stack (faster for insert/pop from both ends).

```python
# Stack using list
stack = []
stack.append(1)   # push
stack.append(2)
stack.append(3)
print(f"Stack: {stack}")
print(f"Pop: {stack.pop()}")   # 3
print(f"Peek: {stack[-1]}")    # 2
print(f"Is empty: {len(stack) == 0}")
print(f"Size: {len(stack)}")

# Stack using deque
from collections import deque
stack_d = deque()
stack_d.append(10)
stack_d.append(20)
print(f"Deque stack pop: {stack_d.pop()}")  # 20
print(f"Deque stack peek: {stack_d[-1]}")   # 10
```

## 2. Balanced Parentheses

The classic stack problem: match opening/closing brackets.

```python
def is_balanced(s):
    """Check if parentheses are balanced."""
    pairs = {')': '(', ']': '[', '}': '{'}
    stack = []
    for char in s:
        if char in '([{':
            stack.append(char)
        else:
            if not stack or stack[-1] != pairs[char]:
                return False
            stack.pop()
    return len(stack) == 0

# Test
tests = ["()", "()[]{}", "(]", "([)]", "{[]}"]
for t in tests:
    print(f"'{t}': {is_balanced(t)}")
# Time: O(n), Space: O(n)
```

## 3. Monotonic Stack

A **monotonic stack** maintains elements in increasing or decreasing order. Used for problems like Next Greater Element.

- **Monotonic Increasing**: stack top is smallest
- **Monotonic Decreasing**: stack top is largest

```python
def next_greater_element(nums):
    """Return next greater element for each position (-1 if none)."""
    n = len(nums)
    result = [-1] * n
    stack = []  # Will store indices, monotonic decreasing
    for i in range(n):
        # While current > element at stack top
        while stack and nums[i] > nums[stack[-1]]:
            idx = stack.pop()
            result[idx] = nums[i]
        stack.append(i)
    return result

def next_smaller_element(nums):
    """Return next smaller element for each position (-1 if none)."""
    n = len(nums)
    result = [-1] * n
    stack = []  # Monotonic increasing
    for i in range(n):
        while stack and nums[i] < nums[stack[-1]]:
            idx = stack.pop()
            result[idx] = nums[i]
        stack.append(i)
    return result

nums = [2, 1, 5, 6, 2, 3]
print(f"Original: {nums}")
print(f"Next Greater: {next_greater_element(nums)}")
print(f"Next Smaller: {next_smaller_element(nums)}")
# Time: O(n), Space: O(n)
```

## 4. Min Stack

Design a stack that supports push, pop, top, and retrieving the minimum element in O(1).

```python
class MinStack:
    """Stack that supports getMin() in O(1)."""
    def __init__(self):
        self.stack = []
        self.min_stack = []  # Track mins

    def push(self, val):
        self.stack.append(val)
        # Push to min_stack if it's the new minimum
        if not self.min_stack or val <= self.min_stack[-1]:
            self.min_stack.append(val)

    def pop(self):
        if self.stack:
            val = self.stack.pop()
            if self.min_stack and val == self.min_stack[-1]:
                self.min_stack.pop()
            return val

    def top(self):
        return self.stack[-1] if self.stack else None

    def getMin(self):
        return self.min_stack[-1] if self.min_stack else None

ms = MinStack()
ms.push(-2)
ms.push(0)
ms.push(-3)
print(f"Min: {ms.getMin()}")  # -3
ms.pop()
print(f"Top: {ms.top()}")     # 0
print(f"Min: {ms.getMin()}")  # -2
# Time: O(1) for all operations, Space: O(n)
```

## 5. Stock Span Problem

Given stock prices, find the span (consecutive days price was <= today's price). Classic monotonic stack problem.

```python
def stock_span(prices):
    """Return span for each day."""
    n = len(prices)
    span = [1] * n
    stack = []  # Indices with decreasing prices
    for i in range(n):
        while stack and prices[i] >= prices[stack[-1]]:
            stack.pop()
        # If stack empty, all previous are <= current
        span[i] = i + 1 if not stack else i - stack[-1]
        stack.append(i)
    return span

prices = [100, 80, 60, 70, 60, 75, 85]
print(f"Prices: {prices}")
print(f"Spans: {stock_span(prices)}")
# Time: O(n), Space: O(n)
```

## 6. Largest Rectangle in Histogram

Find the largest rectangle that can be formed in a histogram. For each bar, find the boundary where bars are shorter.

```python
def largest_rectangle_area(heights):
    """Return area of largest rectangle in histogram."""
    stack = []  # Indices of increasing heights
    max_area = 0
    n = len(heights)
    for i in range(n + 1):
        # When i == n, use height 0 to flush stack
        h = heights[i] if i < n else 0
        while stack and h < heights[stack[-1]]:
            height = heights[stack.pop()]
            width = i if not stack else i - stack[-1] - 1
            max_area = max(max_area, height * width)
        stack.append(i)
    return max_area

# Test
print(largest_rectangle_area([2, 1, 5, 6, 2, 3]))  # 10
print(largest_rectangle_area([2, 4]))              # 4
# Time: O(n), Space: O(n)
```

### LeetCode 20 - Valid Parentheses

**Problem:** Given a string s containing just '()[]{}', determine if it is valid.

```python
def isValid(s: str) -> bool:
    pairs = {')': '(', ']': '[', '}': '{'}
    stack = []
    for c in s:
        if c in '([{':
            stack.append(c)
        else:
            if not stack or stack[-1] != pairs[c]:
                return False
            stack.pop()
    return len(stack) == 0

# Test
print(isValid("()"))        # True
print(isValid("()[]{}"))    # True
print(isValid("(]"))        # False
print(isValid("([)]"))      # False
# Time: O(n), Space: O(n)
```

### LeetCode 155 - Min Stack

**Problem:** Design a stack that supports push, pop, top, and retrieving the minimum element in O(1).

```python
class MinStack:
    def __init__(self):
        self.stack = []
        self.min_stack = []

    def push(self, val):
        self.stack.append(val)
        if not self.min_stack or val <= self.min_stack[-1]:
            self.min_stack.append(val)

    def pop(self):
        if self.stack:
            val = self.stack.pop()
            if self.min_stack and val == self.min_stack[-1]:
                self.min_stack.pop()

    def top(self):
        return self.stack[-1]

    def getMin(self):
        return self.min_stack[-1]

# Time: O(1) each, Space: O(n)
```

### LeetCode 496 - Next Greater Element I

**Problem:** Find the next greater element for each element of nums1 in nums2.

```python
def nextGreaterElement(nums1, nums2):
    nge = {}
    stack = []
    for num in nums2:
        while stack and num > stack[-1]:
            nge[stack.pop()] = num
        stack.append(num)
    while stack:
        nge[stack.pop()] = -1
    return [nge[num] for num in nums1]

# Test
print(nextGreaterElement([4,1,2], [1,3,4,2]))  # [-1,3,-1]
print(nextGreaterElement([2,4], [1,2,3,4]))    # [3,-1]
# Time: O(n + m), Space: O(m)
```

### LeetCode 84 - Largest Rectangle in Histogram

```python
def largestRectangleArea(heights):
    stack = []
    max_area = 0
    n = len(heights)
    for i in range(n + 1):
        h = heights[i] if i < n else 0
        while stack and h < heights[stack[-1]]:
            height = heights[stack.pop()]
            width = i if not stack else i - stack[-1] - 1
            max_area = max(max_area, height * width)
        stack.append(i)
    return max_area

# Test
print(largestRectangleArea([2,1,5,6,2,3]))  # 10
print(largestRectangleArea([2,4]))          # 4
# Time: O(n), Space: O(n)
```

|---|---|---|---|
| Balanced Parentheses | Stack + Map | O(n) | Match closing with stack top |
