# Trees in Python

Trees are hierarchical data structures. Binary trees are the most common - each node has at most 2 children.

## 1. Binary Tree Node

A tree node has `val`, `left`, and `right`.

```python
class TreeNode:
    """Binary tree node."""
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def build_sample_tree():
    """Build a sample binary tree:
          1
         / \
        2   3
       / \   \
      4   5   6
    """
    return TreeNode(1,
        TreeNode(2, TreeNode(4), TreeNode(5)),
        TreeNode(3, None, TreeNode(6))
    )

root = build_sample_tree()
print(f"Root: {root.val}")
print(f"Left child: {root.left.val}")
print(f"Right child: {root.right.val}")
```

## 2. Tree Traversals (Recursive)

- **Preorder**: Root -> Left -> Right
- **Inorder**: Left -> Root -> Right
- **Postorder**: Left -> Right -> Root

```python
def inorder(root):
    """Inorder traversal: Left -> Root -> Right."""
    result = []
    def dfs(node):
        if node:
            dfs(node.left)
            result.append(node.val)
            dfs(node.right)
    dfs(root)
    return result

def preorder(root):
    """Preorder traversal: Root -> Left -> Right."""
    result = []
    def dfs(node):
        if node:
            result.append(node.val)
            dfs(node.left)
            dfs(node.right)
    dfs(root)
    return result

def postorder(root):
    """Postorder traversal: Left -> Right -> Root."""
    result = []
    def dfs(node):
        if node:
            dfs(node.left)
            dfs(node.right)
            result.append(node.val)
    dfs(root)
    return result

root = build_sample_tree()
print(f"Inorder: {inorder(root)}")     # [4, 2, 5, 1, 3, 6]
print(f"Preorder: {preorder(root)}")   # [1, 2, 4, 5, 3, 6]
print(f"Postorder: {postorder(root)}") # [4, 5, 2, 6, 3, 1]
# Time: O(n), Space: O(n)
```

## 3. Iterative Traversals

Using explicit stacks instead of recursion.

```python
def inorder_iterative(root):
    """Iterative inorder using stack."""
    result = []
    stack = []
    curr = root
    while curr or stack:
        while curr:
            stack.append(curr)
            curr = curr.left
        curr = stack.pop()
        result.append(curr.val)
        curr = curr.right
    return result

def preorder_iterative(root):
    """Iterative preorder using stack."""
    if not root:
        return []
    result = []
    stack = [root]
    while stack:
        node = stack.pop()
        result.append(node.val)
        if node.right:
            stack.append(node.right)
        if node.left:
            stack.append(node.left)
    return result

root = build_sample_tree()
print(f"Inorder iterative: {inorder_iterative(root)}")
print(f"Preorder iterative: {preorder_iterative(root)}")
# Time: O(n), Space: O(n)
```

## 4. Level Order Traversal (BFS)

Traverse level by level using a queue.

```python
from collections import deque

def level_order(root):
    """Level order traversal returning list of levels."""
    if not root:
        return []
    result = []
    q = deque([root])
    while q:
        level = []
        for _ in range(len(q)):
            node = q.popleft()
            level.append(node.val)
            if node.left:
                q.append(node.left)
            if node.right:
                q.append(node.right)
        result.append(level)
    return result

def level_order_flat(root):
    """Level order as flat list."""
    if not root:
        return []
    result = []
    q = deque([root])
    while q:
        node = q.popleft()
        result.append(node.val)
        if node.left:
            q.append(node.left)
        if node.right:
            q.append(node.right)
    return result

root = build_sample_tree()
print(f"Level order: {level_order(root)}")     # [[1], [2, 3], [4, 5, 6]]
print(f"Level order flat: {level_order_flat(root)}")  # [1, 2, 3, 4, 5, 6]
# Time: O(n), Space: O(n)
```

## 5. Binary Search Tree

BST property: left subtree < node < right subtree (for all nodes).

```python
def bst_insert(root, val):
    """Insert into BST."""
    if not root:
        return TreeNode(val)
    if val < root.val:
        root.left = bst_insert(root.left, val)
    else:
        root.right = bst_insert(root.right, val)
    return root

def bst_search(root, val):
    """Search in BST."""
    if not root or root.val == val:
        return root
    if val < root.val:
        return bst_search(root.left, val)
    return bst_search(root.right, val)

def bst_min(root):
    """Find minimum value in BST."""
    while root and root.left:
        root = root.left
    return root.val if root else None

def bst_max(root):
    """Find maximum value in BST."""
    while root and root.right:
        root = root.right
    return root.val if root else None

def bst_delete(root, val):
    """Delete from BST."""
    if not root:
        return None
    if val < root.val:
        root.left = bst_delete(root.left, val)
    elif val > root.val:
        root.right = bst_delete(root.right, val)
    else:
        if not root.left:
            return root.right
        if not root.right:
            return root.left
        # Two children: replace with inorder successor
        successor = root.right
        while successor.left:
            successor = successor.left
        root.val = successor.val
        root.right = bst_delete(root.right, successor.val)
    return root

bst = None
for v in [5, 3, 7, 2, 4, 6, 8]:
    bst = bst_insert(bst, v)
print(f"Inorder: {inorder(bst)}")
print(f"Min: {bst_min(bst)}, Max: {bst_max(bst)}")
print(f"Search 4: {bst_search(bst, 4) is not None}")
bst = bst_delete(bst, 3)
print(f"After delete 3: {inorder(bst)}")
```

## 6. Tree Height & Diameter

Height = max distance from root to leaf. Diameter = longest path between any two nodes.

```python
def tree_height(root):
    """Return height of tree (max depth)."""
    if not root:
        return 0
    return 1 + max(tree_height(root.left), tree_height(root.right))

def tree_diameter(root):
    """Return diameter of tree."""
    diameter = 0
    def height(node):
        nonlocal diameter
        if not node:
            return 0
        left = height(node.left)
        right = height(node.right)
        diameter = max(diameter, left + right)
        return 1 + max(left, right)
    height(root)
    return diameter

root = build_sample_tree()
print(f"Height: {tree_height(root)}")      # 3
print(f"Diameter: {tree_diameter(root)}")  # 3 (4->5->2->1->3->6 or 4->2->1->3->6)
```

## 7. Validate BST

A binary tree is a valid BST if: left subtree values < node value < right subtree values for all nodes.

```python
def is_valid_bst(root):
    """Check if binary tree is a valid BST."""
    def validate(node, low, high):
        if not node:
            return True
        if not (low < node.val < high):
            return False
        return validate(node.left, low, node.val) and \
               validate(node.right, node.val, high)
    return validate(root, float('-inf'), float('inf'))

# Valid BST
bst = None
for v in [5, 3, 7, 2, 4, 6, 8]:
    bst = bst_insert(bst, v)
print(f"Valid BST: {is_valid_bst(bst)}")

# Invalid: swap 7 and 4
invalid = TreeNode(5, TreeNode(3, TreeNode(2), TreeNode(7)), TreeNode(4))
print(f"Invalid BST: {is_valid_bst(invalid)}")
# Time: O(n), Space: O(n)
```

## 8. Invert Binary Tree

Swap left and right children of every node.

```python
def invert_tree(root):
    """Invert binary tree (mirror)."""
    if not root:
        return None
    root.left, root.right = root.right, root.left
    invert_tree(root.left)
    invert_tree(root.right)
    return root

root = build_sample_tree()
print(f"Original inorder: {inorder(root)}")
invert_tree(root)
print(f"Inverted inorder: {inorder(root)}")  # [6, 3, 1, 5, 2, 4]
# Time: O(n), Space: O(n)
```

## 9. Lowest Common Ancestor

LCA of two nodes p and q is the deepest node that is an ancestor of both.

```python
def lowest_common_ancestor(root, p, q):
    """Find LCA of p and q in binary tree."""
    if not root or root == p or root == q:
        return root
    left = lowest_common_ancestor(root.left, p, q)
    right = lowest_common_ancestor(root.right, p, q)
    if left and right:
        return root
    return left if left else right

def lca_bst(root, p, q):
    """Find LCA in BST (more efficient)."""
    while root:
        if p.val < root.val and q.val < root.val:
            root = root.left
        elif p.val > root.val and q.val > root.val:
            root = root.right
        else:
            return root
    return None

root = build_sample_tree()
p, q = root.left.left, root.left.right  # 4 and 5
lca = lowest_common_ancestor(root, p, q)
print(f"LCA of 4 and 5: {lca.val}")  # 2
# Time: O(n) for BT, O(log n) for BST, Space: O(n)
```

### LeetCode 94 - Binary Tree Inorder Traversal

**Problem:** Return inorder traversal of a binary tree.

```python
def inorderTraversal(root):
    result = []
    stack = []
    curr = root
    while curr or stack:
        while curr:
            stack.append(curr)
            curr = curr.left
        curr = stack.pop()
        result.append(curr.val)
        curr = curr.right
    return result
# Time: O(n), Space: O(n)
```

### LeetCode 102 - Binary Tree Level Order Traversal

**Problem:** Return level order traversal (BFS) of a binary tree.

```python
from collections import deque

def levelOrder(root):
    if not root:
        return []
    result = []
    q = deque([root])
    while q:
        level = []
        for _ in range(len(q)):
            node = q.popleft()
            level.append(node.val)
            if node.left:
                q.append(node.left)
            if node.right:
                q.append(node.right)
        result.append(level)
    return result
# Time: O(n), Space: O(n)
```

### LeetCode 98 - Validate Binary Search Tree

```python
def isValidBST(root):
    def validate(node, low, high):
        if not node:
            return True
        if not (low < node.val < high):
            return False
        return validate(node.left, low, node.val) and \
               validate(node.right, node.val, high)
    return validate(root, float('-inf'), float('inf'))
# Time: O(n), Space: O(n)
```

### LeetCode 236 - Lowest Common Ancestor of a Binary Tree

**Problem:** Find LCA of two nodes in a binary tree.

```python
def lowestCommonAncestor(root, p, q):
    if not root or root == p or root == q:
        return root
    left = lowestCommonAncestor(root.left, p, q)
    right = lowestCommonAncestor(root.right, p, q)
    if left and right:
        return root
    return left if left else right
# Time: O(n), Space: O(n)
```

## Summary

| Traversal | Order | Recursive | Iterative |
|---|---|---|---|
| Preorder | Root -> L -> R | Simple | Stack (push right first) |
| Inorder | L -> Root -> R | Simple | Stack (go left completely) |
| Postorder | L -> R -> Root | Simple | Two stacks / reversed |
| Level Order | BFS | Queue | Queue |

| Operation | BST Time | BT Time |
|---|---|---|
| Search | O(log n) avg, O(n) worst | O(n) |
| Insert | O(log n) avg, O(n) worst | O(1) at given node |
| Delete | O(log n) avg, O(n) worst | O(1) with reference |
| LCA | O(log n) avg | O(n) |
| Validate | O(n) | O(n) |
