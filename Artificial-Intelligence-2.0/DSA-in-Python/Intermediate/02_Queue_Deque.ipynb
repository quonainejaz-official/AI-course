## 1. Queue Implementations

Using `list` as a queue is inefficient (pop(0) is O(n)). Use `collections.deque` for O(1) operations from both ends.

```python
from collections import deque

# Efficient queue using deque
q = deque()
q.append(1)     # enqueue (right)
q.append(2)
q.append(3)
print(f"Queue: {q}")
print(f"Dequeue: {q.popleft()}")  # 1 - O(1)
print(f"Front: {q[0]}")           # 2 - peek
print(f"Size: {len(q)}")
print(f"Is empty: {len(q) == 0}")

# Thread-safe queue
from queue import Queue
tq = Queue()
tq.put(10)
tq.put(20)
print(f"Thread-safe queue get: {tq.get()}")  # 10
print(f"Queue empty: {tq.empty()}")
```

## 2. Deque for Double-Ended Operations

Deque supports O(1) append/pop from both ends - useful for palindrome checking, sliding window, etc.

```python
from collections import deque

d = deque()

# Add to both ends
d.append('right')       # Add to right
d.appendleft('left')    # Add to left
print(f"After appends: {d}")

# Remove from both ends
print(f"Pop right: {d.pop()}")
print(f"Pop left: {d.popleft()}")

# Extend both ends
d.extend([1, 2, 3])        # Right extend
d.extendleft([-1, 0])       # Left extend (reversed order)
print(f"After extends: {d}")

# Rotation
d.rotate(1)   # Move everything right by 1
print(f"After rotate(1): {d}")
d.rotate(-2)  # Move everything left by 2
print(f"After rotate(-2): {d}")

# Max length deque
d2 = deque(maxlen=3)
d2.extend([1, 2, 3, 4, 5])  # Only keeps last 3
print(f"Maxlen deque: {d2}")  # [3, 4, 5]
```

## 3. Sliding Window Maximum

Given an array and window size k, find the maximum in each window. Use a deque to store **indices** of elements in decreasing order.

```python
from collections import deque

def max_sliding_window(nums, k):
    """Return maximum in each sliding window of size k."""
    if not nums or k <= 0:
        return []
    dq = deque()  # Stores indices, values in decreasing order
    result = []
    for i, num in enumerate(nums):
        # Remove indices outside window
        while dq and dq[0] <= i - k:
            dq.popleft()
        # Remove smaller values (maintain decreasing order)
        while dq and nums[dq[-1]] <= num:
            dq.pop()
        dq.append(i)
        # Window is ready; record max
        if i >= k - 1:
            result.append(nums[dq[0]])
    return result

nums = [1, 3, -1, -3, 5, 3, 6, 7]
print(f"Array: {nums}")
print(f"Sliding window max (k=3): {max_sliding_window(nums, 3)}")
# Time: O(n), Space: O(k)
```

## 4. First Negative Number in Every Window

Find the first negative number in each window of size k. If none, return 0.

```python
from collections import deque

def first_negative_in_window(nums, k):
    """Return first negative in each window."""
    dq = deque()  # Indices of negative numbers
    result = []
    for i, num in enumerate(nums):
        if num < 0:
            dq.append(i)
        # Remove indices outside window
        while dq and dq[0] <= i - k:
            dq.popleft()
        # Window ready
        if i >= k - 1:
            result.append(nums[dq[0]] if dq else 0)
    return result

nums = [12, -1, -7, 8, -15, 30, 16, 28]
print(f"Array: {nums}")
print(f"First negative in window (k=3): {first_negative_in_window(nums, 3)}")
# Time: O(n), Space: O(k)
```

## 5. Circular Queue

A circular queue (ring buffer) reuses space by wrapping around. Implemented with an array and front/rear pointers.

```python
class CircularQueue:
    """Circular queue with fixed capacity."""
    def __init__(self, k):
        self.data = [None] * k
        self.capacity = k
        self.front = 0
        self.rear = -1
        self.size = 0

    def enqueue(self, val):
        if self.is_full():
            return False
        self.rear = (self.rear + 1) % self.capacity
        self.data[self.rear] = val
        self.size += 1
        return True

    def dequeue(self):
        if self.is_empty():
            return False
        self.front = (self.front + 1) % self.capacity
        self.size -= 1
        return True

    def front_val(self):
        return self.data[self.front] if not self.is_empty() else -1

    def rear_val(self):
        return self.data[self.rear] if not self.is_empty() else -1

    def is_empty(self):
        return self.size == 0

    def is_full(self):
        return self.size == self.capacity

cq = CircularQueue(3)
print(f"Enqueue 1: {cq.enqueue(1)}")
print(f"Enqueue 2: {cq.enqueue(2)}")
print(f"Enqueue 3: {cq.enqueue(3)}")
print(f"Enqueue 4 (should fail): {cq.enqueue(4)}")
print(f"Front: {cq.front_val()}")
print(f"Rear: {cq.rear_val()}")
print(f"Dequeue: {cq.dequeue()}")
print(f"Enqueue 4: {cq.enqueue(4)}")
print(f"Front: {cq.front_val()}")
```

## 6. Priority Queue Introduction

Python's `heapq` module implements a min-heap. Priority queues are covered in depth in the Heap notebook.

```python
import heapq

# Min-heap (default)
pq = []
heapq.heappush(pq, 30)
heapq.heappush(pq, 10)
heapq.heappush(pq, 20)
print(f"Heap: {pq}")
print(f"Pop min: {heapq.heappop(pq)}")  # 10
print(f"Peek min: {pq[0]}")            # 20

# Max-heap via negation
max_pq = []
for x in [3, 1, 4, 1, 5]:
    heapq.heappush(max_pq, -x)
print(f"Pop max: {-heapq.heappop(max_pq)}")  # 5
```

### LeetCode 239 - Sliding Window Maximum

**Problem:** Return max in each sliding window of size k. Must be O(n).

```python
from collections import deque

def maxSlidingWindow(nums, k):
    dq = deque()
    result = []
    for i, num in enumerate(nums):
        while dq and dq[0] <= i - k:
            dq.popleft()
        while dq and nums[dq[-1]] <= num:
            dq.pop()
        dq.append(i)
        if i >= k - 1:
            result.append(nums[dq[0]])
    return result

# Test
print(maxSlidingWindow([1,3,-1,-3,5,3,6,7], 3))  # [3,3,5,5,6,7]
print(maxSlidingWindow([1], 1))                  # [1]
# Time: O(n), Space: O(k)
```

### LeetCode 933 - Number of Recent Calls

**Problem:** Count the number of requests in the time range [t-3000, t]. Use a queue.

```python
from collections import deque

class RecentCounter:
    def __init__(self):
        self.q = deque()

    def ping(self, t):
        self.q.append(t)
        while self.q[0] < t - 3000:
            self.q.popleft()
        return len(self.q)

rc = RecentCounter()
print(rc.ping(1))     # 1
print(rc.ping(100))   # 2
print(rc.ping(3001))  # 3
print(rc.ping(3002))  # 4
# Time: O(1) amortized per call, Space: O(n)
```
