# Handling Missing Data

**Learning Objectives:**
- Detect missing values with `isna()`, `isnull()`, `notna()`
- Remove missing values with `dropna()`
- Fill missing values with `fillna()` (value, method, ffill, bfill)
- Interpolate missing values with `interpolate()`
- Replace arbitrary values with `replace()`
- Understand NaN behavior in math operations

## 1. Setup – Dataset with Missing Values

We'll simulate a real-world dataset that has `NaN` (Not a Number) entries.

```python
import pandas as pd
import numpy as np

np.random.seed(42)
data = {
    "ID": range(1, 11),
    "Age": [25, np.nan, 35, 42, np.nan, 29, 31, np.nan, 27, 38],
    "Salary": [50000, 60000, np.nan, 80000, 75000, np.nan, 82000, 65000, 72000, np.nan],
    "Department": ["IT", "HR", "IT", np.nan, "Finance", "IT", "HR", "Finance", np.nan, "IT"]
}
df = pd.DataFrame(data)
print(df)
```

## 2. Detecting Missing Values

- `isna()` / `isnull()` – return `True` where value is missing (aliases)
- `notna()` – inverse of `isna()`

```python
# isna() shows True for missing values
print("isna():")
print(df.isna())

# Count missing per column
print("\nMissing count per column:")
print(df.isna().sum())

# notna() – the opposite
print("\nnotna():")
print(df.notna())
```

## 3. Dropping Missing Values – `dropna()`

Removes rows (default) or columns that contain NaN.

```python
# Drop rows with any NaN
df_drop_any = df.dropna()
print("dropna() – rows with any NaN removed:")
print(df_drop_any)

# Drop rows where all values are NaN
df_drop_all = df.dropna(how="all")
print("\ndropna(how='all') – only rows where ALL NaN are removed:")
print(df_drop_all)

# Drop columns with any NaN
df_drop_col = df.dropna(axis=1)
print("\ndropna(axis=1) – columns with NaN removed:")
print(df_drop_col)

# Drop rows with fewer than 3 non-NA values
df_thresh = df.dropna(thresh=3)
print("\ndropna(thresh=3) – keep rows with >= 3 non-NA:")
print(df_thresh)
```

## 4. Filling Missing Values – `fillna()`

Replace NaN with a constant value or use a filling method.

```python
# Fill with a constant
df_fill_const = df.copy()
df_fill_const["Department"] = df_fill_const["Department"].fillna("Unknown")
print("Department NaN filled with 'Unknown':")
print(df_fill_const)

# Fill numeric columns with mean
df_fill_mean = df.copy()
df_fill_mean["Age"] = df_fill_mean["Age"].fillna(df["Age"].mean())
df_fill_mean["Salary"] = df_fill_mean["Salary"].fillna(df["Salary"].mean())
print("\nNumeric NaN filled with column mean:")
print(df_fill_mean)
```

```python
# Forward fill (ffill): propagate last valid observation forward
df_ffill = df.copy()
df_ffill["Department"] = df_ffill["Department"].ffill()
print("Forward fill on Department:")
print(df_ffill)

# Backward fill (bfill): use next valid observation
df_bfill = df.copy()
df_bfill["Department"] = df_bfill["Department"].bfill()
print("\nBackward fill on Department:")
print(df_bfill)
```

## 5. Interpolation – `interpolate()`

Fills NaN by interpolating between valid values (useful for time series).

```python
# Create a Series with a gap
s = pd.Series([10, 20, np.nan, np.nan, 50, 60])
print("Original:", s.values)

s_interp = s.interpolate()
print("Interpolated:", s_interp.values)
```

## 6. Replacing Values – `replace()`

Replace arbitrary values (not just NaN) with something else.

```python
df_rep = df.copy()

# Replace specific value
df_rep["Department"] = df_rep["Department"].replace("HR", "Human Resources")

# Replace NaN with a string (alternative to fillna)
df_rep["Department"] = df_rep["Department"].fillna("Unknown")

print(df_rep)
```

## 7. Math Operations with NaN

Most operations propagate NaN. Use `skipna=True` (default) to exclude NaN.

```python
print("Sum of Age:", df["Age"].sum())          # NaN excluded
print("Mean of Age:", df["Age"].mean())        # NaN excluded
print("Cumsum:", df["Age"].cumsum().values)    # NaN propagates
```

## 8. Practice Problems

```python
# Problem 1: In the df above, fill ALL missing numeric values with the
#            median of their respective column. Print the result.
df_med = df.copy()
for col in ["Age", "Salary"]:
    df_med[col] = df_med[col].fillna(df_med[col].median())
print("Filled with median:")
print(df_med)
```

```python
# Problem 2: Drop only the rows where BOTH Age and Salary are NaN.
#            Keep other rows even if one column is missing.
df_subset = df.dropna(subset=["Age", "Salary"], how="all")
print(df_subset)
```

```python
# Problem 3: Use interpolate() on a Series with 3 consecutive NaN
#            values in the middle. Observe the linear fill.
s = pd.Series([1.0, np.nan, np.nan, np.nan, 5.0])
print("Interpolated:", s.interpolate().values)
```

## 9. Summary

| Method | Purpose |
|--------|---------|
| `isna()` / `isnull()` | Detect missing values |
| `notna()` | Detect non-missing values |
| `dropna()` | Remove rows/cols with NaN |
| `fillna(value)` | Replace NaN with a value |
| `fillna(method='ffill')` | Forward fill |
| `fillna(method='bfill')` | Backward fill |
| `interpolate()` | Fill by linear interpolation |
| `replace(old, new)` | Replace arbitrary values |
