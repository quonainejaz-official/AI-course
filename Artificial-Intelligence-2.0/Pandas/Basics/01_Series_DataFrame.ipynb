## 1. What is Pandas?

Pandas is a Python library for **data manipulation and analysis**. It provides two primary data structures:

- **Series** – 1D labeled array (like a column)
- **DataFrame** – 2D labeled data structure (like a spreadsheet or SQL table)

> Pandas is built on top of **NumPy**, so we usually import both.

````python
# Import libraries
import pandas as pd
import numpy as np

print(f"Pandas version: {pd.__version__}")
print(f"NumPy version : {np.__version__}")
````

## 2. Creating a Series

A `Series` is a one-dimensional array with **labels** (the index).

````python
# From a list
s_list = pd.Series([10, 20, 30, 40], name="scores")
print("From list:\n", s_list, sep="")

# From a dict (keys become index)
s_dict = pd.Series({"a": 100, "b": 200, "c": 300})
print("\nFrom dict:\n", s_dict, sep="")

# From a scalar (repeated to fill length)
s_scalar = pd.Series(5.0, index=range(4))
print("\nFrom scalar:\n", s_scalar, sep="")

# From a NumPy ndarray
arr = np.random.randn(5)
s_ndarray = pd.Series(arr, index=["a", "b", "c", "d", "e"])
print("\nFrom ndarray:")
print(s_ndarray)
````

## 3. Creating a DataFrame

A `DataFrame` is a 2D structure with **rows** and **columns**.

````python
# From a dict of lists/Series (keys become columns)
df_dict = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [25, 30, 35],
    "Salary": [70000, 80000, 90000]
})
print("From dict of lists:")
print(df_dict)

# From a list of dicts (each dict is a row)
df_list_of_dicts = pd.DataFrame([
    {"City": "NYC", "Temp": 32},
    {"City": "LA", "Temp": 75},
    {"City": "Chicago", "Temp": 20}
])
print("\nFrom list of dicts:")
print(df_list_of_dicts)

# From a NumPy array
df_np = pd.DataFrame(
    np.random.randint(0, 100, size=(4, 3)),
    columns=["Math", "Science", "English"]
)
print("\nFrom NumPy array:")
print(df_np)
````

## 4. Reading from a CSV file

Pandas can read CSV, Excel, JSON, SQL, and many other formats (covered in depth later).

````python
# Create a small CSV in-memory using StringIO and read it
from io import StringIO

csv_data = """product,price,quantity
Widget,10.99,100
Gadget,24.99,50
Doohickey,5.49,200
"""
df_csv = pd.read_csv(StringIO(csv_data))
print("From CSV:\n", df_csv, sep="")
````

## 5. Exploring a DataFrame

Once data is loaded, use these methods for a quick overview.

````python
# Create a larger example DataFrame
np.random.seed(42)
df = pd.DataFrame({
    "A": np.random.randn(100),
    "B": np.random.randint(0, 100, 100),
    "C": np.random.choice(["X", "Y", "Z"], 100),
    "D": np.random.uniform(10, 20, 100)
})

# First / last rows
print("head():\n", df.head(3))
print("\ntail():\n", df.tail(3))
````

````python
# Structural info
print("Shape :", df.shape)
print("Columns:", df.columns.tolist())
print("\nData types:")
print(df.dtypes)

print("\n--- info() ---")
df.info()
````

````python
# Summary statistics
print("describe():")
print(df.describe())
````

## 6. The Index

Every Series and DataFrame has an **index** – a label for each row.

````python
print("Default index:", df.index[:5])

# Set a custom index
df_custom = df.copy()
df_custom.index = [f"row_{i}" for i in range(len(df_custom))]
print("\nCustom index:")
print(df_custom.head(3))
````

## 8. Key Takeaways

| Concept | What It Does |
|---------|-------------|
| `Series` | 1D labeled array |
| `DataFrame` | 2D labeled table |
| `head(n)` / `tail(n)` | View first / last *n* rows |
| `info()` | Column dtypes + non-null counts |
| `describe()` | Summary statistics for numeric columns |
| `dtypes` | Data type of each column |
| `shape` | Tuple `(rows, columns)` |
| `columns` | Column labels |
| `index` | Row labels |
