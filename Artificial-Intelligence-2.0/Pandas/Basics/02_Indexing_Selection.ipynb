# Indexing and Selection

**Learning Objectives:**
- Use `[]` operator for column / row selection
- Use `.loc` (label-based) and `.iloc` (integer-based) indexing
- Access individual values with `.at` / `.iat`
- Boolean indexing with conditions
- Use `isin()`, `between()`, and `query()`
- Set values and update data conditionally

## 1. Setup – Sample DataFrame

We'll use a sales dataset throughout this notebook.

```python
import pandas as pd
import numpy as np

np.random.seed(42)
df = pd.DataFrame({
    "OrderID": range(101, 111),
    "Product": np.random.choice(["Laptop", "Phone", "Tablet", "Monitor"], 10),
    "Price": np.random.randint(200, 1500, 10),
    "Quantity": np.random.randint(1, 5, 10),
    "City": np.random.choice(["NYC", "LA", "Chicago", "Houston"], 10)
})
print(df)
```

## 2. The `[]` Operator

- `df["column"]` selects a column (returns a Series)
- `df[["col1", "col2"]]` selects multiple columns (returns a DataFrame)
- `df[start:stop]` slices rows by integer position

```python
# Single column
print("Single column (Series):")
print(df["Product"].head())

# Multiple columns
print("\nMultiple columns (DataFrame):")
print(df[["Product", "Price"]].head())

# Row slicing
print("\nFirst 3 rows via []:")
print(df[:3])
```

## 3. `.loc` – Label-Based Indexing

Syntax: `df.loc[row_labels, column_labels]`

- Accepts single labels, lists, slices, or boolean arrays
- **Includes** the endpoint of a slice

```python
# Set a meaningful index first
df_idx = df.set_index("OrderID")
print(df_idx)

# Select a single row by label
print("\nRow with label 104:")
print(df_idx.loc[104])

# Select rows and specific columns
print("\nRows 101-103, columns Product and Price:")
print(df_idx.loc[101:103, ["Product", "Price"]])
```

## 4. `.iloc` – Integer-Based Indexing

Syntax: `df.iloc[row_positions, column_positions]`

- Uses 0-based integer positions
- **Excludes** the endpoint of a slice

```python
# iloc[0:3, 0:2]
print("iloc[0:3, 0:2]:")
print(df.iloc[0:3, 0:2])

# All rows, columns at positions 1 and 3
print("\nAll rows, cols [1, 3]:")
print(df.iloc[:, [1, 3]])

# Single cell
print("\nRow 2, col 4 (Quantity):", df.iloc[2, 4])
```

## 5. `.at` / `.iat` – Fast Scalar Access

Use `.at` (label-based) and `.iat` (integer-based) for **single values**.

```python
# at: label-based
print("Price for row index 2:", df.at[2, "Price"])

# iat: integer-based (row 5, col 2 = Price)
print("Value at [5, 2]:", df.iat[5, 2])
```

## 6. Boolean Indexing

Pass a boolean Series to select rows where the condition is `True`.

```python
# Filter: Price > 800
print("Orders with Price > 800:")
print(df[df["Price"] > 800])

# Multiple conditions: use & (and), | (or), ~ (not) – wrap each in ()
print("\nLaptops with Price > 1000:")
print(df[(df["Product"] == "Laptop") & (df["Price"] > 1000)])
```

## 7. `isin()` and `between()`

Convenience methods for common filtering patterns.

```python
# isin(): filter rows where value is in a set
print("Orders in NYC or LA:")
print(df[df["City"].isin(["NYC", "LA"])])

# between(): inclusive range check
print("\nOrders with Price between 500 and 1000:")
print(df[df["Price"].between(500, 1000)])
```

## 8. `query()` – Expressive String Conditions

Write filters as a string expression – often cleaner for complex conditions.

```python
# Equivalent to boolean indexing above
print("Using query():")
print(df.query("Price > 800"))

print("\nComplex query:")
print(df.query("Product == 'Laptop' and Price > 500"))

# Use @ to reference Python variables
min_price = 600
print("\nUsing @variable:")
print(df.query("Price > @min_price"))
```

## 9. Setting Values Conditionally

Combine indexing with assignment to update data.

```python
# Create a copy to modify
df_mod = df.copy()

# Set a new column based on condition
df_mod["Discount"] = 0.0
df_mod.loc[df_mod["Price"] > 1000, "Discount"] = 0.1

print("With discount applied:")
print(df_mod[["Product", "Price", "Discount"]])
```

## 11. Summary

| Method | When to Use |
|--------|------------|
| `[]` | Quick column / row slice |
| `.loc[]` | Label-based, explicit rows and cols |
| `.iloc[]` | Integer-position-based |
| `.at[]` / `.iat[]` | Fast single-value access |
| Boolean Series | Any conditional filter |
| `isin()`, `between()` | Membership / range checks |
| `query()` | Clean string-based conditions |
