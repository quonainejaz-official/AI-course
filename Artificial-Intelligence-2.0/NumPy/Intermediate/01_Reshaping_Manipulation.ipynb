# Reshaping & Array Manipulation

Reshaping, stacking, splitting, and transposing are essential skills for preparing data for machine learning models.

**Key functions:** `reshape, resize, flatten, ravel, transpose, T, swapaxes, moveaxis, concatenate, vstack, hstack, dstack, split, vsplit, hsplit, newaxis, expand_dims, squeeze`

## reshape() and resize()

```python
import numpy as np

arr = np.arange(12)
print("Original:", arr, "shape:", arr.shape)

# reshape (returns a view if possible, otherwise a copy)
r1 = arr.reshape(3, 4)
print("\nreshape(3,4):\n", r1)

# -1 infers the dimension automatically
r2 = arr.reshape(2, -1)
print("\nreshape(2,-1):\n", r2)

# resize (modifies in-place or returns new array)
r3 = np.resize(arr, (2, 4))
print("\nresize(2,4):\n", r3)

# resize can repeat data if target size > original
r4 = np.resize(arr, (3, 6))
print("\nresize(3,6) - repeats data:\n", r4)
```

## flatten() vs ravel()

- `flatten()` always returns a **copy**
- `ravel()` returns a **view** (if possible) — more memory efficient

```python
mat = np.array([[1, 2, 3], [4, 5, 6]])

flat = mat.flatten()
rav = mat.ravel()

print("flatten:", flat)
print("ravel:", rav)

# Modifying flat does NOT change original
flat[0] = 999
print("\nAfter modifying flat[0]=999:")
print("mat:\n", mat)
print("flat:", flat)

# Modifying ravel DOES change original (it's a view)
rav[0] = 999
print("\nAfter modifying rav[0]=999:")
print("mat:\n", mat)
```

## Transpose, swapaxes, moveaxis

```python
mat = np.array([[1, 2, 3], [4, 5, 6]])
print("Original:", mat.shape)
print("mat:\n", mat)

# Transpose (2,3) -> (3,2)
print("\nmat.T:\n", mat.T)
print("mat.T shape:", mat.T.shape)

# 3D example
arr3d = np.arange(24).reshape(2, 3, 4)
print("\n3D arr shape:", arr3d.shape)

# swapaxes: swap two axes
swapped = arr3d.swapaxes(0, 2)
print("swapaxes(0,2) shape:", swapped.shape)

# moveaxis: move axis to new position
moved = np.moveaxis(arr3d, 0, -1)
print("moveaxis(0,-1) shape:", moved.shape)
```

## Concatenate, vstack, hstack, dstack

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])

print("a:\n", a)
print("b:\n", b)

# Concatenate along axis 0 (rows)
c0 = np.concatenate([a, b], axis=0)
print("\nconcatenate axis=0:\n", c0)

# Concatenate along axis 1 (columns)
c1 = np.concatenate([a, b], axis=1)
print("concatenate axis=1:\n", c1)

# vstack (vertical stack = axis 0)
vs = np.vstack([a, b])
print("\nvstack:\n", vs)

# hstack (horizontal stack = axis 1)
hs = np.hstack([a, b])
print("hstack:\n", hs)

# dstack (depth stack = axis 2)
ds = np.dstack([a, b])
print("\ndstack shape:", ds.shape)
print("dstack:\n", ds)
```

## Splitting Arrays

```python
arr = np.arange(16).reshape(4, 4)
print("Original:\n", arr)

# Split into 2 equal parts along axis 0
s0 = np.split(arr, 2, axis=0)
print("\nsplit(arr, 2, axis=0):")
for i, s in enumerate(s0):
    print(f"  Part {i}:\n{s}")

# vsplit (split along axis 0)
vsp = np.vsplit(arr, 2)
print("\nvsplit(arr, 2):")
for i, s in enumerate(vsp):
    print(f"  Part {i}:\n{s}")

# hsplit (split along axis 1)
hsp = np.hsplit(arr, 2)
print("\nhsplit(arr, 2):")
for i, s in enumerate(hsp):
    print(f"  Part {i}:\n{s}")
```

## newaxis, expand_dims, squeeze

Add or remove dimensions of size 1.

```python
arr = np.array([1, 2, 3])  # shape (3,)

# Add new axis at position 0 -> shape (1, 3)
row_vec = arr[np.newaxis, :]
print("newaxis at 0:", row_vec.shape)

# Add new axis at position 1 -> shape (3, 1)
col_vec = arr[:, np.newaxis]
print("newaxis at 1:", col_vec.shape)
print("col_vec:\n", col_vec)

# expand_dims (equivalent)
exp = np.expand_dims(arr, axis=1)
print("\nexpand_dims(arr, 1):", exp.shape)

# squeeze - removes dimensions of size 1
arr_2d = np.array([[[1, 2, 3]]])  # shape (1, 1, 3)
print("\nBefore squeeze:", arr_2d.shape)
squeezed = np.squeeze(arr_2d)
print("After squeeze:", squeezed.shape)  # (3,)
```

## Problem 1: Reshape Images as Flat Vectors

A batch of grayscale images (4 images, each 3x3) — flatten each image into a vector.

```python
# Simulate batch of 4 grayscale images: (4, 3, 3)
images = np.random.randint(0, 256, (4, 3, 3), dtype=np.uint8)
print("Image batch shape:", images.shape)
print("First image:\n", images[0])

# Flatten each image (keep batch dimension)
flat = images.reshape(4, -1)
print("\nFlattened shape:", flat.shape)
print("First flattened image:", flat[0])

# Restore: reshape back
restored = flat.reshape(-1, 3, 3)
print("\nRestored shape:", restored.shape)
print("Match:", np.all(images == restored))

# Practical: ML models expect flat feature vectors
print(f"\nEach image has {flat.shape[1]} features")
```

## Problem 2: Split and Merge Datasets

Split a dataset into train/test (80/20), then merge predictions back.

```python
# Full dataset: 100 samples, 5 features
data = np.arange(500).reshape(100, 5)
labels = np.arange(100)

# 80/20 split
n_train = 80
X_train, X_test = np.split(data, [n_train], axis=0)
y_train, y_test = np.split(labels, [n_train], axis=0)

print(f"X_train: {X_train.shape}, y_train: {y_train.shape}")
print(f"X_test:  {X_test.shape}, y_test:  {y_test.shape}")

# Simulate predictions
y_pred = np.random.randint(0, 100, size=y_test.shape)

# Stack predictions with original test data
augmented = np.column_stack([X_test, y_pred])
print("\nTest data with predictions:\n", augmented[:3])
print("Augmented shape:", augmented.shape)
```

## Problem 3: Broadcasting with newaxis

Use `newaxis` to compute pairwise differences efficiently.

```python
points = np.array([[1, 2],
                   [4, 5],
                   [7, 8]])
print("Points:\n", points)

# Compute pairwise differences using broadcasting
# (3,1,2) - (1,3,2) -> (3,3,2)
diffs = points[:, np.newaxis, :] - points[np.newaxis, :, :]
print("\nPairwise differences:")
print(diffs.shape)
print(diffs)

# Pairwise Euclidean distances
dist = np.sqrt(np.sum(diffs**2, axis=-1))
print("\nPairwise distances:\n", np.round(dist, 2))
```

## Summary

- `reshape` changes shape (view when possible), `resize` can duplicate data
- `flatten` = copy, `ravel` = view (preferred for memory)
- `T` / `transpose` flips axes; `swapaxes` / `moveaxis` for arbitrary reordering
- `concatenate`, `vstack`, `hstack`, `dstack` join arrays
- `split`, `vsplit`, `hsplit` divide arrays
- `newaxis` / `expand_dims` / `squeeze` manage singleton dimensions
