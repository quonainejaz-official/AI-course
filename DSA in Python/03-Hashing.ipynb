# Hashing - DSA Patterns

Hashing uses hash functions to map data to fixed-size values. It enables O(1) average time for insert, delete, and search.

## 1. Hash Map Implementation

```python
class HashMap:
    """Simple hash map implementation using chaining."""
    
    def __init__(self, size=10):
        self.size = size
        self.table = [[] for _ in range(size)]
    
    def _hash(self, key):
        """Hash function."""
        return hash(key) % self.size
    
    def put(self, key, value):
        """Insert or update key-value pair."""
        index = self._hash(key)
        for i, (k, v) in enumerate(self.table[index]):
            if k == key:
                self.table[index][i] = (key, value)
                return
        self.table[index].append((key, value))
    
    def get(self, key):
        """Get value by key."""
        index = self._hash(key)
        for k, v in self.table[index]:
            if k == key:
                return v
        raise KeyError(key)
    
    def delete(self, key):
        """Delete key-value pair."""
        index = self._hash(key)
        for i, (k, v) in enumerate(self.table[index]):
            if k == key:
                del self.table[index][i]
                return
        raise KeyError(key)
    
    def __contains__(self, key):
        """Check if key exists."""
        index = self._hash(key)
        return any(k == key for k, v in self.table[index])


# Example
hm = HashMap()
hm.put('apple', 1)
hm.put('banana', 2)
hm.put('cherry', 3)
print(f"apple: {hm.get('apple')}")
print(f"banana: {hm.get('banana')}")
print(f"'grape' in map: {'grape' in hm}")
print(f"'apple' in map: {'apple' in hm}")
```

## 2. Count Distinct Elements

```python
def count_distinct(arr):
    """Count distinct elements using hash set.
    Time: O(n), Space: O(n)"""
    return len(set(arr))


def count_distinct_elements_k_window(arr, k):
    """Count distinct elements in every window of size k.
    Time: O(n)"""
    from collections import Counter
    
    result = []
    window = Counter()
    
    for i in range(len(arr)):
        window[arr[i]] += 1
        
        if i >= k:
            left = arr[i - k]
            window[left] -= 1
            if window[left] == 0:
                del window[left]
        
        if i >= k - 1:
            result.append(len(window))
    
    return result


# Examples
arr = [1, 2, 2, 1, 3, 1, 4]
print(f"Array: {arr}")
print(f"Distinct elements: {count_distinct(arr)}")
print(f"Distinct in windows of size 3: {count_distinct_elements_k_window(arr, 3)}")
```

## 3. Two Sum and Variants

```python
def two_sum(arr, target):
    """Find two numbers that add up to target (LeetCode 1).
    Time: O(n)"""
    num_map = {}
    
    for i, num in enumerate(arr):
        complement = target - num
        if complement in num_map:
            return [num_map[complement], i]
        num_map[num] = i
    
    return []


def three_sum(arr):
    """Find all unique triplets that sum to zero (LeetCode 15).
    Time: O(n^2)"""
    arr.sort()
    result = []
    
    for i in range(len(arr) - 2):
        if i > 0 and arr[i] == arr[i - 1]:
            continue
        
        left, right = i + 1, len(arr) - 1
        while left < right:
            total = arr[i] + arr[left] + arr[right]
            if total < 0:
                left += 1
            elif total > 0:
                right -= 1
            else:
                result.append([arr[i], arr[left], arr[right]])
                while left < right and arr[left] == arr[left + 1]:
                    left += 1
                while left < right and arr[right] == arr[right - 1]:
                    right -= 1
                left += 1
                right -= 1
    
    return result


# Examples
arr = [2, 7, 11, 15]
target = 9
print(f"Two Sum for {arr} with target {target}: {two_sum(arr, target)}")

arr = [-1, 0, 1, 2, -1, -4]
print(f"Three Sum for {arr}: {three_sum(arr)}")
```

## 4. Frequency Counting

```python
from collections import Counter, defaultdict


def frequency_count(arr):
    """Count frequency of each element."""
    return dict(Counter(arr))


def top_k_frequent(arr, k):
    """Find k most frequent elements (LeetCode 347).
    Time: O(n) using bucket sort"""
    count = Counter(arr)
    
    # Bucket sort
    buckets = [[] for _ in range(len(arr) + 1)]
    for num, freq in count.items():
        buckets[freq].append(num)
    
    result = []
    for freq in range(len(buckets) - 1, 0, -1):
        for num in buckets[freq]:
            result.append(num)
            if len(result) == k:
                return result
    
    return result


def first_non_repeating(s):
    """Find first non-repeating character (LeetCode 387).
    Time: O(n)"""
    count = Counter(s)
    
    for i, char in enumerate(s):
        if count[char] == 1:
            return i
    return -1


# Examples
arr = [1, 1, 2, 2, 2, 3]
print(f"Frequency count: {frequency_count(arr)}")
print(f"Top 2 frequent in {arr}: {top_k_frequent(arr, 2)}")
print(f"First non-repeating in 'leetcode': {first_non_repeating('leetcode')}")
```

## 5. Advanced Hashing Problems

```python
def subarray_sum_equals_k(arr, k):
    """Count subarrays with sum equal to k (LeetCode 560).
    Time: O(n)"""
    count = 0
    prefix_sum = 0
    sum_map = {0: 1}
    
    for num in arr:
        prefix_sum += num
        if prefix_sum - k in sum_map:
            count += sum_map[prefix_sum - k]
        sum_map[prefix_sum] = sum_map.get(prefix_sum, 0) + 1
    
    return count


def longest_consecutive_sequence(arr):
    """Find length of longest consecutive sequence (LeetCode 128).
    Time: O(n)"""
    num_set = set(arr)
    max_length = 0
    
    for num in num_set:
        if num - 1 not in num_set:
            current_num = num
            current_length = 1
            
            while current_num + 1 in num_set:
                current_num += 1
                current_length += 1
            
            max_length = max(max_length, current_length)
    
    return max_length


def four_sum(arr, target):
    """Find all unique quadruplets that sum to target (LeetCode 18).
    Time: O(n^3)"""
    arr.sort()
    result = []
    
    for i in range(len(arr) - 3):
        if i > 0 and arr[i] == arr[i - 1]:
            continue
        for j in range(i + 1, len(arr) - 2):
            if j > i + 1 and arr[j] == arr[j - 1]:
                continue
            
            left, right = j + 1, len(arr) - 1
            while left < right:
                total = arr[i] + arr[j] + arr[left] + arr[right]
                if total < target:
                    left += 1
                elif total > target:
                    right -= 1
                else:
                    result.append([arr[i], arr[j], arr[left], arr[right]])
                    while left < right and arr[left] == arr[left + 1]:
                        left += 1
                    while left < right and arr[right] == arr[right - 1]:
                        right -= 1
                    left += 1
                    right -= 1
    
    return result


# Examples
arr = [1, 1, 1]
print(f"Subarrays with sum 2: {subarray_sum_equals_k(arr, 2)}")

arr = [100, 4, 200, 1, 3, 2]
print(f"Longest consecutive sequence: {longest_consecutive_sequence(arr)}")
```
