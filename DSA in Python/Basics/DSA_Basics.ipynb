## What are Data Structures and Algorithms?

**Data Structure**: A way of organizing and storing data so that it can be accessed and modified efficiently.

**Algorithm**: A step-by-step procedure or set of rules to solve a specific problem.

### Common Data Structures:
- **Arrays**: Contiguous memory locations
- **Linked Lists**: Nodes connected by pointers
- **Stacks**: LIFO (Last In, First Out)
- **Queues**: FIFO (First In, First Out)
- **Trees**: Hierarchical data structure
- **Graphs**: Nodes connected by edges
- **Hash Tables**: Key-value pairs with fast lookup

### Common Algorithms:
- **Sorting**: Bubble, Selection, Insertion, Merge, Quick, etc.
- **Searching**: Linear, Binary
- **Graph Traversal**: BFS, DFS
- **Dynamic Programming**: Memoization, Tabulation
- **Greedy**: Local optimal choices
- **Backtracking**: Exploring all possibilities

## Time and Space Complexity

### Big O Notation
Big O describes the **upper bound** (worst-case) of an algorithm's runtime.

### Omega Notation (Ω)
Omega describes the **lower bound** (best-case) of an algorithm's runtime.

### Theta Notation (Θ)
Theta describes the **tight bound** (average-case) when upper and lower bounds are the same.

### Common Complexities:
| Complexity | Name | Example |
|------------|------|---------|
| O(1) | Constant | Array index access |
| O(log n) | Logarithmic | Binary search |
| O(n) | Linear | Linear search |
| O(n log n) | Linearithmic | Merge sort, Quick sort |
| O(n^2) | Quadratic | Bubble sort, Selection sort |
| O(2^n) | Exponential | Recursive Fibonacci |
| O(n!) | Factorial | Permutations |

### Space Complexity
Space complexity measures the total memory an algorithm uses relative to input size.

**Auxiliary Space**: Extra space used by the algorithm (not including input).
**Total Space**: Input space + Auxiliary space.

```python
# Example: Time Complexity Analysis

# O(1) - Constant Time
def get_first_element(arr):
    return arr[0]

# O(n) - Linear Time
def linear_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1

# O(n^2) - Quadratic Time
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr

# O(log n) - Logarithmic Time
def binary_search(arr, target):
    left, right = 0, len(arr) - 1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1
```

## Two Pointers Technique

The two pointers technique uses two pointers (indices) to traverse an array or list.

### When to use:
- Finding pairs in a sorted array
- Reversing arrays or linked lists
- Detecting cycles
- Partitioning arrays

### Types:
1. **Opposite ends**: One pointer at start, one at end
2. **Same direction**: Both pointers move in the same direction
3. **Fast and slow**: One moves faster than the other

```python
# Two Pointers: Opposite Ends
# Find a pair with given sum in a sorted array
def two_sum_sorted(arr, target):
    left, right = 0, len(arr) - 1
    while left < right:
        current_sum = arr[left] + arr[right]
        if current_sum == target:
            return [left, right]
        elif current_sum < target:
            left += 1
        else:
            right -= 1
    return [-1, -1]

# Example
arr = [2, 7, 11, 15]
print(two_sum_sorted(arr, 9))

# Two Pointers: Same Direction (Remove Duplicates)
def remove_duplicates(arr):
    if not arr:
        return 0
    write_index = 1
    for i in range(1, len(arr)):
        if arr[i] != arr[i - 1]:
            arr[write_index] = arr[i]
            write_index += 1
    return write_index

arr = [1, 1, 2, 2, 3, 3, 3, 4]
new_len = remove_duplicates(arr)
print(f"New length: {new_len}, Array: {arr[:new_len]}")

# Two Pointers: Fast and Slow (Cycle Detection)
def has_cycle(head):
    if not head or not head.next:
        return False
    slow = head
    fast = head.next
    while slow != fast:
        if not fast or not fast.next:
            return False
        slow = slow.next
        fast = fast.next.next
    return True
```

## Sliding Window Technique

The sliding window technique is used to efficiently process subarrays or substrings.

### When to use:
- Finding subarrays with certain properties
- Fixed-size window problems
- Variable-size window problems

### Types:
1. **Fixed-size window**: Window size remains constant
2. **Variable-size window**: Window size changes based on conditions

```python
# Sliding Window: Fixed Size
# Find maximum sum of any subarray of size k
def max_sum_subarray(arr, k):
    if len(arr) < k:
        return -1
    # Calculate first window sum
    window_sum = sum(arr[:k])
    max_sum = window_sum
    # Slide the window
    for i in range(k, len(arr)):
        window_sum = window_sum + arr[i] - arr[i - k]
        max_sum = max(max_sum, window_sum)
    return max_sum

arr = [2, 1, 5, 1, 3, 2]
print(f"Max sum of size 3: {max_sum_subarray(arr, 3)}")

# Sliding Window: Variable Size
# Find smallest subarray with sum >= target
def smallest_subarray_with_sum(arr, target):
    min_length = float('inf')
    window_sum = 0
    left = 0
    for right in range(len(arr)):
        window_sum += arr[right]
        while window_sum >= target:
            min_length = min(min_length, right - left + 1)
            window_sum -= arr[left]
            left += 1
    return min_length if min_length != float('inf') else 0

arr = [2, 3, 1, 2, 4, 3]
print(f"Smallest subarray with sum >= 7: {smallest_subarray_with_sum(arr, 7)}")

# Sliding Window: Longest Substring Without Repeating Characters
def length_of_longest_substring(s):
    char_set = set()
    left = 0
    max_length = 0
    for right in range(len(s)):
        while s[right] in char_set:
            char_set.remove(s[left])
            left += 1
        char_set.add(s[right])
        max_length = max(max_length, right - left + 1)
    return max_length

print(f"Longest substring without repeating: {length_of_longest_substring('abcabcbb')}")
```

## Strings

Strings are sequences of characters.

### Common String Operations:
| Operation | Time Complexity | Description |
|-----------|----------------|-------------|
| Access by index | O(1) | `s[i]` |
| Length | O(1) | `len(s)` |
| Concatenation | O(n) | `s1 + s2` |
| Slicing | O(k) | `s[i:j]` where k = j-i |
| Search (substring) | O(n*m) | `s.find(sub)` |

### String Methods:
- `s.upper()`, `s.lower()` - case conversion
- `s.strip()` - remove whitespace
- `s.split()` - split into list
- `s.replace(old, new)` - replace substrings
- `s.isalnum()`, `s.isalpha()` - check character types

```python
# String Operations

# Creating strings
s1 = "Hello"
s2 = 'World'
s3 = """Multi-line string"""

# String methods
s = "  Hello World  "
print(f"Upper: '{s.upper()}'")
print(f"Lower: '{s.lower()}'")
print(f"Strip: '{s.strip()}'")
print(f"Split: {s.split()}")
print(f"Replace: '{s.replace('World', 'Python')}'")

# String formatting
name = "Alice"
age = 30
print(f"Name: {name}, Age: {age}")
print("Name: {}, Age: {}".format(name, age))
print("Name: %s, Age: %d" % (name, age))

# String immutability
s = "Hello"
# s[0] = 'h'  # This would raise an error
s = 'h' + s[1:]  # Create new string
print(s)

# Check palindrome
def is_palindrome(s):
    left, right = 0, len(s) - 1
    while left < right:
        if s[left] != s[right]:
            return False
        left += 1
        right -= 1
    return True

print(f"Is 'racecar' a palindrome? {is_palindrome('racecar')}")
print(f"Is 'hello' a palindrome? {is_palindrome('hello')}")

# Anagram check
def is_anagram(s1, s2):
    if len(s1) != len(s2):
        return False
    return sorted(s1) == sorted(s2)

print(f"Is 'listen' anagram of 'silent'? {is_anagram('listen', 'silent')}")
```

## Linked Lists

A linked list is a linear data structure where elements are stored in nodes, each pointing to the next node.

### Types of Linked Lists:
1. **Singly Linked List**: Each node points to the next node
2. **Doubly Linked List**: Each node points to both next and previous nodes
3. **Circular Linked List**: Last node points back to the first node

### Linked List vs Array:
| Feature | Array | Linked List |
|---------|-------|-------------|
| Memory | Contiguous | Non-contiguous |
| Access | O(1) by index | O(n) by traversal |
| Insertion (beginning) | O(n) | O(1) |
| Deletion (beginning) | O(n) | O(1) |
| Size | Fixed (static) | Dynamic |

### Operations:
| Operation | Time Complexity |
|-----------|----------------|
| Access | O(n) |
| Search | O(n) |
| Insert (beginning) | O(1) |
| Insert (end) | O(n) |
| Delete (beginning) | O(1) |
| Delete (end) | O(n) |

```python
# Singly Linked List
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class SinglyLinkedList:
    def __init__(self):
        self.head = None
        self.size = 0

    def insert_at_beginning(self, val):
        new_node = ListNode(val)
        new_node.next = self.head
        self.head = new_node
        self.size += 1

    def insert_at_end(self, val):
        new_node = ListNode(val)
        if not self.head:
            self.head = new_node
        else:
            current = self.head
            while current.next:
                current = current.next
            current.next = new_node
        self.size += 1

    def insert_at_index(self, index, val):
        if index < 0 or index > self.size:
            raise IndexError("Index out of bounds")
        if index == 0:
            self.insert_at_beginning(val)
        else:
            new_node = ListNode(val)
            current = self.head
            for _ in range(index - 1):
                current = current.next
            new_node.next = current.next
            current.next = new_node
            self.size += 1

    def delete(self, val):
        if not self.head:
            return
        if self.head.val == val:
            self.head = self.head.next
            self.size -= 1
            return
        current = self.head
        while current.next:
            if current.next.val == val:
                current.next = current.next.next
                self.size -= 1
                return
            current = current.next

    def search(self, val):
        current = self.head
        index = 0
        while current:
            if current.val == val:
                return index
            current = current.next
            index += 1
        return -1

    def display(self):
        elements = []
        current = self.head
        while current:
            elements.append(str(current.val))
            current = current.next
        print(" -> ".join(elements))

# Test Singly Linked List
sll = SinglyLinkedList()
sll.insert_at_end(1)
sll.insert_at_end(2)
sll.insert_at_end(3)
sll.insert_at_beginning(0)
sll.display()
print(f"Size: {sll.size}")
print(f"Search 2: index {sll.search(2)}")
```

```python
# Doubly Linked List
class DListNode:
    def __init__(self, val=0, prev=None, next=None):
        self.val = val
        self.prev = prev
        self.next = next

class DoublyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None
        self.size = 0

    def insert_at_beginning(self, val):
        new_node = DListNode(val)
        if not self.head:
            self.head = new_node
            self.tail = new_node
        else:
            new_node.next = self.head
            self.head.prev = new_node
            self.head = new_node
        self.size += 1

    def insert_at_end(self, val):
        new_node = DListNode(val)
        if not self.tail:
            self.head = new_node
            self.tail = new_node
        else:
            new_node.prev = self.tail
            self.tail.next = new_node
            self.tail = new_node
        self.size += 1

    def delete(self, val):
        current = self.head
        while current:
            if current.val == val:
                if current.prev:
                    current.prev.next = current.next
                else:
                    self.head = current.next
                if current.next:
                    current.next.prev = current.prev
                else:
                    self.tail = current.prev
                self.size -= 1
                return
            current = current.next

    def display_forward(self):
        elements = []
        current = self.head
        while current:
            elements.append(str(current.val))
            current = current.next
        print("Forward: " + " <-> ".join(elements))

    def display_backward(self):
        elements = []
        current = self.tail
        while current:
            elements.append(str(current.val))
            current = current.prev
        print("Backward: " + " <-> ".join(elements))

# Test Doubly Linked List
dll = DoublyLinkedList()
dll.insert_at_end(1)
dll.insert_at_end(2)
dll.insert_at_end(3)
dll.display_forward()
dll.display_backward()
dll.delete(2)
dll.display_forward()
```

```python
# Circular Linked List
class CListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class CircularLinkedList:
    def __init__(self):
        self.head = None
        self.size = 0

    def insert_at_end(self, val):
        new_node = CListNode(val)
        if not self.head:
            self.head = new_node
            new_node.next = self.head
        else:
            current = self.head
            while current.next != self.head:
                current = current.next
            current.next = new_node
            new_node.next = self.head
        self.size += 1

    def display(self):
        if not self.head:
            print("Empty list")
            return
        elements = []
        current = self.head
        while True:
            elements.append(str(current.val))
            current = current.next
            if current == self.head:
                break
        print(" -> ".join(elements) + " -> (back to head)")

# Test Circular Linked List
cll = CircularLinkedList()
cll.insert_at_end(1)
cll.insert_at_end(2)
cll.insert_at_end(3)
cll.display()
```

## Stacks

A stack is a linear data structure that follows the **LIFO (Last In, First Out)** principle.

### Stack Operations:
| Operation | Time Complexity | Description |
|-----------|----------------|-------------|
| push | O(1) | Add element to top |
| pop | O(1) | Remove element from top |
| peek/top | O(1) | Get top element |
| isEmpty | O(1) | Check if stack is empty |
| size | O(1) | Get number of elements |

### Applications:
- Function call stack
- Undo operations
- Expression evaluation
- Backtracking algorithms
- Browser history

```python
# Stack Implementation using List
class Stack:
    def __init__(self):
        self.items = []

    def push(self, item):
        self.items.append(item)

    def pop(self):
        if self.is_empty():
            raise IndexError("Stack is empty")
        return self.items.pop()

    def peek(self):
        if self.is_empty():
            raise IndexError("Stack is empty")
        return self.items[-1]

    def is_empty(self):
        return len(self.items) == 0

    def size(self):
        return len(self.items)

    def display(self):
        print(self.items)

# Test Stack
stack = Stack()
stack.push(1)
stack.push(2)
stack.push(3)
stack.display()
print(f"Peek: {stack.peek()}")
print(f"Pop: {stack.pop()}")
stack.display()

# Using Python's collections.deque for stack
from collections import deque

stack = deque()
stack.append(1)  # push
stack.append(2)
stack.pop()  # pop
print(f"Deque stack: {stack}")

# Stack Applications: Valid Parentheses
def is_valid_parentheses(s):
    stack = []
    mapping = {')': '(', '}': '{', ']': '['}
    for char in s:
        if char in mapping:
            if not stack or stack.pop() != mapping[char]:
                return False
        else:
            stack.append(char)
    return not stack

print(f"Valid '()': {is_valid_parentheses('()')}")
print(f"Valid '()[]{{}}': {is_valid_parentheses('()[]{}')}")
print(f"Valid '(]': {is_valid_parentheses('(]')}")
```

## Queues

A queue is a linear data structure that follows the **FIFO (First In, First Out)** principle.

### Queue Operations:
| Operation | Time Complexity | Description |
|-----------|----------------|-------------|
| enqueue | O(1) | Add element to rear |
| dequeue | O(1) | Remove element from front |
| front/peek | O(1) | Get front element |
| isEmpty | O(1) | Check if queue is empty |
| size | O(1) | Get number of elements |

### Types of Queues:
1. **Simple Queue**: Basic FIFO queue
2. **Circular Queue**: Rear connects to front
3. **Priority Queue**: Elements have priorities
4. **Deque (Double-ended Queue)**: Insert/delete from both ends

### Applications:
- Task scheduling (CPU processes)
- BFS graph traversal
- Print job queue
- Web server request queue

```python
# Queue Implementation using List
class Queue:
    def __init__(self):
        self.items = []

    def enqueue(self, item):
        self.items.append(item)

    def dequeue(self):
        if self.is_empty():
            raise IndexError("Queue is empty")
        return self.items.pop(0)

    def front(self):
        if self.is_empty():
            raise IndexError("Queue is empty")
        return self.items[0]

    def is_empty(self):
        return len(self.items) == 0

    def size(self):
        return len(self.items)

    def display(self):
        print(self.items)

# Test Queue
queue = Queue()
queue.enqueue(1)
queue.enqueue(2)
queue.enqueue(3)
queue.display()
print(f"Front: {queue.front()}")
print(f"Dequeue: {queue.dequeue()}")
queue.display()

# Using collections.deque for efficient queue
from collections import deque

queue = deque()
queue.append(1)  # enqueue
queue.append(2)
queue.popleft()  # dequeue
print(f"Deque queue: {queue}")

# Circular Queue Implementation
class CircularQueue:
    def __init__(self, size):
        self.size = size
        self.queue = [None] * size
        self.front = -1
        self.rear = -1

    def enqueue(self, item):
        if (self.rear + 1) % self.size == self.front:
            raise OverflowError("Queue is full")
        if self.is_empty():
            self.front = 0
        self.rear = (self.rear + 1) % self.size
        self.queue[self.rear] = item

    def dequeue(self):
        if self.is_empty():
            raise IndexError("Queue is empty")
        item = self.queue[self.front]
        if self.front == self.rear:
            self.front = -1
            self.rear = -1
        else:
            self.front = (self.front + 1) % self.size
        return item

    def is_empty(self):
        return self.front == -1

    def is_full(self):
        return (self.rear + 1) % self.size == self.front

    def display(self):
        if self.is_empty():
            print("Empty")
            return
        if self.rear >= self.front:
            print(self.queue[self.front:self.rear + 1])
        else:
            print(self.queue[self.front:] + self.queue[:self.rear + 1])

# Test Circular Queue
cq = CircularQueue(5)
cq.enqueue(1)
cq.enqueue(2)
cq.enqueue(3)
cq.display()
cq.dequeue()
cq.display()
```

## Recursion

Recursion is a technique where a function calls itself to solve smaller instances of the same problem.

### Key Elements of Recursion:
1. **Base Case**: The condition that stops the recursion
2. **Recursive Case**: The function calls itself with a smaller input

### When to use Recursion:
- Tree traversals
- Graph traversals (DFS)
- Divide and conquer algorithms
- Backtracking problems
- Dynamic programming

### Time and Space Complexity:
- Each recursive call adds a frame to the call stack
- Space complexity includes the call stack depth
- Tail recursion can be optimized to O(1) space (Python doesn't do this)

```python
# Recursion Examples

# 1. Factorial
def factorial(n):
    if n <= 1:  # Base case
        return 1
    return n * factorial(n - 1)  # Recursive case

print(f"5! = {factorial(5)}")

# 2. Fibonacci (inefficient - O(2^n))
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

print(f"Fibonacci(10) = {fibonacci(10)}")

# 3. Power function
def power(base, exp):
    if exp == 0:
        return 1
    if exp % 2 == 0:
        half = power(base, exp // 2)
        return half * half
    else:
        return base * power(base, exp - 1)

print(f"2^10 = {power(2, 10)}")

# 4. Sum of array
def sum_array(arr, index=0):
    if index == len(arr):
        return 0
    return arr[index] + sum_array(arr, index + 1)

print(f"Sum of [1, 2, 3, 4, 5] = {sum_array([1, 2, 3, 4, 5])}")

# 5. Reverse string
def reverse_string(s):
    if len(s) <= 1:
        return s
    return reverse_string(s[1:]) + s[0]

print(f"Reverse 'hello' = '{reverse_string('hello')}'")

# 6. Tower of Hanoi
def tower_of_hanoi(n, source, destination, auxiliary):
    if n == 1:
        print(f"Move disk 1 from {source} to {destination}")
        return
    tower_of_hanoi(n - 1, source, auxiliary, destination)
    print(f"Move disk {n} from {source} to {destination}")
    tower_of_hanoi(n - 1, auxiliary, destination, source)

print("Tower of Hanoi (3 disks):")
tower_of_hanoi(3, 'A', 'C', 'B')

# 7. Binary search (recursive)
def binary_search_recursive(arr, target, left=0, right=None):
    if right is None:
        right = len(arr) - 1
    if left > right:
        return -1
    mid = (left + right) // 2
    if arr[mid] == target:
        return mid
    elif arr[mid] < target:
        return binary_search_recursive(arr, target, mid + 1, right)
    else:
        return binary_search_recursive(arr, target, left, mid - 1)

arr = [1, 3, 5, 7, 9, 11]
print(f"Binary search for 7: index {binary_search_recursive(arr, 7)}")
```

## Searching

Searching algorithms find the position of a target element in a data structure.

### Linear Search
- **Time Complexity**: O(n)
- **Space Complexity**: O(1)
- Works on both sorted and unsorted arrays

### Binary Search
- **Time Complexity**: O(log n)
- **Space Complexity**: O(1) iterative, O(log n) recursive
- Only works on **sorted** arrays

### Binary Search Variants:
- Find first occurrence
- Find last occurrence
- Find rotation point
- Search in rotated sorted array

```python
# Linear Search
def linear_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1

arr = [4, 2, 7, 1, 9, 3]
print(f"Linear search for 7: index {linear_search(arr, 7)}")

# Binary Search (Iterative)
def binary_search(arr, target):
    left, right = 0, len(arr) - 1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1

arr = [1, 3, 5, 7, 9, 11, 13]
print(f"Binary search for 7: index {binary_search(arr, 7)}")

# Binary Search: Find First Occurrence
def find_first(arr, target):
    left, right = 0, len(arr) - 1
    result = -1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            result = mid
            right = mid - 1  # Continue searching left
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return result

# Binary Search: Find Last Occurrence
def find_last(arr, target):
    left, right = 0, len(arr) - 1
    result = -1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            result = mid
            left = mid + 1  # Continue searching right
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return result

arr = [1, 2, 2, 2, 3, 4, 5]
print(f"First occurrence of 2: {find_first(arr, 2)}")
print(f"Last occurrence of 2: {find_last(arr, 2)}")
```

## Sorting

Sorting algorithms arrange elements in a specific order (ascending or descending).

### Sorting Algorithms:

#### Bubble Sort
- **Time Complexity**: O(n^2) worst, O(n) best
- **Space Complexity**: O(1)
- **Stable**: Yes

#### Selection Sort
- **Time Complexity**: O(n^2)
- **Space Complexity**: O(1)
- **Stable**: No

#### Insertion Sort
- **Time Complexity**: O(n^2) worst, O(n) best
- **Space Complexity**: O(1)
- **Stable**: Yes
- **In-place**: Yes

```python
# Bubble Sort
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        if not swapped:
            break
    return arr

arr = [64, 34, 25, 12, 22, 11, 90]
print(f"Bubble Sort: {bubble_sort(arr.copy())}")

# Selection Sort
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr

arr = [64, 25, 12, 22, 11]
print(f"Selection Sort: {selection_sort(arr.copy())}")

# Insertion Sort
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr

arr = [12, 11, 13, 5, 6]
print(f"Insertion Sort: {insertion_sort(arr.copy())}")

# Comparison of sorting algorithms
import time
import random

def time_sort(sort_func, arr):
    start = time.time()
    sort_func(arr.copy())
    return time.time() - start

arr = random.sample(range(1, 1000), 100)
print(f"\nTiming on 100 random elements:")
print(f"Bubble Sort: {time_sort(bubble_sort, arr):.6f}s")
print(f"Selection Sort: {time_sort(selection_sort, arr):.6f}s")
print(f"Insertion Sort: {time_sort(insertion_sort, arr):.6f}s")
print(f"Python's Timsort: {time_sort(sorted, arr):.6f}s")
```

### 1. Two Sum (LeetCode #1)
**Difficulty**: Easy

Given an array of integers `nums` and an integer `target`, return indices of the two numbers such that they add up to `target`.

**Approach**: Use a hash map to store the complement (target - current number) and its index.

```python
# Two Sum (LeetCode #1)
def two_sum(nums, target):
    num_map = {}
    for i, num in enumerate(nums):
        complement = target - num
        if complement in num_map:
            return [num_map[complement], i]
        num_map[num] = i
    return []

# Test cases
print(f"Two Sum([2,7,11,15], 9): {two_sum([2,7,11,15], 9)}")
print(f"Two Sum([3,2,4], 6): {two_sum([3,2,4], 6)}")
print(f"Two Sum([3,3], 6): {two_sum([3,3], 6)}")

# Time Complexity: O(n)
# Space Complexity: O(n)
```

### 2. Remove Duplicates from Sorted Array (LeetCode #26)
**Difficulty**: Easy

Given a sorted array `nums`, remove the duplicates in-place such that each element appears only once. Return the new length.

**Approach**: Use two pointers - one to track the position of the last unique element, and another to scan through the array.

```python
# Remove Duplicates from Sorted Array (LeetCode #26)
def remove_duplicates(nums):
    if not nums:
        return 0
    write_index = 1
    for i in range(1, len(nums)):
        if nums[i] != nums[i - 1]:
            nums[write_index] = nums[i]
            write_index += 1
    return write_index

# Test cases
nums1 = [1, 1, 2]
k1 = remove_duplicates(nums1)
print(f"Remove Duplicates([1,1,2]): k={k1}, nums={nums1[:k1]}")

nums2 = [0, 0, 1, 1, 1, 2, 2, 3, 3, 4]
k2 = remove_duplicates(nums2)
print(f"Remove Duplicates([0,0,1,1,1,2,2,3,3,4]): k={k2}, nums={nums2[:k2]}")

# Time Complexity: O(n)
# Space Complexity: O(1)
```

### 3. Valid Parentheses (LeetCode #20)
**Difficulty**: Easy

Given a string `s` containing just the characters '(', ')', '{', '}', '[' and ']', determine if the input string is valid.

**Approach**: Use a stack. When we see an opening bracket, push it. When we see a closing bracket, check if it matches the top of the stack.

```python
# Valid Parentheses (LeetCode #20)
def is_valid(s):
    stack = []
    mapping = {')': '(', '}': '{', ']': '['}
    for char in s:
        if char in mapping:
            if not stack or stack.pop() != mapping[char]:
                return False
        else:
            stack.append(char)
    return not stack

# Test cases
print(f"Valid Parentheses('()'): {is_valid('()')}")
print(f"Valid Parentheses('()[]{{}}'): {is_valid('()[]{}')}")
print(f"Valid Parentheses('(]'): {is_valid('(]')}")
print(f"Valid Parentheses('([)]'): {is_valid('([)]')}")
print(f"Valid Parentheses('{{}}'): {is_valid('{{}}')}")

# Time Complexity: O(n)
# Space Complexity: O(n)
```

### 4. Linked List Cycle (LeetCode #141)
**Difficulty**: Easy

Given `head`, the head of a linked list, determine if the linked list has a cycle in it.

**Approach**: Use Floyd's Cycle Detection Algorithm (fast and slow pointers). If there's a cycle, the fast pointer will eventually meet the slow pointer.

```python
# Linked List Cycle (LeetCode #141)
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def has_cycle(head):
    if not head or not head.next:
        return False
    slow = head
    fast = head.next
    while slow != fast:
        if not fast or not fast.next:
            return False
        slow = slow.next
        fast = fast.next.next
    return True

# Test cases
# Create a list with cycle: 3 -> 2 -> 0 -> -4 -> (back to 2)
node1 = ListNode(3)
node2 = ListNode(2)
node3 = ListNode(0)
node4 = ListNode(-4)
node1.next = node2
node2.next = node3
node3.next = node4
node4.next = node2  # Creates cycle
print(f"Has cycle (with cycle): {has_cycle(node1)}")

# Create a list without cycle: 1 -> 2 -> 3
node_a = ListNode(1)
node_b = ListNode(2)
node_c = ListNode(3)
node_a.next = node_b
node_b.next = node_c
print(f"Has cycle (no cycle): {has_cycle(node_a)}")

# Time Complexity: O(n)
# Space Complexity: O(1)
```

### 5. Reverse Linked List (LeetCode #206)
**Difficulty**: Easy

Given the `head` of a singly linked list, reverse the list, and return the reversed list.

**Approach**: Iteratively reverse the pointers. Use three pointers: prev, current, and next.

```python
# Reverse Linked List (LeetCode #206)
def reverse_linked_list(head):
    prev = None
    current = head
    while current:
        next_node = current.next
        current.next = prev
        prev = current
        current = next_node
    return prev

# Helper function to print linked list
def print_list(head):
    elements = []
    current = head
    while current:
        elements.append(str(current.val))
        current = current.next
    print(" -> ".join(elements))

# Test case
node1 = ListNode(1)
node2 = ListNode(2)
node3 = ListNode(3)
node4 = ListNode(4)
node5 = ListNode(5)
node1.next = node2
node2.next = node3
node3.next = node4
node4.next = node5

print("Original list:")
print_list(node1)

reversed_head = reverse_linked_list(node1)
print("Reversed list:")
print_list(reversed_head)

# Time Complexity: O(n)
# Space Complexity: O(1)
```

### 6. Middle of the Linked List (LeetCode #876)
**Difficulty**: Easy

Given the `head` of a singly linked list, return the middle node of the linked list. If there are two middle nodes, return the second middle node.

**Approach**: Use fast and slow pointers. When the fast pointer reaches the end, the slow pointer will be at the middle.

```python
# Middle of the Linked List (LeetCode #876)
def middle_node(head):
    slow = head
    fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
    return slow

# Test case
node1 = ListNode(1)
node2 = ListNode(2)
node3 = ListNode(3)
node4 = ListNode(4)
node5 = ListNode(5)
node1.next = node2
node2.next = node3
node3.next = node4
node4.next = node5

print("List: 1 -> 2 -> 3 -> 4 -> 5")
middle = middle_node(node1)
print(f"Middle node: {middle.val}")

# Test with even number of nodes
node_a = ListNode(1)
node_b = ListNode(2)
node_c = ListNode(3)
node_d = ListNode(4)
node_e = ListNode(5)
node_f = ListNode(6)
node_a.next = node_b
node_b.next = node_c
node_c.next = node_d
node_d.next = node_e
node_e.next = node_f

print("\nList: 1 -> 2 -> 3 -> 4 -> 5 -> 6")
middle = middle_node(node_a)
print(f"Middle node: {middle.val}")

# Time Complexity: O(n)
# Space Complexity: O(1)
```

### 7. Merge Two Sorted Lists (LeetCode #21)
**Difficulty**: Easy

Given the heads of two sorted linked lists `list1` and `list2`, merge the two lists into one sorted list. Return the head of the merged list.

**Approach**: Use a dummy node and two pointers. Compare nodes and attach the smaller one.

```python
# Merge Two Sorted Lists (LeetCode #21)
def merge_two_lists(list1, list2):
    dummy = ListNode(0)
    current = dummy
    while list1 and list2:
        if list1.val <= list2.val:
            current.next = list1
            list1 = list1.next
        else:
            current.next = list2
            list2 = list2.next
        current = current.next
    current.next = list1 if list1 else list2
    return dummy.next

# Helper function to create linked list from list
def create_linked_list(arr):
    if not arr:
        return None
    head = ListNode(arr[0])
    current = head
    for val in arr[1:]:
        current.next = ListNode(val)
        current = current.next
    return head

# Test case
list1 = create_linked_list([1, 2, 4])
list2 = create_linked_list([1, 3, 4])

print("List 1: 1 -> 2 -> 4")
print("List 2: 1 -> 3 -> 4")
merged = merge_two_lists(list1, list2)
print("Merged list:")
print_list(merged)

# Time Complexity: O(n + m)
# Space Complexity: O(1)
```

### 8. Palindrome Linked List (LeetCode #234)
**Difficulty**: Easy

Given the `head` of a singly linked list, return `true` if it is a palindrome.

**Approach**: Find the middle, reverse the second half, and compare with the first half.

```python
# Palindrome Linked List (LeetCode #234)
def is_palindrome(head):
    if not head or not head.next:
        return True
    
    # Find middle
    slow = head
    fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
    
    # Reverse second half
    prev = None
    current = slow
    while current:
        next_node = current.next
        current.next = prev
        prev = current
        current = next_node
    
    # Compare first and second half
    first_half = head
    second_half = prev
    while second_half:
        if first_half.val != second_half.val:
            return False
        first_half = first_half.next
        second_half = second_half.next
    return True

# Test cases
list1 = create_linked_list([1, 2, 2, 1])
print(f"Is [1,2,2,1] a palindrome? {is_palindrome(list1)}")

list2 = create_linked_list([1, 2])
print(f"Is [1,2] a palindrome? {is_palindrome(list2)}")

list3 = create_linked_list([1])
print(f"Is [1] a palindrome? {is_palindrome(list3)}")

# Time Complexity: O(n)
# Space Complexity: O(1)
```
