# Trees - DSA Patterns

Trees are hierarchical data structures with a root node and child nodes.

## Topics Covered

2. Tree Traversals (Inorder, Preorder, Postorder, Level-order)
3. Binary Search Tree (BST)
4. Lowest Common Ancestor
5. Tree Diameter and Height

## 1. Binary Tree Implementation

```python
class TreeNode:
    """Definition for binary tree node."""
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


# Helper function to create a binary tree from list
def create_tree(values):
    """Create binary tree from level-order list."""
    if not values:
        return None
    
    root = TreeNode(values[0])
    queue = [root]
    i = 1
    
    while i < len(values):
        node = queue.pop(0)
        if i < len(values) and values[i] is not None:
            node.left = TreeNode(values[i])
            queue.append(node.left)
        i += 1
        if i < len(values) and values[i] is not None:
            node.right = TreeNode(values[i])
            queue.append(node.right)
        i += 1
    
    return root


# Example
root = create_tree([1, 2, 3, 4, 5, 6, 7])
print("Binary tree created with values: [1, 2, 3, 4, 5, 6, 7]")
```

## 2. Tree Traversals

```python
def inorder_traversal(root):
    """Inorder traversal: Left -> Root -> Right.
    Time: O(n)"""
    result = []
    
    def traverse(node):
        if node:
            traverse(node.left)
            result.append(node.val)
            traverse(node.right)
    
    traverse(root)
    return result


def preorder_traversal(root):
    """Preorder traversal: Root -> Left -> Right.
    Time: O(n)"""
    result = []
    
    def traverse(node):
        if node:
            result.append(node.val)
            traverse(node.left)
            traverse(node.right)
    
    traverse(root)
    return result


def postorder_traversal(root):
    """Postorder traversal: Left -> Right -> Root.
    Time: O(n)"""
    result = []
    
    def traverse(node):
        if node:
            traverse(node.left)
            traverse(node.right)
            result.append(node.val)
    
    traverse(root)
    return result


def level_order_traversal(root):
    """Level-order (BFS) traversal.
    Time: O(n)"""
    if not root:
        return []
    
    result = []
    queue = [root]
    
    while queue:
        level = []
        level_size = len(queue)
        
        for _ in range(level_size):
            node = queue.pop(0)
            level.append(node.val)
            if node.left:
                queue.append(node.left)
            if node.right:
                queue.append(node.right)
        
        result.append(level)
    
    return result


# Example
root = create_tree([1, 2, 3, 4, 5, 6, 7])
print(f"Inorder:    {inorder_traversal(root)}")
print(f"Preorder:   {preorder_traversal(root)}")
print(f"Postorder:  {postorder_traversal(root)}")
print(f"Level-order: {level_order_traversal(root)}")
```

## 3. Binary Search Tree (BST)

```python
class BST:
    """Binary Search Tree implementation."""
    
    def __init__(self):
        self.root = None
    
    def insert(self, val):
        """Insert a value into BST."""
        self.root = self._insert(self.root, val)
    
    def _insert(self, node, val):
        if not node:
            return TreeNode(val)
        if val < node.val:
            node.left = self._insert(node.left, val)
        elif val > node.val:
            node.right = self._insert(node.right, val)
        return node
    
    def search(self, val):
        """Search for a value in BST."""
        return self._search(self.root, val)
    
    def _search(self, node, val):
        if not node or node.val == val:
            return node
        if val < node.val:
            return self._search(node.left, val)
        return self._search(node.right, val)
    
    def inorder(self):
        """Return inorder traversal (sorted order)."""
        result = []
        self._inorder(self.root, result)
        return result
    
    def _inorder(self, node, result):
        if node:
            self._inorder(node.left, result)
            result.append(node.val)
            self._inorder(node.right, result)


# Example
bst = BST()
for val in [5, 3, 7, 1, 4, 6, 8]:
    bst.insert(val)
print(f"BST inorder (sorted): {bst.inorder()}")
print(f"Search 4: {bst.search(4).val}")
print(f"Search 9: {bst.search(9)}")
```

## 4. Lowest Common Ancestor

```python
def lowest_common_ancestor(root, p, q):
    """Find LCA of two nodes (LeetCode 236).
    Time: O(n)"""
    if not root or root == p or root == q:
        return root
    
    left = lowest_common_ancestor(root.left, p, q)
    right = lowest_common_ancestor(root.right, p, q)
    
    if left and right:
        return root
    return left or right


# Example
root = create_tree([3, 5, 1, 6, 2, 0, 8, None, None, 7, 4])
p = root.left  # Node 5
q = root.right  # Node 1
lca = lowest_common_ancestor(root, p, q)
print(f"LCA of 5 and 1: {lca.val}")
```

## 5. Tree Diameter and Height

```python
def tree_height(root):
    """Calculate height of binary tree.
    Time: O(n)"""
    if not root:
        return 0
    return 1 + max(tree_height(root.left), tree_height(root.right))


def tree_diameter(root):
    """Find diameter of binary tree (LeetCode 543).
    Time: O(n)"""
    self.diameter = 0
    
    def height(node):
        if not node:
            return 0
        left_height = height(node.left)
        right_height = height(node.right)
        self.diameter = max(self.diameter, left_height + right_height)
        return 1 + max(left_height, right_height)
    
    height(root)
    return self.diameter


def is_balanced(root):
    """Check if tree is balanced (LeetCode 110).
    Time: O(n)"""
    def check_height(node):
        if not node:
            return 0
        
        left_height = check_height(node.left)
        if left_height == -1:
            return -1
        
        right_height = check_height(node.right)
        if right_height == -1:
            return -1
        
        if abs(left_height - right_height) > 1:
            return -1
        
        return 1 + max(left_height, right_height)
    
    return check_height(root) != -1


# Example
root = create_tree([1, 2, 3, 4, 5])
print(f"Height: {tree_height(root)}")
print(f"Diameter: {tree_diameter(root)}")
print(f"Is balanced: {is_balanced(root)}")
```
