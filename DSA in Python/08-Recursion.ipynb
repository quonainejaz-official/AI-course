# Recursion - DSA Patterns

Recursion is a technique where a function calls itself. This notebook covers essential recursion patterns.

## 1. Basic Recursion Concepts

```python
def factorial(n):
    """Calculate factorial recursively.
    Time: O(n)"""
    if n <= 1:
        return 1
    return n * factorial(n - 1)


def fibonacci(n):
    """Calculate nth Fibonacci number.
    Time: O(2^n) - inefficient"""
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)


def fibonacci_memo(n, memo={}):
    """Fibonacci with memoization.
    Time: O(n)"""
    if n in memo:
        return memo[n]
    if n <= 1:
        return n
    memo[n] = fibonacci_memo(n - 1, memo) + fibonacci_memo(n - 2, memo)
    return memo[n]


# Examples
print(f"factorial(5) = {factorial(5)}")
print(f"fibonacci(10) = {fibonacci(10)}")
print(f"fibonacci_memo(30) = {fibonacci_memo(30)}")
```

## 2. Subset and Subsequence Generation

```python
def subsets(nums):
    """Generate all subsets (LeetCode 78).
    Time: O(2^n)"""
    result = []
    
    def backtrack(start, current):
        result.append(current[:])
        
        for i in range(start, len(nums)):
            current.append(nums[i])
            backtrack(i + 1, current)
            current.pop()
    
    backtrack(0, [])
    return result


def subsets_with_duplicate(nums):
    """Generate subsets with duplicates (LeetCode 90)."""
    result = []
    nums.sort()
    
    def backtrack(start, current):
        result.append(current[:])
        
        for i in range(start, len(nums)):
            if i > start and nums[i] == nums[i - 1]:
                continue
            current.append(nums[i])
            backtrack(i + 1, current)
            current.pop()
    
    backtrack(0, [])
    return result


def subsequences(nums):
    """Generate all subsequences."""
    result = []
    
    def backtrack(start, current):
        if current:
            result.append(current[:])
        
        for i in range(start, len(nums)):
            current.append(nums[i])
            backtrack(i + 1, current)
            current.pop()
    
    backtrack(0, [])
    return result


# Examples
print(f"Subsets of [1,2,3]: {subsets([1, 2, 3])}")
print(f"\nSubsets with duplicates [1,2,2]: {subsets_with_duplicate([1, 2, 2])}")
```

## 3. Backtracking

```python
def permutations(nums):
    """Generate all permutations (LeetCode 46).
    Time: O(n * n!)"""
    result = []
    
    def backtrack(current, remaining):
        if not remaining:
            result.append(current[:])
            return
        
        for i in range(len(remaining)):
            current.append(remaining[i])
            backtrack(current, remaining[:i] + remaining[i + 1:])
            current.pop()
    
    backtrack([], nums)
    return result


def permutations_unique(nums):
    """Generate unique permutations (LeetCode 47)."""
    result = []
    nums.sort()
    
    def backtrack(current, remaining):
        if not remaining:
            result.append(current[:])
            return
        
        for i in range(len(remaining)):
            if i > 0 and remaining[i] == remaining[i - 1]:
                continue
            current.append(remaining[i])
            backtrack(current, remaining[:i] + remaining[i + 1:])
            current.pop()
    
    backtrack([], nums)
    return result


def combination_sum(candidates, target):
    """Find combinations that sum to target (LeetCode 39).
    Time: O(2^n)"""
    result = []
    
    def backtrack(start, current, remaining):
        if remaining == 0:
            result.append(current[:])
            return
        if remaining < 0:
            return
        
        for i in range(start, len(candidates)):
            current.append(candidates[i])
            backtrack(i, current, remaining - candidates[i])
            current.pop()
    
    backtrack(0, [], target)
    return result


# Examples
print(f"Permutations of [1,2,3]: {permutations([1, 2, 3])}")
print(f"\nCombination sum [2,3,6,7] target 7: {combination_sum([2, 3, 6, 7], 7)}")
```

## 4. Maze Problems

```python
def solve_maze(maze):
    """Solve maze using backtracking."""
    n = len(maze)
    solution = [[0] * n for _ in range(n)]
    
    def is_safe(x, y):
        return (0 <= x < n and 0 <= y < n and maze[x][y] == 1)
    
    def solve(x, y):
        if x == n - 1 and y == n - 1:
            solution[x][y] = 1
            return True
        
        if is_safe(x, y):
            solution[x][y] = 1
            
            if solve(x + 1, y):  # Move down
                return True
            if solve(x, y + 1):  # Move right
                return True
            
            solution[x][y] = 0  # Backtrack
            return False
        
        return False
    
    if solve(0, 0):
        return solution
    return None


# Example
maze = [
    [1, 0, 0, 0],
    [1, 1, 0, 1],
    [0, 1, 0, 0],
    [1, 1, 1, 1]
]
solution = solve_maze(maze)
if solution:
    print("Maze solution:")
    for row in solution:
        print(row)
else:
    print("No solution exists")
```

```python
def rat_in_maze(maze):
    """Find path for rat from top-left to bottom-right.
    1 = path, 0 = wall"""
    n = len(maze)
    path = []
    
    def is_safe(x, y):
        return (0 <= x < n and 0 <= y < n and maze[x][y] == 1)
    
    def solve(x, y, direction):
        if x == n - 1 and y == n - 1:
            path.append(direction)
            return True
        
        if is_safe(x, y):
            maze[x][y] = 0  # Mark visited
            
            if solve(x + 1, y, 'D'):  # Down
                path.append(direction)
                return True
            if solve(x, y + 1, 'R'):  # Right
                path.append(direction)
                return True
            if solve(x - 1, y, 'U'):  # Up
                path.append(direction)
                return True
            if solve(x, y - 1, 'L'):  # Left
                path.append(direction)
                return True
            
            maze[x][y] = 1  # Backtrack
            return False
        
        return False
    
    if solve(0, 0, ''):
        return ''.join(reversed(path))
    return None


# Example
maze = [
    [1, 0, 0, 0],
    [1, 1, 0, 1],
    [0, 1, 0, 0],
    [1, 1, 1, 1]
]
path = rat_in_maze(maze)
print(f"Path: {path}")
```
