## 1. Pattern Matching - KMP Algorithm

Knuth-Morris-Pratt algorithm for efficient substring search.

**Time:** O(n + m) | **Space:** O(m) for LPS array

```python
def compute_lps(pattern):
    """Compute Longest Proper Prefix which is also Suffix array."""
    m = len(pattern)
    lps = [0] * m
    length = 0
    i = 1
    
    while i < m:
        if pattern[i] == pattern[length]:
            length += 1
            lps[i] = length
            i += 1
        else:
            if length != 0:
                length = lps[length - 1]
            else:
                lps[i] = 0
                i += 1
    
    return lps


def kmp_search(text, pattern):
    """KMP pattern matching algorithm.
    Time: O(n + m)"""
    n = len(text)
    m = len(pattern)
    lps = compute_lps(pattern)
    
    i = 0  # index for text
    j = 0  # index for pattern
    results = []
    
    while i < n:
        if pattern[j] == text[i]:
            i += 1
            j += 1
        
        if j == m:
            results.append(i - j)
            j = lps[j - 1]
        elif i < n and pattern[j] != text[i]:
            if j != 0:
                j = lps[j - 1]
            else:
                i += 1
    
    return results


# Example
text = "ABABDABACDABABCABAB"
pattern = "ABABCABAB"
print(f"Text: {text}")
print(f"Pattern: {pattern}")
print(f"Pattern found at indices: {kmp_search(text, pattern)}")
```

```python
def str_str(haystack, needle):
    """Find first occurrence of needle in haystack (LeetCode 28)."""
    if not needle:
        return 0
    
    for i in range(len(haystack) - len(needle) + 1):
        if haystack[i:i + len(needle)] == needle:
            return i
    return -1


# Example
print(f"First occurrence of 'sadbutsad' in 'sad': {str_str('sadbutsad', 'sad')}")
print(f"First occurrence of 'leetcode' in 'leeto': {str_str('leetcode', 'leeto')}")
```

## 2. Anagrams

Two strings are anagrams if they contain the same characters with same frequencies.

```python
from collections import Counter


def is_anagram(s, t):
    """Check if two strings are anagrams.
    Time: O(n)"""
    if len(s) != len(t):
        return False
    return Counter(s) == Counter(t)


def group_anagrams(strs):
    """Group strings that are anagrams of each other (LeetCode 49).
    Time: O(n * k log k) where k is max string length"""
    anagram_map = {}
    
    for s in strs:
        sorted_s = ''.join(sorted(s))
        if sorted_s not in anagram_map:
            anagram_map[sorted_s] = []
        anagram_map[sorted_s].append(s)
    
    return list(anagram_map.values())


# Examples
print(f"Is 'anagram' anagram of 'nagaram'? {is_anagram('anagram', 'nagaram')}")
print(f"Is 'rat' anagram of 'car'? {is_anagram('rat', 'car')}")
print(f"\nGroup anagrams ['eat','tea','tan','ate','nat','bat']:")
print(group_anagrams(['eat', 'tea', 'tan', 'ate', 'nat', 'bat']))
```

```python
def find_anagrams(s, p):
    """Find all start indices of p's anagrams in s (LeetCode 430).
    Time: O(n) using sliding window"""
    result = []
    p_count = Counter(p)
    s_count = Counter()
    
    for i in range(len(s)):
        # Add character to window
        s_count[s[i]] += 1
        
        # Remove character from left if window size exceeds
        if i >= len(p):
            left_char = s[i - len(p)]
            if s_count[left_char] == 1:
                del s_count[left_char]
            else:
                s_count[left_char] -= 1
        
        # Compare counts
        if s_count == p_count:
            result.append(i - len(p) + 1)
    
    return result


# Example
s = "cbaebabacd"
p = "abc"
print(f"String: {s}")
print(f"Pattern: {p}")
print(f"Start indices of anagrams: {find_anagrams(s, p)}")
```

## 3. String Isomorphism

Two strings are isomorphic if characters in one can be replaced to get the other.

```python
def is_isomorphic(s, t):
    """Check if two strings are isomorphic (LeetCode 205).
    Time: O(n)"""
    if len(s) != len(t):
        return False
    
    s_to_t = {}
    t_to_s = {}
    
    for i in range(len(s)):
        if s[i] in s_to_t:
            if s_to_t[s[i]] != t[i]:
                return False
        else:
            if t[i] in t_to_s:
                return False
            s_to_t[s[i]] = t[i]
            t_to_s[t[i]] = s[i]
    
    return True


# Examples
print(f"Is 'egg' isomorphic to 'add'? {is_isomorphic('egg', 'add')}")
print(f"Is 'foo' isomorphic to 'bar'? {is_isomorphic('foo', 'bar')}")
print(f"Is 'paper' isomorphic to 'title'? {is_isomorphic('paper', 'title')}")
```

## 4. Palindrome Problems

```python
def is_palindrome(s):
    """Check if string is palindrome (alphanumeric only)."""
    cleaned = ''.join(c.lower() for c in s if c.isalnum())
    return cleaned == cleaned[::-1]


def longest_palindromic_substring(s):
    """Find longest palindromic substring (LeetCode 5).
    Time: O(n^2) using expand around center"""
    if not s:
        return ""
    
    start = 0
    max_len = 1
    
    def expand_around_center(left, right):
        while left >= 0 and right < len(s) and s[left] == s[right]:
            left -= 1
            right += 1
        return left + 1, right - 1
    
    for i in range(len(s)):
        # Odd length palindromes
        l1, r1 = expand_around_center(i, i)
        if r1 - l1 + 1 > max_len:
            start = l1
            max_len = r1 - l1 + 1
        
        # Even length palindromes
        l2, r2 = expand_around_center(i, i + 1)
        if r2 - l2 + 1 > max_len:
            start = l2
            max_len = r2 - l2 + 1
    
    return s[start:start + max_len]


# Examples
print(f"Is 'A man a plan a canal Panama' palindrome? {is_palindrome('A man a plan a canal Panama')}")
print(f"Longest palindrome in 'babad': {longest_palindromic_substring('babad')}")
print(f"Longest palindrome in 'cbbd': {longest_palindromic_substring('cbbd')}")
```

```python
def longest_palindrome(s):
    """Longest palindromic subsequence (LeetCode 516).
    Time: O(n^2)"""
    n = len(s)
    dp = [[0] * n for _ in range(n)]
    
    # Every single character is a palindrome of length 1
    for i in range(n):
        dp[i][i] = 1
    
    # Fill the table
    for length in range(2, n + 1):
        for i in range(n - length + 1):
            j = i + length - 1
            if s[i] == s[j]:
                dp[i][j] = dp[i + 1][j - 1] + 2
            else:
                dp[i][j] = max(dp[i + 1][j], dp[i][j - 1])
    
    return dp[0][n - 1]


# Example
s = "bbbab"
print(f"Longest palindromic subsequence of '{s}': {longest_palindrome(s)}")
```

## 5. String Manipulation

```python
def reverse_words(s):
    """Reverse words in a string (LeetCode 151)."""
    words = s.split()
    return ' '.join(words[::-1])


def reverse_string(s):
    """Reverse a string in-place using list."""
    s_list = list(s)
    left, right = 0, len(s_list) - 1
    while left < right:
        s_list[left], s_list[right] = s_list[right], s_list[left]
        left += 1
        right -= 1
    return ''.join(s_list)


def count_vowels_consonants(s):
    """Count vowels and consonants in a string."""
    vowels = 'aeiouAEIOU'
    v_count = c_count = 0
    
    for char in s:
        if char.isalpha():
            if char in vowels:
                v_count += 1
            else:
                c_count += 1
    
    return v_count, c_count


# Examples
print(f"Reverse words 'the sky is blue': '{reverse_words('the sky is blue')}'")
print(f"Reverse 'hello': '{reverse_string('hello')}'")
v, c = count_vowels_consonants('Hello World')
print(f"Vowels: {v}, Consonants: {c} in 'Hello World'")
```

```python
def longest_common_prefix(strs):
    """Find longest common prefix among array of strings (LeetCode 14)."""
    if not strs:
        return ""
    
    prefix = strs[0]
    for s in strs[1:]:
        while not s.startswith(prefix):
            prefix = prefix[:-1]
            if not prefix:
                return ""
    return prefix


def longest_substring_without_repeating(s):
    """Longest substring without repeating characters (LeetCode 3).
    Time: O(n)"""
    char_index = {}
    max_length = 0
    start = 0
    
    for end in range(len(s)):
        if s[end] in char_index:
            start = max(start, char_index[s[end]] + 1)
        char_index[s[end]] = end
        max_length = max(max_length, end - start + 1)
    
    return max_length


# Examples
print(f"Longest common prefix ['flower','flow','flight']: '{longest_common_prefix(['flower', 'flow', 'flight'])}'")
print(f"Longest substring without repeating 'abcabcbb': {longest_substring_without_repeating('abcabcbb')}")
print(f"Longest substring without repeating 'bbbbb': {longest_substring_without_repeating('bbbbb')}")
```
